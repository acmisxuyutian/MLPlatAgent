```

import pandas as pd

# Define file paths
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/cirrhosis_patient/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/cirrhosis_patient/test.csv'

# Load datasets
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display basic information about the datasets
print("Training Set Info:")
print(train_df.info())
print("\nTraining Set Description:")
print(train_df.describe())
print("\nTest Set Info:")
print(test_df.info())
print("\nTest Set Description:")
print(test_df.describe())
```

```

# Check for missing values
print("Training Set Missing Values:")
print(train_df.isnull().sum())
print("\nTest Set Missing Values:")
print(test_df.isnull().sum())

# Encode categorical variables
categorical_features = ['Drug', 'Sex', 'Ascites', 'Hepatomegaly', 'Spiders', 'Edema', 'Status']
train_df = pd.get_dummies(train_df, columns=categorical_features, drop_first=True)
test_df = pd.get_dummies(test_df, columns=categorical_features, drop_first=True)

# Display the updated datasets
print("\nTraining Set Info after encoding:")
print(train_df.info())
print("\nTest Set Info after encoding:")
print(test_df.info())

```

```


from sklearn.preprocessing import StandardScaler

# Define numerical features
numerical_features = ['Bilirubin', 'Cholesterol', 'Albumin', 'Copper', 'Alk_Phos', 'SGOT', 'Tryglicerides', 'Platelets', 'Prothrombin', 'Stage', 'N_Days', 'Age']

# Initialize the scaler
scaler = StandardScaler()

# Fit and transform the training dataset
train_df[numerical_features] = scaler.fit_transform(train_df[numerical_features])

# Transform the test dataset
test_df[numerical_features] = scaler.transform(test_df[numerical_features])

# Display the training and test datasets after normalization
print("Training Set after normalization:")
print(train_df.head())
print("\nTest Set after normalization:")
print(test_df.head())


```

```


from sklearn.model_selection import train_test_split

# Define the features and target variable
features = train_df.drop(columns=['id', 'Status_CL', 'Status_D'])
target = train_df['Status_D']  # Assuming 'Status_D' is the binary target variable

# Split the dataset into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(features, target, test_size=0.2, random_state=42)

# Display the shape of the split datasets
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_val shape:", X_val.shape)
print("y_val shape:", y_val.shape)


```

```



from sklearn.linear_model import LogisticRegression
from sklearn.metrics import log_loss

# Initialize the logistic regression model
log_reg_model = LogisticRegression(max_iter=1000)

# Train the model
log_reg_model.fit(X_train, y_train)

# Evaluate the model on the validation set
y_val_pred_proba = log_reg_model.predict_proba(X_val)[:, 1]
validation_logloss = log_loss(y_val, y_val_pred_proba)

# Display the validation logloss
print("Validation Logloss:", validation_logloss)


```

```



# Prepare the test set
X_test = test_df.drop(columns=['id', 'Status_CL', 'Status_D'])

# Predict probabilities on the test set
y_test_pred_proba = log_reg_model.predict_proba(X_test)[:, 1]

# There is no 'Status_D' in the test set, so we cannot compute logloss directly.
# Instead, we can save the predictions for further analysis or submission if needed.
test_predictions = pd.DataFrame({
    'id': test_df['id'],
    'Status_D_prob': y_test_pred_proba
})

print("Test predictions saved to test_predictions.csv")
test_predictions.to_csv('test_predictions.csv', index=False)

# If you had the actual Status_D values, you could compute logloss as follows:
# from sklearn.metrics import log_loss
# y_true = test_df['Status_D']
# test_logloss = log_loss(y_true, y_test_pred_proba)
# print("Test Logloss:", test_logloss)

```
