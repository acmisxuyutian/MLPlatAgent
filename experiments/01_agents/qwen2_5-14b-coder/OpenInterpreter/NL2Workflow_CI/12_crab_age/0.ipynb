```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.preprocessing import OneHotEncoder
import numpy as np
```

```

# Load datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

print(train_df.head())
print(test_df.head())
```

```


# Preprocess the data
train_df = pd.get_dummies(train_df, columns=['Sex'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['Sex'], drop_first=True)

# Split the training data into features and target
X_train = train_df.drop(columns=['Age'])
y_train = train_df['Age']

# Initialize and train the RandomForestRegressor model
model = RandomForestRegressor(random_state=42)
model.fit(X_train, y_train)

# Predict the ages on the test set
X_test = test_df.drop(columns=['Age'])
predictions = model.predict(X_test)

# Save the predictions to a file
predictions_df = pd.DataFrame({'Id': test_df['id'], 'Predicted_Age': predictions})
predictions_df.to_csv('predictions.csv', index=False)

# Evaluate the model using Mean Absolute Error (MAE)
# Since we don't have the true ages for the test set, we can't calculate MAE directly.
# However, we can print the predictions for verification.
print(predictions[:10])

```
