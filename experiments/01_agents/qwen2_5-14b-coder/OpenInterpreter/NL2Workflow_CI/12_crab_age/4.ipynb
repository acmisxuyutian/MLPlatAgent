```

import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import mean_absolute_error
```

```


train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

print(train_data.head())
print(test_data.head())

```

```



# One-hot encode the 'Sex' column
encoder = OneHotEncoder(sparse=False, drop='first')
sex_encoded = encoder.fit_transform(train_data[['Sex']])

# Add the encoded columns to the training data
for i, category in enumerate(encoder.get_feature_names_out(['Sex'])):
    train_data[category] = sex_encoded[:, i]

# Repeat the same process for the test set
sex_encoded_test = encoder.transform(test_data[['Sex']])
for i, category in enumerate(encoder.get_feature_names_out(['Sex'])):
    test_data[category] = sex_encoded_test[:, i]

# Drop the original 'Sex' column
train_data.drop(columns=['Sex'], inplace=True)
test_data.drop(columns=['Sex'], inplace=True)

# Split the data into features and target variable
X_train = train_data.drop(columns=['Age'])
y_train = train_data['Age']
X_test = test_data.drop(columns=['Age'])

print(X_train.head())
print(y_train.head())
print(X_test.head())




```

```




# One-hot encode the 'Sex' column manually
sex_encoded = pd.get_dummies(train_data['Sex'], drop_first=True)
train_data = pd.concat([train_data, sex_encoded], axis=1).drop(columns=['Sex'])

# Repeat the same process for the test set
sex_encoded_test = pd.get_dummies(test_data['Sex'], drop_first=True)
test_data = pd.concat([test_data, sex_encoded_test], axis=1).drop(columns=['Sex'])

# Split the data into features and target variable
X_train = train_data.drop(columns=['Age'])
y_train = train_data['Age']
X_test = test_data.drop(columns=['Age'])

print(X_train.head())
print(y_train.head())
print(X_test.head())

```

```





# Initialize the Random Forest Regressor
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)

# Train the model
rf_model.fit(X_train, y_train)
print("Model training complete.")


```

```


# Make predictions on the test set
y_pred = rf_model.predict(X_test)

# Calculate the Mean Absolute Error
mae = mean_absolute_error(y_test, y_pred)
print(f"Mean Absolute Error: {mae}")

# Save the predictions to a CSV file
predictions_df = pd.DataFrame({'id': test_data['id'], 'Predicted_Age': y_pred})
predictions_df.to_csv('predicted_ages.csv', index=False)
print("Predictions saved to 'predicted_ages.csv'.")




```
