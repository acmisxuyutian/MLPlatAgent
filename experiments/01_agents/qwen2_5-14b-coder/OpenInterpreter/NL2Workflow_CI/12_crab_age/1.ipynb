```

# Step 1: 读取训练集和测试集数据
import pandas as pd

train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# 打印数据集的前几行以确认读取正确
print(train_df.head())
print(test_df.head())
```

```


# Step 2: Preprocess the data by performing one-hot encoding on the 'Sex' column
train_encoded = pd.get_dummies(train_df, columns=['Sex'])
test_encoded = pd.get_dummies(test_df, columns=['Sex'])

# Split the training data into features and target variable
X_train = train_encoded.drop(columns=['id', 'Age'])
y_train = train_encoded['Age']

# Display the first few rows of the encoded training data
print(X_train.head())
print(y_train.head())

```

```



# Step 3: Train the random forest model
from sklearn.ensemble import RandomForestRegressor

# Initialize the model
rf_model = RandomForestRegressor(random_state=42)

# Fit the model to the training data
rf_model.fit(X_train, y_train)

# Print a message to indicate the model has been trained
print("Model trained successfully!")


```

```




# Step 4: Preprocess the test data by performing the same one-hot encoding on the 'Sex' column
# Ensure the columns in test_encoded are the same as in X_train
missing_cols = set(X_train.columns) - set(test_encoded.columns)
for c in missing_cols:
    test_encoded[c] = 0

test_encoded = test_encoded[X_train.columns]

# Split the test data into features
X_test = test_encoded.drop(columns=['id', 'Age'])

# Use the trained model to make predictions on the test set
predictions = rf_model.predict(X_test)

# Print the first few predictions to verify
print(predictions[:5])




```
