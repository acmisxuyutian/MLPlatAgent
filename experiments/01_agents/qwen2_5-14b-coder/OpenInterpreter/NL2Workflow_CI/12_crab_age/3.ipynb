```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import mean_absolute_error

# Load datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

# Check for missing values
print(train_data.isnull().sum())
print(test_data.isnull().sum())

# One-hot encode 'Sex' column
encoder = OneHotEncoder(sparse_output=False, drop='first')
train_encoded = encoder.fit_transform(train_data[['Sex']])
test_encoded = encoder.transform(test_data[['Sex']])

# Get the feature names
feature_names = encoder.get_feature_names_out(['Sex'])

# Convert encoded data to DataFrame
train_encoded_df = pd.DataFrame(train_encoded, columns=feature_names)
test_encoded_df = pd.DataFrame(test_encoded, columns=feature_names)

# Concatenate with the original data
train_data = pd.concat([train_data, train_encoded_df], axis=1).drop(columns=['Sex'])
test_data = pd.concat([test_data, test_encoded_df], axis=1).drop(columns=['Sex'])

# Display the first few rows of the processed training data
train_data.head()
```

```


# Prepare features and target
X_train = train_data.drop(columns=['id', 'Age'])
y_train = train_data['Age']

X_test = test_data.drop(columns=['id'])

# Display the first few rows of X_train and X_test
print('X_train first few rows:')
print(X_train.head())
print('\nX_test first few rows:')
print(X_test.head())

```

```



# Train Random Forest Regressor
rf_model = RandomForestRegressor(random_state=42)
rf_model.fit(X_train, y_train)

```

```



# Make predictions on the test data
y_pred = rf_model.predict(X_test)

# Calculate Mean Absolute Error (MAE)
mae = mean_absolute_error(y_test, y_pred)

# Print the MAE
print(f'Mean Absolute Error: {mae}')

```

```



# Drop the 'Age' column from the test data
X_test = X_test.drop(columns=['Age'])

# Make predictions on the test data
y_pred = rf_model.predict(X_test)

# Display the first few predictions
print('Predictions first few rows:')
print(y_pred[:5])


```

```



# Save predictions to a CSV file
output_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/predictions.csv'
output_df = pd.DataFrame({'id': test_data['id'], 'Age': y_pred})
output_df.to_csv(output_path, index=False)

print(f'Predictions saved to {output_path}')


```
