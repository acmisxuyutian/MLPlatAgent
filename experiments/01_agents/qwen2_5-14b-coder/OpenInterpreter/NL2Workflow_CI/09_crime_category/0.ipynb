```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import log_loss
```

```

train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)
```

```


# Label encode the target variable
label_encoder = LabelEncoder()
train_df['Category'] = label_encoder.fit_transform(train_df['Category'])

# Separate features and target
X_train = train_df.drop('Category', axis=1)
y_train = train_df['Category']

# Split the training data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# Preprocess the test data
test_df['Category'] = label_encoder.transform(test_df['Category'])
X_test = test_df.drop('Category', axis=1)
y_test = test_df['Category']

```

```



# Initialize the model
model = RandomForestClassifier(n_estimators=100, random_state=42)

# Train the model
model.fit(X_train, y_train)


```

```


# Identify categorical columns
categorical_cols = train_df.select_dtypes(include=['object', 'category']).columns

# Encode categorical columns
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder()

# Apply encoding to training and test data
train_df[categorical_cols] = encoder.fit_transform(train_df[categorical_cols])
test_df[categorical_cols] = encoder.transform(test_df[categorical_cols])

# Repeat the preprocessing steps
train_df['Category'] = label_encoder.fit_transform(train_df['Category'])

# Separate features and target
X_train = train_df.drop('Category', axis=1)
y_train = train_df['Category']

# Split the training data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# Preprocess the test data
test_df['Category'] = label_encoder.transform(test_df['Category'])
X_test = test_df.drop('Category', axis=1)
y_test = test_df['Category']



```

```



# Initialize the model
model = RandomForestClassifier(n_estimators=100, random_state=42)

# Train the model
model.fit(X_train, y_train)


```

```



# Make predictions on the validation set
y_val_pred = model.predict_proba(X_val)

# Calculate logloss on the validation set
val_logloss = log_loss(y_val, y_val_pred)
print(f'Validation Logloss: {val_logloss}')

# Make predictions on the test set
y_test_pred = model.predict_proba(X_test)

# Save the predictions to a CSV file
predictions_df = pd.DataFrame(y_test_pred, columns=label_encoder.classes_)
predictions_df.to_csv('crime_category_predictions.csv', index=False)

# Print the first few rows of the predictions
print(predictions_df.head())

```
