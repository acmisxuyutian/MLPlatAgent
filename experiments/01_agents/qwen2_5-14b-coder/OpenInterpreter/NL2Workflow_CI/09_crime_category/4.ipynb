```

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import log_loss
```

```

train_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/train.csv')
test_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/test.csv')

print("Training Data Shape:", train_data.shape)
print("Testing Data Shape:", test_data.shape)
```

```


# Encode the target variable in the training data
label_encoder = LabelEncoder()
train_data['Category'] = label_encoder.fit_transform(train_data['Category'])

# Identify categorical and numerical columns
categorical_cols = train_data.select_dtypes(include=['object']).columns.tolist()
numerical_cols = train_data.select_dtypes(include=['int64', 'float64']).drop(columns=['Category']).columns.tolist()

# One-hot encode categorical columns
train_encoded = pd.get_dummies(train_data, columns=categorical_cols, drop_first=True)

# Split the training data into features and target variable
X_train = train_encoded.drop(columns=['Category'])
y_train = train_encoded['Category']

# Encode the categorical columns in the test data using the same encoder
for col in categorical_cols:
    test_data[col] = label_encoder.transform(test_data[col])
test_encoded = pd.get_dummies(test_data, columns=categorical_cols, drop_first=True)

# Ensure that the test data has the same columns as the training data
X_test = test_encoded.reindex(columns=X_train.columns, fill_value=0)

# Standardize the numerical features
scaler = StandardScaler()
X_train[numerical_cols] = scaler.fit_transform(X_train[numerical_cols])
X_test[numerical_cols] = scaler.transform(X_test[numerical_cols])

```
