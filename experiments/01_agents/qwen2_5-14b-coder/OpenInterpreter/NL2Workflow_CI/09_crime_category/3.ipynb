```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import log_loss
from sklearn.ensemble import RandomForestClassifier

# 加载训练数据
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/train.csv'
train_df = pd.read_csv(train_data_path)

# 加载测试数据
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/test.csv'
test_df = pd.read_csv(test_data_path)

# 显示训练数据的前5行
print(train_df.head())

# 显示测试数据的前5行
print(test_df.head())
```

```


# 将分类变量进行编码
le = LabelEncoder()

train_df['Category'] = le.fit_transform(train_df['Category'])
train_df['DayOfWeek'] = le.fit_transform(train_df['DayOfWeek'])
train_df['PdDistrict'] = le.fit_transform(train_df['PdDistrict'])

test_df['DayOfWeek'] = le.fit_transform(test_df['DayOfWeek'])
test_df['PdDistrict'] = le.fit_transform(test_df['PdDistrict'])

# 分离特征和标签
X_train = train_df.drop('Category', axis=1)
y_train = train_df['Category']

X_test = test_df.drop('Category', axis=1)

# 显示前5行
print(X_train.head())
print(X_test.head())

```

```



# 训练模型
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 用测试数据进行预测
y_pred_prob = model.predict_proba(X_test)

# 显示预测结果的前5行
print(y_pred_prob[:5])


```

```


# 计算logloss
y_test = test_df['Category'].values
logloss_value = log_loss(y_test, y_pred_prob)
print(f'Logloss: {logloss_value}')

```
