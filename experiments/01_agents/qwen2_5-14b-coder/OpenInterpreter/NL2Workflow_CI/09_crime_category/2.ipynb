```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import log_loss

# 1. 加载数据
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

print("训练数据加载完成，前5行：")
print(train_data.head())

print("测试数据加载完成，前5行：")
print(test_data.head())

# 2. 数据预处理
# 假设最后一列是标签，其余是特征
X_train = train_data.iloc[:, :-1]
y_train = train_data.iloc[:, -1]

# 编码分类变量
label_encoder = LabelEncoder()
y_train = label_encoder.fit_transform(y_train)

# 3. 训练模型
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

print("模型训练完成。")

# 4. 进行预测
# 假设测试数据的最后一列是特征
X_test = test_data.iloc[:, :-1]

# 预测概率
y_pred_proba = model.predict_proba(X_test)

print("预测完成。")

# 5. 保存预测结果
output_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/predictions.csv'
pd.DataFrame(y_pred_proba, columns=label_encoder.classes_).to_csv(output_path, index=False)

print(f"预测结果已保存到 {output_path}")

# 6. 评估模型
# 计算logloss
y_true = label_encoder.transform(test_data.iloc[:, -1])  # 测试集的真实标签
logloss_value = log_loss(y_true, y_pred_proba)

print(f"模型在测试集上的logloss是: {logloss_value}")
```
