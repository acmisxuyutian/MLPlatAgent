```

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score
import xgboost as xgb
```

```


# 读取训练集和测试集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/smoker_status_Bio_Signals/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/smoker_status_Bio_Signals/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 查看训练集和测试集的前几行
print(train_df.head())
print(test_df.head())
```

```



# 检查缺失值
print(train_df.isnull().sum())
print(test_df.isnull().sum())

# 填充或删除缺失值（根据实际情况决定）
# 例如，删除有缺失值的行
train_df = train_df.dropna()
test_df = test_df.dropna()

# 分离特征和目标变量
X_train = train_df.drop(columns=['id', 'smoking'])
y_train = train_df['smoking']
X_test = test_df.drop(columns=['id', 'smoking'])
y_test = test_df['smoking']

# 特征标准化
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

```

```


# 配置XGBoost模型
params = {
    'objective': 'binary:logistic',
    'eval_metric': 'auc',
    'learning_rate': 0.1,
    'booster': 'dart',
    'max_depth': 6,
    'subsample': 0.8,
    'colsample_bytree': 0.8
}

# 将数据转换为DMatrix
dtrain = xgb.DMatrix(X_train, label=y_train)
dtest = xgb.DMatrix(X_test, label=y_test)

# 训练模型
num_rounds = 100
model = xgb.train(params, dtrain, num_rounds)

# 预测
y_pred_proba = model.predict(dtest)

# 计算AUC-ROC
auc_roc = roc_auc_score(y_test, y_pred_proba)
print(f"AUC-ROC: {auc_roc:.4f}")

```
