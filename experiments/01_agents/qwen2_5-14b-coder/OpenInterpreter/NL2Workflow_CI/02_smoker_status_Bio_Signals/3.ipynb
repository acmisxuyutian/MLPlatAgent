```

import pandas as pd

# 加载训练集
train_csv_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/smoker_status_Bio_Signals/train.csv'
train_data = pd.read_csv(train_csv_path)

# 加载测试集
test_csv_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/smoker_status_Bio_Signals/test.csv'
test_data = pd.read_csv(test_csv_path)

# 查看数据结构
print(train_data.head())
print(test_data.head())
```

```


import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# 检查缺失值
print(train_data.isnull().sum())
print(test_data.isnull().sum())

# 如果有缺失值，可以选择填充或删除
# 这里我们选择填充缺失值
train_data = train_data.fillna(train_data.mean())
test_data = test_data.fillna(test_data.mean())

# 分离特征和目标变量
X_train = train_data.drop(columns='smoking')
y_train = train_data['smoking']
X_test = test_data.drop(columns='smoking')
y_test = test_data['smoking']

# 特征标准化
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# 查看前5行数据
print(X_train[:5])
print(y_train[:5])

```

```


import xgboost as xgb
from sklearn.metrics import roc_auc_score

# 设置XGBoost参数
params = {
    'objective': 'binary:logistic',
    'booster': 'dart',
    'learning_rate': 0.1,
    'eval_metric': 'auc'
}

# 创建DMatrix
dtrain = xgb.DMatrix(X_train, label=y_train)

# 训练模型
model = xgb.train(params, dtrain, num_boost_round=100)

# 在训练集上评估模型
y_train_pred_proba = model.predict(dtrain)
train_auc = roc_auc_score(y_train, y_train_pred_proba)
print(f'Training AUC-ROC: {train_auc:.4f}')


```

```



# 创建DMatrix for test data
dtest = xgb.DMatrix(X_test, label=y_test)

# 在测试集上评估模型
y_test_pred_proba = model.predict(dtest)
test_auc = roc_auc_score(y_test, y_test_pred_proba)
print(f'Testing AUC-ROC: {test_auc:.4f}')

# 输出模型的详细信息
print(model)

```
