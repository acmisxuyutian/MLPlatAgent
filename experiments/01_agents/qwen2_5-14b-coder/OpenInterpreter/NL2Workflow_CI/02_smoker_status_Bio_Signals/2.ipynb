```

import pandas as pd

# 加载训练集和测试集
train_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/smoker_status_Bio_Signals/train.csv')
test_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/smoker_status_Bio_Signals/test.csv')
```

```

# 检查缺失值
print("Missing values in train data:")
print(train_data.isnull().sum())

print("Missing values in test data:")
print(test_data.isnull().sum())

# 填充缺失值（这里我们简单地用每列的中位数填充）
train_data.fillna(train_data.median(), inplace=True)
test_data.fillna(train_data.median(), inplace=True)

# 分离特征和目标变量
X_train = train_data.drop(columns=['smoker'])  # 假设'smoker'是目标变量
y_train = train_data['smoker']

X_test = test_data.drop(columns=['smoker'])  # 假设'smoker'是目标变量
y_test = test_data['smoker']
```

```

# 查看训练集的列名
print(train_data.columns.tolist())
```

```


# 分离特征和目标变量
X_train = train_data.drop(columns=['smoking'])  # 'smoking' is the target variable, so we drop it
y_train = train_data['smoking']

X_test = test_data.drop(columns=['smoking'])  # 'smoking' is the target variable, so we drop it
y_test = test_data['smoking']

```

```


import xgboost as xgb
from sklearn.metrics import roc_auc_score

# 创建DART模型
dtrain = xgb.DMatrix(X_train, label=y_train)
params = {
    'objective': 'binary:logistic',
    'eval_metric': 'auc',
    'eta': 0.1,
    'booster': 'dart'
}

# 训练模型
model = xgb.train(params, dtrain)

```

```



# 创建测试集的DMatrix
dtest = xgb.DMatrix(X_test)

# 进行预测
y_pred_proba = model.predict(dtest)

# 计算AUC-ROC
auc_roc = roc_auc_score(y_test, y_pred_proba)

# 输出AUC-ROC评分
print(f'AUC-ROC Score: {auc_roc}')


```
