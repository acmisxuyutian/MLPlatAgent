```

import pandas as pd

# 读取训练集和测试集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/obesity_CVD_risk/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/obesity_CVD_risk/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

print(train_data.head())
print(test_data.head())
```

```

# 检查并处理缺失值
print(train_data.isnull().sum())
print(test_data.isnull().sum())

# 编码分类变量
# 使用 one-hot 编码对分类变量进行编码
train_data_encoded = pd.get_dummies(train_data, drop_first=True)
test_data_encoded = pd.get_dummies(test_data, drop_first=True)

print(train_data_encoded.head())
print(test_data_encoded.head())

# 分离特征和目标变量
X_train = train_data_encoded.drop('NObeyesdad', axis=1)
y_train = train_data_encoded['NObeyesdad']
X_test = test_data_encoded.drop('NObeyesdad', axis=1)
y_test = test_data_encoded['NObeyesdad']

print(X_train.head())
print(y_train.head())
print(X_test.head())
print(y_test.head())
```

```

# 重新编码分类变量
train_data_encoded = pd.get_dummies(train_data, drop_first=True, columns=['Gender', 'MTRANS', 'CAEC', 'SMOKE', 'SCC', 'CALC'])
test_data_encoded = pd.get_dummies(test_data, drop_first=True, columns=['Gender', 'MTRANS', 'CAEC', 'SMOKE', 'SCC', 'CALC'])

# 检查编码后的列
print(train_data_encoded.columns)
print(test_data_encoded.columns)

# 分离特征和目标变量
X_train = train_data_encoded.drop('NObeyesdad', axis=1)
y_train = train_data_encoded['NObeyesdad']
X_test = test_data_encoded.drop('NObeyesdad', axis=1)
y_test = test_data_encoded['NObeyesdad']

print(X_train.head())
print(y_train.head())
print(X_test.head())
print(y_test.head())
```

```


from sklearn.feature_selection import SelectKBest, chi2
from sklearn.preprocessing import LabelEncoder

# 编码目标变量
le = LabelEncoder()
y_train_encoded = le.fit_transform(y_train)
y_test_encoded = le.transform(y_test)

# 特征选择
selector = SelectKBest(score_func=chi2, k=10)
X_train_selected = selector.fit_transform(X_train, y_train_encoded)
X_test_selected = selector.transform(X_test)

# 获取选择的特征名
selected_features = X_train.columns[selector.get_support()]

print("Selected Features:", selected_features)

```

```


# 检查数据类型
print(train_data_encoded.dtypes)

# 检查非数值列
non_numeric_cols = train_data_encoded.select_dtypes(exclude=['int64', 'float64']).columns
print(non_numeric_cols)

# 修复非数值列
for col in non_numeric_cols:
    train_data_encoded[col] = train_data_encoded[col].astype('category').cat.codes
    test_data_encoded[col] = test_data_encoded[col].astype('category').cat.codes

print(train_data_encoded.head())
print(test_data_encoded.head())

# 重新分离特征和目标变量
X_train = train_data_encoded.drop('NObeyesdad', axis=1)
y_train = train_data_encoded['NObeyesdad']
X_test = test_data_encoded.drop('NObeyesdad', axis=1)
y_test = test_data_encoded['NObeyesdad']

print(X_train.head())
print(y_train.head())
print(X_test.head())
print(y_test.head())

```

```



from sklearn.feature_selection import SelectKBest, chi2
from sklearn.preprocessing import LabelEncoder

# 编码目标变量
le = LabelEncoder()
y_train_encoded = le.fit_transform(y_train)
y_test_encoded = le.transform(y_test)

# 特征选择
selector = SelectKBest(score_func=chi2, k=10)
X_train_selected = selector.fit_transform(X_train, y_train_encoded)
X_test_selected = selector.transform(X_test)

# 获取选择的特征名
selected_features = X_train.columns[selector.get_support()]

print("Selected Features:", selected_features)


```

```



from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# 初始化模型
model = RandomForestClassifier(n_estimators=100, random_state=42)

# 训练模型
model.fit(X_train_selected, y_train_encoded)

# 预测
y_pred = model.predict(X_test_selected)

# 计算准确率
accuracy = accuracy_score(y_test_encoded, y_pred)
print(f"Model Accuracy: {accuracy * 100:.2f}%")


```
