```

import pandas as pd

# 读取训练集
train_df = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/obesity_CVD_risk/train.csv')

# 读取测试集
test_df = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/obesity_CVD_risk/test.csv')

# 打印数据的前几行
print(train_df.head())
print(test_df.head())
```

```

# 检查训练集和测试集的缺失值
print("训练集缺失值情况:")
print(train_df.isnull().sum())

print("\n测试集缺失值情况:")
print(test_df.isnull().sum())

# 检查训练集和测试集的重复值
print("\n训练集重复值情况:")
print(train_df.duplicated().sum())

print("\n测试集重复值情况:")
print(test_df.duplicated().sum())

# 查看数据的基本统计信息
print("\n训练集基本统计信息:")
print(train_df.describe(include='all'))

print("\n测试集基本统计信息:")
print(test_df.describe(include='all'))
```

```


# 特征构建
from sklearn.preprocessing import LabelEncoder

# 创建年龄的平方
train_df['Age_Squared'] = train_df['Age'] ** 2
test_df['Age_Squared'] = test_df['Age'] ** 2

# 创建身高的平方
train_df['Height_Squared'] = train_df['Height'] ** 2
test_df['Height_Squared'] = test_df['Height'] ** 2

# 创建体重的平方
train_df['Weight_Squared'] = train_df['Weight'] ** 2
test_df['Weight_Squared'] = test_df['Weight'] ** 2

# 创建BMI（体重指数）
train_df['BMI'] = train_df['Weight'] / (train_df['Height'] ** 2)
test_df['BMI'] = test_df['Weight'] / (test_df['Height'] ** 2)

# 对分类特征进行编码
label_encoders = {}
for column in train_df.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    train_df[column] = le.fit_transform(train_df[column])
    test_df[column] = le.transform(test_df[column])
    label_encoders[column] = le

# 打印处理后的数据
print(train_df.head())
print(test_df.head())

```

```

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 目标变量和特征矩阵
X_train = train_df.drop(['id', 'NObeyesdad'], axis=1)
y_train = train_df['NObeyesdad']

# 计算相关性矩阵
correlation_matrix = X_train.corrwith(y_train).sort_values(ascending=False)

# 绘制相关性矩阵图
plt.figure(figsize=(10, 6))
sns.barplot(x=correlation_matrix.values, y=correlation_matrix.index)
plt.title('Feature Correlation with Obesity Risk')
plt.show()

# 打印相关性矩阵
print(correlation_matrix)

```
