```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score
import joblib

# 读取训练数据集和测试数据集
train_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/train.csv')
test_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/test.csv')

# 检查数据集的前几行
print("训练数据集的前几行：")
print(train_data.head())
print("\n测试数据集的前几行：")
print(test_data.head())

# 分离特征和目标变量
X_train = train_data.drop('健康状况', axis=1)
y_train = train_data['健康状况']
X_test = test_data.drop('健康状况', axis=1)
y_test = test_data['健康状况']

# 输出训练集和测试集的形状
print("训练集的特征和目标变量形状：", X_train.shape, y_train.shape)
print("测试集的特征和目标变量形状：", X_test.shape, y_test.shape)

# 训练随机森林分类器
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# 预测测试集
y_pred = model.predict(X_test)

# 计算测试集上的 F1 分数
f1 = f1_score(y_test, y_pred)
print(f"F1 Score on test set: {f1:.4f}")

# 保存模型
joblib.dump(model, 'horse_health_model.pkl')
print("模型已保存为 horse_health_model.pkl")
```

```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score
import joblib

# 读取训练数据集和测试数据集
train_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/train.csv')
test_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/test.csv')

# 检查数据集的前几行
print("训练数据集的前几行：")
print(train_data.head())
print("\n测试数据集的前几行：")
print(test_data.head())

# 分离特征和目标变量
X_train = train_data.drop('outcome', axis=1)
y_train = train_data['outcome']
X_test = test_data.drop('outcome', axis=1)
y_test = test_data['outcome']

# 输出训练集和测试集的形状
print("训练集的特征和目标变量形状：", X_train.shape, y_train.shape)
print("测试集的特征和目标变量形状：", X_test.shape, y_test.shape)

# 训练随机森林分类器
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# 预测测试集
y_pred = model.predict(X_test)

# 计算测试集上的 F1 分数
f1 = f1_score(y_test, y_pred, average='weighted')
print(f"F1 Score on test set: {f1:.4f}")

# 保存模型
joblib.dump(model, 'horse_health_model.pkl')
print("模型已保存为 horse_health_model.pkl")

```

```


import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score
import joblib
from sklearn.preprocessing import LabelEncoder

# 读取训练数据集和测试数据集
train_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/train.csv')
test_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/test.csv')

# 检查数据集的前几行
print("训练数据集的前几行：")
print(train_data.head())
print("\n测试数据集的前几行：")
print(test_data.head())

# 分离特征和目标变量
X_train = train_data.drop('outcome', axis=1)
y_train = train_data['outcome']
X_test = test_data.drop('outcome', axis=1)
y_test = test_data['outcome']

# 使用LabelEncoder对 categorical columns进行编码
categorical_columns = ['surgery', 'peripheral_pulse', 'mucous_membrane', 'capillary_refill_time']
le = LabelEncoder()

for col in categorical_columns:
    X_train[col] = le.fit_transform(X_train[col])
    X_test[col] = le.transform(X_test[col])

# 输出训练集和测试集的形状
print("训练集的特征和目标变量形状：", X_train.shape, y_train.shape)
print("测试集的特征和目标变量形状：", X_test.shape, y_test.shape)

# 训练随机森林分类器
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# 预测测试集
y_pred = model.predict(X_test)

# 计算测试集上的 F1 分数
f1 = f1_score(y_test, y_pred, average='weighted')
print(f"F1 Score on test set: {f1:.4f}")

# 保存模型
joblib.dump(model, 'horse_health_model.pkl')
print("模型已保存为 horse_health_model.pkl")

```
