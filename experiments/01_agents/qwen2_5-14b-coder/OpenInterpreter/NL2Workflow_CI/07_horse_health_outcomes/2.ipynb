```

import pandas as pd

# 读取训练集和测试集
train_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/train.csv')
test_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/test.csv')

# 查看数据结构
print(train_data.head())
print(test_data.head())
```

```

# Preprocess the data
train_data = train_data.dropna()  # Removing rows with missing values
test_data = test_data.dropna()

# Define features and target
X_train = train_data.drop(columns=['outcome'])
y_train = train_data['outcome']

X_test = test_data.drop(columns=['outcome'])
y_test = test_data['outcome']

# Convert categorical variables to numerical
X_train = pd.get_dummies(X_train, drop_first=True)
X_test = pd.get_dummies(X_test, drop_first=True)

# Display the processed data
print(X_train.head())
print(y_train.head())
```

```


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score

# Initialize the Random Forest Classifier
rf_model = RandomForestClassifier()

# Train the model
rf_model.fit(X_train, y_train)

# Predict on the test data
y_pred = rf_model.predict(X_test)

# Calculate and print the F1 score
f1 = f1_score(y_test, y_pred, average='weighted')
print(f'F1 Score: {f1}')

# Save the predictions to a CSV file
predictions_df = pd.DataFrame({'hospital_number': test_data['hospital_number'], 'prediction': y_pred})
predictions_df.to_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/predictions.csv', index=False)

print('Predictions saved to predictions.csv')

```
