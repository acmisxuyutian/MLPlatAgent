```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score
from sklearn.preprocessing import LabelEncoder

# 读取数据
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

# 查看数据的基本信息
print(train_data.head())
print(test_data.head())

# 预处理数据
# 处理缺失值
train_data.fillna(method='ffill', inplace=True)
test_data.fillna(method='ffill', inplace=True)

# 编码分类变量
label_encoders = {}
for column in train_data.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    train_data[column] = le.fit_transform(train_data[column])
    label_encoders[column] = le

for column in test_data.select_dtypes(include=['object']).columns:
    le = label_encoders.get(column, LabelEncoder())
    test_data[column] = le.transform(test_data[column])

# 分离特征和目标变量
X_train = train_data.drop(columns=['label'])
y_train = train_data['label']
X_test = test_data.drop(columns=['label'])

print(X_train.head())
print(X_test.head())

# 训练模型
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 进行预测
y_pred = model.predict(X_test)

# 保存预测结果
output_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/predictions.csv'
result = pd.DataFrame({'predicted_label': y_pred})
result.to_csv(output_path, index=False)

print(f"预测结果已保存到 {output_path}")

# 计算F1分数
f1 = f1_score(y_test, y_pred, average='weighted')
print(f"F1分数: {f1}")
```
