```

# 第1步：导入必要的库
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, classification_report

# 第2步：加载数据集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

print(train_data.head())
print(test_data.head())
```

```

# 第3步：数据预处理

# 检查缺失值
print(train_data.isnull().sum())
print(test_data.isnull().sum())

# 编码分类变量
label_encoders = {}

for column in train_data.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    train_data[column] = le.fit_transform(train_data[column])
    test_data[column] = le.transform(test_data[column])
    label_encoders[column] = le

# 规范化数值特征
scaler = StandardScaler()
train_data[train_data.select_dtypes(include=['float64', 'int64']).columns] = scaler.fit_transform(train_data.select_dtypes(include=['float64', 'int64']).columns)
test_data[test_data.select_dtypes(include=['float64', 'int64']).columns] = scaler.transform(test_data.select_dtypes(include=['float64', 'int64']).columns)

print(train_data.head())
print(test_data.head())
```

```

import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, classification_report

# Load the datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/horse_horse_outcomes_train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/horse_horse_outcomes_test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

# Step 3: Data Preprocessing

# Check for missing values
print(train_data.isnull().sum())
print(test_data.isnull().sum())

# Encode categorical variables
label_encoders = {}

for column in train_data.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    train_data[column] = le.fit_transform(train_data[column])
    test_data[column] = le.transform(test_data[column])
    label_encoders[column] = le

# Ensure 'outcome' is treated as categorical
# Train data
train_data['outcome'] = train_data['outcome'].astype('category')
# Test data alignment using label encoder
test_data['outcome'] = train_data['outcome'].cat.add_categories([-1]).fillna(-1).astype('category')

# Check the unique values
print(f"Training data 'outcome' unique values: {train_data['outcome'].cat.categories}")
print(f"Test data 'outcome' unique values: {test_data['outcome'].cat.categories}")

# Map -1 to 0 in the test set if necessary
# This ensures that the test set has the same categories as the train set
test_data['outcome'] = test_data['outcome'].replace(-1, 0)

# Separate features and target
X_train = train_data.drop('outcome', axis=1)
y_train = train_data['outcome']

X_test = test_data.drop('outcome', axis=1)
y_test = test_data['outcome']

# Normalize numerical feature
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Step 4: Train the model
clf = RandomForestClassifier(n_estimators=100, random_state=42)
clf.fit(X_train_scaled, y_train)

# Step 5: Predict the test set
y_pred = clf.predict(X_test_scaled)

# Step 6: Save the predictions
test_data['prediction'] = y_pred
predictions_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/predictions.csv'
test_data.to_csv(predictions_path, index=False)

# Step 7: Evaluate the model
f1 = f1_score(y_test, y_pred, average='weighted')
print(f"F1 Score: {f1}")
print(classification_report(y_test, y_pred))
```
