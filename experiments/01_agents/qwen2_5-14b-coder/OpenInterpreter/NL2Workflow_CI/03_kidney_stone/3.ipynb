```

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score
```

```

# Define the file paths
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/kidney_stone/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/kidney_stone/test.csv'

# Load the datasets
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and test datasets
print(train_df.head())
print(test_df.head())
```

```


# Check for missing values
print(train_df.isnull().sum())
print(test_df.isnull().sum())

# Fill missing values with the median
train_df.fillna(train_df.median(), inplace=True)
test_df.fillna(train_df.median(), inplace=True)

# Encode the target variable
encoder = LabelEncoder()
train_df['target'] = encoder.fit_transform(train_df['target'])
test_df['target'] = encoder.transform(test_df['target'])

# Separate features and target
X_train = train_df.drop(columns=['target'])
y_train = train_df['target']
X_test = test_df.drop(columns=['target'])
y_test = test_df['target']

# Scale the features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

```

```


# Split the training data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

print(f'Training set shape: {X_train.shape}, {y_train.shape}')
print(f'Validation set shape: {X_val.shape}, {y_val.shape}')

```

```



# Initialize and train the Random Forest model
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# Make predictions on the validation set
y_val_pred = model.predict_proba(X_val)[:, 1]


```

```



# Calculate AUC-ROC on the validation set
auc_roc_val = roc_auc_score(y_val, y_val_pred)

print(f'AUC-ROC on the validation set: {auc_roc_val:.4f}')


```

```




# Make predictions on the test set
y_test_pred = model.predict_proba(X_test)[:, 1]

# Calculate AUC-ROC on the test set
auc_roc_test = roc_auc_score(y_test, y_test_pred)

print(f'AUC-ROC on the test set: {auc_roc_test:.4f}')

```
