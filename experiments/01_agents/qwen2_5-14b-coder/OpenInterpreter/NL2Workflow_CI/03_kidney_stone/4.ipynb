```

import pandas as pd

train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/kidney_stone/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/kidney_stone/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

print("训练集和测试集加载完成。")
```

```

# 检查缺失值
print("训练集缺失值：")
print(train_data.isnull().sum())

print("\n测试集缺失值：")
print(test_data.isnull().sum())

# 如果有缺失值，可以选择填充或删除
# 示例：删除包含缺失值的行
train_data = train_data.dropna()
test_data = test_data.dropna()

# 分离特征和标签
X_train = train_data.drop(columns=['Target'])  # 替换 'Target' 为实际的标签列名
y_train = train_data['Target']

X_test = test_data.drop(columns=['Target'])  # 替换 'Target' 为实际的标签列名
y_test = test_data['Target']

# 检查分类变量并进行编码
categorical_cols = X_train.select_dtypes(include=['object']).columns
X_train = pd.get_dummies(X_train, columns=categorical_cols)
X_test = pd.get_dummies(X_test, columns=categorical_cols)

# 标准化连续变量
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("数据预处理完成。")
```

```


print("训练集列名：")
print(train_data.columns)

print("\n测试集列名：")
print(test_data.columns)

```

```



# 分离特征和标签
X_train = train_data.drop(columns=['id', 'target'])
y_train = train_data['target']

X_test = test_data.drop(columns=['id', 'target'])
y_test = test_data['target']

# 检查分类变量并进行编码
categorical_cols = X_train.select_dtypes(include=['object']).columns
X_train = pd.get_dummies(X_train, columns=categorical_cols)
X_test = pd.get_dummies(X_test, columns=categorical_cols)

# 标准化连续变量
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("数据预处理完成。")

```

```



from sklearn.model_selection import train_test_split

# 划分训练集和验证集（80% 用于训练，20% 用于验证）
X_train_split, X_val, y_train_split, y_val = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42, stratify=y_train
)

print(f"训练集大小: {len(X_train_split)}")
print(f"验证集大小: {len(X_val)}")
print(f"测试集大小: {len(X_test)}")


```

```




from sklearn.ensemble import RandomForestClassifier

# 初始化随机森林分类器
model = RandomForestClassifier(random_state=42)

# 训练模型
model.fit(X_train_split, y_train_split)

print("模型训练完成。")


```

```





from sklearn.metrics import roc_auc_score

# 预测验证集的概率
y_val_prob = model.predict_proba(X_val)[:, 1]

# 计算AUC-ROC指标
val_auc_roc = roc_auc_score(y_val, y_val_prob)

print(f"验证集上的AUC-ROC: {val_auc_roc:.4f}")



```

```






from sklearn.metrics import roc_auc_score

# 预测测试集的概率
y_test_prob = model.predict_proba(X_test)[:, 1]

# 计算AUC-ROC指标
test_auc_roc = roc_auc_score(y_test, y_test_prob)

print(f"测试集上的AUC-ROC: {test_auc_roc:.4f}")




```
