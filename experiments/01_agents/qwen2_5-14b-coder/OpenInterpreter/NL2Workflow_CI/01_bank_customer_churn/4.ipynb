```

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, roc_auc_score, roc_curve
import matplotlib.pyplot as plt
```

```

# Load the training dataset
train_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/train.csv')

# Load the test dataset
test_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/test.csv')
```

```


# Display the first few rows of the training dataset
print(train_data.head())

# Display the first few rows of the test dataset
print(test_data.head())

# Check for missing values in the training dataset
print(train_data.isnull().sum())

# Check for missing values in the test dataset
print(test_data.isnull().sum())
```

```


# One-hot encode the 'Gender' column
train_data = pd.get_dummies(train_data, columns=['Gender'], drop_first=True)
test_data = pd.get_dummies(test_data, columns=['Gender'], drop_first=True)

# Map 'Yes' and 'No' to 1 and 0 in 'Geography'
train_data['Geography'] = train_data['Geography'].map({'France': 0, 'Germany': 1, 'Spain': 2})
test_data['Geography'] = test_data['Geography'].map({'France': 0, 'Germany': 1, 'Spain': 2})

# Drop unnecessary columns
train_data.drop(['id', 'Surname', 'CustomerId'], axis=1, inplace=True)
test_data.drop(['id', 'Surname', 'CustomerId'], axis=1, inplace=True)

```

```



# Define features and target variable
X_train = train_data.drop('Exited', axis=1)
y_train = train_data['Exited']

# Split the test data
X_test = test_data.drop('Exited', axis=1)
y_test = test_data['Exited']


```

```



# Initialize the StandardScaler
scaler = StandardScaler()

# Fit and transform the training features
X_train_scaled = scaler.fit_transform(X_train)

# Transform the test features
X_test_scaled = scaler.transform(X_test)



```

```




# Initialize the Random Forest Classifier
rf_model = RandomForestClassifier(random_state=42)

# Train the model
rf_model.fit(X_train_scaled, y_train)



```

```





# Make predictions on the test set
y_pred = rf_model.predict(X_test_scaled)

# Calculate the confusion matrix
conf_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(conf_matrix)

# Calculate AUC-ROC
y_pred_proba = rf_model.predict_proba(X_test_scaled)[:, 1]
roc_auc = roc_auc_score(y_test, y_pred_proba)
print(f"AUC-ROC: {roc_auc:.4f}")

# Plot ROC curve
fpr, tpr, _ = roc_curve(y_test, y_pred_proba)
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='blue', label=f'ROC curve (area = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], 'k--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic')
plt.legend(loc="lower right")
plt.show()




```
