```

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, roc_auc_score
```

```

train_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/train.csv')
test_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/test.csv')
```

```


# 了解数据结构
print(train_data.info())
print(test_data.info())

# 处理缺失值（如果有）
train_data.fillna(method='ffill', inplace=True)
test_data.fillna(method='ffill', inplace=True)

# 特征工程（示例：将类别变量转换为数值变量）
for column in train_data.select_dtypes(include=['object']).columns:
    train_data[column] = train_data[column].astype('category').cat.codes
for column in test_data.select_dtypes(include=['object']).columns:
    test_data[column] = test_data[column].astype('category').cat.codes

# 拆分特征和标签
X_train = train_data.drop(columns=['Churn'])
y_train = train_data['Churn']
X_test = test_data.drop(columns=['Churn'])
y_test = test_data['Churn']

```

```



# Correctly split features and target variable using 'Exited' column
X_train = train_data.drop(columns=['Exited'])
y_train = train_data['Exited']
X_test = test_data.drop(columns=['Exited'])
y_test = test_data['Exited']



```

```



from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, roc_auc_score, classification_report

# 初始化随机森林分类器
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)

# 训练模型
rf_classifier.fit(X_train, y_train)


```

```





# 使用模型进行预测
y_pred = rf_classifier.predict(X_test)


```

```


import os

# 保存预测结果
predictions_dir = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/predictions'
os.makedirs(predictions_dir, exist_ok=True)

predictions_file = os.path.join(predictions_dir, 'predictions.csv')
pd.DataFrame({'Predicted': y_pred}).to_csv(predictions_file, index=False)

print(f"Predictions saved to {predictions_file}")




```

```



import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# 计算混淆矩阵
cm = confusion_matrix(y_test, y_pred)

# 绘制混淆矩阵
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot(cmap=plt.cm.Blues)
plt.title('Confusion Matrix')
plt.show()




```
