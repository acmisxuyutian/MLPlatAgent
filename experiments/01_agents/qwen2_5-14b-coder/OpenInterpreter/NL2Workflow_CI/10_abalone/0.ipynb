```

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_log_error

# 1. 数据加载
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark\NL2Workflow\datasets\tabular\abalone\train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark\NL2Workflow\datasets\tabular\abalone\test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 2. 数据预处理
# 独热编码
categorical_features = ['Sex']
ohe = OneHotEncoder(sparse=False)
train_ohe = ohe.fit_transform(train_df[categorical_features])
test_ohe = ohe.transform(test_df[categorical_features])

# 将独热编码后的特征添加回数据框
train_df_encoded = pd.concat([train_df.drop(categorical_features, axis=1), pd.DataFrame(train_ohe)], axis=1)
test_df_encoded = pd.concat([test_df.drop(categorical_features, axis=1), pd.DataFrame(test_ohe)], axis=1)

# 3. 数据划分
X_train = train_df_encoded.drop('Rings', axis=1)
y_train = train_df_encoded['Rings']

X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# 4. 模型训练
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 5. 模型评估
y_val_pred = model.predict(X_val)
y_test_pred = model.predict(test_df_encoded.drop('Rings', axis=1))

val_rmlse = np.sqrt(mean_squared_log_error(y_val, y_val_pred))
test_rmlse = np.sqrt(mean_squared_log_error(test_df_encoded['Rings'], y_test_pred))

# 6. 报告指标
print(f"Validation RMLSE: {val_rmlse}")
print(f"Test RMLSE: {test_rmlse}")
```

```


import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_log_error

# 1. 数据加载
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark\NL2Workflow\datasets\tabular\abalone\train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark\NL2Workflow\datasets\tabular\abalone\test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Convert column names to strings
train_df.columns = train_df.columns.astype(str)
test_df.columns = test_df.columns.astype(str)

# 2. 数据预处理
# 独热编码
categorical_features = ['Sex']
ohe = OneHotEncoder(sparse_output=False)
train_ohe = ohe.fit_transform(train_df[categorical_features])
test_ohe = ohe.transform(test_df[categorical_features])

# 将独热编码后的特征添加回数据框
train_df_encoded = pd.concat([train_df.drop(categorical_features, axis=1), pd.DataFrame(train_ohe)], axis=1)
test_df_encoded = pd.concat([test_df.drop(categorical_features, axis=1), pd.DataFrame(test_ohe)], axis=1)

# 3. 数据划分
X_train = train_df_encoded.drop('Rings', axis=1)
y_train = train_df_encoded['Rings']

X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# 4. 模型训练
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 5. 模型评估
y_val_pred = model.predict(X_val)
y_test_pred = model.predict(test_df_encoded.drop('Rings', axis=1))

val_rmlse = np.sqrt(mean_squared_log_error(y_val, y_val_pred))
test_rmlse = np.sqrt(mean_squared_log_error(test_df_encoded['Rings'], y_test_pred))

# 6. 报告指标
print(f"Validation RMLSE: {val_rmlse}")
print(f"Test RMLSE: {test_rmlse}")

```
