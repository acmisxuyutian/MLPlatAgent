```






from sklearn.preprocessing import OneHotEncoder

# Create OneHotEncoder object
encoder = OneHotEncoder(drop='first')

# Fit and transform the 'Sex' column
train_encoded = encoder.fit_transform(train_df_filtered[['Sex']])
test_encoded = encoder.transform(test_df_filtered[['Sex']])

# Convert the encoded arrays back to DataFrames
train_encoded_df = pd.DataFrame(train_encoded, columns=encoder.get_feature_names_out(['Sex']))
test_encoded_df = pd.DataFrame(test_encoded, columns=encoder.get_feature_names_out(['Sex']))

# Concatenate the original dataframes with the encoded dataframes
train_df_encoded = pd.concat([train_df_filtered.drop(columns=['Sex']), train_encoded_df], axis=1)
test_df_encoded = pd.concat([test_df_filtered.drop(columns=['Sex']), test_encoded_df], axis=1)

# Display the first few rows of the processed training data
train_df_encoded.head()



```

```



from sklearn.preprocessing import StandardScaler

# Select numerical features
num_cols = ['Length', 'Diameter', 'Height', 'Whole weight', 'Whole weight.1', 'Whole weight.2', 'Shell weight']

# Initialize the StandardScaler
scaler = StandardScaler()

# Fit and transform the numerical features
train_df_encoded[num_cols] = scaler.fit_transform(train_df_encoded[num_cols])
test_df_encoded[num_cols] = scaler.transform(test_df_encoded[num_cols])

# Display the first few rows of the standardized training data
train_df_encoded.head()


```
