```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from lightgbm import LGBMRegressor
from sklearn.metrics import mean_squared_log_error
import numpy as np
import os
```

```

# Define file paths
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/test.csv'

# Load datasets
train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

# Display the first few rows of the train dataset
print(train_data.head())
```

```


# One-hot encode the 'Sex' column
train_data_encoded = pd.get_dummies(train_data, columns=['Sex'])
test_data_encoded = pd.get_dummies(test_data, columns=['Sex'])

# Display the first few rows of the encoded train dataset
print(train_data_encoded.head())

```

```


# Drop the 'id' column
train_data_encoded.drop('id', axis=1, inplace=True)
test_data_encoded.drop('id', axis=1, inplace=True)

# Separate features and target variable
X_train = train_data_encoded.drop('Rings', axis=1)
y_train = train_data_encoded['Rings']
X_test = test_data_encoded.drop('Rings', axis=1)
y_test = test_data_encoded['Rings']

# Display the shape of the datasets
print(f"X_train shape: {X_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_test shape: {y_test.shape}")


```

```



# Initialize and train the LightGBM model
model = LGBMRegressor()
model.fit(X_train, y_train)

# Display the model
print(model)


```

```




# Make predictions on the test set
y_pred = model.predict(X_test)

# Calculate RMLSE
rmlse = np.sqrt(mean_squared_log_error(y_test, y_pred))

# Display the RMLSE
print(f"RMLSE: {rmlse}")



```
