```

# 加载训练集和测试集
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

print("数据已加载。")
print("训练集的前5行：")
print(train_df.head())
print("\n测试集的前5行：")
print(test_df.head())
```

```


# 检查缺失值
print("训练集缺失值：")
print(train_df.isnull().sum())
print("\n测试集缺失值：")
print(test_df.isnull().sum())

#选择特征和标签
features = train_df.drop(columns=['id', 'cost'])
target = train_df['cost']

test_features = test_df.drop(columns=['id'])

print("特征和标签已选择。")

```

```


# 分割训练集和验证集
X_train, X_val, y_train, y_val = train_test_split(features, target, test_size=0.2, random_state=42)

# 初始化随机森林回归模型
model = RandomForestRegressor(n_estimators=100, random_state=42)

# 训练模型
model.fit(X_train, y_train)

print("模型已训练。")

```

```



# 对测试集进行预测
test_predictions = model.predict(test_features)

# 计算RMLSE
rmlse = np.sqrt(mean_squared_log_error(test_df['cost'], test_predictions))

print(f"RMLSE: {rmlse}")


```

```




# 对测试集进行预测
test_predictions = model.predict(test_features)

# 计算RMLSE
rmlse = np.sqrt(mean_squared_log_error(test_df['cost'], test_predictions))

print(f"RMLSE: {rmlse}")



```
