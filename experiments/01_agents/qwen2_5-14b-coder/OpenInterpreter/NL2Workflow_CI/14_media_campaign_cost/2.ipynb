```

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_log_error
import os

# 1. Load data
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

print("Data loaded successfully.")

# 2. Data preprocessing
# Check for missing values
missing_values = train_data.isnull().sum()
print("Missing values in train set:\n", missing_values)

# Fill missing values (assuming mean imputation for simplicity)
train_data.fillna(train_data.mean(), inplace=True)

# Separate features and target
X_train = train_data.drop('media_campaign_cost', axis=1)
y_train = train_data['media_campaign_cost']

# Standardize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)

print("Data preprocessing completed.")

# 3. Build and train model
model = LinearRegression()
model.fit(X_train_scaled, y_train)

print("Model trained successfully.")

# 4. Make predictions
# Preprocess test data
test_data.fillna(test_data.mean(), inplace=True)
X_test = test_data.drop('media_campaign_cost', axis=1)
X_test_scaled = scaler.transform(X_test)

predictions = model.predict(X_test_scaled)

print("Predictions made successfully.")

# 5. Save predictions to file
output_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/predictions.csv'
predictions_df = pd.DataFrame({'predicted_media_campaign_cost': predictions})
predictions_df.to_csv(output_path, index=False)

print(f"Predictions saved to {output_path}")

# 6. Calculate and report RMLSE
def rmlse(y_true, y_pred):
    return np.sqrt(mean_squared_log_error(y_true, y_pred))

true_cost = test_data['media_campaign_cost']
rmlse_value = rmlse(true_cost, predictions)

print(f"RMLSE on test set: {rmlse_value}")
```
