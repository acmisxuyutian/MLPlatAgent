```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_log_error
import numpy as np
```

```


train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

print(f"训练集的形状: {train_df.shape}")
print(f"测试集的形状: {test_df.shape}")

```

```



# 检查缺失值
print(train_df.isnull().sum())
print(test_df.isnull().sum())

# 如果有缺失值，可以考虑填充或删除
# 例如，填充缺失值为0
train_df.fillna(0, inplace=True)
test_df.fillna(0, inplace=True)


```

```


# 检查数据类型
print(train_df.dtypes)
print(test_df.dtypes)

# 编码分类变量
# 例如，如果'low_fat'和'recyclable_package'是分类变量
train_df['low_fat'] = train_df['low_fat'].astype('category').cat.codes
test_df['low_fat'] = test_df['low_fat'].astype('category').cat.codes

train_df['recyclable_package'] = train_df['recyclable_package'].astype('category').cat.codes
test_df['recyclable_package'] = test_df['recyclable_package'].astype('category').cat.codes


```

```



# 分隔特征和目标变量
X_train = train_df.drop('cost', axis=1)
y_train = train_df['cost']

# 初始化并训练模型
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

print("模型训练完成。")

```

```




# 准备测试集特征
X_test = test_df.drop('cost', axis=1)

# 进行预测
y_pred = model.predict(X_test)

# 计算RMLSE
def rmlse(y_true, y_pred):
    return np.sqrt(np.mean(np.power(np.log1p(y_true) - np.log1p(y_pred), 2)))

rmlse_score = rmlse(test_df['cost'], y_pred)
print(f"模型在测试集上的RMLSE: {rmlse_score}")

```

```





# 保存预测结果
output_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/predictions.csv'
output_df = pd.DataFrame({'cost': y_pred})
output_df.to_csv(output_path, index=False)

print(f"预测结果已保存到: {output_path}")


```
