```

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_log_error
import joblib
```

```


# Load the datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

# Display the first few rows of the training dataset
print(train_data.head())

# Display the first few rows of the test dataset
print(test_data.head())

```

```



# Check for missing values
print(train_data.isnull().sum())
print(test_data.isnull().sum())

# Fill missing values with the mean of the respective columns
train_data.fillna(train_data.mean(), inplace=True)
test_data.fillna(train_data.mean(), inplace=True)

# Drop the 'id' column as it is not relevant for prediction
train_data.drop('id', axis=1, inplace=True)
test_data.drop('id', axis=1, inplace=True)

# Separate the features and target variable
X_train = train_data.drop('cost', axis=1)
y_train = train_data['cost']

# Initialize the StandardScaler
scaler = StandardScaler()

# Fit and transform the training and test features
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(test_data)


```

```




# Separate the features and target variable
X_train = train_data.drop('cost', axis=1)
y_train = train_data['cost']
X_test = test_data.drop('cost', axis=1)

# Initialize the StandardScaler
scaler = StandardScaler()

# Fit and transform the training and test features
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Print the shapes to ensure they match
print("X_train_scaled shape:", X_train_scaled.shape)
print("X_test_scaled shape:", X_test_scaled.shape)

```

```





# Initialize the Random Forest Regressor
model = RandomForestRegressor(n_estimators=100, random_state=42)

# Train the model
model.fit(X_train_scaled, y_train)

# Print the model's feature importances
print("Feature importances:", model.feature_importances_)


```

```





# Make predictions on the test data
y_pred = model.predict(X_test_scaled)

# Save the predictions to a CSV file
predictions_df = pd.DataFrame({'id': test_data['id'], 'cost': y_pred})
predictions_df.to_csv('media_campaign_cost_predictions.csv', index=False)

# Calculate the RMLSE
rmlse = np.sqrt(mean_squared_log_error(test_data['cost'], y_pred))
print("RMLSE:", rmlse)


```
