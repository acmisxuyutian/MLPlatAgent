```

import pandas as pd

# 读取训练集和测试集
train_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/train.csv')
test_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/test.csv')

# 查看数据的基本信息
print(train_data.head())
print(test_data.head())
```

```

# 特征构建
train_data['fruitmass_seeds_ratio'] = train_data['fruitmass'] / train_data['seeds']
test_data['fruitmass_seeds_ratio'] = test_data['fruitmass'] / test_data['seeds']

# 特征选择
import numpy as np

# 计算特征与目标变量之间的相关性
correlation_matrix = train_data.corr()
print(correlation_matrix['yield'].abs().sort_values(ascending=False))
```

```


from sklearn.model_selection import train_test_split

# 选择关键特征
features = ['fruitset', 'seeds', 'fruitmass', 'fruitmass_seeds_ratio']
X_train = train_data[features]
y_train = train_data['yield']

# 进行抽样处理
X_train_sample, X_val_sample, y_train_sample, y_val_sample = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

```

```

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

# 初始化模型
model = RandomForestRegressor(n_estimators=100, random_state=42)

# 训练模型
model.fit(X_train_sample, y_train_sample)

# 预测验证集
y_pred = model.predict(X_val_sample)

# 计算MAE
mae = mean_absolute_error(y_val_sample, y_pred)
print(f"Validation MAE: {mae}")

```

```


# 准备测试集数据
X_test = test_data[features]
y_test = test_data['yield']

# 预测测试集
y_test_pred = model.predict(X_test)

# 计算测试集上的MAE
test_mae = mean_absolute_error(y_test, y_test_pred)
print(f"Test MAE: {test_mae}")


```

```


print(f"Model Performance on Test Set: MAE = {test_mae}")

```
