```

import pandas as pd

# 加载训练集和测试集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

print(train_df.head())
print(test_df.head())
```

```

# 检查缺失值
print(train_df.isnull().sum())
print(test_df.isnull().sum())
```

```


# 选择特征
features = [col for col in train_df.columns if col not in ['id', 'yield']]
target = 'yield'

X_train = train_df[features]
y_train = train_df[target]
X_test = test_df[features]
y_test = test_df[target]

print(X_train.head())
print(y_train.head())

```

```



from sklearn.model_selection import train_test_split

# 在训练集中进行随机采样
X_sample, X_unused, y_sample, y_unused = train_test_split(X_train, y_train, test_size=0.8, random_state=42)

print(X_sample.shape)
print(y_sample.shape)


```

```




from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

# 初始化线性回归模型
model = LinearRegression()

# 使用采样数据进行训练
model.fit(X_sample, y_sample)

print("Model trained.")


```

```





# 使用测试集进行预测
y_pred = model.predict(X_test)

# 计算MAE
mae = mean_absolute_error(y_test, y_pred)

print(f"Mean Absolute Error (MAE) on the test set: {mae}")



```
