```



import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import roc_auc_score
import lightgbm as lgb

# Load data
train_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/reservation_cancellation/train.csv')
test_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/reservation_cancellation/test.csv')

# Check for missing values
print(train_data.isnull().sum())
print(test_data.isnull().sum())

# Encode categorical variables
label_encoders = {}
categorical_columns = ['type_of_meal_plan', 'room_type_reserved', 'market_segment_type', 'istribution_channel']

for column in categorical_columns:
    if column in train_data.columns and column in test_data.columns:
        le = LabelEncoder()
        train_data[column] = le.fit_transform(train_data[column])
        test_data[column] = le.transform(test_data[column])
        label_encoders[column] = le

# Split features and target variable
X_train = train_data.drop(columns=['id', 'booking_status'])
y_train = train_data['booking_status']
X_test = test_data.drop(columns=['id'])

# Check data types
print(X_train.dtypes)
print(X_test.dtypes)

# Display first few rows of processed data
print(X_train.head())
print(y_train.head())
print(X_test.head())

# Train LightGBM model
train_set = lgb.Dataset(X_train, label=y_train)
params = {
    'objective': 'binary',
    'metric': 'auc'
}
num_round = 100
bst = lgb.train(params, train_set, num_round)

# Make predictions
y_pred_proba = bst.predict(X_test, num_iteration=bst.best_iteration)

# Save predictions
predictions_df = pd.DataFrame({'id': test_data['id'], 'booking_status': y_pred_proba})
predictions_df.to_csv('reservation_cancellation_predictions.csv', index=False)

# Evaluate model
from sklearn.metrics import roc_auc_score
train_set = lgb.Dataset(X_train, label=y_train)
bst = lgb.train(params, train_set, num_round)
y_pred_proba_train = bst.predict(X_train, num_iteration=bst.best_iteration)
auc_roc = roc_auc_score(y_train, y_pred_proba_train)
print('AUC-ROC score:', auc_roc)


```
