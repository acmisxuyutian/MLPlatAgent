```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/04_diabetes/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/04_diabetes/test.csv'

train_data = pd.read_csv(train_data_path)
test_data = pd.read_csv(test_data_path)

# 1. Drop duplicate rows from the dataset
train_data.drop_duplicates(inplace=True)
test_data.drop_duplicates(inplace=True)

# Display the first few rows of the train data to verify
train_data.head()
```

```


# 2. One-hot encoding for gender and smoking_history
train_data_encoded = pd.get_dummies(train_data, columns=['gender', 'smoking_history'])
test_data_encoded = pd.get_dummies(test_data, columns=['gender', 'smoking_history'])

# Display the first few rows of the encoded train data to verify
train_data_encoded.head()

```

```



# 3. Rank the most important features
X_train = train_data_encoded.drop('diabetes', axis=1)
y_train = train_data_encoded['diabetes']

# Initialize and train a random forest classifier
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)

# Get feature importances
importances = rf.feature_importances_
feature_names = X_train.columns
feature_importances = pd.Series(importances, index=feature_names).sort_values(ascending=False)

# Display the feature importances
feature_importances


```

```



# 4. Build a random forest classifier and fit it on the training set
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)

# 5. Make predictions on the test set
X_test = test_data_encoded.drop('diabetes', axis=1)
y_test = test_data_encoded['diabetes']
y_pred_proba = rf.predict_proba(X_test)[:, 1]
y_pred = rf.predict(X_test)

# 6. Compute area under ROC curve
roc_auc = roc_auc_score(y_test, y_pred_proba)
print(f"Area under ROC curve: {roc_auc}")

# 7. Show the confusion matrix
conf_matrix = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', xticklabels=['No Diabetes', 'Diabetes'], yticklabels=['No Diabetes', 'Diabetes'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()

```
