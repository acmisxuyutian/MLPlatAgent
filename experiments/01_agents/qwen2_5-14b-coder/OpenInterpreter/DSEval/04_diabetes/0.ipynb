````

import pandas as pd

# Load the datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/04_diabetes/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/04_diabetes/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Drop duplicate rows
train_df = train_df.drop_duplicates()
test_df = test_df.drop_duplicates()

# Print the shapes to ensure duplicates are dropped
print(f"Training data shape: {train_df.shape}")
print(f"Test data shape: {test_df.shape}")

# Show the first few rows of the datasets
print(train_df.head())
print(test_df.head())
````

````

# One-hot encode the "gender" and "smoking_history" columns
train_df_encoded = pd.get_dummies(train_df, columns=['gender', 'smoking_history'])
test_df_encoded = pd.get_dummies(test_df, columns=['gender', 'smoking_history'])

# Print the shapes to ensure encoding is done correctly
print(f"Training data shape after encoding: {train_df_encoded.shape}")
print(f"Test data shape after encoding: {test_df_encoded.shape}")

# Show the first few rows of the encoded datasets
print(train_df_encoded.head())
print(test_df_encoded.head())

````

````


import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import SelectFromModel

# Define the features and target variable
X_train = train_df_encoded.drop('diabetes', axis=1)
y_train = train_df_encoded['diabetes']

# Initialize and fit the Random Forest Classifier
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)

# Get feature importances
feature_importances = pd.Series(rf.feature_importances_, index=X_train.columns).sort_values(ascending=False)

# Print the feature importances
print(feature_importances)


````

````



# Initialize and fit the Random Forest Classifier
rf = RandomForestClassifier(n_estimators=100, random_state=42, class_weight='balanced')
rf.fit(X_train, y_train)

# Prepare the test set
X_test = test_df_encoded.drop('diabetes', axis=1)
y_test = test_df_encoded['diabetes']

# Make predictions on the test set
y_pred = rf.predict(X_test)

# Print the predictions
print(y_pred)


````

````




from sklearn.metrics import roc_auc_score, confusion_matrix, classification_report

# Compute the probability estimates for the positive class
y_pred_proba = rf.predict_proba(X_test)[:, 1]

# Compute the AUC-ROC
auc_roc = roc_auc_score(y_test, y_pred_proba)
print(f"AUC-ROC: {auc_roc}")

# Show the confusion matrix
cm = confusion_matrix(y_test, y_pred)
print(f"Confusion Matrix:\n{cm}")

# Show the classification report
cr = classification_report(y_test, y_pred)
print(f"Classification Report:\n{cr}")


````
