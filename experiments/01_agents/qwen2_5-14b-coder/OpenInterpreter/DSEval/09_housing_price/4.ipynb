```

import pandas as pd

# Define the paths
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/09_housing_price/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/09_housing_price/test.csv'

# Load the data
housing_price_prediction_train = pd.read_csv(train_data_path)
housing_price_prediction_test = pd.read_csv(test_data_path)

# Display the first few rows of the data
print(housing_price_prediction_train.head())
print(housing_price_prediction_test.head())
```

```

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

# Define features and target variable
X_train = housing_price_prediction_train.drop('price', axis=1)
y_train = housing_price_prediction_train['price']

# Define features for the test set
X_test = housing_price_prediction_test.drop('price', axis=1)
y_test = housing_price_prediction_test['price']

# Initialize the Linear Regression model
model = LinearRegression()

# Fit the model on the training data
model.fit(X_train, y_train)

# Display the model's coefficients
print("Model coefficients:", model.coef_)
print("Model intercept:", model.intercept_)
```

```


from sklearn.preprocessing import LabelEncoder

# List of columns that need to be encoded
cols_to_encode = ['prefarea', 'furnishingstatus']

# Initialize Label Encoder
label_encoder = LabelEncoder()

# Encode the columns
for col in cols_to_encode:
    housing_price_prediction_train[col] = label_encoder.fit_transform(housing_price_prediction_train[col])
    housing_price_prediction_test[col] = label_encoder.transform(housing_price_prediction_test[col])

# Define features and target variable again
X_train = housing_price_prediction_train.drop('price', axis=1)
y_train = housing_price_prediction_train['price']

# Define features for the test set
X_test = housing_price_prediction_test.drop('price', axis=1)
y_test = housing_price_prediction_test['price']

# Initialize the Linear Regression model
model = LinearRegression()

# Fit the model on the training data
model.fit(X_train, y_train)

# Display the model's coefficients
print("Model coefficients:", model.coef_)
print("Model intercept:", model.intercept_)

```

```



# Make predictions on the test set
y_pred = model.predict(X_test)

# Calculate RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

# Display the RMSE
print("RMSE:", rmse)

```
