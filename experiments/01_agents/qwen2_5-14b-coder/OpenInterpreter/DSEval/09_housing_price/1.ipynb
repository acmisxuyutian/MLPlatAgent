```

import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

# Step 1: Load the data
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/09_housing_price/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/09_housing_price/test.csv'

train_data = pd.read_csv(train_data_path)
test_data = pd.read_csv(test_data_path)

# Display the first few rows of the train and test data
print("Training data head:")
print(train_data.head())

print("\nTesting data head:")
print(test_data.head())

# Step 2: Prepare the features and target variables
X_train = train_data.drop('price', axis=1)  # Assuming 'price' is the target column
y_train = train_data['price']

X_test = test_data.drop('price', axis=1)
y_test = test_data['price']

# Step 3: Build and fit the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Step 4: Make predictions on the test set
y_pred = model.predict(X_test)

# Step 5: Evaluate the model performance using RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
```

```


import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler, OneHotEncoder
import numpy as np

# Paths to the datasets
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/09_housing_price/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/09_housing_price/test.csv'

# Load the data
train_data = pd.read_csv(train_data_path)
test_data = pd.read_csv(test_data_path)

# Categorical columns to encode
categorical_columns = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea', 'furnishingstatus']

# One-hot encode the categorical columns
train_data_encoded = pd.get_dummies(train_data, columns=categorical_columns, drop_first=True)
test_data_encoded = pd.get_dummies(test_data, columns=categorical_columns, drop_first=True)

# Prepare the features and target variables
X_train = train_data_encoded.drop('price', axis=1)
y_train = train_data_encoded['price']

X_test = test_data_encoded.drop('price', axis=1)
y_test = test_data_encoded['price']

# Feature Scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Build and fit the Random Forest Regressor
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train_scaled, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test_scaled)

# Evaluate the model performance using RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")

# Feature Importance
feature_importances = model.feature_importances_
feature_names = X_train.columns
feature_importance_df = pd.DataFrame({'Feature': feature_names, 'Importance': feature_importances})
feature_importance_df = feature_importance_df.sort_values(by='Importance', ascending=False)

# Display the feature importance
print("\nFeature Importances:")
print(feature_importance_df)


```

```



import pandas as pd
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler, OneHotEncoder
import numpy as np

# Paths to the datasets
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/09_housing_price/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/09_housing_price/test.csv'

# Load the data
train_data = pd.read_csv(train_data_path)
test_data = pd.read_csv(test_data_path)

# Categorical columns to encode
categorical_columns = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea', 'furnishingstatus']

# One-hot encode the categorical columns
train_data_encoded = pd.get_dummies(train_data, columns=categorical_columns, drop_first=True)
test_data_encoded = pd.get_dummies(test_data, columns=categorical_columns, drop_first=True)

# Prepare the features and target variables
X_train = train_data_encoded.drop('price', axis=1)
y_train = train_data_encoded['price']

X_test = test_data_encoded.drop('price', axis=1)
y_test = test_data_encoded['price']

# Feature Scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Build and fit the Gradient Boosting Regressor
model = GradientBoostingRegressor(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42)
model.fit(X_train_scaled, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test_scaled)

# Evaluate the model performance using RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")

# Feature Importance
feature_importances = model.feature_importances_
feature_names = X_train.columns
feature_importance_df = pd.DataFrame({'Feature': feature_names, 'Importance': feature_importances})
feature_importance_df = feature_importance_df.sort_values(by='Importance', ascending=False)

# Display the feature importance
print("\nFeature Importances:")
print(feature_importance_df)




```

```



import pandas as pd
import matplotlib.pyplot as plt

# Paths to the datasets
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/09_housing_price/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/09_housing_price/test.csv'

# Load the data
train_data = pd.read_csv(train_data_path)
test_data = pd.read_csv(test_data_path)

# Combine train and test data for outlier detection
data = pd.concat([train_data, test_data])

# Visualize the distribution of each numerical feature using boxplots
numerical_features = ['area', 'bedrooms', 'bathrooms', 'stories', 'parking', 'price']
plt.figure(figsize=(15, 10))

for i, feature in enumerate(numerical_features):
    plt.subplot(2, 3, i + 1)
    plt.boxplot(data[feature])
    plt.title(feature)

plt.tight_layout()
plt.show()

```
