```

import pandas as pd

# Define the paths
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/09_housing_price/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/09_housing_price/test.csv'

# Load the datasets
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the datasets
print("Training Data:")
print(train_df.head())
print("\nTest Data:")
print(test_df.head())
```

```

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

# Assuming 'price' is the target variable and the rest are features
X_train = train_df.drop('price', axis=1)
y_train = train_df['price']

# Initialize the Linear Regression model
model = LinearRegression()

# Fit the model on the training data
model.fit(X_train, y_train)

# Prepare the test data without the target variable
X_test = test_df.drop('price', axis=1)

# Make predictions on the test data
y_pred = model.predict(X_test)

# Calculate RMSE on the test data
rmse = np.sqrt(mean_squared_error(test_df['price'], y_pred))

# Print the RMSE
print(f"The Root Mean Square Error (RMSE) of the model is: {rmse}")
```

```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Define the paths
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/09_housing_price/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/09_housing_price/test.csv'

# Load the datasets
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Define categorical columns
categorical_cols = ['prefarea', 'furnishingstatus']

# Create a column transformer to apply one-hot encoding to the categorical columns
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(), categorical_cols)
    ],
    remainder='passthrough'  # keep the numerical columns as is
)

# Initialize the Linear Regression model
linear_regression = LinearRegression()

# Create a pipeline that first transforms the data and then fits the model
model_pipe = Pipeline(steps=[('preprocessor', preprocessor),
                             ('model', linear_regression)])

# Split the data into features and target
X_train = train_df.drop('price', axis=1)
y_train = train_df['price']

# Fit the model on the training data
model_pipe.fit(X_train, y_train)

# Prepare the test data without the target variable
X_test = test_df.drop('price', axis=1)

# Make predictions on the test data
y_pred = model_pipe.predict(X_test)

# Calculate RMSE on the test data
rmse = np.sqrt(mean_squared_error(test_df['price'], y_pred))

# Print the RMSE
print(f"The Root Mean Square Error (RMSE) of the model is: {rmse}")
```

```


import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Define the paths
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/09_housing_price/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/09_housing_price/test.csv'

# Load the datasets
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Define categorical columns
categorical_cols = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea', 'furnishingstatus']

# Create a column transformer to apply one-hot encoding to the categorical columns
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(), categorical_cols)
    ],
    remainder='passthrough'  # keep the numerical columns as is
)

# Initialize the Linear Regression model
linear_regression = LinearRegression()

# Create a pipeline that first transforms the data and then fits the model
model_pipe = Pipeline(steps=[('preprocessor', preprocessor),
                             ('model', linear_regression)])

# Split the data into features and target
X_train = train_df.drop('price', axis=1)
y_train = train_df['price']

# Fit the model on the training data
model_pipe.fit(X_train, y_train)

# Prepare the test data without the target variable
X_test = test_df.drop('price', axis=1)

# Make predictions on the test data
y_pred = model_pipe.predict(X_test)

# Calculate RMSE on the test data
rmse = np.sqrt(mean_squared_error(test_df['price'], y_pred))

# Print the RMSE
print(f"The Root Mean Square Error (RMSE) of the model is: {rmse}")

```
