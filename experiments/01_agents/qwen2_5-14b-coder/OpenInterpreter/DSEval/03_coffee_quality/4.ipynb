```

# Step 1: Load the datasets
import pandas as pd

train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/03_coffee_quality/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/03_coffee_quality/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

print("Data loaded.")
```

```

import numpy as np

# Step 2: Drop duplicate rows and specified columns
columns_to_drop = [
    "ID", "ICO Number", "Owner", "Region", "Certification Contact", "Certification Address", 
    "Farm Name", "Lot Number", "Mill", "ICO Number", "Producer", "Company", "Expiration", 
    "Harvest Year", "Unnamed: 0", "Number of Bags", "Bag Weight", "In-Country Partner", 
    "Country of Origin", "Processing Method", "Grading Date", "Variety", "Status", "Defects", 
    "Uniformity", "Clean Cup", "Sweetness", "Certification Body"
]

train_df = train_df.drop_duplicates().drop(columns=columns_to_drop)
test_df = test_df.drop(columns=columns_to_drop)

print("Dropped specified columns and duplicates.")

# Step 3: Perform one-hot encoding for the "Color" column
train_df = pd.get_dummies(train_df, columns=['Color'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['Color'], drop_first=True)

print("One-hot encoded 'Color' column.")

# Step 4: Train a RandomForestRegressor model
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

# Separate features and target variable
X_train = train_df.drop(columns=['Total Cup Points'])
y_train = train_df['Total Cup Points']

X_test = test_df.drop(columns=['Total Cup Points'])
y_test = test_df['Total Cup Points']

# Ensure both train and test sets have the same columns
for col in X_train.columns:
    if col not in X_test.columns:
        X_test[col] = 0

# Train the model
model = RandomForestRegressor(random_state=42)
model.fit(X_train, y_train)

print("Model trained.")

# Step 5: Make predictions on the test set
predictions = model.predict(X_test)

print("Predictions made.")

# Step 6: Evaluate the model using mean-squared error
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, predictions)
print(f"Mean Squared Error: {mse}")
```
