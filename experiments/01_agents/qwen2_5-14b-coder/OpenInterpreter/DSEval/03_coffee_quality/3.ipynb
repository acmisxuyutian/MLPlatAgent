```

import pandas as pd

# Define file paths
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/03_coffee_quality/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/03_coffee_quality/test.csv'

# Load the data
train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

# Print initial information
print("Train data shape:", train_data.shape)
print("Test data shape:", test_data.shape)

# Drop duplicates
train_data.drop_duplicates(inplace=True)
test_data.drop_duplicates(inplace=True)

# Columns to drop
columns_to_drop = [
    "ID", "ICO Number", "Owner", "Region", "Certification Contact", 
    "Certification Address", "Farm Name", "Lot Number", "Mill", "ICO Number", 
    "Producer", "Company", "Expiration", "Harvest Year", "Unnamed: 0", 
    "Number of Bags", "Bag Weight", "In-Country Partner", "Country of Origin", 
    "Processing Method", "Grading Date", "Variety", "Status", "Defects", 
    "Uniformity", "Clean Cup", "Sweetness", "Certification Body"
]

# Drop the specified columns
train_data.drop(columns=columns_to_drop, inplace=True)
test_data.drop(columns=columns_to_drop, inplace=True)

# Print updated information
print("Train data shape after cleaning:", train_data.shape)
print("Test data shape after cleaning:", test_data.shape)
```

```

# Verify columns in the train and test datasets
print("Train data columns:", train_data.columns.tolist())
print("Test data columns:", test_data.columns.tolist())
```

```


# Remove 'Unnamed: 0' from the list of columns to drop
columns_to_drop = [
    "ID", "ICO Number", "Owner", "Region", "Certification Contact", 
    "Certification Address", "Farm Name", "Lot Number", "Mill", "Producer", 
    "Company", "Expiration", "Harvest Year", 
    "Number of Bags", "Bag Weight", "In-Country Partner", "Country of Origin", 
    "Processing Method", "Grading Date", "Variety", "Status", "Defects", 
    "Uniformity", "Clean Cup", "Sweetness", "Certification Body"
]

# Drop the specified columns
train_data.drop(columns=columns_to_drop, inplace=True)
test_data.drop(columns=columns_to_drop, inplace=True)

# Print updated information
print("Train data shape after cleaning:", train_data.shape)
print("Test data shape after cleaning:", test_data.shape)

# Perform one-hot encoding for the categorical column 'Color'
train_data = pd.get_dummies(train_data, columns=['Color'], drop_first=True)
test_data = pd.get_dummies(test_data, columns=['Color'], drop_first=True)

# Print updated information after one-hot encoding
print("Train data shape after one-hot encoding:", train_data.shape)
print("Test data shape after one-hot encoding:", test_data.shape)

# Separate features and target variable
X_train = train_data.drop(columns=['Total Cup Points'])
y_train = train_data['Total Cup Points']

X_test = test_data.drop(columns=['Total Cup Points'])
y_test = test_data['Total Cup Points']

# Import RandomForestRegressor
from sklearn.ensemble import RandomForestRegressor

# Initialize the regressor
regressor = RandomForestRegressor(random_state=42)

# Train the model
regressor.fit(X_train, y_train)

# Make predictions on the test set
y_pred = regressor.predict(X_test)

# Import mean_squared_error
from sklearn.metrics import mean_squared_error

# Evaluate the mean-squared error
mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:", mse)
```
