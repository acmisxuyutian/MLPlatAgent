```

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import os
```

```

# Load the training and test datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/03_coffee_quality/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/03_coffee_quality/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

print(train_df.head())  # Check the first few rows of the training dataset
print(test_df.head())   # Check the first few rows of the test dataset
```

```


# Drop duplicated rows from the training dataset
train_df = train_df.drop_duplicates()
print(f"Number of rows after dropping duplicates: {len(train_df)}")

```

```


# Define the columns to drop
columns_to_drop = [
    "ID", "ICO Number", "Owner", "Region", "Certification Contact", "Certification Address", 
    "Farm Name", "Lot Number", "Mill", "ICO Number", "Producer", "Company", "Expiration", 
    "Harvest Year", "Unnamed: 0", "Number of Bags", "Bag Weight", "In-Country Partner", 
    "Country of Origin", "Processing Method", "Grading Date", "Variety", "Status", "Defects", 
    "Uniformity", "Clean Cup", "Sweetness", "Certification Body"
]

# Drop the specified columns from both datasets
train_df = train_df.drop(columns=columns_to_drop)
test_df = test_df.drop(columns=columns_to_drop)

print(train_df.head())  # Check the first few rows of the training dataset after dropping columns
print(test_df.head())   # Check the first few rows of the test dataset after dropping columns

```

```



# Define the columns to drop, excluding "Unnamed: 0" which was not found
columns_to_drop = [
    "ID", "ICO Number", "Owner", "Region", "Certification Contact", "Certification Address", 
    "Farm Name", "Lot Number", "Mill", "ICO Number", "Producer", "Company", "Expiration", 
    "Harvest Year", "Number of Bags", "Bag Weight", "In-Country Partner", 
    "Country of Origin", "Processing Method", "Grading Date", "Variety", "Status", "Defects", 
    "Uniformity", "Clean Cup", "Sweetness", "Certification Body"
]

# Drop the specified columns from both datasets
train_df = train_df.drop(columns=columns_to_drop)
test_df = test_df.drop(columns=columns_to_drop)

print(train_df.head())  # Check the first few rows of the training dataset after dropping columns
print(test_df.head())   # Check the first few rows of the test dataset after dropping columns


```

```




# One-hot encode the "Color" column
train_df_encoded = pd.get_dummies(train_df, columns=["Color"], drop_first=True)
test_df_encoded = pd.get_dummies(test_df, columns=["Color"], drop_first=True)

print(train_df_encoded.head())  # Check the first few rows of the training dataset after one-hot encoding
print(test_df_encoded.head())   # Check the first few rows of the test dataset after one-hot encoding



```

```





# Define the features and target variable
X_train = train_df_encoded.drop(columns=["Total Cup Points"])
y_train = train_df_encoded["Total Cup Points"]

# Initialize and train the RandomForestRegressor
regressor = RandomForestRegressor(n_estimators=100, random_state=42)
regressor.fit(X_train, y_train)

print("Model trained successfully.")


```

```




# Define the features in the test set
X_test = test_df_encoded.drop(columns=["Total Cup Points"])

# Make predictions on the test set
y_pred = regressor.predict(X_test)

print("Predictions made successfully.")



```

```



# Calculate the mean-squared error
mse = mean_squared_error(test_df_encoded["Total Cup Points"], y_pred)

print(f"Mean Squared Error on the test set: {mse:.4f}")


```
