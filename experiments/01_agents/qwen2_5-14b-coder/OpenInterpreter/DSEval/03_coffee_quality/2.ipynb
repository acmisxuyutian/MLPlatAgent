```

import pandas as pd

train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/03_coffee_quality/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/03_coffee_quality/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

# Display the first few rows to check the data
print(train_data.head())
print(test_data.head())
```

```


# Remove duplicated rows from the train and test datasets
train_data.drop_duplicates(inplace=True)
test_data.drop_duplicates(inplace=True)

# Verify the number of rows after dropping duplicates
print(f"Train data after dropping duplicates: {len(train_data)} rows")
print(f"Test data after dropping duplicates: {len(test_data)} rows")

```

```




# List of columns to drop
columns_to_drop = [
    "ID", "ICO Number", "Owner", "Region", "Certification Contact", "Certification Address",
    "Farm Name", "Lot Number", "Mill", "ICO Number", "Producer", "Company", "Expiration",
    "Harvest Year", "Unnamed: 0", "Number of Bags", "Bag Weight", "In-Country Partner",
    "Country of Origin", "Processing Method", "Grading Date", "Variety", "Status", "Defects",
    "Uniformity", "Clean Cup", "Sweetness", "Certification Body"
]

# Drop columns from both datasets
train_data.drop(columns=columns_to_drop, inplace=True)
test_data.drop(columns=columns_to_drop, inplace=True)

# Verify the remaining columns
print(train_data.head())
print(test_data.head())


```

```



# Check the columns in the training dataset
print(train_data.columns)

# Check the columns in the test dataset
print(test_data.columns)


```

```





# Updated list of columns to drop
columns_to_drop = [
    "ID", "ICO Number", "Owner", "Region", "Certification Contact", "Certification Address",
    "Farm Name", "Lot Number", "Mill", "ICO Number", "Producer", "Company", "Expiration",
    "Harvest Year", "Number of Bags", "Bag Weight", "In-Country Partner",
    "Country of Origin", "Processing Method", "Grading Date", "Variety", "Status", "Defects",
    "Uniformity", "Clean Cup", "Sweetness", "Certification Body"
]

# Drop columns from both datasets
train_data.drop(columns=columns_to_drop, inplace=True)
test_data.drop(columns=columns_to_drop, inplace=True)

# Verify the remaining columns
print(train_data.head())
print(test_data.head())



```

```




# One-hot encode the "Color" column
train_data_encoded = pd.get_dummies(train_data, columns=["Color"])
test_data_encoded = pd.get_dummies(train_data, columns=["Color"])

# Verify the new columns after one-hot encoding
print(train_data_encoded.head())
print(test_data_encoded.head())



```

```





# Identify the features and target variable
features = train_data_encoded.drop(columns=["Total Cup Points"])
target = train_data_encoded["Total Cup Points"]

# Split the test data into features and target variable
test_features = test_data_encoded.drop(columns=["Total Cup Points"])
test_target = test_data_encoded["Total Cup Points"]

# Verify the split
print(features.head())
print(target.head())


```

```





# Import the necessary library
from sklearn.ensemble import RandomForestRegressor

# Initialize the RandomForestRegressor model
model = RandomForestRegressor(random_state=42)

# Train the model on the training data
model.fit(features, target)

# Verify the model training
print("Model trained successfully")




```

```






# Make predictions on the test data
predictions = model.predict(test_features)

# Verify the predictions
print(predictions[:10])  # Print the first 10 predictions


```

```






# Import the necessary library
from sklearn.metrics import mean_squared_error

# Calculate the mean-squared error (MSE)
mse = mean_squared_error(test_target, predictions)

# Print the MSE
print(f"Mean Squared Error on the test set: {mse}")





```
