```

import pandas as pd

# Load the datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/10_life_expectancy/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/10_life_expectancy/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the training data
train_df.head()
```

```


# Define a function to categorize GDP into bins
def categorize_gdp(gdp):
    if gdp <= 1000:
        return "Low GDP"
    elif 1000 < gdp <= 10000:
        return "Medium GDP"
    else:
        return "High GDP"

# Apply the function to create the GDP_bins feature
train_df['GDP_bins'] = train_df['GDP'].apply(categorize_gdp)
test_df['GDP_bins'] = test_df['GDP'].apply(categorize_gdp)

# Display the first few rows of the training data with the new feature
train_df.head()
```

```



# Define the target variable and features
target = 'Life expectancy'
features = [col for col in train_df.columns if col != target]

# Split the data into training and test sets
X_train = train_df[features]
y_train = train_df[target]
X_test = test_df[features]
y_test = test_df[target]

# Display the first few rows of the training features
X_train.head()

```

```



# Ensure column names are consistent by renaming the columns in the test dataset
test_df = test_df.rename(columns={'Life expectancy': 'Life expectancy '})

# Split the data into training and test sets again
X_train = train_df[features]
y_train = train_df[target]
X_test = test_df[features]
y_test = test_df[target]

# Display the first few rows of the training features
X_train.head()



```

```




# Import necessary libraries
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Initialize the Linear Regression model
model = LinearRegression()

# Fit the model on the training data
model.fit(X_train, y_train)

# Print the model coefficients
print("Model coefficients:")
print(model.coef_)


```

```





# Drop non-numeric columns
X_train_numeric = X_train.select_dtypes(include=['number'])
X_test_numeric = X_test.select_dtypes(include=['number'])

# Fit the model on the training data using only numeric features
model.fit(X_train_numeric, y_train)

# Print the model coefficients
print("Model coefficients:")
print(model.coef_)



```

```





# Make predictions on the test set
y_pred = model.predict(X_test_numeric)

# Print the first few predictions
print("First few predictions:")
print(y_pred[:5])




```

```






# Calculate the R-squared score
r2 = r2_score(y_test, y_pred)

# Print the R-squared score
print("R-squared score:")
print(r2)



```
