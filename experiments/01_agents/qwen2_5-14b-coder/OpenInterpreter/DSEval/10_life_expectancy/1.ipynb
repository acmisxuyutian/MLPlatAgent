```

import pandas as pd

# Define the paths
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\10_life_expectancy\\train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\10_life_expectancy\\test.csv'

# Load the data
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and test data
print(train_df.head())
print(test_df.head())
```

```

# Define the GDP binning function
def create_gdp_bins(row):
    if row['GDP'] <= 1000:
        return 'Low GDP'
    elif 1000 < row['GDP'] <= 10000:
        return 'Medium GDP'
    else:
        return 'High GDP'

# Apply the function to create GDP_bins for both datasets
train_df['GDP_bins'] = train_df.apply(create_gdp_bins, axis=1)
test_df['GDP_bins'] = test_df.apply(create_gdp_bins, axis=1)

# Display the first few rows of the training and test data with the new GDP_bins feature
print(train_df[['GDP', 'GDP_bins']].head())
print(test_df[['GDP', 'GDP_bins']].head())
```

```


# Drop the original GDP feature from both datasets
train_df = train_df.drop(columns=['GDP'])
test_df = test_df.drop(columns=['GDP'])

# Select features and target variable
features = ['Income composition of resources', 'Schooling', 'GDP_bins']
target = 'Life expectancy'

# Encode categorical variable 'GDP_bins'
train_df = pd.get_dummies(train_df, columns=['GDP_bins'])
test_df = pd.get_dummies(test_df, columns=['GDP_bins'])

# Ensure that both train and test datasets have the same columns
train_columns = train_df.columns.tolist()
test_columns = test_df.columns.tolist()
missing_columns = [col for col in train_columns if col not in test_columns]

for col in missing_columns:
    test_df[col] = 0

# Selecting features and target for training and test datasets
X_train = train_df[features + ['GDP_bins_Low GDP', 'GDP_bins_Medium GDP', 'GDP_bins_High GDP']]
y_train = train_df[target]
X_test = test_df[features + ['GDP_bins_Low GDP', 'GDP_bins_Medium GDP', 'GDP_bins_High GDP']]
y_test = test_df[target]

# Display the first few rows of the prepared training and test data
print(X_train.head())
print(y_train.head())
print(X_test.head())
print(y_test.head())

```

```


import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Define the paths
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\10_life_expectancy\\train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\10_life_expectancy\\test.csv'

# Load the data
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Check if 'GDP' column exists
if 'GDP' not in train_df.columns or 'GDP' not in test_df.columns:
    raise ValueError("'GDP' column is missing in one of the datasets.")

# Define the GDP binning function
def create_gdp_bins(row):
    if row['GDP'] <= 1000:
        return 'Low GDP'
    elif 1000 < row['GDP'] <= 10000:
        return 'Medium GDP'
    else:
        return 'High GDP'

# Apply the function to create GDP_bins for both datasets
train_df['GDP_bins'] = train_df.apply(create_gdp_bins, axis=1)
test_df['GDP_bins'] = test_df.apply(create_gdp_bins, axis=1)

# Display the first few rows of the training and test data with the new GDP_bins feature
print(train_df[['GDP', 'GDP_bins']].head())
print(test_df[['GDP', 'GDP_bins']].head())

# One-hot encode 'GDP_bins'
train_df = pd.get_dummies(train_df, columns=['GDP_bins'])
test_df = pd.get_dummies(test_df, columns=['GDP_bins'])

# Ensure that both train and test datasets have the same columns
train_columns = train_df.columns.tolist()
test_columns = test_df.columns.tolist()
missing_columns = [col for col in train_columns if col not in test_columns]

for col in missing_columns:
    test_df[col] = 0

# Sort columns to ensure alignment
test_df = test_df[test_df.columns.intersection(train_df.columns)]

# Select features and target variable
features = ['GDP', 'Income composition of resources', 'Schooling', 'GDP_bins_Low GDP', 'GDP_bins_Medium GDP', 'GDP_bins_High GDP']
target = 'Life expectancy'

X_train = train_df[features]
y_train = train_df[target]
X_test = test_df[features]
y_test = test_df[target]

# Display the first few rows of the prepared training and test data
print(X_train.head())
print(y_train.head())
print(X_test.head())
print(y_test.head())

# Build and train the Linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model using R-squared score
r2 = r2_score(y_test, y_pred)

print(f"R-squared score: {r2}")

```
