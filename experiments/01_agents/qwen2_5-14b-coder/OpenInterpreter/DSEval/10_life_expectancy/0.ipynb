```

# Define the paths to the training and test data
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/10_life_expectancy/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/10_life_expectancy/test.csv'

# Load the training and test data
train_data = pd.read_csv(train_data_path)
test_data = pd.read_csv(test_data_path)

# Display the first few rows of the training data
print("Training Data:")
print(train_data.head())

# Display the first few rows of the test data
print("\nTest Data:")
print(test_data.head())

# Check for missing values in the training data
print("\nMissing values in training data:")
print(train_data.isnull().sum())

# Check for missing values in the test data
print("\nMissing values in test data:")
print(test_data.isnull().sum())
```

```


# Define the GDP bins
def categorize_gdp(gdp):
    if gdp <= 1000:
        return "Low GDP"
    elif 1000 < gdp <= 10000:
        return "Medium GDP"
    else:
        return "High GDP"

# Apply the GDP bins to the training and test data
train_data['GDP_bins'] = train_data['GDP'].apply(categorize_gdp)
test_data['GDP_bins'] = test_data['GDP'].apply(categorize_gdp)

# Display the first few rows of the training and test data with the new GDP_bins feature
print("Training Data with GDP_bins:")
print(train_data.head())

print("\nTest Data with GDP_bins:")
print(test_data.head())
```

```


# Select the features and target variable for training
X_train = train_data[['GDP', 'Schooling']]
y_train = train_data['Life expectancy']

# Initialize the Linear Regression model
model = LinearRegression()

# Fit the model on the training data
model.fit(X_train, y_train)

# Print the coefficients of the model
print("Model Coefficients:")
print("Slope (GDP):", model.coef_[0])
print("Intercept:", model.intercept_)


```

```




# Re-select the features and target variable for training
X_train = train_data[['GDP', 'Schooling']]
y_train = train_data['Life expectancy ']  # Use the correct column name

# Initialize the Linear Regression model
model = LinearRegression()

# Fit the model on the training data
model.fit(X_train, y_train)

# Print the coefficients of the model
print("Model Coefficients:")
print("Slope (GDP):", model.coef_[0])
print("Intercept:", model.intercept_)



```

```




# Select the features for the test set
X_test = test_data[['GDP', 'Schooling']]

# Make predictions on the test set
y_pred = model.predict(X_test)

# Display the first few predicted values
print("Predicted Life Expectancy on Test Set:")
print(y_pred[:5])



```

```





# Select the actual life expectancy values from the test set
y_true = test_data['Life expectancy ']

# Calculate the R-squared score
r2 = r2_score(y_true, y_pred)

# Print the R-squared score
print("R-squared Score:")
print(r2)



```
