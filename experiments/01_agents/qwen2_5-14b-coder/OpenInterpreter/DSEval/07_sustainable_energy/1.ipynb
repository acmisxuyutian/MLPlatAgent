```




# Step 1: Load the data
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/07_sustainable_energy/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/07_sustainable_energy/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

# Step 2: Remove commas from numeric columns
def remove_commas_from_numeric(df):
    for column in df.columns:
        if df[column].dtype == object:
            try:
                df[column] = df[column].str.replace(',', '').astype(float)
            except ValueError:
                continue
    return df

train_data = remove_commas_from_numeric(train_data)
test_data = remove_commas_from_numeric(test_data)

# Step 3: Split the data into features and target variable
features = [col for col in train_data.columns if col not in ['Entity', 'Year', 'Renewable energy share in the total final energy consumption (%)']]
target = 'Renewable energy share in the total final energy consumption (%)'

X_train = train_data[features]
y_train = train_data[target]
X_test = test_data[features]
y_test = test_data[target]

# Step 4: Handle missing values using SimpleImputer for both X and y
imputer = SimpleImputer(strategy='mean')

# Impute X
X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)

# Impute y
imputer_y = SimpleImputer(strategy='mean')
y_train_imputed = imputer_y.fit_transform(y_train.values.reshape(-1, 1)).flatten()
y_test_imputed = imputer_y.transform(y_test.values.reshape(-1, 1)).flatten()

# Step 5: Build and fit the linear regression model
model = LinearRegression()
model.fit(X_train_imputed, y_train_imputed)

# Step 6: Make predictions on the test set
predictions = model.predict(X_test_imputed)

# Step 7: Evaluate the model using RMSE
rmse = np.sqrt(mean_squared_error(y_test_imputed, predictions))
print(f"RMSE: {rmse}")



```
