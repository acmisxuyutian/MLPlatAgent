```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np
```

```


# Load training data
train_data = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\07_sustainable_energy\\train.csv')

# Load test data
test_data = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\07_sustainable_energy\\test.csv')

# Display the first few rows of the training data
print(train_data.head())

# Display the first few rows of the test data
print(test_data.head())

# Check the columns in the training data
print(train_data.columns)

# Check the columns in the test data
print(test_data.columns)

```

```



# Remove 'Entity' and 'Year' columns from both datasets
train_data = train_data.drop(['Entity', 'Year'], axis=1)
test_data = test_data.drop(['Entity', 'Year'], axis=1)

# Select features and target variable from the training data
X_train = train_data.drop(['Renewable energy share in the total final energy consumption (%)'], axis=1)
y_train = train_data['Renewable energy share in the total final energy consumption (%)']

# Initialize and fit the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions on the test data
X_test = test_data.drop(['Renewable energy share in the total final energy consumption (%)'], axis=1)
y_pred = model.predict(X_test)

# Evaluate the model's performance using RMSE
rmse = np.sqrt(mean_squared_error(test_data['Renewable energy share in the total final energy consumption (%)'], y_pred))
print(f'RMSE: {rmse:.4f}')


```
