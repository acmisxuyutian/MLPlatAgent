```

import pandas as pd

# Load the training and test datasets
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/07_sustainable_energy/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/07_sustainable_energy/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of each dataset to verify
print(train_df.head())
print(test_df.head())
```

```


# Step 2: Prepare the features and target variable

# Define features and target variable
features = [col for col in train_df.columns if col not in ['Entity', 'Year', 'Renewable energy share in the total final energy consumption (%)']]
target = 'Renewable energy share in the total final energy consumption (%)'

# Separate features and target in both training and test datasets
X_train = train_df[features]
y_train = train_df[target]

X_test = test_df[features]
y_test = test_df[target]

# Display the first few rows of the features and target variable to verify
print(X_train.head())
print(y_train.head())

```

```



# Step 3: Build and fit a linear regression model

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Initialize the linear regression model
lr_model = LinearRegression()

# Fit the model on the training data
lr_model.fit(X_train, y_train)
```

```



# Clean the data by removing commas from numeric columns
def clean_data(df):
    for column in df.select_dtypes(include=['object']):
        df[column] = df[column].str.replace(',', '')
    return df

# Clean the training and test datasets
train_df_cleaned = clean_data(train_df)
test_df_cleaned = clean_data(test_df)

# Prepare the features and target variable again with the cleaned data
X_train_cleaned = train_df_cleaned[features]
y_train_cleaned = train_df_cleaned[target]

X_test_cleaned = test_df_cleaned[features]
y_test_cleaned = test_df_cleaned[target]

# Build and fit a linear regression model again
lr_model_cleaned = LinearRegression()
lr_model_cleaned.fit(X_train_cleaned, y_train_cleaned)


```

```


# Handle NaN values by imputation
from sklearn.impute import SimpleImputer

# Initialize the imputer
imputer = SimpleImputer(strategy='mean')

# Fit and transform the training data
X_train_imputed = imputer.fit_transform(X_train_cleaned)

# Transform the test data
X_test_imputed = imputer.transform(X_test_cleaned)

# Build and fit a linear regression model again with imputed data
lr_model_imputed = LinearRegression()
lr_model_imputed.fit(X_train_imputed, y_train_cleaned)



```

```




# Handle NaN values in the target variable by imputation
y_train_imputed = imputer.fit_transform(y_train_cleaned.values.reshape(-1, 1)).flatten()
y_test_imputed = imputer.transform(y_test_cleaned.values.reshape(-1, 1)).flatten()

# Build and fit a linear regression model again with imputed data
lr_model_imputed = LinearRegression()
lr_model_imputed.fit(X_train_imputed, y_train_imputed)

```

```




# Make predictions on the test data
y_pred = lr_model_imputed.predict(X_test_imputed)

# Evaluate the model using RMSE
from sklearn.metrics import mean_squared_error
import numpy as np

rmse = np.sqrt(mean_squared_error(y_test_imputed, y_pred))
print(f'RMSE: {rmse}')

```
