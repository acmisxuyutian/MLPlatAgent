```

import pandas as pd

# Load the training data
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/07_sustainable_energy/train.csv'
train_df = pd.read_csv(train_data_path)

# Load the test data
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/07_sustainable_energy/test.csv'
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the training data
print(train_df.head())
```

```


# Define the target column and feature columns
target_column = 'Renewable energy share in the total final energy consumption (%)'
feature_columns = [col for col in train_df.columns if col not in ['Entity', 'Year', target_column]]

# Split the training data into features and target
X_train = train_df[feature_columns]
y_train = train_df[target_column]

# Split the test data into features and target
X_test = test_df[feature_columns]
y_test = test_df[target_column]

# Display the first few rows of the features training data
print(X_train.head())

```

```

from sklearn.linear_model import LinearRegression

# Initialize the linear regression model
model = LinearRegression()

# Fit the model on the training data
model.fit(X_train, y_train)

# Print the model coefficients
print("Model coefficients:", model.coef_)
print("Model intercept:", model.intercept_)


```
