```

import pandas as pd

# Load the training and test datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\05_patient_profile\\train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\05_patient_profile\\test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)
```

```


from sklearn.preprocessing import OneHotEncoder

# Identify categorical columns (excluding "Disease")
categorical_cols = train_data.select_dtypes(include=['object']).columns.tolist()
categorical_cols.remove('Disease')

# One-hot encode the categorical columns
encoder = OneHotEncoder(sparse=False, drop='first')
train_encoded = encoder.fit_transform(train_data[categorical_cols])
test_encoded = encoder.transform(test_data[categorical_cols])

# Create a DataFrame from the encoded features
encoded_feature_names = encoder.get_feature_names_out(categorical_cols)
train_encoded_df = pd.DataFrame(train_encoded, columns=encoded_feature_names)
test_encoded_df = pd.DataFrame(test_encoded, columns=encoded_feature_names)

# Concatenate the encoded features with the remaining columns
train_data = pd.concat([train_data.drop(columns=categorical_cols), train_encoded_df], axis=1)
test_data = pd.concat([test_data.drop(columns=categorical_cols), test_encoded_df], axis=1)

```

```



from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

# Initialize the logistic regression models
model1 = LogisticRegression(max_iter=10000)
model2 = LogisticRegression(max_iter=10000)

# Train the models
model1.fit(X_train, y_train)
model2.fit(X_train, y_train)

# Predict on the test set
y_pred1 = model1.predict(X_test)
y_pred2 = model2.predict(X_test)

# Evaluate the models
print('Model 1 - Classification Report:')
print(classification_report(train_data['Outcome Variable_Positive'], model1.predict(X_train)))
print(f'Accuracy: {accuracy_score(train_data["Outcome Variable_Positive"], model1.predict(X_train))}')

print('\nModel 2 - Classification Report:')
print(classification_report(train_data['Outcome Variable_Positive'], model2.predict(X_train)))
print(f'Accuracy: {accuracy_score(train_data["Outcome Variable_Positive"], model2.predict(X_train))}')



```
