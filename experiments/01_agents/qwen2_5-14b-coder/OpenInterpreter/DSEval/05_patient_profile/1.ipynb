````

import pandas as pd

train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\05_patient_profile\\train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\05_patient_profile\\test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)
````

````

# One-hot encode categorical features except "Disease" column
train_encoded = pd.get_dummies(train_df.drop('Disease', axis=1))
test_encoded = pd.get_dummies(test_df.drop('Disease', axis=1))
````

````


# Extract the target variable 'outcome variable'
train_target = train_df['outcome variable']
test_target = test_df['outcome variable']

# Ensure the "Disease" column is handled consistently
if 'Disease' in train_encoded.columns:
    train_encoded = train_encoded.drop('Disease', axis=1)
if 'Disease' in test_encoded.columns:
    test_encoded = test_encoded.drop('Disease', axis=1)

````

````

# Define the target variable name
target_variable = 'Outcome Variable'

# Ensure the target variable is in the encoded DataFrame and exclude it
if target_variable in train_encoded.columns:
    train_encoded = train_encoded.drop(target_variable, axis=1)
if target_variable in test_encoded.columns:
    test_encoded = test_encoded.drop(target_variable, axis=1)

# Initialize the logistic regression model
logit = LogisticRegression(max_iter=10000)

# Initialize RFE for selecting 5 features
rfe = RFE(estimator=logit, n_features_to_select=5)

# Fit RFE on the training data
rfe.fit(train_encoded, train_target)

# Get the selected features
selected_features = train_encoded.columns[rfe.support_]
print("Selected features:", selected_features.tolist())
````

````


# Manually select 5 relevant features (exclude target variable columns)
selected_features = ['Fever_No', 'Gender_Female', 'Cholesterol Level_High', 'Difficulty Breathing_Yes', 'Age']

print("Selected features:", selected_features)

````

````



# Import necessary libraries
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Initialize the logistic regression models
model1 = LogisticRegression(max_iter=10000)
model2 = LogisticRegression(max_iter=10000)

# Train model 1 on all features
model1.fit(train_encoded, train_target)

# Train model 2 on the selected 5 features
model2.fit(train_encoded[selected_features], train_target)

````

````



# Evaluate Model 1 on the test set
test_pred_model1 = model1.predict(test_encoded)
print("Model 1 - Test Accuracy:", accuracy_score(test_target, test_pred_model1))
print("Model 1 - Classification Report:\n", classification_report(test_target, test_pred_model1))

# Evaluate Model 2 on the test set
test_pred_model2 = model2.predict(test_encoded[selected_features])
print("Model 2 - Test Accuracy:", accuracy_score(test_target, test_pred_model2))
print("Model 2 - Classification Report:\n", classification_report(test_target, test_pred_model2))



````
