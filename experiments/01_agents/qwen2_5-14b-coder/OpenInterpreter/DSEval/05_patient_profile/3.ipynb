```

import pandas as pd

# Define the file paths
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/05_patient_profile/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/05_patient_profile/test.csv'

# Load the datasets
train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

# Display the first few rows of the training dataset
train_data.head()
```

```


from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
import numpy as np

# Define the categorical columns to be one-hot encoded
categorical_columns = ['Fever', 'Cough', 'Fatigue', 'Difficulty Breathing', 'Gender', 'Blood Pressure', 'Cholesterol Level']
drop_columns = ['Disease']

# Create a column transformer to perform one-hot encoding
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(), categorical_columns)
    ],
    remainder='passthrough'
)

# Combine the preprocessor with the logistic regression model
model1 = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(max_iter=1000))
])

# Fit the model on the training data
X_train = train_data.drop(columns=['Outcome Variable'] + drop_columns)
y_train = train_data['Outcome Variable']
model1.fit(X_train, y_train)

# Display the feature names after one-hot encoding
feature_names = model1.named_steps['preprocessor'].get_feature_names_out(categorical_columns)
print("Feature names after one-hot encoding:", feature_names)

```
