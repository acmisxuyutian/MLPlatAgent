```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score
```

```

# Define paths to the datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/test.csv'

# Load datasets
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training dataset
print("First few rows of the training dataset:")
print(train_df.head())

# Drop rows with missing values and duplicates
train_df.dropna(inplace=True)
train_df.drop_duplicates(inplace=True)

# Display the first few rows after preprocessing
print("\nFirst few rows of the training dataset after preprocessing:")
print(train_df.head())
```

```


# Define a function to categorize BMI
def categorize_bmi(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif 18.5 <= bmi < 25:
        return "Normal weight"
    elif 25 <= bmi < 30:
        return "Overweight"
    else:
        return "Obesity"

# Calculate BMI and create BMI_Category feature
train_df['BMI'] = train_df['Weight_kg'] / ((train_df['Height_cm'] / 100) ** 2)
train_df['BMI_Category'] = train_df['BMI'].apply(categorize_bmi).astype('category')

# Display the first few rows after adding BMI category
print("\nFirst few rows of the training dataset after adding BMI Category:")
print(train_df.head())

```

```



# Define a function to categorize BMI
def categorize_bmi(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif 18.5 <= bmi < 25:
        return "Normal weight"
    elif 25 <= bmi < 30:
        return "Overweight"
    else:
        return "Obesity"

# Calculate BMI and create BMI_Category feature
train_df['BMI'] = train_df['Weight_(kg)'] / ((train_df['Height_(cm)'] / 100) ** 2)
train_df['BMI_Category'] = train_df['BMI'].apply(categorize_bmi).astype('category')

# Display the first few rows after adding BMI category
print("\nFirst few rows of the training dataset after adding BMI Category:")
print(train_df.head())


```

```



# Separate features and target variable
X_train = train_df.drop(columns=['Heart_Disease', 'BMI'])  # Exclude 'Heart_Disease' and 'BMI'
y_train = train_df['Heart_Disease']

# Initialize the Random Forest Classifier
model = RandomForestClassifier(random_state=42)

# Fit the model to the training data
model.fit(X_train, y_train)

print("Model trained successfully.")


```

```


# Display the data types of the columns
print("Data types before encoding:")
print(train_df.dtypes)

# One-hot encode all categorical columns
train_df_encoded = pd.get_dummies(train_df, drop_first=True)

# Display the data types after encoding
print("\nData types after encoding:")
print(train_df_encoded.dtypes)

# Display the first few rows after encoding
print("\nFirst few rows after encoding:")
print(train_df_encoded.head())

```

```




# Separate features and target variable
X_train_encoded = train_df_encoded.drop(columns=['Heart_Disease_Yes'])  # Exclude 'Heart_Disease_Yes'
y_train_encoded = train_df_encoded['Heart_Disease_Yes']

# Load and preprocess the test dataset
test_df.dropna(inplace=True)
test_df.drop_duplicates(inplace=True)

# One-hot encode the test dataset similarly
test_df_encoded = pd.get_dummies(test_df, drop_first=True)

# Ensure both datasets have the same columns
missing_cols = set(X_train_encoded.columns) - set(test_df_encoded.columns)
for c in missing_cols:
    test_df_encoded[c] = 0
test_df_encoded = test_df_encoded[X_train_encoded.columns]

# Initialize the Random Forest Classifier
model = RandomForestClassifier(random_state=42)

# Fit the model to the training data
model.fit(X_train_encoded, y_train_encoded)
print("Model trained successfully.")

# Make predictions on the test set
y_pred_proba = model.predict_proba(test_df_encoded)[:, 1]

# Compute Area Under ROC Curve
roc_auc = roc_auc_score(test_df['Heart_Disease'], y_pred_proba)
print(f"Area Under ROC Curve: {roc_auc.round(4)}")



```
