```

# Step 1: Load the Data
import pandas as pd

train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)
print("Data loaded successfully")

# Step 2: Data Cleaning
# Drop rows with missing values and duplicates
train_data = train_data.dropna().drop_duplicates()
test_data = test_data.dropna().drop_duplicates()

print("Data cleaned successfully")

# Step 3: Feature Construction
def categorize_bmi(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif 18.5 <= bmi < 25:
        return "Normal weight"
    elif 25 <= bmi < 30:
        return "Overweight"
    else:
        return "Obesity"

# Add BMI_Category column to both train and test datasets
train_data['BMI'] = train_data['Weight'] / (train_data['Height'] / 100) ** 2
test_data['BMI'] = test_data['Weight'] / (test_data['Height'] / 100) ** 2

train_data['BMI_Category'] = train_data['BMI'].apply(categorize_bmi)
test_data['BMI_Category'] = test_data['BMI'].apply(categorize_bmi)

# Convert BMI_Category to category dtype
train_data['BMI_Category'] = train_data['BMI_Category'].astype('category')
test_data['BMI_Category'] = test_data['BMI_Category'].astype('category')

print("Feature construction completed")

# Step 4: Model Construction and Training
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

# Encode the target variable
label_encoder = LabelEncoder()
train_data['Heart_Disease'] = label_encoder.fit_transform(train_data['Heart_Disease'])

# Split the training data into features and target
X_train = train_data.drop(columns=['Heart_Disease', 'BMI'])
y_train = train_data['Heart_Disease']

# Initialize and train the model
model = RandomForestClassifier()
model.fit(X_train, y_train)

print("Model trained successfully")

# Step 5: Make Predictions and Compute ROC AUC
# Split the test data into features and target
X_test = test_data.drop(columns=['Heart_Disease', 'BMI'])
y_test = test_data['Heart_Disease']

# Make predictions
y_pred_proba = model.predict_proba(X_test)[:, 1]

# Compute AUC
auc = roc_auc_score(y_test, y_pred_proba)
print(f"Area under ROC curve: {auc}")
```

```


# Import necessary libraries
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score
from sklearn.preprocessing import LabelEncoder

# Function to categorize BMI
def categorize_bmi(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif 18.5 <= bmi < 25:
        return "Normal weight"
    elif 25 <= bmi < 30:
        return "Overweight"
    else:
        return "Obesity"

# Load the data
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

# Step 2: Data Cleaning
# Drop rows with missing values and duplicates
train_data = train_data.dropna().drop_duplicates()
test_data = test_data.dropna().drop_duplicates()

# Step 3: Feature Construction
# Calculate BMI
train_data['BMI'] = train_data['Weight_(kg)'] / (train_data['Height_(cm)'] / 100) ** 2
test_data['BMI'] = test_data['Weight_(kg)'] / (test_data['Height_(cm)'] / 100) ** 2

# Categorize BMI
train_data['BMI_Category'] = train_data['BMI'].apply(categorize_bmi)
test_data['BMI_Category'] = test_data['BMI'].apply(categorize_bmi)

# Encode the BMI_Category column
bmi_enc = LabelEncoder()
train_data['BMI_Category'] = bmi_enc.fit_transform(train_data['BMI_Category'])
test_data['BMI_Category'] = bmi_enc.transform(test_data['BMI_Category'])

# List of non-numeric columns to encode
non_numeric_columns = [
    'General_Health', 'Checkup', 'Exercise', 'Skin_Cancer', 'Other_Cancer',
    'Depression', 'Diabetes', 'Arthritis', 'Sex', 'Age_Category', 'Smoking_History',
    'BMI_Category'
]

# Encode non-numeric columns
label_encoders = {}
for column in non_numeric_columns:
    le = LabelEncoder()
    train_data[column] = le.fit_transform(train_data[column])
    test_data[column] = le.transform(test_data[column])
    label_encoders[column] = le

print("Feature construction and encoding completed")

# Step 4: Model Construction and Training
# Encode the target variable
label_encoder = LabelEncoder()
train_data['Heart_Disease'] = label_encoder.fit_transform(train_data['Heart_Disease'])

# Split the training data into features and target
X_train = train_data.drop(columns=['Heart_Disease'])
y_train = train_data['Heart_Disease']

# Initialize and train the model
model = RandomForestClassifier()
model.fit(X_train, y_train)

print("Model trained successfully")

# Step 5: Make Predictions and Compute ROC AUC
# Split the test data into features and target
X_test = test_data.drop(columns=['Heart_Disease'])
y_test = test_data['Heart_Disease']

# Make predictions
y_pred_proba = model.predict_proba(X_test)[:, 1]

# Compute AUC
auc = roc_auc_score(y_test, y_pred_proba)
print(f"Area under ROC curve: {auc}")

# Save the results for future reference
results = {
    'AUC': auc,
    'Train Data Shape': X_train.shape,
    'Test Data Shape': X_test.shape
}

import json
with open('model_results.json', 'w') as f:
    json.dump(results, f)

print("Results saved to model_results.json")

```
