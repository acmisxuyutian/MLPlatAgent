```

import pandas as pd

# Load training data
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/train.csv'
train_data = pd.read_csv(train_path)

# Load test data
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/test.csv'
test_data = pd.read_csv(test_path)

print(train_data.head())
print(test_data.head())
```

```


# Drop rows with missing values
train_data.dropna(inplace=True)
test_data.dropna(inplace=True)

# Drop duplicates
train_data.drop_duplicates(inplace=True)
test_data.drop_duplicates(inplace=True)

print(train_data.shape)
print(test_data.shape)

```

```


# Define the BMI categories
def categorize_bmi(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif 18.5 <= bmi < 25:
        return "Normal weight"
    elif 25 <= bmi < 30:
        return "Overweight"
    else:
        return "Obesity"

# Apply the categorization to both training and test datasets
train_data['BMI_Category'] = train_data['BMI'].apply(categorize_bmi)
test_data['BMI_Category'] = test_data['BMI'].apply(categorize_bmi)

# Convert the "BMI_Category" column to category dtype
train_data['BMI_Category'] = train_data['BMI_Category'].astype('category')
test_data['BMI_Category'] = test_data['BMI_Category'].astype('category')

print(train_data.head())
print(test_data.head())

```

```



from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

# Define the features and target variable
X_train = train_data.drop(columns=['Heart_Disease'])
y_train = train_data['Heart_Disease']

X_test = test_data.drop(columns=['Heart_Disease'])
y_test = test_data['Heart_Disease']

# Ensure the target variable is numeric
y_train = y_train.astype('int')
y_test = y_test.astype('int')

# Initialize the model
model = RandomForestClassifier(random_state=42)

# Fit the model
model.fit(X_train, y_train)

print("Model trained successfully.")


```

```



# Check the unique values in the target variable
print(train_data['Heart_Disease'].unique())
print(test_data['Heart_Disease'].unique())

# Replace 'No' with a valid integer (e.g., 0)
train_data['Heart_Disease'] = train_data['Heart_Disease'].replace('No', 0)
test_data['Heart_Disease'] = test_data['Heart_Disease'].replace('No', 0)

# Ensure the target variable is numeric
train_data['Heart_Disease'] = train_data['Heart_Disease'].astype('int')
test_data['Heart_Disease'] = test_data['Heart_Disease'].astype('int')

print(train_data['Heart_Disease'].unique())
print(test_data['Heart_Disease'].unique())

```

```

from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Identify categorical columns
categorical_columns = train_data.select_dtypes(include=['object']).columns.tolist()

# Define the transformations
preprocessor = ColumnTransformer(
    transformers=[
        ('num', 'passthrough', train_data.select_dtypes(exclude=['object']).columns),
        ('cat', OneHotEncoder(), categorical_columns)
    ]
)

# Define the pipeline with the preprocessor and the model
pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                           ('model', RandomForestClassifier(random_state=42))])

# Fit the pipeline
pipeline.fit(X_train, y_train)

print("Pipeline trained successfully.")
```
