```

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score

# Load training and test data
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/01_bank_customer_churn/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/01_bank_customer_churn/test.csv'

train = pd.read_csv(train_data_path)
test = pd.read_csv(test_data_path)
```

```


# Define the columns to drop
uninteresting_columns = ['CustomerId', 'Complain', 'Surname']

# Drop these columns from both train and test sets
train = train.drop(columns=uninteresting_columns)
test = test.drop(columns=uninteresting_columns)

```

```



# Define the columns to encode
columns_to_encode = ['Geography', 'Card Type']

# Initialize the OneHotEncoder
encoder = OneHotEncoder(sparse_output=False)

# Concatenate encoded columns with the original data
train_encoded = encoder.fit_transform(train[columns_to_encode])
train_encoded_df = pd.DataFrame(train_encoded, columns=encoder.get_feature_names_out(columns_to_encode))

# Drop the original columns from the training data
train = train.drop(columns=columns_to_encode)

# Repeat the process for the test data
test_encoded = encoder.transform(test[columns_to_encode])
test_encoded_df = pd.DataFrame(test_encoded, columns=encoder.get_feature_names_out(columns_to_encode))

# Drop the original columns from the test data
test = test.drop(columns=columns_to_encode)

# Concatenate the encoded columns back with the main dataframes
train = pd.concat([train, train_encoded_df], axis=1)
test = pd.concat([test, test_encoded_df], axis=1)


```

```


# Separate features and target variable for training set
X_train = train.drop(columns=['Exited'])
y_train = train['Exited']

# Separate features and target variable for test set
X_test = test.drop(columns=['Exited'])
y_test = test['Exited']

```

```



# Initialize the XGBoost classifier
xgb_model = XGBClassifier(use_label_encoder=False, eval_metric='logloss')

# Train the model
xgb_model.fit(X_train, y_train)


```

```



# Encode the 'Gender' column
gender_encoded = pd.get_dummies(train['Gender'], prefix='Gender')
train = pd.concat([train, gender_encoded], axis=1)
train = train.drop(columns=['Gender'])

# Repeat the process for the test data
gender_encoded_test = pd.get_dummies(test['Gender'], prefix='Gender')
test = pd.concat([test, gender_encoded_test], axis=1)
test = test.drop(columns=['Gender'])

# Ensure both train and test data have the same columns
train, test = train.align(test, fill_value=0)

# Separate features and target variable again for training and test sets
X_train = train.drop(columns=['Exited'])
y_train = train['Exited']

X_test = test.drop(columns=['Exited'])
y_test = test['Exited']


```

```




# Initialize the XGBoost classifier
xgb_model = XGBClassifier(use_label_encoder=False, eval_metric='logloss')

# Train the model
xgb_model.fit(X_train, y_train)



```

```




# Make predictions on the test set
y_pred_proba = xgb_model.predict_proba(X_test)[:, 1]

# Compute the area under ROC curve
auc = roc_auc_score(y_test, y_pred_proba)

print(f"Area under ROC curve: {auc:.4f}")

```
