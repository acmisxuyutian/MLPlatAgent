```

import pandas as pd

# Define file paths
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/01_bank_customer_churn/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/01_bank_customer_churn/test.csv'

# Load the data
train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

# Define columns to drop
columns_to_drop = ['CustomerId', 'Complain', 'Surname']

# Drop the columns
train_data = train_data.drop(columns=columns_to_drop)
test_data = test_data.drop(columns=columns_to_drop)

# Display the first few rows of the cleaned data
print(train_data.head())
print(test_data.head())
```

```


# Perform one-hot encoding
train_encoded = pd.get_dummies(train_data)
test_encoded = pd.get_dummies(test_data)

# Display the first few rows of the one-hot encoded data
print(train_encoded.head())
print(test_encoded.head())

```

```



# Install XGBoost if it's not already installed
!pip install xgboost

import xgboost as xgb
from sklearn.metrics import roc_auc_score

# Define X (features) and y (target) for training
X_train = train_encoded.drop('Exited', axis=1)
y_train = train_encoded['Exited']

# Train the XGBoost model
model = xgb.XGBClassifier(eval_metric='logloss')
model.fit(X_train, y_train)

# Display the feature importances
print(model.feature_importances_)


```

```




# Define X (features) for testing
X_test = test_encoded.drop('Exited', axis=1)
y_test = test_encoded['Exited']

# Make predictions on the test set
y_pred_prob = model.predict_proba(X_test)[:, 1]

# Display the first few predictions
print(y_pred_prob[:5])


```

```



# Compute AUC
auc = roc_auc_score(y_test, y_pred_prob)
print(f"AUC: {auc}")

```
