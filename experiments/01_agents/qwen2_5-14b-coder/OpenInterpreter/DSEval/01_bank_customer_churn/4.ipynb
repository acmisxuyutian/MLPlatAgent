```

import pandas as pd

# Load the training and test data
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/01_bank_customer_churn/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/01_bank_customer_churn/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)
```

```

# Drop the columns "CustomerId", "Complain", and "Surname"
columns_to_drop = ["CustomerId", "Complain", "Surname"]

train_data_cleaned = train_data.drop(columns=columns_to_drop)
test_data_cleaned = test_data.drop(columns=columns_to_drop)
```

```


# Perform one-hot encoding for "Geography" and "Card Type"
train_data_encoded = pd.get_dummies(train_data_cleaned, columns=["Geography", "Card Type"])
test_data_encoded = pd.get_dummies(test_data_cleaned, columns=["Geography", "Card Type"])

```

```


import xgboost as xgb
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import train_test_split

# Define the features and target variable
X_train = train_data_encoded.drop(columns=["Exited"])
y_train = train_data_encoded["Exited"]

# Define the test set
X_test = test_data_encoded.drop(columns=["Exited"])

# Train the XGBoost model
model = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss')
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_proba = model.predict_proba(X_test)[:, 1]

```

```


# Map Gender column to numerical values
gender_mapping = {'Male': 0, 'Female': 1}

X_train['Gender'] = X_train['Gender'].map(gender_mapping)
X_test['Gender'] = X_test['Gender'].map(gender_mapping)

```

```



# Train the XGBoost model
model = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss')
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_proba = model.predict_proba(X_test)[:, 1]


```

```





# Define the true labels for the test set
# Note: We need to load the true labels from the test data since they are not provided separately.
# Assuming the 'Exited' column has the true labels.
y_test = test_data['Exited']

# Compute the area under the ROC curve
auc_roc = roc_auc_score(y_test, y_pred_proba)

print(f"Area Under ROC Curve: {auc_roc}")


```
