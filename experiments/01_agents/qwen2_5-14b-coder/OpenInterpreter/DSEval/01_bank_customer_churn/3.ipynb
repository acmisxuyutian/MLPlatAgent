```

import pandas as pd

train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/01_bank_customer_churn/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/01_bank_customer_churn/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

print(train_data.head())
print(test_data.head())
```

```


# Step 2: Drop unnecessary columns
columns_to_drop = ['CustomerId', 'Complain', 'Surname']

train_data = train_data.drop(columns=columns_to_drop)
test_data = test_data.drop(columns=columns_to_drop)

print(train_data.head())
print(test_data.head())

```

```



# Step 3: One-hot encode "Geography" and "Card Type"
train_data_encoded = pd.get_dummies(train_data, columns=['Geography', 'Card Type'])
test_data_encoded = pd.get_dummies(test_data, columns=['Geography', 'Card Type'])

print(train_data_encoded.head())
print(test_data_encoded.head())


```

```






# Step 4: Separate features and target variable
X_train = train_data_encoded.drop(columns=['Exited'])
y_train = train_data_encoded['Exited']

X_test = test_data_encoded.drop(columns=['Exited'])
y_test = test_data_encoded['Exited']

print(X_train.head())
print(y_train.head())


```

```



# Step 5: Construct and train an XGBoost model
import xgboost as xgb

# Convert data to DMatrix
dtrain = xgb.DMatrix(X_train, label=y_train)
dtest = xgb.DMatrix(X_test, label=y_test)

# Define the parameters
params = {
    'max_depth': 3,
    'eta': 0.1,
    'objective': 'binary:logistic'
}

# Train the model
num_rounds = 100
bst = xgb.train(params, dtrain, num_rounds)

```

```




# Convert categorical columns to category type
train_data_encoded['Gender'] = train_data_encoded['Gender'].astype('category')
test_data_encoded['Gender'] = test_data_encoded['Gender'].astype('category')

# Step 5: Construct and train an XGBoost model
import xgboost as xgb

# Convert data to DMatrix with enable_categorical set to True
dtrain = xgb.DMatrix(X_train, label=y_train, enable_categorical=True)
dtest = xgb.DMatrix(X_test, label=y_test, enable_categorical=True)

# Define the parameters
params = {
    'max_depth': 3,
    'eta': 0.1,
    'objective': 'binary:logistic'
}

# Train the model
num_rounds = 100
bst = xgb.train(params, dtrain, num_rounds)


```

```





# Step 5: Construct and train an XGBoost model
import xgboost as xgb

# Convert data to DMatrix with enable_categorical set to True
dtrain = xgb.DMatrix(train_data_encoded.drop(columns=['Exited']), label=train_data_encoded['Exited'], enable_categorical=True)
dtest = xgb.DMatrix(test_data_encoded.drop(columns=['Exited']), label=test_data_encoded['Exited'], enable_categorical=True)

# Define the parameters
params = {
    'max_depth': 3,
    'eta': 0.1,
    'objective': 'binary:logistic'
}

# Train the model
num_rounds = 100
bst = xgb.train(params, dtrain, num_rounds)

# Step 6: Make predictions on the test set
y_pred_prob = bst.predict(dtest)

# Step 7: Compute the area under the ROC curve
from sklearn.metrics import roc_auc_score

y_test = test_data_encoded['Exited']
roc_auc = roc_auc_score(y_test, y_pred_prob)

print(f'Area under the ROC curve: {roc_auc}')




```
