```

import pandas as pd

# Load the data
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/01_bank_customer_churn/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/01_bank_customer_churn/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Drop the specified columns
columns_to_drop = ['CustomerId', 'Complain', 'Surname']
train_df = train_df.drop(columns=columns_to_drop)
test_df = test_df.drop(columns=columns_to_drop)

# Display the first few rows of the train and test data
print("Train Data Head:")
print(train_df.head())
print("\nTest Data Head:")
print(test_df.head())
```

```


# One-hot encoding for 'Geography' and 'Card Type'
train_encoded_df = pd.get_dummies(train_df, columns=['Geography', 'Card Type'])
test_encoded_df = pd.get_dummies(test_df, columns=['Geography', 'Card Type'])

# Display the first few rows of the encoded train and test data
print("Encoded Train Data Head:")
print(train_encoded_df.head())
print("\nEncoded Test Data Head:")
print(test_encoded_df.head())

# Split the data into features and target variable
X_train = train_encoded_df.drop('Exited', axis=1)
y_train = train_encoded_df['Exited']
X_test = test_encoded_df.drop('Exited', axis=1)
y_test = test_encoded_df['Exited']

# Display the first few rows of the feature and target sets
print("Feature Train Data Head:")
print(X_train.head())
print("\nTarget Train Data Head:")
print(y_train.head())

print("\nFeature Test Data Head:")
print(X_test.head())
print("\nTarget Test Data Head:")
print(y_test.head())

```

```



from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score

# Initialize the XGBoost classifier
xgb_clf = XGBClassifier(use_label_encoder=False, eval_metric='logloss')

# Train the model
xgb_clf.fit(X_train, y_train)

# Make predictions on the test set
y_pred_proba = xgb_clf.predict_proba(X_test)[:, 1]

# Compute the AUC
auc = roc_auc_score(y_test, y_pred_proba)

# Display the AUC
print(f"AUC: {auc}")


```
