```

import pandas as pd

# Load the training data
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/08_heart_attack/train.csv'
train_df = pd.read_csv(train_data_path)

# Load the test data
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/08_heart_attack/test.csv'
test_df = pd.read_csv(test_data_path)

# Map the 'Diet' column values
diet_mapping = {'Healthy': 1, 'Average': 2, 'Unhealthy': 3}
train_df['Diet'] = train_df['Diet'].map(diet_mapping)
test_df['Diet'] = test_df['Diet'].map(diet_mapping)

# Check the first few rows of the training data to ensure the transformation is correct
print(train_df.head())
```

```

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score

# Define the features and target variable
X_train = train_df.drop(columns='Heart Attack Risk')
y_train = train_df['Heart Attack Risk']
X_test = test_df.drop(columns='Heart Attack Risk')
y_test = test_df['Heart Attack Risk']

# Initialize and fit the RandomForest model
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_proba = model.predict_proba(X_test)[:, 1]

# Calculate the ROC AUC score
roc_auc = roc_auc_score(y_test, y_pred_proba)

# Return the ROC AUC score in a tuple
print((roc_auc,))
```

```




from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# Initialize the one-hot encoder
oe = OneHotEncoder(handle_unknown='ignore')

# Fit and transform the 'Continent' column for the training set
continent_encoded_train = oe.fit_transform(X_train[['Continent']]).toarray()

# Transform the 'Continent' column for the test set
continent_encoded_test = oe.transform(X_test[['Continent']]).toarray()

# Get the feature names for the encoded columns
continent_feature_names = oe.get_feature_names_out(['Continent'])

# Convert the encoded arrays to DataFrames and concatenate them with the original DataFrames
X_train_encoded = pd.concat([X_train.drop(columns=['Continent']), pd.DataFrame(continent_encoded_train, columns=continent_feature_names)], axis=1)
X_test_encoded = pd.concat([X_test.drop(columns=['Continent']), pd.DataFrame(continent_encoded_test, columns=continent_feature_names)], axis=1)

# Initialize and fit the RandomForest model
model = RandomForestClassifier(random_state=42)
model.fit(X_train_encoded, y_train)

# Make predictions on the test set
y_pred_proba = model.predict_proba(X_test_encoded)[:, 1]

# Calculate the ROC AUC score
roc_auc = roc_auc_score(y_test, y_pred_proba)

# Return the ROC AUC score in a tuple
print((roc_auc,))


```

```


from sklearn.preprocessing import LabelEncoder, OneHotEncoder
import pandas as pd

# Identify string columns in X_train and X_test
string_columns = X_train.select_dtypes(include=['object']).columns

# Create a copy of the original DataFrames to avoid modifying them
X_train_encoded = X_train.copy()
X_test_encoded = X_test.copy()

# Initialize encoders
label_encoders = {}

# Encode string columns using label encoding
for column in string_columns:
    if X_train_encoded[column].nunique() <= 2:
        # Binary columns are better encoded with LabelEncoder
        le = LabelEncoder()
        X_train_encoded[column] = le.fit_transform(X_train_encoded[column])
        X_test_encoded[column] = le.transform(X_test_encoded[column])
        label_encoders[column] = le
    else:
        # Non-binary columns are better encoded with OneHotEncoder
        oe = OneHotEncoder(handle_unknown='ignore')
        encoded_train = oe.fit_transform(X_train_encoded[[column]]).toarray()
        encoded_test = oe.transform(X_test_encoded[[column]]).toarray()
        feature_names = oe.get_feature_names_out([column])
        X_train_encoded = pd.concat([X_train_encoded.drop(columns=[column]), pd.DataFrame(encoded_train, columns=feature_names)], axis=1)
        X_test_encoded = pd.concat([X_test_encoded.drop(columns=[column]), pd.DataFrame(encoded_test, columns=feature_names)], axis=1)
        label_encoders[column] = oe

# Initialize and fit the RandomForest model
model = RandomForestClassifier(random_state=42)
model.fit(X_train_encoded, y_train)

# Make predictions on the test set
y_pred_proba = model.predict_proba(X_test_encoded)[:, 1]

# Calculate the ROC AUC score
roc_auc = roc_auc_score(y_test, y_pred_proba)

# Return the ROC AUC score in a tuple
print((roc_auc,))

```

```

# Select numeric columns
numeric_columns = X_train_encoded.select_dtypes(include=['number']).columns
X_train_selected = X_train_encoded[numeric_columns]
X_test_selected = X_test_encoded[numeric_columns]

# Initialize and fit the RandomForest model with selected features
model = RandomForestClassifier(random_state=42)
model.fit(X_train_selected, y_train)

# Make predictions on the test set
y_pred_proba = model.predict_proba(X_test_selected)[:, 1]

# Calculate the ROC AUC score
roc_auc = roc_auc_score(y_test, y_pred_proba)

# Return the ROC AUC score in a tuple
print((roc_auc,))
```

```


import matplotlib.pyplot as plt

# If the Random Forest model hasn't been re-fitted with the selected features, do so
if not hasattr(model, 'feature_importances_'):
    model.fit(X_train_selected, y_train)

# Extract feature importances
feature_importances = model.feature_importances_

# Create a DataFrame for easier analysis
importance_df = pd.DataFrame({
    'Feature': numeric_columns,
    'Importance': feature_importances
})

# Sort the DataFrame by importance
importance_df = importance_df.sort_values(by='Importance', ascending=False)

# Print the feature importances
print(importance_df)

# Plot the feature importances
plt.figure(figsize=(10, 6))
plt.barh(importance_df['Feature'], importance_df['Importance'], color='skyblue')
plt.xlabel('Feature Importance')
plt.title('Feature Importances in Random Forest Model')
plt.show()

```
