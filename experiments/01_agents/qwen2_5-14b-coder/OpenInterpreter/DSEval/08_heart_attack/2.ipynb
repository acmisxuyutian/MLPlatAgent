```

import pandas as pd

train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\08_heart_attack\\train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\08_heart_attack\\test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

print(train_data.head())
print(test_data.head())
```

```


# Mapping the 'Diet' column
diet_mapping = {
    'Healthy': 1,
    'Average': 2,
    'Unhealthy': 3
}

train_data['Diet'] = train_data['Diet'].map(diet_mapping)
test_data['Diet'] = test_data['Diet'].map(diet_mapping)

print(train_data.head())
print(test_data.head())

```

```

# Select the features and target variable
features = train_data.drop(columns=['Patient ID', 'Heart Attack Risk'])
train_target = train_data['Heart Attack Risk']
test_features = test_data.drop(columns=['Patient ID', 'Heart Attack Risk'])
test_target = test_data['Heart Attack Risk']

print(features.head())
print(train_target.head())
```

```


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

# Initialize the RandomForest model
rf_model = RandomForestClassifier(random_state=42)

# Fit the model on the training data
rf_model.fit(features, train_target)

# Make predictions on the test set
test_predictions = rf_model.predict(test_features)
test_predictions_proba = rf_model.predict_proba(test_features)[:, 1]

# Evaluate the model using ROC AUC
roc_auc = roc_auc_score(test_target, test_predictions_proba)

# Print the ROC AUC score
print(f'ROC AUC Score: {roc_auc}')

```

```



# Mapping the 'Sex' column
sex_mapping = {
    'Male': 0,
    'Female': 1
}

train_data['Sex'] = train_data['Sex'].map(sex_mapping)
test_data['Sex'] = test_data['Sex'].map(sex_mapping)

# Re-select the features to update the feature sets
features = train_data.drop(columns=['Patient ID', 'Heart Attack Risk'])
train_target = train_data['Heart Attack Risk']
test_features = test_data.drop(columns=['Patient ID', 'Heart Attack Risk'])
test_target = test_data['Heart Attack Risk']

print(features.head())
print(train_target.head())

```

```



# One-hot encode categorical columns
features = pd.get_dummies(features, columns=['Country', 'Continent', 'Hemisphere'])
test_features = pd.get_dummies(test_features, columns=['Country', 'Continent', 'Hemisphere'])

print(features.head())
print(test_features.head())


```
