```

# Step 1: Load the data
import pandas as pd

train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\08_heart_attack\\train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\08_heart_attack\\test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the dataframes
print(train_df.head())
print(test_df.head())
```

```

# Step 2: Convert category values
# Define the mapping for the Diet column
diet_mapping = {
    'Healthy': 1,
    'Average': 2,
    'Unhealthy': 3
}

# Apply the mapping to the Diet column in both train and test DataFrames
train_df['Diet'] = train_df['Diet'].map(diet_mapping)
test_df['Diet'] = test_df['Diet'].map(diet_mapping)

# Display the first few rows of the dataframes after conversion
print(train_df.head())
print(test_df.head())
```

```


# Verify the Diet column conversion by checking unique values
unique_diet_values_train = train_df['Diet'].unique()
unique_diet_values_test = test_df['Diet'].unique()

print("Unique Diet values in training set:", unique_diet_values_train)
print("Unique Diet values in test set:", unique_diet_values_test)

# Step 3: Prepare the data for model training
from sklearn.model_selection import train_test_split

# Define features and target variable
X_train = train_df.drop(columns=['Patient ID', 'Heart Attack Risk'])
y_train = train_df['Heart Attack Risk']
X_test = test_df.drop(columns=['Patient ID', 'Heart Attack Risk'])
y_test = test_df['Heart Attack Risk']

# Display the first few rows of X_train and X_test
print(X_train.head())
print(X_test.head())

```

```



# Step 4: Build and train the RandomForest model
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

# Initialize the RandomForest model
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)

# Fit the model on the training data
rf_model.fit(X_train, y_train)

# Step 5: Make predictions on the test set
y_pred_prob = rf_model.predict_proba(X_test)[:, 1]  # Probability of the positive class

# Step 6: Evaluate the model performance using ROC AUC score
roc_auc = roc_auc_score(y_test, y_pred_prob)

print(f"ROC AUC score: {roc_auc:.4f}")


```

```




# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import GridSearchCV

# Load the data
train_df = pd.read_csv('train_data.csv')
test_df = pd.read_csv('test_data.csv')

# Define the mapping for the Diet column
diet_mapping = {
    'Healthy': 1,
    'Average': 2,
    'Unhealthy': 3
}

# Apply the mapping to the Diet column in both train and test DataFrames
train_df['Diet'] = train_df['Diet'].map(diet_mapping)
test_df['Diet'] = test_df['Diet'].map(diet_mapping)

# Perform one-hot encoding for the 'Sex' column if it exists
if 'Sex' in train_df.columns:
    train_df = pd.get_dummies(train_df, columns=['Sex'], drop_first=True)
if 'Sex' in test_df.columns:
    test_df = pd.get_dummies(test_df, columns=['Sex'], drop_first=True)

# Drop the 'Hemisphere' column if it exists
if 'Hemisphere' in train_df.columns:
    train_df = train_df.drop(columns=['Hemisphere'])
if 'Hemisphere' in test_df.columns:
    test_df = test_df.drop(columns=['Hemisphere'])

# Create interaction terms
train_df['AgeDiet'] = train_df['Age'] * train_df['Diet']
test_df['AgeDiet'] = test_df['Age'] * test_df['Diet']

train_df['AgeCholesterol'] = train_df['Age'] * train_df['Cholesterol']
test_df['AgeCholesterol'] = test_df['Age'] * test_df['Cholesterol']

train_df['AgeHeartRate'] = train_df['Age'] * train_df['Heart Rate']
test_df['AgeHeartRate'] = test_df['Age'] * test_df['Heart Rate']

# Display the first few rows of the dataframes after conversion
print(train_df.head())
print(test_df.head())

# Step 3: Prepare the data for model training
# Define features and target variable
X_train = train_df.drop(columns=['Patient ID', 'Heart Attack Risk'])
y_train = train_df['Heart Attack Risk']
X_test = test_df.drop(columns=['Patient ID', 'Heart Attack Risk'])
y_test = test_df['Heart Attack Risk']

# Display the first few rows of X_train and X_test
print(X_train.head())
print(X_test.head())

# Step 4: Build and train the RandomForest model with hyperparameter tuning
# Define the hyperparameter grid
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# Initialize the GridSearchCV
grid_search = GridSearchCV(estimator=RandomForestClassifier(random_state=42), param_grid=param_grid, cv=5, scoring='roc_auc')

# Fit the GridSearchCV on the training data
grid_search.fit(X_train, y_train)

# Get the best model
best_rf_model = grid_search.best_estimator_

# Print the best hyperparameters
print("Best hyperparameters: ", grid_search.best_params_)

# Make predictions on the test set
y_pred_prob = best_rf_model.predict_proba(X_test)[:, 1]  # Probability of the positive class

# Step 6: Evaluate the model performance using ROC AUC score
roc_auc = roc_auc_score(y_test, y_pred_prob)

print(f"ROC AUC score: {roc_auc:.4f}")



```
