```

import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import train_test_split
```

```

train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/08_heart_attack/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/08_heart_attack/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)
```

```


# Map the 'Diet' values to 1, 2, 3
diet_mapping = {
    'Healthy': 1,
    'Average': 2,
    'Unhealthy': 3
}

train_df['Diet'] = train_df['Diet'].map(diet_mapping)
test_df['Diet'] = test_df['Diet'].map(diet_mapping)

```

```



# Separate features and target variable
X_train = train_df.drop('Heart Attack Risk', axis=1)
y_train = train_df['Heart Attack Risk']

# Initialize and train the RandomForest model
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)


```

```



# Display the data types of the training and test datasets
print(train_df.dtypes)
print(test_df.dtypes)

# Example: Drop or encode non-numeric columns if necessary
non_numeric_cols = train_df.select_dtypes(object).columns
print("Non-numeric columns found:", non_numeric_cols)

# For this example, let's assume we drop the non-numeric columns
train_df = train_df.drop(non_numeric_cols, axis=1)
test_df = test_df.drop(non_numeric_cols, axis=1)


```

```




# Drop the non-numeric columns
non_numeric_cols = ['Patient ID', 'Sex', 'Country', 'Continent', 'Hemisphere']
train_df = train_df.drop(non_numeric_cols, axis=1)
test_df = test_df.drop(non_numeric_cols, axis=1)



```

```




# Verify the columns in the training and test datasets
print("Training columns:", train_df.columns)
print("Test columns:", test_df.columns)

# Only drop columns that actually exist in the dataframe
valid_cols = set(train_df.columns).intersection(non_numeric_cols)
print(" Columns to drop:", valid_cols)

# Drop the valid columns
if valid_cols:
    train_df = train_df.drop(valid_cols, axis=1)
    test_df = test_df.drop(valid_cols, axis=1)



```

```




# Separate features and target variable for training and test sets
X_train = train_df.drop('Heart Attack Risk', axis=1)
y_train = train_df['Heart Attack Risk']

X_test = test_df.drop('Heart Attack Risk', axis=1)
y_test = test_df['Heart Attack Risk']



```

```





# Initialize and train the RandomForest model
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)




```

```





# Make predictions on the test data
y_pred = rf_model.predict(X_test)




```

```



from sklearn.metrics import classification_report, accuracy_score

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)

print(f"Accuracy: {accuracy:.2f}")
print("Classification Report:")
print(report)




```
