```

import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, confusion_matrix
from sklearn.model_selection import train_test_split
```

```


train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/06_employee/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/06_employee/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

print(f"Training data shape: {train_data.shape}")
print(f"Test data shape: {test_data.shape}")

```

```



# Function to calculate years in company
def calculate_years_in_company(df):
    df['YearsInCompany'] = 2023 - df['JoiningYear']
    return df

train_data = calculate_years_in_company(train_data)
test_data = calculate_years_in_company(test_data)

print(train_data.head())
print(test_data.head())


```

```


# Identify categorical columns
categorical_cols = ['Education']

# One-hot encode categorical columns for training data
encoder = OneHotEncoder(sparse=False, drop='first')
encoded_train = encoder.fit_transform(train_data[categorical_cols])
encoded_cols = encoder.get_feature_names_out(categorical_cols)

# Create a DataFrame with encoded columns and concatenate with the rest of the training data
encoded_train_df = pd.DataFrame(encoded_train, columns=encoded_cols)
train_data_encoded = pd.concat([train_data.drop(columns=categorical_cols), encoded_train_df], axis=1)

# Perform one-hot encoding for test data
encoded_test = encoder.transform(test_data[categorical_cols])
encoded_test_df = pd.DataFrame(encoded_test, columns=encoded_cols)
test_data_encoded = pd.concat([test_data.drop(columns=categorical_cols), encoded_test_df], axis=1)

print(train_data_encoded.head())
print(test_data_encoded.head())

```

```



# Define features and target variable
features = ['JoiningYear', 'City', 'YearsInCompany', 'Education_Masters', 'Education_PHD']
target = 'LeaveOrNot'

X_train = train_data_encoded[features]
y_train = train_data_encoded[target]

# Initialize the Random Forest Classifier
rf_classifier = RandomForestClassifier(random_state=42)

# Train the model
rf_classifier.fit(X_train, y_train)

print("Model training complete.")


```
