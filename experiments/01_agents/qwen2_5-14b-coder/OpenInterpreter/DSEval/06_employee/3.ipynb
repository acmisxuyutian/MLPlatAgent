````

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, confusion_matrix, classification_report
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Load the data
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark\DSEval\datasets\06_employee\train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark\DSEval\datasets\06_employee\test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)
````

````

# Feature construction: create 'YearsInCompany' feature
train_data['YearsInCompany'] = 2023 - train_data['JoiningYear']
test_data['YearsInCompany'] = 2023 - test_data['JoiningYear']

# Identify categorical columns for one-hot encoding
categorical_cols = train_data.select_dtypes(include=['object']).columns.tolist()

# Define the preprocessing for categorical variables
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ])

# Define the pipeline
pipe = Pipeline(steps=[('preprocessor', preprocessor),
                      ('classifier', RandomForestClassifier(random_state=42))])

# Fit the pipeline on the training data
pipe.fit(train_data.drop(columns=['LeaveOrNot']), train_data['LeaveOrNot'])

# Make predictions on the test set
test_predictions = pipe.predict_proba(test_data.drop(columns=['LeaveOrNot']))[:, 1]

# Compute the ROC AUC score
roc_auc = roc_auc_score(test_data['LeaveOrNot'], test_predictions)
print(f"ROC AUC Score: {roc_auc}")

# Compute the confusion matrix
conf_matrix = confusion_matrix(test_data['LeaveOrNot'], pipe.predict(test_data.drop(columns=['LeaveOrNot'])))
print("Confusion Matrix:")
print(conf_matrix)

# Classification report
class_report = classification_report(test_data['LeaveOrNot'], pipe.predict(test_data.drop(columns=['LeaveOrNot'])))
print("Classification Report:")
print(class_report)
````
