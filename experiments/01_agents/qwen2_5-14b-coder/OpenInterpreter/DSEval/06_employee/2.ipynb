```

import pandas as pd

# Read the training data
train_data = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/06_employee/train.csv')

# Read the test data
test_data = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/06_employee/test.csv')

# Display the first few rows of the training data
train_data.head()
```

```


# Create the "YearsInCompany" feature
train_data['YearsInCompany'] = 2023 - train_data['JoiningYear']
test_data['YearsInCompany'] = 2023 - test_data['JoiningYear']

# Display the first few rows of the updated training data
train_data.head()

```

```



# Encode categorical columns using one-hot encoding
train_encoded = pd.get_dummies(train_data, drop_first=True)
test_encoded = pd.get_dummies(test_data, drop_first=True)

# Display the first few rows of the encoded training data
train_encoded.head()

```

```



from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score, confusion_matrix, classification_report
import numpy as np

# Separate features and target variable
X_train = train_encoded.drop(columns='LeaveOrNot')
y_train = train_encoded['LeaveOrNot']
X_test = test_encoded.drop(columns='LeaveOrNot')
y_test = test_encoded['LeaveOrNot']

# Initialize the Random Forest Classifier
rf_classifier = RandomForestClassifier(random_state=42)

# Train the classifier
rf_classifier.fit(X_train, y_train)


```

```



# Make predictions on the test set
y_pred = rf_classifier.predict(X_test)
y_pred_proba = rf_classifier.predict_proba(X_test)[:, 1]

# Display the first few predictions
y_pred[:10]


```

```




# Compute the area under the ROC curve
roc_auc = roc_auc_score(y_test, y_pred_proba)

# Display the ROC AUC score
print(f"AUC-ROC: {roc_auc:.2f}")


```

```




# Compute the confusion matrix
cm = confusion_matrix(y_test, y_pred)

# Display the confusion matrix
print("Confusion Matrix:")
print(cm)



```
