```

import pandas as pd

# Load training data
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/06_employee/train.csv'
train_data = pd.read_csv(train_path)

# Load test data
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/06_employee/test.csv'
test_data = pd.read_csv(test_path)

print(train_data.head())
print(test_data.head())
```

```

# Feature Construction: Adding YearsInCompany
train_data['YearsInCompany'] = 2023 - train_data['JoiningYear']
test_data['YearsInCompany'] = 2023 - test_data['JoiningYear']

# Display the updated datasets
print(train_data.head())
print(test_data.head())
```

```


# Identify categorical columns
categorical_columns = ['Education']

# One-hot encode categorical columns
train_encoded = pd.get_dummies(train_data, columns=categorical_columns)
test_encoded = pd.get_dummies(test_data, columns=categorical_columns)

# Display the encoded datasets
print(train_encoded.head())
print(test_encoded.head())

```

```


from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score, confusion_matrix

# Define features and target variable
# Assuming 'LeaveOrNot' is the target column
X_train = train_encoded.drop(columns=['LeaveOrNot'])
y_train = train_encoded['LeaveOrNot']

# Initialize the Random Forest Classifier
rf_classifier = RandomForestClassifier()

# Fit the model
rf_classifier.fit(X_train, y_train)

# Display the model
print("Model trained successfully.")

```
