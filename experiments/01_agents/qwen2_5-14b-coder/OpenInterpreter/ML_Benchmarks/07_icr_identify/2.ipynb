```

import pandas as pd

# Load training data
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/ml_benchmark/07_icr_identify/split_train.csv'
train_data = pd.read_csv(train_path)

# Load evaluation data
eval_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/ml_benchmark/07_icr_identify/split_eval.csv'
eval_data = pd.read_csv(eval_path)

# Display the first few rows of the training data
print(train_data.head())
```

```

# Display basic information about the dataset
print(train_data.info())
print(eval_data.info())

# Check for missing values
print(train_data.isnull().sum())
print(eval_data.isnull().sum())

# Display summary statistics for numerical features
print(train_data.describe())

# Display summary statistics for categorical features, if any
# (This assumes that categorical columns have been identified, but they are not clearly visible)

# Pairplot of a subset of features to visualize relationships
import seaborn as sns
import matplotlib.pyplot as plt

# Select a subset of columns for pairing (for better visualization and performance)
selected_columns = list(train_data.columns[:5]) + ['Class']
sns.pairplot(train_data[selected_columns])
plt.show()
```
