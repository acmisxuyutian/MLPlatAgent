```

import os
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Define the paths to the dataset files
train_data_path = os.path.join(os.getcwd(), '..', 'datasets', 'train.csv')
eval_data_path = os.path.join(os.getcwd(), '..', 'datasets', 'eval.csv')

# Check if the dataset files exist
if not os.path.exists(train_data_path) or not os.path.exists(eval_data_path):
    print(f"Error: One or both dataset files do not exist.")
    print(f"Train CSV Path: {train_data_path}")
    print(f"Eval CSV Path: {eval_data_path}")
else:
    # Load the train and eval data
    train_data = pd.read_csv(train_data_path)
    eval_data = pd.read_csv(eval_data_path)

    # Display the first few rows of the train and eval data
    print(train_data.head())
    print(eval_data.head())

    # Display the data types of the train and eval data
    print(train_data.dtypes)
    print(eval_data.dtypes)

    # Check for missing values in the train and eval data
    print(train_data.isnull().sum())
    print(eval_data.isnull().sum())

    # Display the shape of the train and eval data
    print(train_data.shape)
    print(eval_data.shape)

    # Identify numeric and object columns in the train data
    numeric_cols = train_data.select_dtypes(include=['float64', 'int64']).columns
    object_cols = train_data.select_dtypes(include=['object']).columns

    # Display the number of numeric and object columns in the train data
    print(f"Number of numeric columns: {len(numeric_cols)}")
    print(f"Number of object columns: {len(object_cols)}")

    # Separate features and target in train data
    X_train = train_data.drop('Class', axis=1)
    y_train = train_data['Class']

    # Separate features and target in eval data
    X_eval = eval_data.drop('Class', axis=1)
    y_eval = eval_data['Class']

    # Initialize SimpleImputer for numeric columns
    imputer_numeric = SimpleImputer(strategy='median')
    numeric_cols = X_train.select_dtypes(include=['float64', 'int64']).columns
    X_train[numeric_cols] = imputer_numeric.fit_transform(X_train[numeric_cols])
    X_eval[numeric_cols] = imputer_numeric.transform(X_eval[numeric_cols])

    # Initialize SimpleImputer for object columns
    imputer_object = SimpleImputer(strategy='most_frequent')
    object_cols = X_train.select_dtypes(include=['object']).columns
    X_train[object_cols] = imputer_object.fit_transform(X_train[object_cols])
    X_eval[object_cols] = imputer_object.transform(X_eval[object_cols])

    # Encode categorical variables in both datasets
    X_train = pd.get_dummies(X_train, columns=object_cols, drop_first=True)
    X_eval = pd.get_dummies(X_eval, columns=object_cols, drop_first=True)

    # Display the first few rows of the processed train and eval data
    print(X_train.head())
    print(y_train.head())
    print(X_eval.head())
    print(y_eval.head())

    # Now proceed with building and evaluating the model
    # Initialize the logistic regression model
    model = LogisticRegression(max_iter=10000)

    # Fit the model on the training data
    model.fit(X_train, y_train)

    # Make predictions on the evaluation data
    y_pred = model.predict(X_eval)

    # Evaluate the model
    accuracy = accuracy_score(y_eval, y_pred)
    report = classification_report(y_eval, y_pred)

    print(f'Accuracy: {accuracy}')
    print(f'Classification Report:\n{report}')

```
