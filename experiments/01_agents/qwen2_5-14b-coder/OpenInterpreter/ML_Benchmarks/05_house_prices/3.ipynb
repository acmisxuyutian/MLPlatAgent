```

# Step 1: Load the datasets
import pandas as pd

train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\05_house_prices\\split_train.csv'
eval_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\05_house_prices\\split_eval.csv'

train_df = pd.read_csv(train_data_path)
eval_df = pd.read_csv(eval_data_path)

print(train_df.head())
print(eval_df.head())
```

```

# Step 2: Perform EDA

# Check for missing values
print(train_df.isnull().sum())
print(eval_df.isnull().sum())

# Visualize the distribution of SalePrice
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(train_df['SalePrice'], bins=50, kde=True)
plt.title('Distribution of SalePrice')
plt.xlabel('SalePrice')
plt.ylabel('Frequency')
plt.show()
```
