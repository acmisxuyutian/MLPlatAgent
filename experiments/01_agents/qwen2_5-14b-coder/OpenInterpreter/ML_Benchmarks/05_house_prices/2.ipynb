```

import pandas as pd

# Load the training and evaluation datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\/ml_benchmark/05_house_prices/split_train.csv'
eval_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\/ml_benchmark/05_house_prices/split_eval.csv'

train_data = pd.read_csv(train_path)
eval_data = pd.read_csv(eval_path)

print("Training data shape:", train_data.shape)
print("Evaluation data shape:", eval_data.shape)
```

```


import seaborn as sns
import matplotlib.pyplot as plt

# Display the first few rows of the training data
print(train_data.head())

# Display basic statistics
print(train_data.describe())

# Check for missing values
print(train_data.isnull().sum())

# Visualize the distribution of the target variable
plt.figure(figsize=(10, 6))
sns.histplot(train_data['SalePrice'], bins=30, kde=True)
plt.title('Distribution of SalePrice')
plt.xlabel('Sale Price')
plt.ylabel('Frequency')
plt.show()
```
