```

import pandas as pd

# Load the datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\04_titanic\\split_train.csv'
eval_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\04_titanic\\split_eval.csv'

train_data = pd.read_csv(train_path)
eval_data = pd.read_csv(eval_path)

# Display the first few rows of the datasets
print(train_data.head())
print(eval_data.head())

# Display basic statistics
print(train_data.describe())
print(eval_data.describe())

# Display information about the dataset
print(train_data.info())
print(eval_data.info())
```

```


# Fill missing values in the 'Age' column with the median
train_data['Age'].fillna(train_data['Age'].median(), inplace=True)
eval_data['Age'].fillna(eval_data['Age'].median(), inplace=True)

# Fill missing values in the 'Embarked' column with the mode
train_data['Embarked'].fillna(train_data['Embarked'].mode()[0], inplace=True)

# Drop the 'Cabin' column due to high number of missing values
train_data.drop(columns=['Cabin'], inplace=True)
eval_data.drop(columns=['Cabin'], inplace=True)

# Display the first few rows after handling missing values
print(train_data.head())
print(eval_data.head())

# Display information about the dataset after handling missing values
print(train_data.info())
print(eval_data.info())

```

```



# Encode categorical variables
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
train_data['Sex'] = le.fit_transform(train_data['Sex'])
eval_data['Sex'] = le.transform(eval_data['Sex'])

train_data['Embarked'] = le.fit_transform(train_data['Embarked'])
eval_data['Embarked'] = le.transform(eval_data['Embarked'])

# Extract titles from the 'Name' column
train_data['Title'] = train_data['Name'].str.extract(' ([A-Za-z]+)\.', expand=False)
eval_data['Title'] = eval_data['Name'].str.extract(' ([A-Za-z]+)\.', expand=False)

# Map titles to categories
title_mapping = {"Mr": 1, "Miss": 2, "Mrs": 3, "Master": 4, "Dr": 5, "Rev": 6, "Col": 7, "Major": 8, "Mlle": 9, "Countess": 10, "Ms": 11, "Lady": 12, "Jonkheer": 13, "Don": 14, "Sir": 15, "Capt": 16, "Mme": 17}
train_data['Title'] = train_data['Title'].map(title_mapping).fillna(0)
eval_data['Title'] = eval_data['Title'].map(title_mapping).fillna(0)

# Convert titles to integers
train_data['Title'] = train_data['Title'].astype(int)
eval_data['Title'] = eval_data['Title'].astype(int)

# Drop the 'Name' and 'Ticket' columns
train_data.drop(columns=['Name', 'Ticket'], inplace=True)
eval_data.drop(columns=['Name', 'Ticket'], inplace=True)

# Display the first few rows after feature engineering
print(train_data.head())
print(eval_data.head())

# Display information about the dataset after feature engineering
print(train_data.info())
print(eval_data.info())


```

```



# Split the data into features and target
X_train = train_data.drop(columns=['Survived'])
y_train = train_data['Survived']
X_eval = eval_data.drop(columns=['Survived'])
y_eval = eval_data['Survived']

# Import and train the Random Forest model
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)

# Predict on the eval dataset
y_pred = rf.predict(X_eval)

# Calculate and print the accuracy
accuracy = accuracy_score(y_eval, y_pred)
print(f'Accuracy: {accuracy:.2f}')


```
