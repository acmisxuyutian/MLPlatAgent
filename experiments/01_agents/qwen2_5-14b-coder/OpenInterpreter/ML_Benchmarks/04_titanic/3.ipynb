```



import pandas as pd
import numpy as np

# Load the datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\04_titanic\\split_train.csv'
eval_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\04_titanic\\split_eval.csv'

train_df = pd.read_csv(train_data_path)
eval_df = pd.read_csv(eval_data_path)

# Handle missing values in Age
median_age_train = train_df['Age'].median()
train_df = train_df.assign(Age=train_df['Age'].fillna(median_age_train))

median_age_eval = eval_df['Age'].median()
eval_df = eval_df.assign(Age=eval_df['Age'].fillna(median_age_eval))

# Handle missing values in Cabin by creating a new column indicating presence of a cabin
train_df = train_df.assign(Has_Cabin=train_df['Cabin'].notnull().astype(int))
eval_df = eval_df.assign(Has_Cabin=eval_df['Cabin'].notnull().astype(int))

# Drop the Cabin column as it's no longer useful
train_df = train_df.drop(columns='Cabin')
eval_df = eval_df.drop(columns='Cabin')

# Handle missing values in Embarked
mode_embarked_train = train_df['Embarked'].mode()[0]
train_df = train_df.assign(Embarked=train_df['Embarked'].fillna(mode_embarked_train))

mode_embarked_eval = eval_df['Embarked'].mode()[0]
eval_df = eval_df.assign(Embarked=eval_df['Embarked'].fillna(mode_embarked_eval))

# Check for missing values again
print(train_df.isnull().sum())
print(eval_df.isnull().sum())

# Encode categorical variables
train_df['Sex'] = train_df['Sex'].map({'female': 0, 'male': 1})
eval_df['Sex'] = eval_df['Sex'].map({'female': 0, 'male': 1})

train_df['Embarked'] = train_df['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})
eval_df['Embarked'] = eval_df['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})

# Check the first few rows of the preprocessed data
print(train_df.head())
print(eval_df.head())



```

```




# Create a new feature: FamilySize
train_df = train_df.assign(FamilySize=train_df['SibSp'] + train_df['Parch'] + 1)
eval_df = eval_df.assign(FamilySize=eval_df['SibSp'] + eval_df['Parch'] + 1)

# Drop the SibSp and Parch columns as they are no longer needed
train_df = train_df.drop(columns=['SibSp', 'Parch'])
eval_df = eval_df.drop(columns=['SibSp', 'Parch'])

# Check the first few rows of the updated data
print(train_df.head())
print(eval_df.head())


```

```





from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Define features and target variable
features = ['Pclass', 'Sex', 'Age', 'Fare', 'Embarked', 'Has_Cabin', 'FamilySize']
target = 'Survived'

X_train = train_df[features]
y_train = train_df[target]

X_eval = eval_df[features]
y_eval = eval_df[target]

# Train the logistic regression model
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

# Make predictions on the evaluation data
predictions = model.predict(X_eval)

# Evaluate the model's accuracy
accuracy = accuracy_score(y_eval, predictions)
print(f"Accuracy: {accuracy * 100:.2f}%")


```

```



from sklearn.ensemble import RandomForestClassifier

# Train the Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Make predictions on the evaluation data
rf_predictions = rf_model.predict(X_eval)

# Evaluate the model's accuracy
rf_accuracy = accuracy_score(y_eval, rf_predictions)
print(f"Random Forest Accuracy: {rf_accuracy * 100:.2f}%")



```

```




from sklearn.model_selection import GridSearchCV

# Define the parameter grid
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10]
}

# Set up GridSearchCV
rf_cv = GridSearchCV(RandomForestClassifier(random_state=42), param_grid, cv=5, scoring='accuracy', n_jobs=-1)

# Perform the grid search
rf_cv.fit(X_train, y_train)

# Best parameters and best score
print(f"Best Parameters: {rf_cv.best_params_}")
print(f"Best Cross-Validation Accuracy: {rf_cv.best_score_ * 100:.2f}%")

# Evaluate the best model on the evaluation data
best_rf_model = rf_cv.best_estimator_
best_rf_predictions = best_rf_model.predict(X_eval)
best_rf_accuracy = accuracy_score(y_eval, best_rf_predictions)
print(f"Best Model Accuracy on Eval Data: {best_rf_accuracy * 100:.2f}%")




```

```




import matplotlib.pyplot as plt
import seaborn as sns

# Get feature importances from the best model
feature_importances = best_rf_model.feature_importances_

# Create a DataFrame for easy visualization
importance_df = pd.DataFrame({
    'Feature': features,
    'Importance': feature_importances
})

# Sort the DataFrame by importance
importance_df = importance_df.sort_values(by='Importance', ascending=False)

# Plot feature importances
plt.figure(figsize=(10, 6))
sns.barplot(x='Importance', y='Feature', data=importance_df)
plt.title('Feature Importances')
plt.show()

# Print the feature importances
print(importance_df)

```
