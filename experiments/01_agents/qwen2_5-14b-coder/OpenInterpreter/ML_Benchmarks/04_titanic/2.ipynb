```

import pandas as pd

# 读取数据
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/ml_benchmark/04_titanic/split_train.csv'
eval_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/ml_benchmark/04_titanic/split_eval.csv'

# 读取数据
train_data = pd.read_csv(train_data_path)
eval_data = pd.read_csv(eval_data_path)

# 查看数据的前几行
print("训练数据前5行:")
print(train_data.head())

print("\n评估数据前5行:")
print(eval_data.head())
```

```


# 处理缺失值
# 'Age' 列缺失值可以用中位数填充
train_data['Age'].fillna(train_data['Age'].median(), inplace=True)
eval_data['Age'].fillna(eval_data['Age'].median(), inplace=True)

# 'Embarked' 列缺失值可以用众数填充
train_data['Embarked'].fillna(train_data['Embarked'].mode()[0], inplace=True)
eval_data['Embarked'].fillna(eval_data['Embarked'].mode()[0], inplace=True)

# 'Cabin' 列缺失值较多，可以直接删除
train_data.drop(columns=['Cabin'], inplace=True)
eval_data.drop(columns=['Cabin'], inplace=True)

# 'Fare' 列也有缺失值，可以用中位数填充
train_data['Fare'].fillna(train_data['Fare'].median(), inplace=True)
eval_data['Fare'].fillna(eval_data['Fare'].median(), inplace=True)

# 检查缺失值是否处理完毕
print(train_data.isnull().sum())
print(eval_data.isnull().sum())

```

```


# 处理缺失值
# 'Age' 列缺失值可以用中位数填充
train_data.loc[:, 'Age'].fillna(train_data['Age'].median(), inplace=True)
eval_data.loc[:, 'Age'].fillna(eval_data['Age'].median(), inplace=True)

# 'Embarked' 列缺失值可以用众数填充
train_data.loc[:, 'Embarked'].fillna(train_data['Embarked'].mode()[0], inplace=True)
eval_data.loc[:, 'Embarked'].fillna(eval_data['Embarked'].mode()[0], inplace=True)

# 'Cabin' 列缺失值较多，可以直接删除
train_data.drop(columns=['Cabin'], inplace=True)
eval_data.drop(columns=['Cabin'], inplace=True)

# 'Fare' 列也有缺失值，可以用中位数填充
train_data.loc[:, 'Fare'].fillna(train_data['Fare'].median(), inplace=True)
eval_data.loc[:, 'Fare'].fillna(eval_data['Fare'].median(), inplace=True)

# 检查缺失值是否处理完毕
print(train_data.isnull().sum())
print(eval_data.isnull().sum())

```

```



# 处理分类特征
# 'Sex' 和 'Embarked' 列需要转换为数值形式
train_data = pd.get_dummies(train_data, columns=['Sex', 'Embarked'], drop_first=True)
eval_data = pd.get_dummies(eval_data, columns=['Sex', 'Embarked'], drop_first=True)

# 确保训练数据和评估数据的特征一致
common_columns = train_data.columns.intersection(eval_data.columns)
train_data = train_data[common_columns]
eval_data = eval_data[common_columns]

# 查看处理后的数据
print(train_data.head())
print(eval_data.head())


```

```



# 特征选择
# 选择特征和目标变量
X_train = train_data.drop(columns=['Survived', 'PassengerId'])
y_train = train_data['Survived']
X_eval = eval_data.drop(columns=['Survived', 'PassengerId'])
y_eval = eval_data['Survived']

# 查看选择后的特征
print(X_train.head())
print(y_train.head())
print(X_eval.head())
print(y_eval.head())




```

```

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# 数据标准化
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

# 对特征进行标准化
X_train_scaled = scaler.fit_transform(X_train)
X_eval_scaled = scaler.transform(X_eval)

# 初始化随机森林分类器
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)

# 训练模型
rf_classifier.fit(X_train_scaled, y_train)

# 预测
y_pred = rf_classifier.predict(X_eval_scaled)

# 计算准确率
accuracy = accuracy_score(y_eval, y_pred)
print(f"模型在评估数据上的准确率为: {accuracy:.2f}")

```

```




# 删除 'Name' 列
X_train = X_train.drop(columns=['Name'])
X_eval = X_eval.drop(columns=['Name'])

# 数据标准化
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_eval_scaled = scaler.transform(X_eval)

# 初始化随机森林分类器
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)

# 训练模型
rf_classifier.fit(X_train_scaled, y_train)

# 预测
y_pred = rf_classifier.predict(X_eval_scaled)

# 计算准确率
accuracy = accuracy_score(y_eval, y_pred)
print(f"模型在评估数据上的准确率为: {accuracy:.2f}")


```

```





# 删除 'Ticket' 列
X_train = X_train.drop(columns=['Ticket'])
X_eval = X_eval.drop(columns=['Ticket'])

# 数据标准化
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_eval_scaled = scaler.transform(X_eval)

# 初始化随机森林分类器
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)

# 训练模型
rf_classifier.fit(X_train_scaled, y_train)

# 预测
y_pred = rf_classifier.predict(X_eval_scaled)

# 计算准确率
accuracy = accuracy_score(y_eval, y_pred)
print(f"模型在评估数据上的准确率为: {accuracy:.2f}")



```

```

from sklearn.model_selection import GridSearchCV

# 定义参数网格
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True, False]
}

# 初始化 GridSearchCV
grid_search = GridSearchCV(estimator=rf_classifier, param_grid=param_grid, cv=5, n_jobs=-1, verbose=2)

# 执行网格搜索
grid_search.fit(X_train_scaled, y_train)

# 输出最佳参数和最佳分数
print("Best Parameters:", grid_search.best_params_)
print("Best Cross-Validation Accuracy:", grid_search.best_score_)

# 使用最佳参数进行预测
best_rf = grid_search.best_estimator_
y_pred_best = best_rf.predict(X_eval_scaled)

# 计算准确率
accuracy_best = accuracy_score(y_eval, y_pred_best)
print(f"模型在评估数据上的准确率为: {accuracy_best:.2f}")

```

```


from sklearn.model_selection import cross_val_score

# 使用最佳模型进行交叉验证
cv_scores = cross_val_score(best_rf, X_train_scaled, y_train, cv=5, scoring='accuracy')

# 输出交叉验证得分
print("交叉验证得分:", cv_scores)
print("平均交叉验证准确率为: {:.2f} ± {:.2f}".format(cv_scores.mean(), cv_scores.std()))


```
