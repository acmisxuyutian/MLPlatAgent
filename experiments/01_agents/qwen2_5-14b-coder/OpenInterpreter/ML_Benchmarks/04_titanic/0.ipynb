```

import pandas as pd

# Load the datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\04_titanic\\split_train.csv'
eval_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\04_titanic\\split_eval.csv'

train_df = pd.read_csv(train_path)
eval_df = pd.read_csv(eval_path)

# Display the first few rows of the datasets
print(train_df.head())
print(eval_df.head())

# Check the data types and basic statistics
print(train_df.info())
print(train_df.describe())
print(eval_df.info())
print(eval_df.describe())

# Check for missing values
print(train_df.isnull().sum())
print(eval_df.isnull().sum())
```

```

# Impute missing values in Age with the median
train_df['Age'].fillna(train_df['Age'].median(), inplace=True)
eval_df['Age'].fillna(eval_df['Age'].median(), inplace=True)

# Drop the Cabin column due to high number of missing values
train_df.drop(columns=['Cabin'], inplace=True)
eval_df.drop(columns=['Cabin'], inplace=True)

# Display the first few rows after preprocessing
print(train_df.head())
print(eval_df.head())

# Check for missing values after preprocessing
print(train_df.isnull().sum())
print(eval_df.isnull().sum())
```

```


from sklearn.preprocessing import LabelEncoder

# Initialize LabelEncoders for categorical columns
label_encoder_sex = LabelEncoder()
label_encoder_embarked = LabelEncoder()

# Fit and transform the 'Sex' column
train_df['Sex'] = label_encoder_sex.fit_transform(train_df['Sex'])
eval_df['Sex'] = label_encoder_sex.transform(eval_df['Sex'])

# Fit and transform the 'Embarked' column
train_df['Embarked'] = label_encoder_embarked.fit_transform(train_df['Embarked'].fillna('S'))
eval_df['Embarked'] = label_encoder_embarked.transform(eval_df['Embarked'].fillna('S'))

# Display the first few rows after encoding
print(train_df.head())
print(eval_df.head())

# Check the data types after encoding
print(train_df.info())
```

```


# Create a new feature 'FamilySize' by summing SibSp and Parch
train_df['FamilySize'] = train_df['SibSp'] + train_df['Parch']
eval_df['FamilySize'] = eval_df['SibSp'] + eval_df['Parch']

# Display the first few rows after feature engineering
print(train_df.head())
print(eval_df.head())

# Check for new feature in the dataset
print(train_df.info())

```

```



from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Define features and target variables
X_train = train_df.drop(columns=['PassengerId', 'Survived', 'Name', 'Ticket'])
y_train = train_df['Survived']

# Split the training data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# Initialize the Random Forest Classifier
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)

# Train the classifier
rf_classifier.fit(X_train, y_train)

# Predict on the validation set
y_val_pred = rf_classifier.predict(X_val)

# Calculate and print the accuracy
val_accuracy = accuracy_score(y_val, y_val_pred)
print(f'Validation Accuracy: {val_accuracy:.2f}')

# Prepare the evaluation data for prediction
X_eval = eval_df.drop(columns=['PassengerId', 'Name', 'Ticket'])

# Predict on the evaluation set
y_eval_pred = rf_classifier.predict(X_eval)

# Calculate and print the accuracy on the evaluation set
eval_accuracy = accuracy_score(eval_df['Survived'], y_eval_pred)
print(f'Evaluation Accuracy: {eval_accuracy:.2f}')

```

```



from sklearn.metrics import accuracy_score

# Assuming 'eval_df' already contains the true 'Survived' labels
# We drop the unnecessary columns from 'eval_df' to match the features used in training
X_eval_full = eval_df.drop(columns=['PassengerId', 'Name', 'Ticket'])

# Make sure 'Survived' is in the correct order
y_eval_true = X_eval_full.pop('Survived')

# Predict on the evaluation set
y_eval_pred = rf_classifier.predict(X_eval_full)

# Calculate and print the accuracy on the evaluation set
eval_accuracy = accuracy_score(y_eval_true, y_eval_pred)
print(f'Evaluation Accuracy: {eval_accuracy:.2f}')


```

```




from sklearn.model_selection import GridSearchCV

# Define the parameter grid for Grid Search
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# Initialize Grid Search with Random Forest classifier
grid_search = GridSearchCV(estimator=rf_classifier, param_grid=param_grid,cv=3, n_jobs=-1, verbose=2)

# Fit Grid Search on the training data
grid_search.fit(X_train, y_train)

# Print the best parameters and the best score
print(f'Best Parameters: {grid_search.best_params_}')
print(f'Best Cross-Validation Score: {grid_search.best_score_:.2f}')

# Evaluate the best model on the evaluation data
best_rf_classifier = grid_search.best_estimator_
y_eval_pred_best = best_rf_classifier.predict(X_eval_full)
eval_accuracy_best = accuracy_score(y_eval_true, y_eval_pred_best)
print(f'Evaluation Accuracy with Best Model: {eval_accuracy_best:.2f}')



```
