```

import pandas as pd

train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/ml_benchmark/08_santander_value/split_train.csv'
eval_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/ml_benchmark/08_santander_value/split_eval.csv'

train_df = pd.read_csv(train_data_path)
eval_df = pd.read_csv(eval_data_path)

# Display the first few rows of the train and eval datasets
print(train_df.head())
print(eval_df.head())
```

```

# Get a summary of the train and eval datasets
print(train_df.info())
print(eval_df.info())

# Display basic statistics of the numeric columns
print(train_df.describe())
print(eval_df.describe())

# Check for missing values
print(train_df.isnull().sum())
print(eval_df.isnull().sum())

# Check for duplicate rows
print(train_df.duplicated().sum())
print(eval_df.duplicated().sum())

# Display the distribution of the target variable
print(train_df['target'].describe())
```

```


# Remove duplicate rows from the train dataset
train_df.drop_duplicates(inplace=True)

print(f"Number of rows after removing duplicates: {len(train_df)}")
```

```



# This step involves creating new features. Let's start with something simple:
# Calculate the sum and mean of all features for each sample

# Drop the target column for feature calculation
features_train = train_df.drop(columns=['target'])
features_eval = eval_df.drop(columns=['target'])

# Calculate sum and mean of features
train_df['feature_sum'] = features_train.sum(axis=1)
train_df['feature_mean'] = features_train.mean(axis=1)

eval_df['feature_sum'] = features_eval.sum(axis=1)
eval_df['feature_mean'] = features_eval.mean(axis=1)

# Display the first few rows of the updated datasets
print(train_df.head())
print(eval_df.head())

```

```


from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_log_error

# Split the data into features and target
X_train = train_df.drop(columns=['target'])
y_train = train_df['target']

X_eval = eval_df.drop(columns=['target'])
y_eval = eval_df['target']

# Split the training data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# Initialize the RandomForestRegressor
model = RandomForestRegressor(n_estimators=100, random_state=42)

# Train the model
model.fit(X_train, y_train)

# Make predictions on the validation set
y_pred = model.predict(X_val)

# Calculate RMSLE on the validation set
rmsle_val = mean_squared_log_error(y_val, y_pred, squared=False)
print(f"RMSLE on validation set: {rmsle_val}")

# Make predictions on the eval set
y_eval_pred = model.predict(X_eval)

# Calculate RMSLE on the eval set
rmsle_eval = mean_squared_log_error(y_eval, y_eval_pred, squared=False)
print(f"RMSLE on eval set: {rmsle_eval}")

```

```



from sklearn.model_selection import GridSearchCV

# Define the parameter grid for hyperparameter tuning
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# Initialize the RandomForestRegressor
model = RandomForestRegressor(random_state=42)

# Initialize GridSearchCV
grid_search = GridSearchCV(estimator=model, param_grid=param_grid, scoring='neg_mean_squared_log_error', cv=3, n_jobs=-1)

# Fit the grid search to the training data
grid_search.fit(X_train, y_train)

# Best parameters and best score
print("Best parameters: ", grid_search.best_params_)
print("Best RMSLE score: ", -grid_search.best_score_)

# Evaluate the best model on the validation set
best_model = grid_search.best_estimator_
y_pred_val = best_model.predict(X_val)
rmsle_val_best = np.sqrt(mean_squared_log_error(y_val, y_pred_val))
print(f'RMSLE on validation set (best model): {rmsle_val_best}')

# Evaluate the best model on the eval set
y_pred_eval_best = best_model.predict(X_eval)
rmsle_eval_best = np.sqrt(mean_squared_log_error(y_eval, y_pred_eval_best))
print(f'RMSLE on eval set (best model): {rmsle_eval_best}')


```
