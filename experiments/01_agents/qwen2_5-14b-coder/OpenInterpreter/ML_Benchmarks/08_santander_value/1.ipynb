```

import pandas as pd

# Load train data
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/ml_benchmark/08_santander_value/split_train.csv'
train_df = pd.read_csv(train_data_path)

# Load eval data
eval_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/ml_benchmark/08_santander_value/split_eval.csv'
eval_df = pd.read_csv(eval_data_path)

# Display the first few rows of the train and eval data
print(train_df.head())
print(eval_df.head())
```

```

import matplotlib.pyplot as plt
import seaborn as sns

# Display basic information about the datasets
print(train_df.info())
print(eval_df.info())

# Plot the distribution of the target variable
plt.figure(figsize=(10, 6))
sns.histplot(train_df['target'], bins=30, kde=True)
plt.title('Target Variable Distribution')
plt.xlabel('Transaction Value')
plt.ylabel('Frequency')
plt.show()

# Check for missing values
print("Train data missing values:\n", train_df.isnull().sum())
print("Eval data missing values:\n", eval_df.isnull().sum())
```
