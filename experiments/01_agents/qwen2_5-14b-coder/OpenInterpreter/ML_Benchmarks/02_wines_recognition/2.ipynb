```

from sklearn.datasets import load_wine
import pandas as pd

# Load the dataset
wine = load_wine()
X = wine.data
y = wine.target

# Create a DataFrame
df = pd.DataFrame(data=X, columns=wine.feature_names)
df['class'] = y

print(df.head())
```

```


import seaborn as sns
import matplotlib.pyplot as plt

# Pairplot of some features
sns.pairplot(df[['alcohol', 'malic_acid', 'ash', 'alcalinity_of_ash', 'magnesium', 'class']], hue='class', diag_kind='kde')
plt.show()

```
