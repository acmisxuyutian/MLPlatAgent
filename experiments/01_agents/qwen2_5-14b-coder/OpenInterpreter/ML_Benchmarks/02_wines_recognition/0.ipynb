```

from sklearn.datasets import load_wine
import pandas as pd

# Load the dataset
wine = load_wine()
data = pd.DataFrame(data=wine.data, columns=wine.feature_names)
data['target'] = wine.target
```

```


import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Perform PCA to reduce the dataset to 2 dimensions
pca = PCA(n_components=2)
data_reduced = pca.fit_transform(data.drop('target', axis=1))

# Plot the data
plt.figure(figsize=(8, 6))
scatter = plt.scatter(data_reduced[:, 0], data_reduced[:, 1], c=data['target'], cmap='viridis')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('Wine Dataset PCA')
plt.colorbar(scatter, label='Wine Class')
plt.show()

```
