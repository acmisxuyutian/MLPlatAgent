```

import os
import pandas as pd

# Normalize the file path
train_data_path = os.path.join('D:\\', 'LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark', 'ml_benchmark\\06_santander_customer\\split_train.csv')
eval_data_path = os.path.join('D:\\', 'LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark', 'ml_benchmark\\06_santander_customer\\split_eval.csv')

# Verify file existence
print(f"Train file exists: {os.path.exists(train_data_path)}")
print(f"Eval file exists: {os.path.exists(eval_data_path)}")

# Load the datasets
try:
    train_df = pd.read_csv(train_data_path)
    eval_df = pd.read_csv(eval_data_path)
    print("Successfully loaded the datasets.")
except Exception as e:
    print(f"Failed to load datasets: {e}")

# Display the first few rows of the train and eval datasets
if train_df is not None and eval_df is not None:
    print(train_df.head())
    print(eval_df.head())
```

```


import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Display basic statistics
print(train_df.describe())

# Display the distribution of the target variable
sns.countplot(x='target', data=train_df)
plt.title('Distribution of Target Variable')
plt.show()

# Display correlation heatmap
plt.figure(figsize=(12, 10))
corr = train_df.corr()
sns.heatmap(corr, annot=False, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()

```
