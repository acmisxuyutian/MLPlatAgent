```

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

# Load datasets
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/ml_benchmark/06_santander_customer/split_train.csv'
eval_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/ml_benchmark/06_santander_customer/split_eval.csv'

train_df = pd.read_csv(train_data_path)
eval_df = pd.read_csv(eval_data_path)
```

```


# Display the first few rows of the training data
print(train_df.head())

# Display some basic information about the training data
print(train_df.info())

# Check for missing values in the training data
print(train_df.isnull().sum())

# Display the target distribution
print(train_df['target'].value_counts(normalize=True))

```

```



# Separate features and target variable
X_train = train_df.drop(columns=['target'])
y_train = train_df['target']

# Normalize the features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)

# Split the training data into training and validation sets
X_train_final, X_val_final, y_train_final, y_val_final = train_test_split(X_train_scaled, y_train, test_size=0.2, random_state=42)

```

```



# Initialize the Random Forest model
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)

# Train the model
rf_model.fit(X_train_final, y_train_final)

# Make predictions on the training and validation sets
y_train_pred_proba = rf_model.predict_proba(X_train_final)[:, 1]
y_val_pred_proba = rf_model.predict_proba(X_val_final)[:, 1]

# Evaluate the model using AUC on the training and validation sets
train_auc = roc_auc_score(y_train_final, y_train_pred_proba)
val_auc = roc_auc_score(y_val_final, y_val_pred_proba)

print(f"Training AUC: {train_auc:.4f}")
print(f"Validation AUC: {val_auc:.4f}")


```

```



from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.model_selection import GridSearchCV

# Feature Selection: Select top 10 features using ANOVA F-value
selector = SelectKBest(score_func=f_classif, k=10)
X_train_selected = selector.fit_transform(X_train_scaled, y_train_final)
X_val_selected = selector.transform(X_val_final)

# Hyperparameter tuning: Define parameter grid for GridSearchCV
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2]
}

# Initialize Random Forest with GridSearchCV for hyperparameter tuning
rf_model_tuned = RandomForestClassifier(random_state=42, n_jobs=-1)
grid_search = GridSearchCV(estimator=rf_model_tuned, param_grid=param_grid, scoring='roc_auc', cv=3, n_jobs=-1)

# Fit the GridSearchCV on the training data
grid_search.fit(X_train_selected, y_train_final)

# Get the best model from GridSearchCV
best_rf_model = grid_search.best_estimator_

# Make predictions on the training and validation sets
y_train_pred_proba_tuned = best_rf_model.predict_proba(X_train_selected)[:, 1]
y_val_pred_proba_tuned = best_rf_model.predict_proba(X_val_selected)[:, 1]

# Evaluate the model using AUC on the training and validation sets
train_auc_tuned = roc_auc_score(y_train_final, y_train_pred_proba_tuned)
val_auc_tuned = roc_auc_score(y_val_final, y_val_pred_proba_tuned)

print(f"Tuned Training AUC: {train_auc_tuned:.4f}")
print(f"Tuned Validation AUC: {val_auc_tuned:.4f}")
print(f"Best Parameters: {grid_search.best_params_}")

```
