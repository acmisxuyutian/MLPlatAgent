```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
```

```

# Step 2: 加载数据
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/reservation_cancellation/train.csv'
train_df = pd.read_csv(train_data_path)

# 查看数据的前几行
print(train_df.head())
```

```

# 检查缺失值
print(train_df.isnull().sum())

# 填充缺失值
imputer = SimpleImputer(strategy='median')
train_df_imputed = pd.DataFrame(imputer.fit_transform(train_df), columns=train_df.columns)

# 编码分类变量
train_df_encoded = pd.get_dummies(train_df_imputed, drop_first=True)

# 划分特征和目标变量
X = train_df_encoded.drop('booking_status', axis=1)
y = train_df_encoded['booking_status']

# 标准化特征
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

```

```


# 拆分数据集
X_train, X_val, y_train, y_val = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

```

```


# 建立模型
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 预测验证集
y_val_pred_proba = model.predict_proba(X_val)[:, 1]


```

```



# 计算AUC-ROC分数
roc_auc = roc_auc_score(y_val, y_val_pred_proba)

# 输出结果
print(f"AUC-ROC Score: {roc_auc:.4f}")

```

```


# 打印最终结果
print(f"Final AUC-ROC Score: {roc_auc:.4f}")

```
