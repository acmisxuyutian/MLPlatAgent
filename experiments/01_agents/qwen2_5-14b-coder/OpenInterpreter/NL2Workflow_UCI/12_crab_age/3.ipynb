```

import pandas as pd

# 读取训练数据集
train_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/train.csv')

# 查看数据的前几行
train_data.head()
```

```


# 查看数据的基本信息
train_data.info()

# 查看数据的统计信息
train_data.describe()

# 查看数据的基本统计信息
train_data.describe(include='all')

# 查看数据的缺失值情况
train_data.isnull().sum()

# 查看分类变量的分布情况
train_data['Sex'].value_counts()

```

```



from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split

# 定义预处理步骤
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), ["Length", "Diameter", "Height", "Weight", "Shucked_Weight", "Viscera_Weight", "Shell_Weight"]),
        ("cat", OneHotEncoder(), ["Sex"]),
    ])

# 将预处理与模型结合
model = Pipeline(steps=[('preprocessor', preprocessor)])

# 分割数据集为训练集和验证集
X_train, X_val, y_train, y_val = train_test_split(train_data[['Sex', 'Length', 'Diameter', 'Height', 'Weight', 'Shucked_Weight', 'Viscera_Weight', 'Shell_Weight']], train_data['Age'], test_size=0.2, random_state=42)

# 数据预处理
X_train_preprocessed = model['preprocessor'].fit_transform(X_train)
X_val_preprocessed = model['preprocessor'].transform(X_val)

# 查看预处理后的数据
print(X_train_preprocessed.shape)
print(X_val_preprocessed.shape)


```

```

from sklearn.ensemble import RandomForestRegressor

# 初始化随机森林回归器
rf_regressor = RandomForestRegressor(n_estimators=100, random_state=42)

# 训练模型
rf_regressor.fit(X_train_preprocessed, y_train)

# 在验证集上进行预测
y_val_pred = rf_regressor.predict(X_val_preprocessed)

# 查看预测结果的前几行
print(y_val_pred[:10])
```

```


from sklearn.metrics import mean_absolute_error

# 计算MAE
mae = mean_absolute_error(y_val, y_val_pred)

print(f"Validation MAE: {mae}")

```
