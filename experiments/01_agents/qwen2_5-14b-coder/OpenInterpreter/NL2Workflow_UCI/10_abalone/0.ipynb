```

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_log_error
import os
```

```

# Define the file path
file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/train.csv'

# Load the data
data = pd.read_csv(file_path)

# Display the first few rows of the dataset
print(data.head())
```

```


# Check for missing values
print(data.isnull().sum())

# Convert the 'Sex' column to categorical variables
data = pd.get_dummies(data, columns=['Sex'], drop_first=True)

# Display the first few rows of the preprocessed dataset
print(data.head())

```

```


# Select features and target variable
features = data.drop(columns=['id', 'Whole weight', 'Whole weight.1', 'Whole weight.2', 'Rings'])
target = data['Rings']

# Display the first few rows of the selected features
print(features.head())
```

```


# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=42)

# Display the size of the training and testing sets
print(f"Training set size: {X_train.shape}")
print(f"Testing set size: {X_test.shape}")

```

```



# Standardize the features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Display the first few rows of the scaled features
print(X_train_scaled[:5])


```

```



# Initialize the SVR model
svr_model = SVR(kernel='rbf')

# Train the model
svr_model.fit(X_train_scaled, y_train)

# Display the training status
print("Model training completed.")


```

```




# Predict the rings (age) on the test set
y_pred = svr_model.predict(X_test_scaled)

# Ensure all predictions are positive (since Rings represent age, which cannot be negative)
y_pred = np.maximum(y_pred, 1e-6)

# Calculate RMLSE
rmlse = np.sqrt(mean_squared_log_error(y_test, y_pred))

# Display the RMLSE
print(f"RMLSE: {rmlse}")


```
