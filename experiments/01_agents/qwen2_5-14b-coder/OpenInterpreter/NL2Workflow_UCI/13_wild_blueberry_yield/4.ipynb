```

import pandas as pd
import numpy as np

# 读取训练集数据
train_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/train.csv')

# 检查缺失值
missing_values = train_data.isnull().sum()
print("Missing values in each column:\n", missing_values)

# 如果有缺失值，可以选择填充或删除
# 这里我们选择使用均值填充数字类型的缺失值
numeric_cols = train_data.select_dtypes(include=[np.number]).columns
train_data[numeric_cols] = train_data[numeric_cols].fillna(train_data[numeric_cols].mean())

# 检查分类变量并进行编码
categorical_cols = train_data.select_dtypes(include=['object']).columns
for col in categorical_cols:
    train_data[col] = train_data[col].astype('category').cat.codes

# 查看处理后的数据
train_data.head()
```
