```


import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import median_absolute_error

# Step 1: 数据加载
train_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/mohs_hardness/train.csv'
train_data = pd.read_csv(train_file_path)

# Step 2: 数据预处理
# 查看数据的基本信息
print(train_data.info())
print(train_data.describe())

# 检查缺失值
print(train_data.isnull().sum())

# 查看所有列名
print(train_data.columns)

# 选择特征和目标变量
features = train_data.drop('Hardness', axis=1)
target = train_data['Hardness']

# 将数据拆分为训练集和验证集
X_train, X_val, y_train, y_val = train_test_split(features, target, test_size=0.2, random_state=42)

# 特征标准化
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)

# Step 3: 模型选择
model = RandomForestRegressor(random_state=42)

# Step 4: 模型训练
model.fit(X_train_scaled, y_train)

# Step 5: 模型评估
y_val_pred = model.predict(X_val_scaled)
medae = median_absolute_error(y_val, y_val_pred)

# 输出模型评估结果
print(f"Model MEDAE on validation set: {medae}")

```
