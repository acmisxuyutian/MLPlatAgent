```

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import median_absolute_error
```

```


train_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/mohs_hardness/train.csv')
print(train_data.head())

```

```


# Drop the 'id' column as it's not needed for modeling
train_data = train_data.drop(columns=['id'])

# Check for missing values
print(train_data.isnull().sum())

# If there are missing values, we can fill them or drop them
# Here, we'll drop them for simplicity
train_data = train_data.dropna()

```

```

# Split the data into features (X) and target (y)
X = train_data.drop(columns=['Hardness'])
y = train_data['Hardness']

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert the scaled features back to a DataFrame for easier manipulation
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

# Display the first few rows of the scaled data
print(X_scaled_df.head())


```

```



# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled_df, y, test_size=0.2, random_state=42)

# Display the shapes of the split data
print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)


```

```



# Initialize the RandomForestRegressor
model = RandomForestRegressor(n_estimators=100, random_state=42)

# Train the model on the training data
model.fit(X_train, y_train)

# Display the model's performance on the training data
print("Model score on training data:", model.score(X_train, y_train))


```

```


# Predict the hardness on the testing data
y_pred = model.predict(X_test)

# Calculate the Median Absolute Error (MAE)
mae = median_absolute_error(y_test, y_pred)
print("Median Absolute Error on the testing data:", mae)


```
