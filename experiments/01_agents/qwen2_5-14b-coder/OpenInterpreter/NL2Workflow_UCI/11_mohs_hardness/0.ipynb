```

# 导入必要的库
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

# 读取数据
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/mohs_hardness/train.csv'
data = pd.read_csv(train_data_path)

# 检查数据的基本信息
print("数据的前五行：")
print(data.head())

# 处理缺失值（如果有）
data = data.dropna()

# 分离特征和目标变量
features = data.drop(columns=['mohs_hardness'])
target = data['mohs_hardness']

# 划分数据集
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=42)

# 选择和训练回归模型
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 在测试集上进行预测
y_pred = model.predict(X_test)

# 计算均方根误差（MRAE）
mrae = mean_absolute_error(y_test, y_pred)
print(f"测试集上的均方根误差（MRAE）: {mrae:.4f}")
```
