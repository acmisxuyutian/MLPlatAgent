```
import pandas as pd

# Load the training and test datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/01_bank_customer_churn/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/01_bank_customer_churn/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the datasets to understand their structure
print("Training Data:")
print(train_df.head())
print("\nTest Data:")
print(test_df.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Check column information for the training data
column_info_train = get_column_info(train_df)
print("Column Information for Training Data:")
print(column_info_train)

# Check column information for the test data
column_info_test = get_column_info(test_df)
print("\nColumn Information for Test Data:")
print(column_info_test)
```

```
# Drop the columns 'CustomerId', 'Complain', and 'Surname' from both datasets
columns_to_drop = ['CustomerId', 'Complain', 'Surname']

# Copy the DataFrames before processing
train_df_processed = train_df.copy()
test_df_processed = test_df.copy()

# Drop the specified columns
train_df_processed = train_df_processed.drop(columns=columns_to_drop)
test_df_processed = test_df_processed.drop(columns=columns_to_drop)

# Display the processed DataFrames
print("Processed Training Data:")
print(train_df_processed.head())
print("\nProcessed Test Data:")
print(test_df_processed.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

column_info = get_column_info(train_df_processed)
print("column_info")
print(column_info)
```

```
from metagpt.tools.libs.data_preprocess import OneHotEncode

# Initialize the OneHotEncode tool for the specified columns
one_hot_encoder = OneHotEncode(features=['Geography', 'Card Type'])

# Fit and transform the training data
train_df_processed = one_hot_encoder.fit_transform(train_df_processed)

# Transform the test data using the same encoder
test_df_processed = one_hot_encoder.transform(test_df_processed)

print("Processed Training Data after One-Hot Encoding:")
print(train_df_processed.head())
print("\nProcessed Test Data after One-Hot Encoding:")
print(test_df_processed.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

column_info = get_column_info(train_df_processed)
print("column_info")
print(column_info)
```

```
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score

# Separate features and target variable
X_train = train_df_processed.drop(columns=['Exited'])
y_train = train_df_processed['Exited']

# Split the training data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# Convert data to DMatrix format for XGBoost
dtrain = xgb.DMatrix(X_train, label=y_train)
dval = xgb.DMatrix(X_val, label=y_val)

# Define hyperparameters
params = {
    'objective': 'binary:logistic',
    'eval_metric': 'auc',
    'max_depth': 6,
    'eta': 0.1,
    'subsample': 0.8,
    'colsample_bytree': 0.8,
    'seed': 42
}

# Train the XGBoost model
num_rounds = 100
bst = xgb.train(params, dtrain, num_rounds, evals=[(dval, 'eval')], early_stopping_rounds=10)

# Make predictions on the test set
dtest = xgb.DMatrix(test_df_processed)
y_pred_proba = bst.predict(dtest)

# Compute the area under the ROC curve
roc_auc = roc_auc_score(test_df_processed['Exited'], y_pred_proba)
print(f"Area under ROC curve: {roc_auc}")
```
