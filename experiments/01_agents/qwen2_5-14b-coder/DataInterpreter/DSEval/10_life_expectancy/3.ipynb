```
import pandas as pd

# Load the training and test datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/10_life_expectancy/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/10_life_expectancy/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the datasets to understand their structure
print("Training Data:")
print(train_df.head())
print("\nTest Data:")
print(test_df.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Check column information for the training data
column_info_train = get_column_info(train_df)
print("Column Information for Training Data:")
print(column_info_train)

# Check column information for the test data
column_info_test = get_column_info(test_df)
print("\nColumn Information for Test Data:")
print(column_info_test)
```

```
from metagpt.tools.libs.feature_engineering import SplitBins

# Define the GDP bins
gdp_bins = SplitBins(cols=['GDP'], strategy='custom', bins=[0, 1000, 10000, float('inf')], labels=['Low GDP', 'Medium GDP', 'High GDP'])

# Fit and transform the training data
train_df = gdp_bins.fit_transform(train_df.copy())

# Transform the test data using the same binning strategy
test_df = gdp_bins.transform(test_df.copy())

# Display the updated training and test data
print("Training Data with GDP_bins:")
print(train_df.head())
print("\nTest Data with GDP_bins:")
print(test_df.head())
```

```
from metagpt.tools.libs.feature_engineering import SplitBins

# Define the GDP bins using the 'quantile' strategy
# This will automatically create bins based on the specified quantiles
# We will then map these bins to the desired labels

gdp_bins = SplitBins(cols=['GDP'], strategy='quantile', q=[0, 0.25, 0.75, 1])

# Fit and transform the training data
temp_train_df = gdp_bins.fit_transform(train_df.copy())

# Transform the test data using the same binning strategy
temp_test_df = gdp_bins.transform(test_df.copy())

# Map the bin identifiers to the desired labels
bin_labels = ['Low GDP', 'Medium GDP', 'High GDP']
train_df['GDP_bins'] = temp_train_df['GDP'].map({0: bin_labels[0], 1: bin_labels[1], 2: bin_labels[2]})
test_df['GDP_bins'] = temp_test_df['GDP'].map({0: bin_labels[0], 1: bin_labels[1], 2: bin_labels[2]})

# Display the updated training and test data
print("Training Data with GDP_bins:")
print(train_df.head())
print("\nTest Data with GDP_bins:")
print(test_df.head())
```

```
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Assuming the data is already loaded and feature construction is done
# train_df and test_df are already defined with 'GDP_bins' feature

# Label encode categorical columns
label_encoders = {}
for col in ['Country', 'Status', 'GDP_bins']:
    le = LabelEncoder()
    train_df[col] = le.fit_transform(train_df[col])
    test_df[col] = le.transform(test_df[col])
    label_encoders[col] = le

# Define features and target
features = [col for col in train_df.columns if col != 'Life expectancy ']
X_train = train_df[features]
y_train = train_df['Life expectancy ']

X_test = test_df[features]
y_test = test_df['Life expectancy']

# Initialize and train the Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model using R-squared score
r2 = r2_score(y_test, y_pred)
print(f"R-squared score: {r2}")
```
