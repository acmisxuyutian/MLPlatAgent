```
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import LabelEncoder
import numpy as np

# Load the data
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/09_housing_price/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/09_housing_price/test.csv'
housing_price_prediction_train = pd.read_csv(train_path)
housing_price_prediction_test = pd.read_csv(test_path)

# Separate features and target
X_train = housing_price_prediction_train.drop('price', axis=1)
y_train = housing_price_prediction_train['price']
X_test = housing_price_prediction_test.drop('price', axis=1)
y_test = housing_price_prediction_test['price']

# Label encode categorical columns
label_encoders = {}
for col in ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea', 'furnishingstatus']:
    le = LabelEncoder()
    X_train[col] = le.fit_transform(X_train[col])
    X_test[col] = le.transform(X_test[col])
    label_encoders[col] = le

# Initialize and fit the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model using RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print(f"RMSE: {rmse}")
```

```
# Ensure the test data is processed in the same way as the training data
for col in ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea', 'furnishingstatus']:
    X_test[col] = label_encoders[col].transform(X_test[col])

# Make predictions on the test set using the trained linear regression model
y_pred = model.predict(X_test)

# Print the predictions
print("Predictions on the test set:")
print(y_pred)
```
