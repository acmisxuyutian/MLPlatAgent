```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/05_patient_profile/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/05_patient_profile/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the datasets
print("Training Data:")
print(train_df.head())
print("\nTest Data:")
print(test_df.head())

# Check the data types of the columns
print("\nData Types - Training Data:")
print(train_df.dtypes)
print("\nData Types - Test Data:")
print(test_df.dtypes)

# Distinguish column types
categorical_cols = train_df.select_dtypes(include=['object', 'category']).columns
numerical_cols = train_df.select_dtypes(include=['int64', 'float64']).columns

print("\nCategorical Columns:", categorical_cols)
print("Numerical Columns:", numerical_cols)

# Summary statistics for numerical columns
print("\nSummary Statistics - Numerical Columns:")
print(train_df[numerical_cols].describe())

# Correlation matrix for numerical columns
plt.figure(figsize=(10, 8))
correlation_matrix = train_df[numerical_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Numerical Features')
plt.show()

# Distribution of the target variable
plt.figure(figsize=(8, 6))
sns.countplot(x=train_df['outcome variable'])
plt.title('Distribution of Outcome Variable')
plt.show()

# Check for missing values
print("\nMissing Values - Training Data:")
print(train_df.isnull().sum())
print("\nMissing Values - Test Data:")
print(test_df.isnull().sum())
```
