```
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score
from metagpt.tools.libs.feature_engineering import KFoldTargetMeanEncoder, GeneralSelection
from metagpt.tools.libs.data_preprocess import RobustScale, OneHotEncode

# Load the cleaned data
train_df_cleaned = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/train.csv')
test_df_cleaned = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/test.csv')

# Drop rows with missing values and duplicates
train_df_cleaned = train_df_cleaned.dropna().drop_duplicates()
test_df_cleaned = test_df_cleaned.dropna().drop_duplicates()

# Categorize BMI
def categorize_bmi(bmi):
    if bmi < 18.5:
        return 'Underweight'
    elif 18.5 <= bmi < 25:
        return 'Normal weight'
    elif 25 <= bmi < 30:
        return 'Overweight'
    else:
        return 'Obesity'

train_df_cleaned['BMI_Category'] = train_df_cleaned['BMI'].apply(categorize_bmi)
test_df_cleaned['BMI_Category'] = test_df_cleaned['BMI'].apply(categorize_bmi)
train_df_cleaned['BMI_Category'] = train_df_cleaned['BMI_Category'].astype('category')
test_df_cleaned['BMI_Category'] = test_df_cleaned['BMI_Category'].astype('category')

# Feature engineering
# Label encode categorical columns
label_encoders = {}
for col in ['General_Health', 'Checkup', 'Exercise', 'Skin_Cancer', 'Other_Cancer', 'Depression', 'Diabetes', 'Arthritis', 'Sex', 'Age_Category', 'Smoking_History', 'BMI_Category']:
    le = LabelEncoder()
    train_df_cleaned[col] = le.fit_transform(train_df_cleaned[col])
    test_df_cleaned[col] = le.transform(test_df_cleaned[col])
    label_encoders[col] = le

# K-fold target mean encoding for categorical features
kfold_encoder = KFoldTargetMeanEncoder(col='General_Health', label='Heart_Disease')
train_df_cleaned = kfold_encoder.fit_transform(train_df_cleaned)
test_df_cleaned = kfold_encoder.transform(test_df_cleaned)

# Robust scaling for numeric features
numeric_features = ['Height_(cm)', 'Weight_(kg)', 'BMI', 'Alcohol_Consumption', 'Fruit_Consumption', 'Green_Vegetables_Consumption', 'FriedPotato_Consumption']
robust_scaler = RobustScale(features=numeric_features)
train_df_cleaned = robust_scaler.fit_transform(train_df_cleaned)
test_df_cleaned = robust_scaler.transform(test_df_cleaned)

# One-hot encoding for categorical features
categorical_features = ['General_Health', 'Checkup', 'Exercise', 'Skin_Cancer', 'Other_Cancer', 'Depression', 'Diabetes', 'Arthritis', 'Sex', 'Age_Category', 'Smoking_History', 'BMI_Category']
one_hot_encoder = OneHotEncode(features=categorical_features)
train_df_cleaned = one_hot_encoder.fit_transform(train_df_cleaned)
test_df_cleaned = one_hot_encoder.transform(test_df_cleaned)

# Drop features with only one unique value
general_selection = GeneralSelection(label_col='Heart_Disease')
train_df_cleaned = general_selection.fit_transform(train_df_cleaned)
test_df_cleaned = general_selection.transform(test_df_cleaned)

# Split the training data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(train_df_cleaned.drop('Heart_Disease', axis=1), train_df_cleaned['Heart_Disease'], test_size=0.2, random_state=42)

# Train a Random Forest model
model = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42)
model.fit(X_train, y_train)

# Make predictions on the validation set
y_val_pred = model.predict_proba(X_val)[:, 1]

# Compute the area under the ROC curve
roc_auc = roc_auc_score(y_val, y_val_pred)
print(f'Validation ROC AUC: {roc_auc}')

# Make predictions on the test set
test_predictions = model.predict_proba(test_df_cleaned.drop('Heart_Disease', axis=1))[:, 1]
```
