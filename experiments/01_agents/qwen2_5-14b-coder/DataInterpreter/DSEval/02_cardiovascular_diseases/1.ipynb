```
import pandas as pd

# Load the training and test datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the datasets to understand their structure
print("Training Data Head:")
print(train_df.head())
print("\nTest Data Head:")
print(test_df.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Check column information for the training data
column_info_train = get_column_info(train_df)
print("Training Data Column Info:")
print(column_info_train)

# Check column information for the test data
column_info_test = get_column_info(test_df)
print("\nTest Data Column Info:")
print(column_info_test)
```

```
import pandas as pd

# Drop rows with missing values and duplicates from the training dataset
train_df_cleaned = train_df.dropna().drop_duplicates()

# Drop rows with missing values and duplicates from the test dataset
test_df_cleaned = test_df.dropna().drop_duplicates()

# Function to categorize BMI
def categorize_bmi(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif 18.5 <= bmi < 25:
        return "Normal weight"
    elif 25 <= bmi < 30:
        return "Overweight"
    else:
        return "Obesity"

# Apply the function to create the BMI_Category feature in the training dataset
train_df_cleaned['BMI_Category'] = train_df_cleaned['BMI'].apply(categorize_bmi)
train_df_cleaned['BMI_Category'] = train_df_cleaned['BMI_Category'].astype('category')

# Apply the function to create the BMI_Category feature in the test dataset
test_df_cleaned['BMI_Category'] = test_df_cleaned['BMI'].apply(categorize_bmi)
test_df_cleaned['BMI_Category'] = test_df_cleaned['BMI_Category'].astype('category')

# Display the first few rows of the cleaned training dataset
print("Cleaned Training Data Head:")
print(train_df_cleaned.head())

# Display the first few rows of the cleaned test dataset
print("\nCleaned Test Data Head:")
print(test_df_cleaned.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

column_info = get_column_info(train_df_cleaned)
print("column_info")
print(column_info)
```

```
# Feature Engineering: Create new features based on existing ones

# Copy the DataFrames to avoid modifying the original data
train_df_processed = train_df_cleaned.copy()
test_df_processed = test_df_cleaned.copy()

# Example of creating a new feature: Age in years (assuming Age_Category is in years)
# This is just an example, you can create more features based on domain knowledge
def extract_age(age_category):
    # Assuming Age_Category is already in a format that can be directly used
    return age_category

train_df_processed['Age'] = train_df_processed['Age_Category'].apply(extract_age)
test_df_processed['Age'] = test_df_processed['Age_Category'].apply(extract_age)

# Example of creating a new feature: Total Vegetable Consumption
train_df_processed['Total_Vegetable_Consumption'] = train_df_processed['Green_Vegetables_Consumption']
test_df_processed['Total_Vegetable_Consumption'] = test_df_processed['Green_Vegetables_Consumption']

# Example of creating a new feature: Alcohol Consumption per day
train_df_processed['Alcohol_Consumption_Per_Day'] = train_df_processed['Alcohol_Consumption']
test_df_processed['Alcohol_Consumption_Per_Day'] = test_df_processed['Alcohol_Consumption']

# Print the head of the processed DataFrames to verify new features
print("Processed Training Data Head:")
print(train_df_processed.head())
print("\nProcessed Test Data Head:")
print(test_df_processed.head())
```

```
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score
from metagpt.tools.libs.feature_engineering import KFoldTargetMeanEncoder
from metagpt.tools.libs.data_preprocess import RobustScale, OneHotEncode

# Load the processed data
train_df_processed = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/train_processed.csv')
test_df_processed = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/test_processed.csv')

# Separate features and target
X_train = train_df_processed.drop(columns=['Heart_Disease'])
y_train = train_df_processed['Heart_Disease']
X_test = test_df_processed.drop(columns=['Heart_Disease'])

# Encode categorical features
categorical_features = ['General_Health', 'Checkup', 'Exercise', 'Skin_Cancer', 'Other_Cancer', 'Depression', 'Diabetes', 'Arthritis', 'Sex', 'Age_Category', 'Smoking_History', 'BMI_Category']
label_encoders = {}

for feature in categorical_features:
    le = LabelEncoder()
    X_train[feature] = le.fit_transform(X_train[feature])
    X_test[feature] = le.transform(X_test[feature])
    label_encoders[feature] = le

# Apply RobustScaler
numeric_features = ['Height_(cm)', 'Weight_(kg)', 'BMI', 'Alcohol_Consumption', 'Fruit_Consumption', 'Green_Vegetables_Consumption', 'FriedPotato_Consumption', 'Total_Vegetable_Consumption', 'Alcohol_Consumption_Per_Day', 'Age']
scaler = RobustScale(features=numeric_features)
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Train a RandomForestClassifier
model = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42)
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_proba = model.predict_proba(X_test)[:, 1]

# Compute the area under the ROC curve
roc_auc = roc_auc_score(test_df_processed['Heart_Disease'], y_pred_proba)
print(f"Area under ROC curve: {roc_auc}")
```
