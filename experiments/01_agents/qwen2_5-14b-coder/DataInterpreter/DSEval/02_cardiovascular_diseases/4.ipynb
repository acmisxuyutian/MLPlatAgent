```
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score
from metagpt.tools.libs.feature_engineering import KFoldTargetMeanEncoder
from metagpt.tools.libs.data_preprocess import RobustScale, OneHotEncode

# Load the cleaned data
train_df_cleaned = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/train.csv').dropna().drop_duplicates()
test_df_cleaned = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/test.csv').dropna().drop_duplicates()

# Feature construction: BMI Category
def categorize_bmi(bmi):
    if bmi < 18.5:
        return 'Underweight'
    elif 18.5 <= bmi < 25:
        return 'Normal weight'
    elif 25 <= bmi < 30:
        return 'Overweight'
    else:
        return 'Obesity'

train_df_cleaned['BMI_Category'] = train_df_cleaned['BMI'].apply(categorize_bmi)
test_df_cleaned['BMI_Category'] = test_df_cleaned['BMI'].apply(categorize_bmi)
train_df_cleaned['BMI_Category'] = train_df_cleaned['BMI_Category'].astype('category')
test_df_cleaned['BMI_Category'] = test_df_cleaned['BMI_Category'].astype('category')

# Label encoding for categorical columns
label_encoders = {}
for col in ['General_Health', 'Checkup', 'Exercise', 'Skin_Cancer', 'Other_Cancer', 'Depression', 'Diabetes', 'Arthritis', 'Sex', 'Age_Category', 'Smoking_History', 'BMI_Category']:
    le = LabelEncoder()
    train_df_cleaned[col] = le.fit_transform(train_df_cleaned[col])
    test_df_cleaned[col] = le.transform(test_df_cleaned[col])
    label_encoders[col] = le

# Define features and target
X_train = train_df_cleaned.drop(columns=['Heart_Disease'])
y_train = train_df_cleaned['Heart_Disease']
X_test = test_df_cleaned.drop(columns=['Heart_Disease'])

# Feature engineering using KFoldTargetMeanEncoder
encoder = KFoldTargetMeanEncoder(col='BMI_Category', label='Heart_Disease')
X_train = encoder.fit_transform(X_train)
X_test = encoder.transform(X_test)

# Robust scaling for numeric features
scaler = RobustScale(features=['Height_(cm)', 'Weight_(kg)', 'BMI', 'Alcohol_Consumption', 'Fruit_Consumption', 'Green_Vegetables_Consumption', 'FriedPotato_Consumption'])
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# One-hot encoding for categorical features
onehot_encoder = OneHotEncode(features=['General_Health', 'Checkup', 'Exercise', 'Skin_Cancer', 'Other_Cancer', 'Depression', 'Diabetes', 'Arthritis', 'Sex', 'Age_Category', 'Smoking_History'])
X_train = onehot_encoder.fit_transform(X_train)
X_test = onehot_encoder.transform(X_test)

# Train a RandomForestClassifier
model = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42)
model.fit(X_train, y_train)

# Predict on the test set
y_pred_proba = model.predict_proba(X_test)[:, 1]

# Compute the area under the ROC curve
roc_auc = roc_auc_score(test_df_cleaned['Heart_Disease'], y_pred_proba)
print(f"Area under ROC curve: {roc_auc}")
```
