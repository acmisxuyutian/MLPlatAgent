```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the training and testing datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and testing datasets
print("Training Data Sample:")
print(train_df.head())
print("\nTesting Data Sample:")
print(test_df.head())

# Display the basic information of the datasets
print("\nTraining Data Info:")
print(train_df.info())
print("\nTesting Data Info:")
print(test_df.info())

# Display the summary statistics of the datasets
print("\nTraining Data Summary Statistics:")
print(train_df.describe())
print("\nTesting Data Summary Statistics:")
print(test_df.describe())

# Distinguish column types
numeric_cols = train_df.select_dtypes(include=[np.number]).columns
categorical_cols = train_df.select_dtypes(include=['object']).columns

print("\nNumeric Columns:", numeric_cols)
print("Categorical Columns:", categorical_cols)

# Visualize the distribution of numeric columns
plt.figure(figsize=(15, 10))
for i, col in enumerate(numeric_cols):
    plt.subplot(3, 3, i + 1)
    sns.histplot(train_df[col], kde=True)
    plt.title(col)
plt.tight_layout()
plt.show()

# Visualize the distribution of categorical columns
plt.figure(figsize=(15, 10))
for i, col in enumerate(categorical_cols):
    plt.subplot(3, 3, i + 1)
    sns.countplot(data=train_df, x=col)
    plt.title(col)
plt.tight_layout()
plt.show()

# Correlation matrix for numeric columns
plt.figure(figsize=(12, 8))
sns.heatmap(train_df[numeric_cols].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Numeric Columns')
plt.show()
```
