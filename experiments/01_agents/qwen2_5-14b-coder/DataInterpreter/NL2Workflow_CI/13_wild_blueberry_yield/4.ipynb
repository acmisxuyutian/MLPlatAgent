```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the datasets
print("Training Data:")
print(train_df.head())
print("\nTesting Data:")
print(test_df.head())

# Check the shape of the datasets
print("\nTraining Data Shape:", train_df.shape)
print("Testing Data Shape:", test_df.shape)

# Check for missing values
print("\nMissing values in Training Data:")
print(train_df.isnull().sum())
print("\nMissing values in Testing Data:")
print(test_df.isnull().sum())

# Check data types
print("\nData types in Training Data:")
print(train_df.dtypes)
print("\nData types in Testing Data:")
print(test_df.dtypes)

# Descriptive statistics
print("\nDescriptive statistics for Training Data:")
print(train_df.describe())

# Visualize the distribution of the target variable
plt.figure(figsize=(10, 6))
sns.histplot(train_df['yield'], bins=30, kde=True)
plt.title('Distribution of Wild Blueberry Yield')
plt.xlabel('Yield')
plt.ylabel('Frequency')
plt.show()

# Correlation matrix
plt.figure(figsize=(12, 8))
correlation_matrix = train_df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()

# Distinguish column types
numeric_cols = train_df.select_dtypes(include=[np.number]).columns
categorical_cols = train_df.select_dtypes(include=['object']).columns

print("\nNumeric Columns:", numeric_cols)
print("Categorical Columns:", categorical_cols)

# Visualize the distribution of numeric features
train_df[numeric_cols].hist(bins=30, figsize=(15, 10))
plt.suptitle('Distribution of Numeric Features')
plt.show()

# Visualize the distribution of categorical features
for col in categorical_cols:
    plt.figure(figsize=(10, 6))
    sns.countplot(data=train_df, x=col)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.show()
```

```
import pandas as pd
import numpy as np
from metagpt.tools.libs.feature_engineering import PolynomialExpansion, CatCross, VarianceBasedSelection

# Load the data
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/test.csv'
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Remove ID columns
train_df = train_df.drop(columns=['id'])
test_df = test_df.drop(columns=['id'])

# Feature Construction
# Polynomial expansion for numeric features
numeric_cols = train_df.select_dtypes(include=[np.number]).columns.tolist()
poly_expansion = PolynomialExpansion(cols=numeric_cols, label_col='yield', degree=2)
train_df = poly_expansion.fit_transform(train_df)
test_df = poly_expansion.transform(test_df)

# Feature Selection
# Variance-based selection
variance_selection = VarianceBasedSelection(label_col='yield', threshold=0.01)
train_df = variance_selection.fit_transform(train_df)
test_df = variance_selection.transform(test_df)

# Display the transformed data
print("Training Data after Feature Engineering:")
print(train_df.head())
print("\nTesting Data after Feature Engineering:")
print(test_df.head())

# Save the processed data for further use
train_df.to_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\processed/train_processed.csv', index=False)
test_df.to_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\processed/test_processed.csv', index=False)
```
