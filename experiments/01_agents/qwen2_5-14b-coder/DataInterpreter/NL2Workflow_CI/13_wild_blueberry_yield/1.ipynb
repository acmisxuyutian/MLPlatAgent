```
import pandas as pd
import numpy as np
from metagpt.tools.libs.feature_engineering import PolynomialExpansion, CatCross, VarianceBasedSelection

# Load the data
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/test.csv'
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Remove ID columns
train_df = train_df.drop(columns=['id'])
test_df = test_df.drop(columns=['id'])

# Feature Construction
# Polynomial expansion for numeric features
numeric_cols = train_df.select_dtypes(include=[np.number]).columns.tolist()
poly_expansion = PolynomialExpansion(cols=numeric_cols, label_col='yield', degree=2)
train_df = poly_expansion.fit_transform(train_df)
test_df = poly_expansion.transform(test_df)

# Feature Selection
# Variance-based selection
variance_selection = VarianceBasedSelection(label_col='yield', threshold=0.01)
train_df = variance_selection.fit_transform(train_df)
test_df = variance_selection.transform(test_df)

# Display the transformed data
print("Training Data after Feature Engineering:")
print(train_df.head())
print("\nTesting Data after Feature Engineering:")
print(test_df.head())
print("\nTraining Data Shape after Feature Engineering:", train_df.shape)
print("Testing Data Shape after Feature Engineering:", test_df.shape)
```

```
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Assuming train_df and test_df are already loaded and preprocessed from previous steps

# Separate features and target variable
X_train = train_df.drop(columns=['yield'])
y_train = train_df['yield']
X_test = test_df.drop(columns=['yield'])
y_test = test_df['yield']

# Define numeric and categorical columns
numeric_cols = X_train.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = X_train.select_dtypes(include=['object']).columns.tolist()

# Preprocessing for numeric columns
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])

# Preprocessing for categorical columns
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('label_encoder', LabelEncoder())
])

# Combine preprocessing steps
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_cols),
        ('cat', categorical_transformer, categorical_cols)
    ])

# Define the model
model = RandomForestRegressor(n_estimators=100, random_state=42)

# Create a pipeline
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', model)
])

# Train the model
pipeline.fit(X_train, y_train)

# Predict on the test set
y_pred = pipeline.predict(X_test)

# Evaluate the model
mae = mean_absolute_error(y_test, y_pred)
print(f"Mean Absolute Error on Test Set: {mae}")
```
