```
import pandas as pd

# Load the training and testing datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the datasets to understand their structure
print("Training Data:")
print(train_df.head())
print("\nTesting Data:")
print(test_df.head())

# Monitor data types per column
print("\nTraining Data Types:")
print(train_df.dtypes)
print("\nTesting Data Types:")
print(test_df.dtypes)
```

```
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Distinguish column types
numeric_cols = train_df.select_dtypes(include=[np.number]).columns
categorical_cols = train_df.select_dtypes(include=['object']).columns

# Summary statistics for numeric columns
print("Numeric Columns Summary Statistics:")
print(train_df[numeric_cols].describe())

# Summary statistics for categorical columns
print("\nCategorical Columns Summary Statistics:")
print(train_df[categorical_cols].describe())

# Correlation matrix for numeric columns
plt.figure(figsize=(12, 8))
correlation_matrix = train_df[numeric_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Numeric Features')
plt.show()

# Distribution of target variable (Age)
plt.figure(figsize=(8, 6))
sns.histplot(train_df['Age'], bins=30, kde=True)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

# Distribution of categorical variable (Sex)
plt.figure(figsize=(8, 6))
sns.countplot(x='Sex', data=train_df)
plt.title('Distribution of Sex')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.show()
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

column_info = get_column_info(train_df)
print("column_info")
print(column_info)
```

```
from metagpt.tools.libs.data_preprocess import OneHotEncode

# Initialize the OneHotEncode tool for the 'Sex' column
one_hot_encoder = OneHotEncode(features=['Sex'])

# Fit and transform the training data
train_df_encoded = one_hot_encoder.fit_transform(train_df.copy())

# Transform the testing data using the same encoder
test_df_encoded = one_hot_encoder.transform(test_df.copy())

# Display the first few rows of the encoded training data
print("Encoded Training Data:")
print(train_df_encoded.head())

# Display the first few rows of the encoded testing data
print("\nEncoded Testing Data:")
print(test_df_encoded.head())

# Display the data types of the encoded training data
print("\nEncoded Training Data Types:")
print(train_df_encoded.dtypes)

# Display the data types of the encoded testing data
print("\nEncoded Testing Data Types:")
print(test_df_encoded.dtypes)
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

column_info = get_column_info(train_df_encoded)
print("column_info")
print(column_info)
```

```
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split

# Assuming train_df_encoded and test_df_encoded are already defined from previous tasks
X_train = train_df_encoded.drop(columns=['id', 'Age'])
y_train = train_df_encoded['Age']

# Initialize and train the Random Forest model
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Prepare the test data
X_test = test_df_encoded.drop(columns=['id', 'Age'])

# Make predictions on the test data
predictions = rf_model.predict(X_test)

# Calculate the Mean Absolute Error
mae = mean_absolute_error(test_df_encoded['Age'], predictions)
print(f"Mean Absolute Error on Test Set: {mae}")

# Save the predictions to a CSV file
predictions_df = pd.DataFrame({'id': test_df_encoded['id'], 'Predicted_Age': predictions})
predictions_df.to_csv('crab_age_predictions.csv', index=False)
```

```
# Evaluate the trained model on the preprocessed testing dataset and calculate the Mean Absolute Error (MAE)

# Ensure the test data is preprocessed in the same way as the training data
X_test = test_df_encoded.drop(columns=['id', 'Age'])
predictions = rf_model.predict(X_test)

# Calculate Mean Absolute Error
mae = mean_absolute_error(test_df_encoded['Age'], predictions)
print(f"Mean Absolute Error on Test Set: {mae}")

# Save the predictions to a CSV file
predictions_df = pd.DataFrame({'id': test_df_encoded['id'], 'Predicted_Age': predictions})
predictions_df.to_csv('crab_age_predictions.csv', index=False)
```
