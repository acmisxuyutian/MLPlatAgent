```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
train_data = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/train.csv')

# Display the first few rows of the dataset
print(train_data.head())

# Get a summary of the dataset
print(train_data.info())

# Check for missing values
print(train_data.isnull().sum())

# Visualize the distribution of the target variable (assuming 'age' is the target)
plt.figure(figsize=(8, 6))
sns.histplot(train_data['age'], bins=30, kde=True)
plt.title('Distribution of Crab Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

# Distinguish column types
numeric_features = train_data.select_dtypes(include=[np.number]).columns
categorical_features = train_data.select_dtypes(include=['object', 'category']).columns

print("Numeric Features:", numeric_features)
print("Categorical Features:", categorical_features)

# Visualize correlation matrix for numeric features
plt.figure(figsize=(12, 8))
corr_matrix = train_data[numeric_features].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numeric Features')
plt.show()

# Visualize the distribution of categorical features
for feature in categorical_features:
    plt.figure(figsize=(8, 6))
    sns.countplot(data=train_data, x=feature)
    plt.title(f'Distribution of {feature}')
    plt.xlabel(feature)
    plt.ylabel('Count')
    plt.show()
```
