```
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target

# Distinguish column types
numeric_features = df.select_dtypes(include=[np.number]).columns
categorical_features = df.select_dtypes(include=['object', 'category']).columns

# Display basic statistics for numeric features
print("Numeric Features Statistics:")
print(df[numeric_features].describe())

# Display basic statistics for categorical features
print("\nCategorical Features Statistics:")
print(df[categorical_features].describe())

# Correlation matrix for numeric features
corr_matrix = df[numeric_features].corr()
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numeric Features')
plt.show()
```
