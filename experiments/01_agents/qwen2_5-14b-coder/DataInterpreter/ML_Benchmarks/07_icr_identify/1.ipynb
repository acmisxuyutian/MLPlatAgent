```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the training dataset
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/ml_benchmark/07_icr_identify/split_train.csv'
train_df = pd.read_csv(train_data_path)

# Display the first few rows of the dataset
print(train_df.head())

# Get basic information about the dataset
print(train_df.info())

# Check for missing values
print(train_df.isnull().sum())

# Check for columns with only one unique value
print(train_df.nunique())

# Distinguish column types
numeric_cols = train_df.select_dtypes(include=[np.number]).columns
categorical_cols = train_df.select_dtypes(include=['object', 'category']).columns

# Display summary statistics for numeric columns
print(train_df[numeric_cols].describe())

# Display value counts for categorical columns
for col in categorical_cols:
    print(train_df[col].value_counts())

# Visualize missing values
plt.figure(figsize=(12, 6))
sns.heatmap(train_df.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Values Heatmap')
plt.show()

# Visualize distribution of numeric columns
train_df[numeric_cols].hist(bins=20, figsize=(20, 15))
plt.suptitle('Distribution of Numeric Columns')
plt.show()

# Visualize correlation matrix for numeric columns
plt.figure(figsize=(16, 12))
correlation_matrix = train_df[numeric_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix')
plt.show()
```

```
from metagpt.tools.libs.data_preprocess import FillMissingValue, StandardScale, LabelEncode

# Load eval data
eval_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/ml_benchmark/07_icr_identify/split_eval.csv'
eval_df = pd.read_csv(eval_data_path)

# Copy the DataFrames to avoid modifying the original data
train_df_copy = train_df.copy()
eval_df_copy = eval_df.copy()

# Handle missing values
numeric_cols = train_df_copy.select_dtypes(include=[np.number]).columns
categorical_cols = train_df_copy.select_dtypes(include=['object', 'category']).columns

# Fill missing values for numeric columns with mean
fill_missing_numeric = FillMissingValue(features=numeric_cols, strategy='mean')
train_df_copy = fill_missing_numeric.fit_transform(train_df_copy)
eval_df_copy = fill_missing_numeric.transform(eval_df_copy)

# Fill missing values for categorical columns with most frequent value
fill_missing_categorical = FillMissingValue(features=categorical_cols, strategy='most_frequent')
train_df_copy = fill_missing_categorical.fit_transform(train_df_copy)
eval_df_copy = fill_missing_categorical.transform(eval_df_copy)

# Encode categorical variables using label encoding
label_encode = LabelEncode(features=categorical_cols)
train_df_copy = label_encode.fit_transform(train_df_copy)
eval_df_copy = label_encode.transform(eval_df_copy)

# Normalize numerical features using standard scaling
standard_scale = StandardScale(features=numeric_cols)
train_df_copy = standard_scale.fit_transform(train_df_copy)
eval_df_copy = standard_scale.transform(eval_df_copy)

# Display the first few rows of the processed data
print(train_df_copy.head())
print(eval_df_copy.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

column_info = get_column_info(train_df_copy)
print("column_info")
print(column_info)
```

```
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score
from sklearn.preprocessing import LabelEncoder

# Assuming train_df_copy and eval_df_copy are already preprocessed
X_train = train_df_copy.drop(columns=['Id', 'Class'])
y_train = train_df_copy['Class']

X_eval = eval_df_copy.drop(columns=['Id', 'Class'])
y_eval = eval_df_copy['Class']

# Initialize and train the RandomForestClassifier
model = RandomForestClassifier(n_estimators=100, max_depth=None, min_samples_split=2, random_state=42)
model.fit(X_train, y_train)

# Predict on the evaluation set
y_pred = model.predict(X_eval)

# Calculate F1 Score
f1 = f1_score(y_eval, y_pred, average='weighted')
print(f'F1 Score on eval data: {f1}')
```
