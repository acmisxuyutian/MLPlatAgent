```
import pandas as pd
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df['target'] = iris.target

# Display the first few rows of the dataset
print(iris_df.head())
```

```
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Distinguish column types
numeric_features = iris_df.select_dtypes(include=[np.number]).columns
categorical_features = iris_df.select_dtypes(include=['object']).columns

# Summary statistics for numeric features
print(iris_df[numeric_features].describe())

# Correlation matrix for numeric features
correlation_matrix = iris_df[numeric_features].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()

# Pairplot for numeric features
sns.pairplot(iris_df, hue='target', diag_kind='kde')
plt.suptitle('Pairplot of Iris Dataset', y=1.02)
plt.show()
```

```
# Plotting the distribution of each feature
plt.figure(figsize=(15, 10))
for i, feature in enumerate(numeric_features):
    plt.subplot(2, 2, i + 1)
    sns.histplot(iris_df[feature], bins=20, kde=True)
    plt.title(f'Distribution of {feature}')
plt.tight_layout()
plt.show()

# Boxplot for each feature by target
plt.figure(figsize=(15, 10))
for i, feature in enumerate(numeric_features):
    plt.subplot(2, 2, i + 1)
    sns.boxplot(x='target', y=feature, data=iris_df)
    plt.title(f'Boxplot of {feature} by Target')
plt.tight_layout()
plt.show()
```
