```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/obesity_CVD_risk/train.csv'
df = pd.read_csv(train_path)

# Display the first few rows of the dataframe
df.head()

# Summary statistics
df.describe(include='all')

# Check for missing values
df.isnull().sum()

# Distinguish column types
numeric_features = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_features = df.select_dtypes(include=[object]).columns.tolist()

# Visualize the distribution of numeric features
for feature in numeric_features:
    plt.figure(figsize=(10, 4))
    sns.histplot(df[feature], kde=True)
    plt.title(f'Distribution of {feature}')
    plt.show()

# Visualize the distribution of categorical features
for feature in categorical_features:
    plt.figure(figsize=(10, 4))
    sns.countplot(data=df, x=feature)
    plt.title(f'Distribution of {feature}')
    plt.xticks(rotation=45)
    plt.show()

# Correlation matrix for numeric features
plt.figure(figsize=(12, 8))
corr_matrix = df[numeric_features].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Numeric Features')
plt.show()
```

```
from metagpt.tools.libs.data_preprocess import FillMissingValue, StandardScale

# Copy the DataFrame to avoid modifying the original data
df_copy = df.copy()

# Handling missing values
# For numeric features, we will use the mean strategy
numeric_features = ['id', 'Age', 'Height', 'Weight', 'FCVC', 'NCP', 'CH2O', 'FAF', 'TUE']
fill_missing_numeric = FillMissingValue(numeric_features, strategy='mean')
df_copy = fill_missing_numeric.fit_transform(df_copy)

# For categorical features, we will use the most_frequent strategy
categorical_features = ['Gender', 'family_history_with_overweight', 'FAVC', 'CAEC', 'SMOKE', 'SCC', 'CALC', 'MTRANS', 'NObeyesdad']
fill_missing_categorical = FillMissingValue(categorical_features, strategy='most_frequent')
df_copy = fill_missing_categorical.fit_transform(df_copy)

# Encoding categorical variables
# We will use OneHotEncode for categorical features
from metagpt.tools.libs.data_preprocess import OneHotEncode
one_hot_encoder = OneHotEncode(categorical_features)
df_copy = one_hot_encoder.fit_transform(df_copy)

# Normalizing numerical features
# We will use StandardScale for numerical features
scaler = StandardScale(numeric_features)
df_copy = scaler.fit_transform(df_copy)

# Display the preprocessed DataFrame
df_copy.head()
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

column_info = get_column_info(df_copy)
print("column_info")
print(column_info)
```

```
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# Define the target and features
target = 'NObeyesdad_Overweight_Level_II'
features = [col for col in df_copy.columns if col != target]

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(df_copy[features], df_copy[target], test_size=0.2, random_state=42)

# Define the model
xgb_model = XGBClassifier(use_label_encoder=False, eval_metric='mlogloss', random_state=42)

# Define the parameter grid for hyperparameter tuning
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.1, 0.2]
}

# Perform grid search with cross-validation
grid_search = GridSearchCV(estimator=xgb_model, param_grid=param_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train, y_train)

# Get the best model
best_model = grid_search.best_estimator_

# Make predictions on the test set
y_pred = best_model.predict(X_test)

# Calculate the accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f'Test Accuracy: {accuracy:.4f}')
```
