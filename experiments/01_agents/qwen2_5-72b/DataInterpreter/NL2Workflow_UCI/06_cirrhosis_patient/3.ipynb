```
from metagpt.tools.libs.data_preprocess import FillMissingValue, StandardScale, OneHotEncode

# Copy the DataFrame to avoid modifying the original data
train_df_copy = train_df.copy()

# Handling missing values
# Separate numerical and categorical features
numerical_features = ['N_Days', 'Age', 'Bilirubin', 'Cholesterol', 'Albumin', 'Copper', 'Alk_Phos', 'SGOT', 'Tryglicerides', 'Platelets', 'Prothrombin', 'Stage']
categorical_features = ['Drug', 'Sex', 'Ascites', 'Hepatomegaly', 'Spiders', 'Edema']

# Impute numerical features with mean
fill_missing_numerical = FillMissingValue(features=numerical_features, strategy='mean')
train_df_processed = fill_missing_numerical.fit_transform(train_df_copy)

# Impute categorical features with most_frequent
fill_missing_categorical = FillMissingValue(features=categorical_features, strategy='most_frequent')
train_df_processed = fill_missing_categorical.fit_transform(train_df_processed)

# Encoding categorical variables
# Using OneHotEncode for categorical columns
one_hot_encoder = OneHotEncode(features=categorical_features)
train_df_processed = one_hot_encoder.fit_transform(train_df_processed)

# Normalizing numerical features
# Using StandardScale for numerical columns
scaler = StandardScale(features=numerical_features)
train_df_processed = scaler.fit_transform(train_df_processed)

# Display the processed DataFrame
train_df_processed.head()
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

column_info = get_column_info(train_df_processed)
print("column_info")
print(column_info)
```

```
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import log_loss
from xgboost import XGBClassifier
from sklearn.preprocessing import LabelEncoder

# Prepare the data
X = train_df_processed.drop(['id', 'Status'], axis=1)
y = train_df_processed['Status']

# Encode the target variable
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

# Initialize the XGBoost classifier
xgb = XGBClassifier(use_label_encoder=False, eval_metric='logloss')

# Define the parameter grid for GridSearchCV
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.1, 0.2],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

# Perform grid search to find the best parameters
grid_search = GridSearchCV(estimator=xgb, param_grid=param_grid, cv=3, scoring='neg_log_loss', n_jobs=-1)
grid_search.fit(X_train, y_train)

# Get the best model
best_model = grid_search.best_estimator_

# Predict probabilities on the test set
y_pred_proba = best_model.predict_proba(X_test)

# Calculate log loss
logloss = log_loss(y_test, y_pred_proba)
print(f'Log Loss on the test set: {logloss}')
```
