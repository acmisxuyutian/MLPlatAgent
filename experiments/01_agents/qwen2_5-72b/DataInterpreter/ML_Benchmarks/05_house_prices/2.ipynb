```
from metagpt.tools.libs.data_preprocess import FillMissingValue, StandardScale

# Load the eval data
eval_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\05_house_prices\\split_eval.csv'
eval_df = pd.read_csv(eval_data_path)

# Copy the dataframes to avoid modifying the original data
train_df_copy = train_df.copy()
eval_df_copy = eval_df.copy()

# Handle missing values
numeric_features = train_df_copy.select_dtypes(include=[np.number]).columns.tolist()
categorical_features = train_df_copy.select_dtypes(include=[object]).columns.tolist()

# Fill missing values for numeric features with median
fill_missing_numeric = FillMissingValue(numeric_features, strategy='median')
train_df_copy = fill_missing_numeric.fit_transform(train_df_copy)
eval_df_copy = fill_missing_numeric.transform(eval_df_copy)

# Fill missing values for categorical features with most frequent value
fill_missing_categorical = FillMissingValue(categorical_features, strategy='most_frequent')
train_df_copy = fill_missing_categorical.fit_transform(train_df_copy)
eval_df_copy = fill_missing_categorical.transform(eval_df_copy)

# Scale numeric features
scaler = StandardScale(numeric_features)
train_df_copy = scaler.fit_transform(train_df_copy)
eval_df_copy = scaler.transform(eval_df_copy)

# Display the first few rows of the preprocessed data
print(train_df_copy.head())
print(eval_df_copy.head())
```

```
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import LabelEncoder
import numpy as np

# Label encoding for categorical features
label_encoders = {}
for col in train_df_copy.select_dtypes(include=[object]).columns:
    le = LabelEncoder()
    train_df_copy[col] = le.fit_transform(train_df_copy[col].astype(str))
    eval_df_copy[col] = le.transform(eval_df_copy[col].astype(str))
    label_encoders[col] = le

# Split the training data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(
    train_df_copy.drop('SalePrice', axis=1), 
    train_df_copy['SalePrice'], 
    test_size=0.2, 
    random_state=42
)

# Define the model
xgb_model = XGBRegressor(objective='reg:squarederror', random_state=42)

# Define the parameter grid for hyperparameter tuning
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.1, 0.2]
}

# Perform grid search with cross-validation
grid_search = GridSearchCV(estimator=xgb_model, param_grid=param_grid, cv=5, scoring='neg_mean_squared_error', n_jobs=-1)
grid_search.fit(X_train, y_train)

# Get the best model
best_model = grid_search.best_estimator_

# Evaluate the best model on the validation set
y_val_pred = best_model.predict(X_val)
rmse_val = np.sqrt(mean_squared_error(y_val, y_val_pred))
print(f'Validation RMSE: {rmse_val}')

# Predict on the evaluation set
y_eval_pred = best_model.predict(eval_df_copy.drop('SalePrice', axis=1))
rmse_eval = np.sqrt(mean_squared_error(eval_df_copy['SalePrice'], y_eval_pred))
print(f'Evaluation RMSE: {rmse_eval}')
```
