```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\07_icr_identify\\split_train.csv'
eval_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\07_icr_identify\\split_eval.csv'

train_df = pd.read_csv(train_data_path)
eval_df = pd.read_csv(eval_data_path)

# Display basic information about the datasets
print("Train Data Info:")
train_df.info()
print("\nEval Data Info:")
eval_df.info()

# Display the first few rows of the datasets
print("\nTrain Data Head:")
print(train_df.head())
print("\nEval Data Head:")
print(eval_df.head())

# Check for missing values
print("\nMissing Values in Train Data:")
print(train_df.isnull().sum())
print("\nMissing Values in Eval Data:")
print(eval_df.isnull().sum())

# Check the distribution of the target variable
print("\nTarget Variable Distribution in Train Data:")
print(train_df['Class'].value_counts(normalize=True))
print("\nTarget Variable Distribution in Eval Data:")
print(eval_df['Class'].value_counts(normalize=True))

# Visualize the distribution of numerical features
numerical_features = train_df.select_dtypes(include=[np.number]).columns.tolist()
numerical_features.remove('Class')

plt.figure(figsize=(15, 25))
for i, feature in enumerate(numerical_features, 1):
    plt.subplot(7, 8, i)
    sns.histplot(train_df[feature], kde=True)
    plt.title(f'Distribution of {feature}')
plt.tight_layout()
plt.show()

# Visualize the correlation matrix for numerical features
plt.figure(figsize=(12, 8))
correlation_matrix = train_df[numerical_features + ['Class']].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```
from metagpt.tools.libs.data_preprocess import FillMissingValue, RobustScale

# Copy the dataframes to avoid modifying the original data
train_df_copy = train_df.copy()
eval_df_copy = eval_df.copy()

# Handling missing values
# For numerical features, use the median to fill missing values
numerical_features = [col for col in train_df_copy.columns if train_df_copy[col].dtype in ['int64', 'float64']]
fill_missing_value = FillMissingValue(features=numerical_features, strategy='median')
train_df_copy = fill_missing_value.fit_transform(train_df_copy)
eval_df_copy = fill_missing_value.transform(eval_df_copy)

# For categorical features, use the most frequent value to fill missing values
categorical_features = ['EJ']
fill_missing_value = FillMissingValue(features=categorical_features, strategy='most_frequent')
train_df_copy = fill_missing_value.fit_transform(train_df_copy)
eval_df_copy = fill_missing_value.transform(eval_df_copy)

# Scaling numerical features using RobustScaler to handle outliers
robust_scale = RobustScale(features=numerical_features)
train_df_copy = robust_scale.fit_transform(train_df_copy)
eval_df_copy = robust_scale.transform(eval_df_copy)

# Display the first few rows of the preprocessed data
print("Preprocessed Train Data Head:")
print(train_df_copy.head())
print("\nPreprocessed Eval Data Head:")
print(eval_df_copy.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Using the preprocessed train data to get the latest column information
column_info = get_column_info(train_df_copy)
print("column_info")
print(column_info)
```

```
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import f1_score
from xgboost import XGBClassifier
from metagpt.tools.libs.data_preprocess import LabelEncode

# Label Encode the categorical column 'EJ'
label_encode = LabelEncode(features=['EJ'])
train_df_copy = label_encode.fit_transform(train_df_copy)
eval_df_copy = label_encode.transform(eval_df_copy)

# Prepare the data for training
X_train = train_df_copy.drop(['Id', 'Class'], axis=1)
y_train = train_df_copy['Class']
X_eval = eval_df_copy.drop(['Id', 'Class'], axis=1)
y_eval = eval_df_copy['Class']

# Initialize the XGBoost classifier
xgb = XGBClassifier(use_label_encoder=False, eval_metric='logloss')

# Define the parameter grid for GridSearchCV
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.1, 0.2],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

# Perform grid search with cross-validation
grid_search = GridSearchCV(estimator=xgb, param_grid=param_grid, cv=3, scoring='f1', n_jobs=-1)
grid_search.fit(X_train, y_train)

# Get the best model
best_model = grid_search.best_estimator_

# Predict on the evaluation set
y_pred = best_model.predict(X_eval)

# Calculate the F1 score
f1 = f1_score(y_eval, y_pred)
print(f"F1 Score on the eval data: {f1}")
```
