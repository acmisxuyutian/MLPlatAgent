```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['target'] = iris.target

# Display the first few rows of the dataframe
df.head()

# Summary statistics
df.describe()

# Check for missing values
df.isnull().sum()

# Distinguish column types
numeric_features = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_features = df.select_dtypes(exclude=[np.number]).columns.tolist()

# Correlation matrix for numeric features
correlation_matrix = df[numeric_features].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()

# Pairplot for numeric features
sns.pairplot(df, hue='target')
plt.show()
```

```
# Since the data and necessary imports are already handled, we can directly proceed to the visualization task.

# Let's create a box plot to visualize the distribution of the numeric features across different target classes.
plt.figure(figsize=(12, 8))
for i, feature in enumerate(numeric_features, 1):
    plt.subplot(2, 2, i)
    sns.boxplot(x='target', y=feature, data=df)
    plt.title(f'Distribution of {feature} by Target')
plt.tight_layout()
plt.show()
```
