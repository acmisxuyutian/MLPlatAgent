```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
train_data = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\06_santander_customer\\split_train.csv')
eval_data = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\06_santander_customer\\split_eval.csv')

# Display the first few rows of the train and eval datasets
print("Train Data:")
display(train_data.head())
print("Eval Data:")
display(eval_data.head())

# Check the shape of the datasets
print("Train Data Shape:", train_data.shape)
print("Eval Data Shape:", eval_data.shape)

# Check for missing values
print("Missing values in Train Data:")
print(train_data.isnull().sum())
print("Missing values in Eval Data:")
print(eval_data.isnull().sum())

# Check the data types of the columns
print("Data types in Train Data:")
print(train_data.dtypes)
print("Data types in Eval Data:")
print(eval_data.dtypes)

# Summary statistics for numerical columns
print("Summary statistics for numerical columns in Train Data:")
display(train_data.describe())

# Distribution of the target variable
plt.figure(figsize=(10, 6))
sns.countplot(x='target', data=train_data)
plt.title('Distribution of Target Variable')
plt.show()

# Correlation matrix for numerical columns
numerical_features = train_data.select_dtypes(include=[np.number]).columns
correlation_matrix = train_data[numerical_features].corr()
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix for Numerical Features')
plt.show()
```

```
from metagpt.tools.libs.data_preprocess import StandardScale

# Copy the datasets to avoid modifying the original data
train_data_copy = train_data.copy()
eval_data_copy = eval_data.copy()

# Since there are no missing values and no categorical variables, we will only standardize the numerical features
# Initialize the StandardScale tool for numerical features
scaler = StandardScale(features=train_data_copy.columns.drop('target'))

# Fit the scaler on the training data and transform both the training and evaluation data
train_data_scaled = scaler.fit_transform(train_data_copy)
eval_data_scaled = scaler.transform(eval_data_copy)

# Display the first few rows of the scaled datasets
print("Scaled Train Data:")
display(train_data_scaled.head())
print("Scaled Eval Data:")
display(eval_data_scaled.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Using the latest DataFrame from the finished tasks
column_info_train = get_column_info(train_data_scaled)
print("Column Info for Scaled Train Data:")
print(column_info_train)

column_info_eval = get_column_info(eval_data_scaled)
print("Column Info for Scaled Eval Data:")
print(column_info_eval)
```

```
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import roc_auc_score
from xgboost import XGBClassifier

# Split the training data into features and target
X_train = train_data_scaled.drop(columns=['target'])
y_train = train_data_scaled['target']

# Split the evaluation data into features and target
X_eval = eval_data_scaled.drop(columns=['target'])
y_eval = eval_data_scaled['target']

# Initialize the XGBoost classifier
xgb = XGBClassifier(use_label_encoder=False, eval_metric='logloss')

# Define the parameter grid for GridSearchCV
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.1, 0.2],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

# Perform grid search with cross-validation
grid_search = GridSearchCV(estimator=xgb, param_grid=param_grid, cv=3, scoring='roc_auc', n_jobs=-1)
grid_search.fit(X_train, y_train)

# Get the best model
best_model = grid_search.best_estimator_

# Predict probabilities on the evaluation set
y_pred_proba = best_model.predict_proba(X_eval)[:, 1]

# Calculate AUC score
auc_score = roc_auc_score(y_eval, y_pred_proba)
print(f"AUC Score on Evaluation Data: {auc_score:.4f}")
```
