```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and testing datasets
print("Training dataset:")
display(train_df.head())
print("Testing dataset:")
display(test_df.head())

# Check the information of the datasets
print("Training dataset info:")
train_df.info()
print("Testing dataset info:")
test_df.info()

# Check for missing values
print("Missing values in training dataset:")
print(train_df.isnull().sum())
print("Missing values in testing dataset:")
print(test_df.isnull().sum())

# Summary statistics
print("Summary statistics for training dataset:")
display(train_df.describe(include='all'))
print("Summary statistics for testing dataset:")
display(test_df.describe(include='all'))

# Distinguish column types
numeric_features = train_df.select_dtypes(include=[np.number]).columns
categorical_features = train_df.select_dtypes(exclude=[np.number]).columns

print("Numeric features:", numeric_features)
print("Categorical features:", categorical_features)

# Visualize the distribution of numeric features
for feature in numeric_features:
    plt.figure(figsize=(10, 6))
    sns.histplot(train_df[feature], kde=True)
    plt.title(f'Distribution of {feature}')
    plt.show()

# Visualize the distribution of categorical features
for feature in categorical_features:
    plt.figure(figsize=(10, 6))
    sns.countplot(data=train_df, x=feature)
    plt.title(f'Distribution of {feature}')
    plt.show()

# Correlation matrix for numeric features
plt.figure(figsize=(12, 8))
correlation_matrix = train_df[numeric_features].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Numeric Features')
plt.show()
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Using train_df for the column information
column_info = get_column_info(train_df)
print("column_info")
print(column_info)
```

```
from metagpt.tools.libs.feature_engineering import PolynomialExpansion, VarianceBasedSelection

# Copy the dataframes to avoid modifying the original ones
train_df_copy = train_df.copy()
test_df_copy = test_df.copy()

# Remove ID columns from feature generation
train_df_copy = train_df_copy.drop(columns=['id'])
test_df_copy = test_df_copy.drop(columns=['id'])

# Perform polynomial expansion on selected numeric features
poly_expander = PolynomialExpansion(cols=['clonesize', 'honeybee', 'bumbles', 'andrena', 'osmia'], label_col='yield', degree=2)
train_df_poly = poly_expander.fit_transform(train_df_copy)
test_df_poly = poly_expander.transform(test_df_copy)

# Perform variance-based feature selection
variance_selector = VarianceBasedSelection(label_col='yield', threshold=0.01)
train_df_final = variance_selector.fit_transform(train_df_poly)
test_df_final = variance_selector.transform(test_df_poly)

# Display the final datasets
print("Final training dataset:")
display(train_df_final.head())
print("Final testing dataset:")
display(test_df_final.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Using the final training and testing datasets
column_info_train = get_column_info(train_df_final)
column_info_test = get_column_info(test_df_final)

print("Column information for the final training dataset:")
print(column_info_train)
print("\nColumn information for the final testing dataset:")
print(column_info_test)
```

```
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error

# Split the final training dataset into features and target
X_train = train_df_final.drop(columns=['yield'])
y_train = train_df_final['yield']

# Split the final testing dataset into features and target
X_test = test_df_final.drop(columns=['yield'])
y_test = test_df_final['yield']

# Standardize the features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Define the model
model = XGBRegressor()

# Define the parameter grid for GridSearchCV
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 4, 5],
    'learning_rate': [0.01, 0.1, 0.2],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

# Perform grid search to find the best hyperparameters
grid_search = GridSearchCV(estimator=model, param_grid=param_grid, cv=5, scoring='neg_mean_absolute_error', n_jobs=-1)
grid_search.fit(X_train_scaled, y_train)

# Get the best model
best_model = grid_search.best_estimator_

# Evaluate the best model on the test set
y_pred = best_model.predict(X_test_scaled)
mae = mean_absolute_error(y_test, y_pred)

print(f"Best parameters: {grid_search.best_params_}")
print(f"Mean Absolute Error on test set: {mae}")
```
