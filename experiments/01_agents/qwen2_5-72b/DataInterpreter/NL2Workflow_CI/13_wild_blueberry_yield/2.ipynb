```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and testing datasets
print("Training Data:")
display(train_df.head())
print("Testing Data:")
display(test_df.head())

# Check the information of the datasets
print("Training Data Info:")
train_df.info()
print("Testing Data Info:")
test_df.info()

# Check for missing values
print("Missing values in Training Data:")
print(train_df.isnull().sum())
print("Missing values in Testing Data:")
print(test_df.isnull().sum())

# Check the distribution of the target variable
plt.figure(figsize=(10, 6))
sns.histplot(train_df['yield'], kde=True)
plt.title('Distribution of Yield in Training Data')
plt.xlabel('Yield')
plt.ylabel('Frequency')
plt.show()

# Check the correlation matrix
plt.figure(figsize=(12, 8))
correlation_matrix = train_df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Training Data')
plt.show()
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Using train_df for the column information
column_info = get_column_info(train_df)
print("column_info")
print(column_info)
```

```
from metagpt.tools.libs.feature_engineering import GeneralSelection, PolynomialExpansion, VarianceBasedSelection
from metagpt.tools.libs.data_preprocess import MinMaxScale

# Copy the dataframes to avoid modifying the original data
train_df_copy = train_df.copy()
test_df_copy = test_df.copy()

# Remove the ID column
train_df_copy.drop(columns=['id'], inplace=True)
test_df_copy.drop(columns=['id'], inplace=True)

# Perform general feature selection to drop all NaN features and features with only one unique value
selector = GeneralSelection(label_col='yield')
train_df_selected = selector.fit_transform(train_df_copy)
test_df_selected = selector.transform(test_df_copy)

# Perform polynomial expansion to add interaction and polynomial features
poly_expander = PolynomialExpansion(cols=train_df_selected.columns.drop('yield').tolist(), label_col='yield', degree=2)
train_df_poly = poly_expander.fit_transform(train_df_selected)
test_df_poly = poly_expander.transform(test_df_selected)

# Perform variance-based feature selection to remove features with low variance
variance_selector = VarianceBasedSelection(label_col='yield', threshold=0.01)
train_df_final = variance_selector.fit_transform(train_df_poly)
test_df_final = variance_selector.transform(test_df_poly)

# Scale the features using MinMaxScaler
scaler = MinMaxScale(features=train_df_final.columns.drop('yield').tolist())
train_df_scaled = scaler.fit_transform(train_df_final)
test_df_scaled = scaler.transform(test_df_final)

# Display the final train and test dataframes
print("Final Training Data:")
display(train_df_scaled.head())
print("Final Testing Data:")
display(test_df_scaled.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Using the final scaled training and testing dataframes
column_info_train = get_column_info(train_df_scaled)
column_info_test = get_column_info(test_df_scaled)

print("Column Info for Final Training Data:")
print(column_info_train)
print("\nColumn Info for Final Testing Data:")
print(column_info_test)
```

```
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import mean_absolute_error
from xgboost import XGBRegressor

# Split the training data into features and target
X_train = train_df_scaled.drop(columns=['yield'])
y_train = train_df_scaled['yield']

# Split the test data into features and target
X_test = test_df_scaled.drop(columns=['yield'])
y_test = test_df_scaled['yield']

# Initialize the XGBoost model
xgb_model = XGBRegressor()

# Define the parameter grid for GridSearchCV
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 4, 5],
    'learning_rate': [0.01, 0.1, 0.2],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

# Perform grid search to find the best hyperparameters
grid_search = GridSearchCV(estimator=xgb_model, param_grid=param_grid, cv=5, scoring='neg_mean_absolute_error', n_jobs=-1)
grid_search.fit(X_train, y_train)

# Get the best model
best_model = grid_search.best_estimator_

# Evaluate the best model on the test set
y_pred = best_model.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)

print(f"Best Model Parameters: {grid_search.best_params_}")
print(f"Mean Absolute Error (MAE) on Test Set: {mae}")
```
