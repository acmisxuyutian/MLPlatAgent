```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/monster/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/monster/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and testing datasets
print("Training Data:")
display(train_df.head())
print("Testing Data:")
display(test_df.head())

# Check the shape of the datasets
print(f"Training Data Shape: {train_df.shape}")
print(f"Testing Data Shape: {test_df.shape}")

# Check for missing values
print("Missing values in Training Data:")
print(train_df.isnull().sum())
print("Missing values in Testing Data:")
print(test_df.isnull().sum())

# Check the data types of the columns
print("Data types in Training Data:")
print(train_df.dtypes)
print("Data types in Testing Data:")
print(test_df.dtypes)

# Summary statistics for numerical columns
print("Summary statistics for numerical columns in Training Data:")
display(train_df.describe())

# Summary statistics for categorical columns
print("Summary statistics for categorical columns in Training Data:")
display(train_df.describe(include=['O']))

# Correlation matrix for numerical columns
plt.figure(figsize=(12, 8))
sns.heatmap(train_df.select_dtypes(include=[np.number]).corr(), annot=True, fmt=".2f")
plt.title("Correlation Matrix for Numerical Columns in Training Data")
plt.show()
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Using the train_df DataFrame from the finished tasks
column_info = get_column_info(train_df)
print("column_info")
print(column_info)
```

```
from metagpt.tools.libs.data_preprocess import LabelEncode, StandardScale

# Copy the datasets to avoid modifying the original data
train_df_copy = train_df.copy()
test_df_copy = test_df.copy()

# Encode categorical variables
label_encoder = LabelEncode(features=['color'])
train_df_encoded = label_encoder.fit_transform(train_df_copy)
test_df_encoded = label_encoder.transform(test_df_copy)

# Scale numerical features
scaler = StandardScale(features=['bone_length', 'rotting_flesh', 'hair_length'])
train_df_scaled = scaler.fit_transform(train_df_encoded)
test_df_scaled = scaler.transform(test_df_encoded)

# Display the first few rows of the preprocessed datasets
print("Preprocessed Training Data:")
display(train_df_scaled.head())
print("Preprocessed Testing Data:")
display(test_df_scaled.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Using the preprocessed training data DataFrame
column_info = get_column_info(train_df_scaled)
print("column_info")
print(column_info)
```

```
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from xgboost import XGBClassifier

# Prepare the data
X_train = train_df_scaled.drop(['id', 'type'], axis=1)
y_train = train_df_scaled['type']
X_test = test_df_scaled.drop(['id', 'type'], axis=1)
y_test = test_df_scaled['type']

# Initialize and train the model
model = XGBClassifier(use_label_encoder=False, eval_metric='mlogloss')
model.fit(X_train, y_train)

# Predict on the test set
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy on Test Set: {accuracy:.4f}")

# Save predictions
test_df_scaled['predicted_type'] = y_pred
test_df_scaled[['id', 'predicted_type']].to_csv('monster_predictions.csv', index=False)
```
