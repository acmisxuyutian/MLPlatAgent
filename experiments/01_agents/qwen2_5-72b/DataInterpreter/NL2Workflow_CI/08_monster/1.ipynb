```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/monster/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/monster/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and testing datasets
print("Training Data:")
display(train_df.head())
print("Testing Data:")
display(test_df.head())

# Check the information of the datasets
print("Training Data Info:")
train_df.info()
print("Testing Data Info:")
test_df.info()

# Check for missing values
print("Missing values in Training Data:")
print(train_df.isnull().sum())
print("Missing values in Testing Data:")
print(test_df.isnull().sum())

# Descriptive statistics
print("Descriptive statistics for Training Data:")
display(train_df.describe(include='all'))

# Distinguish column types
numeric_features = train_df.select_dtypes(include=[np.number]).columns
categorical_features = train_df.select_dtypes(include=[object]).columns

# Visualize the distribution of numeric features
for feature in numeric_features:
    plt.figure(figsize=(10, 6))
    sns.histplot(train_df[feature], kde=True)
    plt.title(f'Distribution of {feature}')
    plt.show()

# Visualize the distribution of categorical features
for feature in categorical_features:
    plt.figure(figsize=(10, 6))
    sns.countplot(data=train_df, x=feature)
    plt.title(f'Distribution of {feature}')
    plt.xticks(rotation=45)
    plt.show()

# Check the correlation matrix for numeric features
plt.figure(figsize=(12, 8))
sns.heatmap(train_df[numeric_features].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Numeric Features')
plt.show()
```

```
from metagpt.tools.libs.data_preprocess import LabelEncode, StandardScale

# Copy the datasets to avoid modifying the original data
train_df_copy = train_df.copy()
test_df_copy = test_df.copy()

# Encode categorical variables
categorical_features = ['color']
label_encoder = LabelEncode(features=categorical_features)
train_df_encoded = label_encoder.fit_transform(train_df_copy)
test_df_encoded = label_encoder.transform(test_df_copy)

# Scale numeric features
numeric_features = ['bone_length', 'rotting_flesh', 'hair_length']
scaler = StandardScale(features=numeric_features)
train_df_scaled = scaler.fit_transform(train_df_encoded)
test_df_scaled = scaler.transform(test_df_encoded)

# Display the first few rows of the preprocessed datasets
print("Preprocessed Training Data:")
display(train_df_scaled.head())
print("Preprocessed Testing Data:")
display(test_df_scaled.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Using the preprocessed training data DataFrame
df = train_df_scaled
column_info = get_column_info(df)
print("column_info")
print(column_info)
```

```
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score
from xgboost import XGBClassifier

# Prepare the data
X_train = train_df_scaled.drop(['id', 'type'], axis=1)
y_train = train_df_scaled['type']
X_test = test_df_scaled.drop(['id', 'type'], axis=1)
y_test = test_df_scaled['type']

# Initialize the model
model = XGBClassifier(use_label_encoder=False, eval_metric='mlogloss')

# Define the parameter grid
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 4, 5],
    'learning_rate': [0.01, 0.1, 0.2]
}

# Perform grid search with cross-validation
grid_search = GridSearchCV(estimator=model, param_grid=param_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train, y_train)

# Get the best model
best_model = grid_search.best_estimator_

# Predict on the test set
y_pred = best_model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Test Accuracy: {accuracy:.4f}")

# Save the predictions
test_df_scaled['predicted_type'] = y_pred
test_df_scaled[['id', 'predicted_type']].to_csv('monster_predictions.csv', index=False)
```

```
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score
from xgboost import XGBClassifier
from sklearn.preprocessing import LabelEncoder

# Label encode the 'type' column
label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

# Prepare the data
X_train = train_df_scaled.drop(['id', 'type'], axis=1)
y_train = y_train_encoded
X_test = test_df_scaled.drop(['id', 'type'], axis=1)
y_test = y_test_encoded

# Initialize the model
model = XGBClassifier(use_label_encoder=False, eval_metric='mlogloss')

# Define the parameter grid
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 4, 5],
    'learning_rate': [0.01, 0.1, 0.2]
}

# Perform grid search with cross-validation
grid_search = GridSearchCV(estimator=model, param_grid=param_grid, cv=5, scoring='accuracy', n_jobs=-1)
grid_search.fit(X_train, y_train)

# Get the best model
best_model = grid_search.best_estimator_

# Predict on the test set
y_pred = best_model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Test Accuracy: {accuracy:.4f}")

# Save the predictions
test_df_scaled['predicted_type'] = label_encoder.inverse_transform(y_pred)
test_df_scaled[['id', 'predicted_type']].to_csv('monster_predictions.csv', index=False)
```
