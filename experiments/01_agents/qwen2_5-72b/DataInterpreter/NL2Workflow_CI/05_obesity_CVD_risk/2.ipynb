```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/obesity_CVD_risk/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/obesity_CVD_risk/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and testing datasets
print("Training dataset:")
display(train_df.head())
print("Testing dataset:")
display(test_df.head())

# Check the information of the datasets
print("Training dataset info:")
train_df.info()
print("Testing dataset info:")
test_df.info()

# Check for missing values
print("Missing values in training dataset:")
print(train_df.isnull().sum())
print("Missing values in testing dataset:")
print(test_df.isnull().sum())

# Summary statistics
print("Summary statistics for training dataset:")
display(train_df.describe(include='all'))

# Distribution of the target variable
sns.countplot(data=train_df, x='NObeyesdad')
plt.title('Distribution of Obesity Risk in Training Dataset')
plt.show()

# Correlation matrix for numerical features
numeric_features = train_df.select_dtypes(include=[np.number]).columns
corr_matrix = train_df[numeric_features].corr()
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix for Numerical Features')
plt.show()
```

```
from metagpt.tools.libs.feature_engineering import CatCount, TargetMeanEncoder, PolynomialExpansion, VarianceBasedSelection, GeneralSelection

# Copy the dataframes to avoid modifying the original data
train_df_copy = train_df.copy()
test_df_copy = test_df.copy()

# Remove ID columns
train_df_copy.drop(columns=['id'], inplace=True)
test_df_copy.drop(columns=['id'], inplace=True)

# Feature Engineering
# 1. Categorical Count Encoding
cat_count = CatCount(col='Gender')
train_df_copy = cat_count.fit_transform(train_df_copy)
test_df_copy = cat_count.transform(test_df_copy)

# 2. Target Mean Encoding
target_mean_encoder = TargetMeanEncoder(col='family_history_with_overweight', label='NObeyesdad')
train_df_copy = target_mean_encoder.fit_transform(train_df_copy)
test_df_copy = target_mean_encoder.transform(test_df_copy)

# 3. Polynomial Expansion
poly_expansion = PolynomialExpansion(cols=['Age', 'Height', 'Weight'], label_col='NObeyesdad', degree=2)
train_df_copy = poly_expansion.fit_transform(train_df_copy)
test_df_copy = poly_expansion.transform(test_df_copy)

# 4. Variance Based Selection
variance_selector = VarianceBasedSelection(label_col='NObeyesdad', threshold=0.01)
train_df_copy = variance_selector.fit_transform(train_df_copy)
test_df_copy = variance_selector.transform(test_df_copy)

# 5. General Selection
general_selector = GeneralSelection(label_col='NObeyesdad')
train_df_copy = general_selector.fit_transform(train_df_copy)
test_df_copy = general_selector.transform(test_df_copy)

# Display the first few rows of the transformed datasets
print("Transformed Training dataset:")
display(train_df_copy.head())
print("Transformed Testing dataset:")
display(test_df_copy.head())
```
