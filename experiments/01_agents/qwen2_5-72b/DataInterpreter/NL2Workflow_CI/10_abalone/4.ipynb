```
import pandas as pd
import numpy as np

# Load the datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and testing datasets
print("Training dataset:")
display(train_df.head())
print("Testing dataset:")
display(test_df.head())

# Check the information of the datasets
print("Training dataset info:")
train_df.info()
print("Testing dataset info:")
test_df.info()

# Check for missing values
print("Missing values in training dataset:")
print(train_df.isnull().sum())
print("Missing values in testing dataset:")
print(test_df.isnull().sum())

# Check the distribution of the target variable
print("Target variable distribution in training dataset:")
print(train_df['Rings'].describe())

# Check the correlation between features and the target variable
print("Correlation matrix for training dataset:")
print(train_df.corr()['Rings'].sort_values(ascending=False))
```
