```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and testing datasets
print("Training Data:")
display(train_df.head())
print("Testing Data:")
display(test_df.head())

# Check the information of the datasets
print("Training Data Info:")
train_df.info()
print("Testing Data Info:")
test_df.info()

# Check for missing values
print("Missing values in Training Data:")
print(train_df.isnull().sum())
print("Missing values in Testing Data:")
print(test_df.isnull().sum())

# Summary statistics for numerical columns
print("Summary statistics for Training Data:")
display(train_df.describe())
print("Summary statistics for Testing Data:")
display(test_df.describe())

# Distribution of the target variable
plt.figure(figsize=(10, 6))
sns.countplot(data=train_df, x='Category')
plt.title('Distribution of Crime Categories in Training Data')
plt.xticks(rotation=90)
plt.show()

# Correlation matrix for numerical columns
plt.figure(figsize=(12, 8))
corr_matrix = train_df.select_dtypes(include=[np.number]).corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix for Numerical Columns in Training Data')
plt.show()
```

```
from metagpt.tools.libs.data_preprocess import LabelEncode, StandardScale

# Copy the dataframes to avoid modifying the original data
train_df_copy = train_df.copy()
test_df_copy = test_df.copy()

# Encode categorical variables
categorical_features = ['DayOfWeek', 'PdDistrict']
label_encoder = LabelEncode(features=categorical_features)
train_df_encoded = label_encoder.fit_transform(train_df_copy)
test_df_encoded = label_encoder.transform(test_df_copy)

# Scale numerical features
numeric_features = ['X', 'Y']
scaler = StandardScale(features=numeric_features)
train_df_scaled = scaler.fit_transform(train_df_encoded)
test_df_scaled = scaler.transform(test_df_encoded)

# Display the first few rows of the preprocessed data
print("Preprocessed Training Data:")
display(train_df_scaled.head())
print("Preprocessed Testing Data:")
display(test_df_scaled.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Using the preprocessed training data DataFrame
column_info = get_column_info(train_df_scaled)
print("column_info")
print(column_info)
```

```
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import log_loss
from xgboost import XGBClassifier
from metagpt.tools.libs.data_preprocess import LabelEncode

# Split the preprocessed training data into features and target
X_train = train_df_scaled.drop('Category', axis=1)
y_train = train_df_scaled['Category']

# Split the preprocessed testing data into features and target
X_test = test_df_scaled.drop('Category', axis=1)
y_test = test_df_scaled['Category']

# Initialize the XGBoost classifier
xgb_model = XGBClassifier(use_label_encoder=False, eval_metric='mlogloss')

# Define the parameter grid for GridSearchCV
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.1, 0.2]
}

# Perform grid search to find the best hyperparameters
grid_search = GridSearchCV(estimator=xgb_model, param_grid=param_grid, cv=3, scoring='neg_log_loss', n_jobs=-1)
grid_search.fit(X_train, y_train)

# Get the best model
best_model = grid_search.best_estimator_

# Predict the probabilities on the test set
y_pred_proba = best_model.predict_proba(X_test)

# Calculate the log loss
logloss = log_loss(y_test, y_pred_proba)
print(f"Log Loss on the test set: {logloss}")

# Save the predictions
test_df['Predicted_Category'] = best_model.predict(X_test)
test_df.to_csv('predictions.csv', index=False)
```
