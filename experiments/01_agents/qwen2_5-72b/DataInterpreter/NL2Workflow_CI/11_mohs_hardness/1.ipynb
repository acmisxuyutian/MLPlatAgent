```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/mohs_hardness/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/mohs_hardness/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and testing datasets
print("Training Data:")
display(train_df.head())
print("Testing Data:")
display(test_df.head())

# Check the information of the datasets
print("Training Data Info:")
train_df.info()
print("Testing Data Info:")
test_df.info()

# Check for missing values
print("Missing values in Training Data:")
print(train_df.isnull().sum())
print("Missing values in Testing Data:")
print(test_df.isnull().sum())

# Descriptive statistics
print("Descriptive statistics for Training Data:")
display(train_df.describe(include='all'))
print("Descriptive statistics for Testing Data:")
display(test_df.describe(include='all'))

# Distinguish column types
numeric_features = train_df.select_dtypes(include=[np.number]).columns
categorical_features = train_df.select_dtypes(exclude=[np.number]).columns

# Visualize the distribution of numeric features
for feature in numeric_features:
    plt.figure(figsize=(10, 6))
    sns.histplot(train_df[feature], kde=True)
    plt.title(f'Distribution of {feature}')
    plt.show()

# Visualize the distribution of categorical features
for feature in categorical_features:
    plt.figure(figsize=(10, 6))
    sns.countplot(data=train_df, x=feature)
    plt.title(f'Distribution of {feature}')
    plt.xticks(rotation=45)
    plt.show()

# Correlation matrix for numeric features
plt.figure(figsize=(12, 8))
sns.heatmap(train_df[numeric_features].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Numeric Features')
plt.show()
```

```
from metagpt.tools.libs.data_preprocess import FillMissingValue, StandardScale

# Copy the datasets to avoid modifying the original data
train_df_copy = train_df.copy()
test_df_copy = test_df.copy()

# Handling missing values
# Since there are no missing values, we can skip this step, but for demonstration, we will initialize the FillMissingValue class
fill_missing = FillMissingValue(features=train_df_copy.columns, strategy='mean')
train_df_processed = fill_missing.fit_transform(train_df_copy)
test_df_processed = fill_missing.transform(test_df_copy)

# Scaling features
# We will scale all numeric features except the label column 'Hardness'
numeric_features_to_scale = [col for col in train_df_copy.columns if col != 'Hardness']
scaler = StandardScale(features=numeric_features_to_scale)
train_df_processed = scaler.fit_transform(train_df_processed)
test_df_processed = scaler.transform(test_df_processed)

# Display the first few rows of the processed datasets
print("Processed Training Data:")
display(train_df_processed.head())
print("Processed Testing Data:")
display(test_df_processed.head())
```

```
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import median_absolute_error

# Prepare the data
X_train = train_df_processed.drop(columns=['id', 'Hardness']).values
y_train = train_df_processed['Hardness'].values
X_test = test_df_processed.drop(columns=['id', 'Hardness']).values
y_test = test_df_processed['Hardness'].values

# Convert to PyTorch tensors
X_train_tensor = torch.tensor(X_train, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train, dtype=torch.float32).view(-1, 1)
X_test_tensor = torch.tensor(X_test, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test, dtype=torch.float32).view(-1, 1)

# Define the neural network model
class HardnessPredictor(nn.Module):
    def __init__(self, input_dim):
        super(HardnessPredictor, self).__init__()
        self.fc1 = nn.Linear(input_dim, 64)
        self.fc2 = nn.Linear(64, 32)
        self.fc3 = nn.Linear(32, 1)
        self.tanh = nn.Tanh()
    
    def forward(self, x):
        x = self.tanh(self.fc1(x))
        x = self.tanh(self.fc2(x))
        x = self.fc3(x)
        return x

# Initialize the model, loss function, and optimizer
input_dim = X_train_tensor.shape[1]
model = HardnessPredictor(input_dim)
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01, momentum=0.9)

# Train the model
num_epochs = 500
for epoch in range(num_epochs):
    model.train()
    optimizer.zero_grad()
    outputs = model(X_train_tensor)
    loss = criterion(outputs, y_train_tensor)
    loss.backward()
    optimizer.step()
    
    if (epoch+1) % 50 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')

# Evaluate the model on the test set
model.eval()
with torch.no_grad():
    y_pred = model(X_test_tensor)
    mae = median_absolute_error(y_test_tensor, y_pred)
    print(f'Median Absolute Error on Test Set: {mae:.4f}')
```
