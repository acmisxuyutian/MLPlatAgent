```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and testing datasets
print("Training Data:")
display(train_df.head())
print("Testing Data:")
display(test_df.head())

# Check the data types and basic statistics
print("Training Data Info:")
train_df.info()
print("Testing Data Info:")
test_df.info()

# Summary statistics for numerical columns
print("Training Data Descriptive Statistics:")
display(train_df.describe())
print("Testing Data Descriptive Statistics:")
display(test_df.describe())

# Check for missing values
print("Missing values in Training Data:")
display(train_df.isnull().sum())
print("Missing values in Testing Data:")
display(test_df.isnull().sum())

# Ensure column names are consistent
train_df.columns = train_df.columns.str.lower()
test_df.columns = test_df.columns.str.lower()

# Visualize the distribution of the target variable (age)
plt.figure(figsize=(10, 6))
sns.histplot(train_df['age'], bins=30, kde=True)
plt.title('Distribution of Age in Training Data')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

# Visualize the correlation matrix for numerical features
plt.figure(figsize=(12, 8))
corr_matrix = train_df.select_dtypes(include=[np.number]).corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Numerical Features in Training Data')
plt.show()
```

```
from metagpt.tools.libs.data_preprocess import OneHotEncode

# Copy the dataframes to avoid modifying the original data
train_df_copy = train_df.copy()
test_df_copy = test_df.copy()

# One-hot encode the categorical feature 'sex'
one_hot_encoder = OneHotEncode(features=['sex'])
train_df_encoded = one_hot_encoder.fit_transform(train_df_copy)
test_df_encoded = one_hot_encoder.transform(test_df_copy)

# Split the data into features and target variables
X_train = train_df_encoded.drop(columns=['age'])
y_train = train_df_encoded['age']
X_test = test_df_encoded.drop(columns=['age'])
y_test = test_df_encoded['age']

# Display the first few rows of the processed data
print("Processed Training Data:")
display(X_train.head())
print("Processed Testing Data:")
display(X_test.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Using the latest processed dataframes
column_info_train = get_column_info(X_train)
column_info_test = get_column_info(X_test)

print("Column Information for Processed Training Data:")
print(column_info_train)
print("\nColumn Information for Processed Testing Data:")
print(column_info_test)
```

```
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

# Initialize the Random Forest Regressor with some hyperparameters
rf_regressor = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42)

# Train the model
rf_regressor.fit(X_train, y_train)

# Predict on the test set
y_pred = rf_regressor.predict(X_test)

# Calculate the Mean Absolute Error
mae = mean_absolute_error(y_test, y_pred)
print(f"Mean Absolute Error on the test set: {mae:.2f}")

# Save the predictions
predictions = pd.DataFrame({'id': test_df['id'], 'predicted_age': y_pred})
predictions.to_csv('crab_age_predictions.csv', index=False)
print("Predictions saved to 'crab_age_predictions.csv'")
```
