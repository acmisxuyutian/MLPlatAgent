```
import pandas as pd
import numpy as np

# Load the datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/05_patient_profile/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/05_patient_profile/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the training and test datasets
print("Training Data:")
display(train_df.head())
print("Test Data:")
display(test_df.head())

# Check the structure of the datasets
print("Training Data Info:")
train_df.info()
print("Test Data Info:")
test_df.info()

# Check for missing values
print("Missing values in Training Data:")
print(train_df.isnull().sum())
print("Missing values in Test Data:")
print(test_df.isnull().sum())

# Check the distribution of the target variable
print("Distribution of the target variable in Training Data:")
print(train_df['Outcome Variable'].value_counts(normalize=True))

# Check the distribution of categorical and numerical features
categorical_features = train_df.select_dtypes(include=['object']).columns.tolist()
numerical_features = train_df.select_dtypes(include=['int64', 'float64']).columns.tolist()

print("Categorical Features:")
print(categorical_features)
print("Numerical Features:")
print(numerical_features)

# Visualize the distribution of numerical features
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 10))
for i, feature in enumerate(numerical_features, 1):
    plt.subplot(3, 3, i)
    sns.histplot(train_df[feature], kde=True)
    plt.title(f'Distribution of {feature}')
plt.tight_layout()
plt.show()

# Visualize the distribution of categorical features
plt.figure(figsize=(15, 10))
for i, feature in enumerate(categorical_features, 1):
    plt.subplot(3, 3, i)
    sns.countplot(x=feature, data=train_df)
    plt.title(f'Distribution of {feature}')
plt.tight_layout()
plt.show()
```

```
from metagpt.tools.libs.data_preprocess import OneHotEncode

# Copy the DataFrames to avoid modifying the original data
train_df_copy = train_df.copy()
test_df_copy = test_df.copy()

# List of categorical features to be one-hot encoded (excluding 'Disease')
categorical_features_to_encode = [col for col in categorical_features if col != 'Disease']

# Initialize the OneHotEncode tool
one_hot_encoder = OneHotEncode(features=categorical_features_to_encode)

# Fit and transform the training data
train_df_encoded = one_hot_encoder.fit_transform(train_df_copy)

# Transform the test data using the fitted model
test_df_encoded = one_hot_encoder.transform(test_df_copy)

# Display the first few rows of the encoded training and test data
print("Encoded Training Data:")
display(train_df_encoded.head())
print("Encoded Test Data:")
display(test_df_encoded.head())
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Using the latest DataFrame from the finished tasks
column_info_train = get_column_info(train_df_encoded)
column_info_test = get_column_info(test_df_encoded)

print("Column Info for Encoded Training Data:")
print(column_info_train)
print("\nColumn Info for Encoded Test Data:")
print(column_info_test)
```

```
from metagpt.tools.libs.feature_engineering import GeneralSelection
from sklearn.feature_selection import SelectKBest, chi2

# Copy the encoded data
train_df_encoded_copy = train_df_encoded.copy()
test_df_encoded_copy = test_df_encoded.copy()

# Remove the label column from the feature selection process
X_train = train_df_encoded_copy.drop(columns=['Outcome Variable_Negative', 'Outcome Variable_Positive'])
X_test = test_df_encoded_copy.drop(columns=['Outcome Variable_Negative', 'Outcome Variable_Positive'])
y_train = train_df_encoded_copy['Outcome Variable_Positive']

# Apply GeneralSelection to remove features with only one unique value
general_selector = GeneralSelection(label_col='Outcome Variable_Positive')
X_train_general = general_selector.fit_transform(X_train)
X_test_general = general_selector.transform(X_test)

# Apply SelectKBest to select the top 5 features
k_best_selector = SelectKBest(score_func=chi2, k=5)
X_train_k_best = k_best_selector.fit_transform(X_train_general, y_train)
X_test_k_best = k_best_selector.transform(X_test_general)

# Get the names of the selected features
selected_features = X_train_general.columns[k_best_selector.get_support()]
print("Selected Features:", selected_features)

# Add the label column back to the selected features
X_train_k_best_df = pd.DataFrame(X_train_k_best, columns=selected_features, index=X_train_general.index)
X_test_k_best_df = pd.DataFrame(X_test_k_best, columns=selected_features, index=X_test_general.index)
X_train_k_best_df['Outcome Variable_Positive'] = y_train
X_test_k_best_df['Outcome Variable_Positive'] = test_df_encoded_copy['Outcome Variable_Positive']

# Display the selected features in the training and test sets
print("Training Data with Selected Features:")
display(X_train_k_best_df.head())
print("Test Data with Selected Features:")
display(X_test_k_best_df.head())
```
