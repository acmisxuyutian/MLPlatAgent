```
import pandas as pd

# Load the datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/03_coffee_quality/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/03_coffee_quality/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Drop duplicated rows
train_df = train_df.drop_duplicates()
test_df = test_df.drop_duplicates()

# Drop specified columns
columns_to_drop = ["ID", "ICO Number", "Owner", "Region", "Certification Contact", "Certification Address", "Farm Name", "Lot Number", "Mill", "ICO Number", "Producer", "Company", "Expiration", "Harvest Year", "Unnamed: 0", "Number of Bags", "Bag Weight", "In-Country Partner", "Country of Origin", "Processing Method", "Grading Date", "Variety", "Status", "Defects", "Uniformity", "Clean Cup", "Sweetness", "Certification Body"]
train_df = train_df.drop(columns=columns_to_drop, errors='ignore')
test_df = test_df.drop(columns=columns_to_drop, errors='ignore')

# Display the first few rows of the processed datasets
train_df.head(), test_df.head()
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

column_info_train = get_column_info(train_df)
column_info_test = get_column_info(test_df)
print("Train column_info")
print(column_info_train)
print("\nTest column_info")
print(column_info_test)
```

```
from metagpt.tools.libs.data_preprocess import OneHotEncode

# Copy the DataFrames to avoid modifying the original ones
train_df_copy = train_df.copy()
test_df_copy = test_df.copy()

# Initialize the OneHotEncode tool with the 'Color' column
one_hot_encoder = OneHotEncode(features=['Color'])

# Fit and transform the training data
train_df_encoded = one_hot_encoder.fit_transform(train_df_copy)

# Transform the test data using the fitted model
test_df_encoded = one_hot_encoder.transform(test_df_copy)

train_df_encoded.head(), test_df_encoded.head()
```

```
from metagpt.tools.libs.data_preprocess import get_column_info

# Using the latest DataFrame after one-hot encoding
column_info_train = get_column_info(train_df_encoded)
column_info_test = get_column_info(test_df_encoded)

print("Train DataFrame column_info")
print(column_info_train)
print("\nTest DataFrame column_info")
print(column_info_test)
```

```
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error

# Define the target and features
target = 'Total Cup Points'
features = [col for col in train_df_encoded.columns if col != target]

# Initialize the model with some hyperparameters
model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42)

# Train the model
model.fit(train_df_encoded[features], train_df_encoded[target])

# Make predictions on the test set
predictions = model.predict(test_df_encoded[features])

# Evaluate the model
mse = mean_squared_error(test_df_encoded[target], predictions)
mse
```

```
from metagpt.tools.libs.data_preprocess import FillMissingValue
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error

# Define the target and features
target = 'Total Cup Points'
features = [col for col in train_df_encoded.columns if col != target]

# Fill missing values in the training and test datasets
fill_missing = FillMissingValue(features=features, strategy='mean')
train_df_encoded = fill_missing.fit_transform(train_df_encoded)
test_df_encoded = fill_missing.transform(test_df_encoded)

# Initialize the model with some hyperparameters
model = RandomForestRegressor(n_estimators=100, max_depth=20, random_state=42)

# Train the model
model.fit(train_df_encoded[features], train_df_encoded[target])

# Make predictions on the test set
predictions = model.predict(test_df_encoded[features])

# Evaluate the model
mse = mean_squared_error(test_df_encoded[target], predictions)
mse
```
