```

import pandas as pd

# 加载训练集
train_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/train.csv'
train_df = pd.read_csv(train_file_path)

# 查看数据集的前几行
print(train_df.head())
```

```


# 检查数据集中是否存在缺失值
print(train_df.isnull().sum())

# 查看每个特征的基本统计信息
print(train_df.describe())
```

```


from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# 分离特征和目标变量
X = train_df.drop(columns=['id', 'yield'])
y = train_df['yield']

# 标准化特征
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 将数据集划分为训练集和测试集
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# 查看标准化后的数据
print(X_train[:5])  # 打印前5行标准化后的特征数据
```

```


from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import cross_val_score

# 定义模型
models = {
    'Linear Regression': LinearRegression(),
    'Random Forest': RandomForestRegressor(random_state=42),
    'XGBoost': XGBRegressor(random_state=42)
}

# 交叉验证选择最佳模型
best_model = None
best_mae = float('inf')

for name, model in models.items():
    scores = cross_val_score(model, X_train, y_train, cv=5, scoring='neg_mean_absolute_error')
    mae = -scores.mean()
    print(f'{name} MAE: {mae:.2f}')
    if mae < best_mae:
        best_mae = mae
        best_model = model

# 使用最佳模型进行训练
best_model.fit(X_train, y_train)

# 在测试集上评估性能
y_pred = best_model.predict(X_test)
test_mae = mean_absolute_error(y_test, y_pred)
print(f'Best Model: {best_model.__class__.__name__}')
print(f'Test Set MAE: {test_mae:.2f}')

```
