```

import pandas as pd

# 加载训练数据
train_csv_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/mohs_hardness/train.csv'
train_data = pd.read_csv(train_csv_path)

# 查看数据基本信息
print("训练数据的基本信息：")
print(train_data.info())
print("\n数据的前几行：")
print(train_data.head())
print("\n数据的描述性统计：")
print(train_data.describe())
```

```


from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 去除 id 列
X = train_data.drop(columns=['id', 'Hardness'])
y = train_data['Hardness']

# 标准化数值特征
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 划分数据集
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

print("数据集划分完成，训练集和测试集的形状：")
print(f"X_train: {X_train.shape}, y_train: {y_train.shape}")
print(f"X_test: {X_test.shape}, y_test: {y_test.shape}")

```

```


from sklearn.linear_model import LinearRegression
from sklearn.metrics import median_absolute_error

# 选择线性回归模型
model = LinearRegression()

# 训练模型
model.fit(X_train, y_train)

# 预测测试集
y_pred = model.predict(X_test)

# 计算中位绝对误差
medae = median_absolute_error(y_test, y_pred)

print(f"线性回归模型的中位绝对误差 (MedAE)：{medae:.4f}")
```

```



from sklearn.ensemble import RandomForestRegressor

# 选择随机森林模型
model = RandomForestRegressor(random_state=42)

# 训练模型
model.fit(X_train, y_train)

# 预测测试集
y_pred = model.predict(X_test)

# 计算中位绝对误差
medae = median_absolute_error(y_test, y_pred)

print(f"随机森林模型的中位绝对误差 (MedAE)：{medae:.4f}")

```
