```

import pandas as pd

# 数据集地址
file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/train.csv'

# 加载数据
data = pd.read_csv(file_path)

# 显示数据的前5行
data.head()
```

```

# 检查数据信息，包括各列的数据类型和非空值计数
data_info = data.info()

# 检查数据描述，包括统计指标如均值、标准差等
data_description = data.describe()

# 检查每个列的缺失值数量
missing_values = data.isna().sum()

data_info, data_description, missing_values
```

```


# 去除不需要的列
data = data.drop(columns=['id', 'CustomerId', 'Surname'])

# 对类别变量进行编码
data['Geography'] = data['Geography'].astype('category').cat.codes
data['Gender'] = data['Gender'].astype('category').cat.codes

# 检查处理后的数据
data.head()

```

```


from sklearn.model_selection import train_test_split

# 定义特征和目标变量
X = data.drop(columns=['Exited'])
y = data['Exited']

# 分割数据集为训练集和测试集
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 检查分割后的数据集大小
X_train.shape, X_test.shape, y_train.shape, y_test.shape

```

```


from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

# 初始化逻辑回归模型
model = LogisticRegression(max_iter=1000, random_state=42)

# 训练模型
model.fit(X_train, y_train)

# 预测测试集的概率
y_pred_proba = model.predict_proba(X_test)[:, 1]

# 计算AUC-ROC分数
auc_roc = roc_auc_score(y_test, y_pred_proba)

auc_roc


```

```



from sklearn.preprocessing import StandardScaler

# 初始化标准化器
scaler = StandardScaler()

# 对训练集和测试集进行标准化
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 重新初始化逻辑回归模型
model = LogisticRegression(max_iter=1000, random_state=42)

# 再次训练模型
model.fit(X_train_scaled, y_train)

# 预测测试集的概率
y_pred_proba = model.predict_proba(X_test_scaled)[:, 1]

# 计算AUC-ROC分数
auc_roc_scaled = roc_auc_score(y_test, y_pred_proba)

auc_roc_scaled


```
