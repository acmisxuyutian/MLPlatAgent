```


import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

# 加载训练数据
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/train.csv'
train_df = pd.read_csv(train_path)

# 检查缺失值
print(train_df.isnull().sum())

# 对分类特征进行编码
label_encoder = LabelEncoder()
train_df['Sex'] = label_encoder.fit_transform(train_df['Sex'])

# 分离特征和目标变量
X = train_df.drop(columns=['id', 'Rings'])
y = train_df['Rings']

# 特征缩放
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 划分训练集和测试集
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# 查看预处理后的数据
print(X_train[:5])
print(y_train[:5])
```

```


from sklearn.ensemble import RandomForestRegressor

# 初始化随机森林回归模型
model = RandomForestRegressor(n_estimators=100, random_state=42)

# 训练模型
model.fit(X_train, y_train)

# 在测试集上进行预测
y_pred = model.predict(X_test)

# 查看预测结果的前几个值
print(y_pred[:5])

```

```



from sklearn.metrics import mean_squared_error
import numpy as np

# 计算 RMLSE
rmlse = np.sqrt(mean_squared_error(np.log1p(y_test), np.log1p(y_pred)))

# 输出 RMLSE
print(f"RMLSE: {rmlse}")

```
