```

import pandas as pd

# 假设CSV文件的第一行是表头
train_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/train.csv'
train_data = pd.read_csv(train_file_path)

# 显示前5行
train_data.head()
```

```

# 更改有重复名称的列名称
train_data.columns = ['id', 'Sex', 'Length', 'Diameter', 'Height',
                      'Whole_weight', 'Shucked_weight', 'Viscera_weight', 'Shell_weight', 'Rings']

# 检查缺失值
missing_values = train_data.isnull().sum()

# 将分类变量(sex)转换为数值变量
train_data['Sex'] = train_data['Sex'].map({'F': 0, 'M': 1, 'I': 2})

# 显示数据的基本信息和描述性统计
print(missing_values)
print(train_data.dtypes)
train_data.describe(include='all')

```

```

# 更改有重复名称的列名称
train_data.columns = ['id', 'Sex', 'Length', 'Diameter', 'Height',
                      'Whole_weight', 'Shucked_weight', 'Viscera_weight', 'Shell_weight', 'Rings']

# 检查缺失值
missing_values = train_data.isnull().sum()

# 将分类变量(sex)转换为数值变量
train_data['Sex'] = train_data['Sex'].map({'F': 0, 'M': 1, 'I': 2})

# 显示数据的基本信息和描述性统计
print(missing_values)
print(train_data.dtypes)
train_data.describe(include='all')

```

```


import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# 检查Rings列的分布
plt.figure(figsize=(12, 6))
sns.histplot(train_data['Rings'], kde=True, bins=30)
plt.title('Distribution of Rings (Original)')
plt.show()

# 对Rings列进行对数转换，并查看转换后的分布
train_data['Rings_Log'] = np.log1p(train_data['Rings'])  # log(1 + x) to handle zero values
plt.figure(figsize=(12, 6))
sns.histplot(train_data['Rings_Log'], kde=True, bins=30)
plt.title('Distribution of Rings (Log Transformed)')
plt.show()

```
