```

import pandas as pd

# 加载训练集
train_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/kidney_stone/train.csv'
train_df = pd.read_csv(train_file_path)

# 查看数据集基本信息
train_df.info()
train_df.head()
```

```


# 描述性统计
train_df.describe(include='all')

```

```


import seaborn as sns
import matplotlib.pyplot as plt

# 设置绘图风格
sns.set(style='whitegrid')

# 绘制箱线图
plt.figure(figsize=(15, 8))
sns.boxplot(data=train_df.drop(['id', 'target'], axis=1))
plt.title('Box Plot of Features')
plt.show()

# 绘制直方图
plt.figure(figsize=(15, 10))
for i, col in enumerate(train_df.drop(['id', 'target'], axis=1).columns):
    plt.subplot(3, 2, i+1)
    sns.histplot(train_df[col], kde=True)
    plt.title(f'Histogram of {col}')
plt.tight_layout()
plt.show()

# 绘制散点图矩阵
sns.pairplot(train_df.drop('id', axis=1), hue='target')
plt.show()

```
