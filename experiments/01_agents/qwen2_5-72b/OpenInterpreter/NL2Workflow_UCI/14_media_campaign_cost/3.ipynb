```

import pandas as pd

# 加载数据
data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/train.csv'
df = pd.read_csv(data_path)

# 显示数据的前几行
df.head()
```

```

# 查看数据描述
print(df.describe())

# 检查缺失值
print(df.isnull().sum())

# 查看目标变量 cost 的分布
import matplotlib.pyplot as plt
plt.hist(df['cost'], bins=50, color='skyblue', edgecolor='black')
plt.xlabel('Cost')
plt.ylabel('Frequency')
plt.title('Distribution of Media Campaign Cost')
plt.show()
```
