```

import pandas as pd

# 加载数据集
file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/train.csv'
df = pd.read_csv(file_path)

# 查看前几行数据以了解数据集
df.head()
```

```


# 检查数据的总体信息
df.info()

# 查看数据的基本统计信息
df.describe()

```

```


# 检查数据的总体信息
df.info()

# 查看数据的基本统计信息
df.describe()

```

```


import matplotlib.pyplot as plt
import seaborn as sns

# 绘制目标变量 cost 的直方图
plt.figure(figsize=(10, 6))
sns.histplot(df['cost'], kde=True)
plt.title('Distribution of Cost')
plt.xlabel('Cost')
plt.ylabel('Frequency')
plt.show()

# 绘制相关矩阵
plt.figure(figsize=(12, 10))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()

# 绘制一些特征与目标变量之间的散点图
plt.figure(figsize=(15, 10))
plt.subplot(2, 3, 1)
sns.scatterplot(x='store_sales', y='cost', data=df)
plt.title('Store Sales vs Cost')

plt.subplot(2, 3, 2)
sns.scatterplot(x='unit_sales', y='cost', data=df)
plt.title('Unit Sales vs Cost')

plt.subplot(2, 3, 3)
sns.scatterplot(x='store_sqft', y='cost', data=df)
plt.title('Store Sqft vs Cost')

plt.subplot(2, 3, 4)
sns.scatterplot(x='total_children', y='cost', data=df)
plt.title('Total Children vs Cost')

plt.subplot(2, 3, 5)
sns.scatterplot(x='gross_weight', y='cost', data=df)
plt.title('Gross Weight vs Cost')

plt.subplot(2, 3, 6)
sns.scatterplot(x='avg_cars_at_home', y='cost', data=df)
plt.title('Avg Cars at Home vs Cost')

plt.tight_layout()
plt.show()

```
