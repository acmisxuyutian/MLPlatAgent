```

import pandas as pd

# 指定训练数据集的路径
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/train.csv'

# 加载训练数据
df_train = pd.read_csv(train_data_path)

# 查看数据的前几行
df_train.head()
```

```

# 查看数据的一般信息
df_train.info()

# 描述性统计
df_train.describe(include='all')

# 检查目标变量 'cost' 的分布
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(df_train['cost'], kde=True)
plt.title('Distribution of Cost')
plt.show()
```
