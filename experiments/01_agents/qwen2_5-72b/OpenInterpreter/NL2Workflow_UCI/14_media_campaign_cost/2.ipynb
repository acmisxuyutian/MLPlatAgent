```

import pandas as pd

# 加载训练数据
train_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/train.csv'
train_data = pd.read_csv(train_file_path)

# 查看数据的前几行
train_data.head()
```

```


# 检查缺失值
train_data.isnull().sum()

```

```


import matplotlib.pyplot as plt
import seaborn as sns

# 查看目标变量 'cost' 的分布
plt.figure(figsize=(10, 6))
sns.histplot(train_data['cost'], kde=True)
plt.title('Distribution of Media Campaign Cost')
plt.xlabel('Cost')
plt.ylabel('Frequency')
plt.show()

# 计算特征之间的相关性
correlation_matrix = train_data.corr()
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()

# 查看目标变量与各特征的相关性
correlation_with_cost = correlation_matrix['cost'].sort_values(ascending=False)
print(correlation_with_cost)
```
