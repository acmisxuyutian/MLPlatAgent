```

import pandas as pd

# 读取训练数据
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/train.csv'
train_data = pd.read_csv(train_data_path)

# 显示数据集的前几行
train_data.head()
```

```


# 检查数据集的基本信息
train_data.info()
```

```


# 描述性统计
train_data.describe(include='all')
```

```


import matplotlib.pyplot as plt
import seaborn as sns

# 绘制数值特征的分布图
numerical_features = ['Length', 'Diameter', 'Height', 'Weight', 'Shucked_Weight', 'Viscera_Weight', 'Shell_Weight']

# 设置图形大小
plt.figure(figsize=(15, 10))

# 绘制分布图
for i, feature in enumerate(numerical_features, 1):
    plt.subplot(2, 4, i)
    sns.histplot(train_data[feature], kde=True)
    plt.title(f'Distribution of {feature}')

plt.tight_layout()
plt.show()

# 绘制箱形图
plt.figure(figsize=(15, 10))

for i, feature in enumerate(numerical_features, 1):
    plt.subplot(2, 4, i)
    sns.boxplot(y=train_data[feature])
    plt.title(f'Boxplot of {feature}')

plt.tight_layout()
plt.show()
```
