```

import pandas as pd

# 加载训练数据集
file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/train.csv'
train_df = pd.read_csv(file_path)

# 查看数据集的基本信息
train_df.info()
# 显示数据集的前五行
train_df.head()
```

```

import matplotlib.pyplot as plt
import seaborn as sns

# 检查每个类别（Category）的数量
category_counts = train_df['Category'].value_counts()

# 目标变量（Category）的分布
plt.figure(figsize=(12, 8))
sns.countplot(y='Category', data=train_df, order=category_counts.index)
plt.title('Distribution of Crime Categories')
plt.show()

# 查看DayOfWeek的分布
plt.figure(figsize=(10, 6))
sns.countplot(x='DayOfWeek', data=train_df)
plt.title('Distribution of Crime by Day of Week')
plt.show()

# 查看PdDistrict的分布
plt.figure(figsize=(10, 6))
sns.countplot(x='PdDistrict', data=train_df)
plt.title('Distribution of Crime by Police District')
plt.xticks(rotation=45)
plt.show()

# 查看地理位置分布（X, Y）
plt.figure(figsize=(10, 10))
sns.scatterplot(x='X', y='Y', data=train_df, alpha=0.3)
plt.title('Geographical Distribution of Crimes')
plt.show()

# 输出经度和纬度的统计描述
train_df[['X', 'Y']].describe()
```
