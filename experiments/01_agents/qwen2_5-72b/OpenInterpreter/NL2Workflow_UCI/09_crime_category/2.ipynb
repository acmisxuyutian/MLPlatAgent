```

import pandas as pd

# 加载训练数据
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/train.csv'
train_df = pd.read_csv(train_data_path)

# 查看数据的前几行
train_df.head()
```

```


# 检查数据中是否有缺失值
train_df.isnull().sum()

```

```


from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split

# 对目标变量进行编码
label_encoder = LabelEncoder()
train_df['Category'] = label_encoder.fit_transform(train_df['Category'])

# 定义特征和目标变量
X = train_df.drop(columns=['Category'])
y = train_df['Category']

# 对分类变量进行 one-hot 编码
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(), ['DayOfWeek', 'PdDistrict'])],
    remainder='passthrough'
)

# 应用预处理
X_encoded = preprocessor.fit_transform(X)

# 将数据集分为训练集和测试集
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

# 查看编码后的数据
X_train[:5], y_train[:5]

```

```



from sklearn.linear_model import LogisticRegression
from sklearn.metrics import log_loss

# 训练逻辑回归模型
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train, y_train)

# 预测测试集的概率
y_pred_proba = model.predict_proba(X_test)

# 计算对数损失
log_loss_value = log_loss(y_test, y_pred_proba)

log_loss_value


```

```



# 增加迭代次数
model = LogisticRegression(max_iter=5000, random_state=42)
model.fit(X_train, y_train)

# 预测测试集的概率
y_pred_proba = model.predict_proba(X_test)

# 计算对数损失
log_loss_value = log_loss(y_test, y_pred_proba)

log_loss_value

```

```



from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# 创建标准化和逻辑回归的流水线
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('logreg', LogisticRegression(max_iter=5000, random_state=42))
])

# 训练模型
pipeline.fit(X_train, y_train)

# 预测测试集的概率
y_pred_proba = pipeline.predict_proba(X_test)

# 计算对数损失
log_loss_value = log_loss(y_test, y_pred_proba)

log_loss_value


```

```




# 创建标准化和逻辑回归的流水线，并禁用中心化操作
pipeline = Pipeline([
    ('scaler', StandardScaler(with_mean=False)),
    ('logreg', LogisticRegression(max_iter=5000, random_state=42))
])

# 训练模型
pipeline.fit(X_train, y_train)

# 预测测试集的概率
y_pred_proba = pipeline.predict_proba(X_test)

# 计算对数损失
log_loss_value = log_loss(y_test, y_pred_proba)

log_loss_value



```

```




# 使用saga优化器
model = LogisticRegression(max_iter=5000, solver='saga', random_state=42)
model.fit(X_train, y_train)

# 预测测试集的概率
y_pred_proba = model.predict_proba(X_test)

# 计算对数损失
log_loss_value = log_loss(y_test, y_pred_proba)

log_loss_value


```

```




from sklearn.model_selection import GridSearchCV

# 定义参数网格
param_grid = {'C': [0.001, 0.01, 0.1, 1, 10, 100]}

# 创建GridSearchCV对象
grid_search = GridSearchCV(
    LogisticRegression(max_iter=10000, solver='saga', random_state=42),
    param_grid,
    cv=5,
    scoring='neg_log_loss',
    n_jobs=-1
)

# 进行网格搜索
grid_search.fit(X_train, y_train)

# 打印最佳参数和最佳得分
best_params = grid_search.best_params_
best_score = -grid_search.best_score_

best_params, best_score



```
