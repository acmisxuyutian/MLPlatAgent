```


import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Load the datasets again for consistency
train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Drop the 'Disease' column as it will not be used in the models
train_df = train_df.drop(columns=['Disease'])
test_df = test_df.drop(columns=['Disease'])

# Identify categorical columns (excluding 'Outcome Variable' which is the target)
categorical_columns = ['Fever', 'Cough', 'Fatigue', 'Difficulty Breathing', 'Gender', 'Blood Pressure', 'Cholesterol Level']

# One-hot encode the categorical columns
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
train_encoded = encoder.fit_transform(train_df[categorical_columns])
test_encoded = encoder.transform(test_df[categorical_columns])

# Create new DataFrames with the one-hot encoded features
train_encoded_df = pd.DataFrame(train_encoded, columns=encoder.get_feature_names_out(categorical_columns))
test_encoded_df = pd.DataFrame(test_encoded, columns=encoder.get_feature_names_out(categorical_columns))

# Combine the encoded features with the numerical feature 'Age'
train_df_final = pd.concat([train_df[['Age']], train_encoded_df], axis=1)
test_df_final = pd.concat([test_df[['Age']], test_encoded_df], axis=1)

# Separate the target variable
train_target = train_df['Outcome Variable']
test_target = test_df['Outcome Variable']

# Display the first few rows of the final training and test datasets
print("Final Training Data:\n", train_df_final.head())
print("\nFinal Test Data:\n", test_df_final.head())
```

```



from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import RFE

# Initialize the logistic regression model
logreg = LogisticRegression(max_iter=1000, solver='liblinear')

# Perform RFE to select the top 5 features
rfe = RFE(estimator=logreg, n_features_to_select=5, step=1)
rfe.fit(train_df_final, train_target)

# Get the selected features
selected_features = train_df_final.columns[rfe.support_]

print("Selected Features:", selected_features)
```

```



from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Initialize the logistic regression model
logreg_all = LogisticRegression(max_iter=1000, solver='liblinear')
logreg_selected = LogisticRegression(max_iter=1000, solver='liblinear')

# Train Model 1 using all features
logreg_all.fit(train_df_final, train_target)

# Train Model 2 using the selected features
logreg_selected.fit(train_df_final[selected_features], train_target)

# Predict on the test set using Model 1
test_predictions_all = logreg_all.predict(test_df_final)

# Predict on the test set using Model 2
test_predictions_selected = logreg_selected.predict(test_df_final[selected_features])

# Evaluate Model 1
accuracy_all = accuracy_score(test_target, test_predictions_all)
report_all = classification_report(test_target, test_predictions_all)

# Evaluate Model 2
accuracy_selected = accuracy_score(test_target, test_predictions_selected)
report_selected = classification_report(test_target, test_predictions_selected)

# Print the evaluation results
print("Model 1 (All Features) - Accuracy:", accuracy_all)
print("Model 1 (All Features) - Classification Report:\n", report_all)

print("\nModel 2 (Top 5 Features) - Accuracy:", accuracy_selected)
print("Model 2 (Top 5 Features) - Classification Report:\n", report_selected)

```
