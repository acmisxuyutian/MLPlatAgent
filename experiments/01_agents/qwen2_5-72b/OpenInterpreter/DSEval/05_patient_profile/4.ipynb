```



import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Load the training and test datasets
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark\DSEval\datasets\05_patient_profile\train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark\DSEval\datasets\05_patient_profile\test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# One-hot encode categorical features (excluding 'Disease')
categorical_features = [col for col in train_df.columns if train_df[col].dtype == 'object' and col != 'Disease']
train_encoded = pd.get_dummies(train_df, columns=categorical_features, drop_first=True)
test_encoded = pd.get_dummies(test_df, columns=categorical_features, drop_first=True)

# Ensure both train and test datasets have the same columns
missing_cols = set(train_encoded.columns) - set(test_encoded.columns)
for c in missing_cols:
    test_encoded[c] = 0
test_encoded = test_encoded[train_encoded.columns]

# Check the columns of the train_encoded DataFrame
print("Columns in train_encoded:")
print(train_encoded.columns)

# Check the unique values in the 'Outcome Variable' column
print("\nUnique values in 'Outcome Variable' column in train_encoded:")
print(train_encoded['Outcome Variable_Positive'].unique())

# Split the data into features and target variable
X_train = train_encoded.drop(columns=['Disease', 'Outcome Variable_Positive'])
y_train = train_encoded['Outcome Variable_Positive'].astype(int)

# Ensure the test data has the same target variable column
X_test = test_encoded.drop(columns=['Disease', 'Outcome Variable_Positive'])
y_test = test_encoded['Outcome Variable_Positive'].astype(int)

# Display the first few rows of the preprocessed data
print("\nPreprocessed Training Data (Features):")
print(X_train.head())
print("\nPreprocessed Training Data (Target):")
print(y_train.head())
print("\nPreprocessed Test Data (Features):")
print(X_test.head())
print("\nPreprocessed Test Data (Target):")
print(y_test.head())

```

```


import pandas as pd
from sklearn.feature_selection import SelectKBest, chi2

# Feature selection using SelectKBest with chi-squared test
kbest = SelectKBest(score_func=chi2, k=5)
kbest.fit(X_train, y_train)

# Get the indices of the top 5 features
top_features_indices = kbest.get_support(indices=True)
top_features = X_train.columns[top_features_indices]

print("Top 5 features:")
print(top_features)

```

```



import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Build logistic regression model1 using all features
model1 = LogisticRegression()
model1.fit(X_train, y_train)

# Predict on the test set
y_pred1 = model1.predict(X_test)

# Evaluate model1
accuracy1 = accuracy_score(y_test, y_pred1)
report1 = classification_report(y_test, y_pred1)

print("Model1 (All Features) Performance:")
print(f"Accuracy: {accuracy1}")
print("Classification Report:\n", report1)

# Build logistic regression model2 using only the top 5 features
X_train_top5 = X_train[top_features]
X_test_top5 = X_test[top_features]

model2 = LogisticRegression()
model2.fit(X_train_top5, y_train)

# Predict on the test set
y_pred2 = model2.predict(X_test_top5)

# Evaluate model2
accuracy2 = accuracy_score(y_test, y_pred2)
report2 = classification_report(y_test, y_pred2)

print("\nModel2 (Top 5 Features) Performance:")
print(f"Accuracy: {accuracy2}")
print("Classification Report:\n", report2)


```
