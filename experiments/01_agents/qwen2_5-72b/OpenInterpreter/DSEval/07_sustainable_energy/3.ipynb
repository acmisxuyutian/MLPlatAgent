```

import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

# Paths to the datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/07_sustainable_energy/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/07_sustainable_energy/test.csv'

# Load the datasets
train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the training data to understand its structure
train_df.head()
```

```


# Define the target variable and features
target = 'Renewable energy share in the total final energy consumption (%)'
features = [col for col in train_df.columns if col not in ['Entity', 'Year', target]]

# Prepare the training data
X_train = train_df[features]
y_train = train_df[target]

# Prepare the test data
X_test = test_df[features]
y_test = test_df[target]

# Display the shapes of the prepared data
X_train.shape, y_train.shape, X_test.shape, y_test.shape

```

```


# Create and fit the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Display the coefficients of the model
model.coef_


```

```


# Function to clean the 'gdp_per_capita' column
def clean_gdp_per_capita(value):
    if isinstance(value, str):
        value = value.replace(',', '')
    return float(value)

# Apply the cleaning function to the 'gdp_per_capita' column in both datasets
train_df['gdp_per_capita'] = train_df['gdp_per_capita'].apply(clean_gdp_per_capita)
test_df['gdp_per_capita'] = test_df['gdp_per_capita'].apply(clean_gdp_per_capita)

# Re-prepare the data after cleaning
X_train = train_df[features]
y_train = train_df[target]

X_test = test_df[features]
y_test = test_df[target]

# Display the shapes of the prepared data to confirm
X_train.shape, y_train.shape, X_test.shape, y_test.shape
```

```



# Create and fit the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Display the coefficients of the model
model.coef_

```

```



# Function to clean all columns that might be numeric
def clean_numeric_columns(df):
    for col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
    return df

# Clean the training and test datasets
train_df_cleaned = clean_numeric_columns(train_df)
test_df_cleaned = clean_numeric_columns(test_df)

# Fill NaN values with the mean of each column
train_df_cleaned.fillna(train_df_cleaned.mean(), inplace=True)
test_df_cleaned.fillna(test_df_cleaned.mean(), inplace=True)

# Re-prepare the data after cleaning
X_train = train_df_cleaned[features]
y_train = train_df_cleaned[target]

X_test = test_df_cleaned[features]
y_test = test_df_cleaned[target]

# Display the shapes of the prepared data to confirm
X_train.shape, y_train.shape, X_test.shape, y_test.shape


```

```




# Create and fit the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Display the coefficients of the model
model.coef_


```

```



# Make predictions on the test set
y_pred = model.predict(X_test)

# Display the first few predictions
y_pred[:5]


```

```




# Calculate the RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

# Display the RMSE
rmse



```
