```

import pandas as pd

# Load the training and test datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\07_sustainable_energy\\train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\07_sustainable_energy\\test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the training and test datasets
train_df.head(), test_df.head()
```

```

# Identify the target variable
target_column = 'Renewable energy share in the total final energy consumption (%)'

# Drop the 'Entity' and 'Year' columns from the features
features = [col for col in train_df.columns if col not in ['Entity', 'Year', target_column]]

# Split the data into features (X) and target (y)
X_train = train_df[features]
y_train = train_df[target_column]

X_test = test_df[features]
y_test = test_df[target_column]

# Display the first few rows of the prepared data
X_train.head(), y_train.head(), X_test.head(), y_test.head()
```

```


from sklearn.linear_model import LinearRegression

# Initialize the linear regression model
model = LinearRegression()

# Fit the model on the training data
model.fit(X_train, y_train)

# Print the coefficients and intercept of the model
model.coef_, model.intercept_

```

```


# Function to clean the data by converting comma-separated numbers to floats
def clean_data(df):
    for col in df.columns:
        if df[col].dtype == 'object':
            df[col] = df[col].str.replace(',', '').astype(float, errors='coerce')
    return df

# Clean the training and test datasets
X_train_cleaned = clean_data(X_train)
X_test_cleaned = clean_data(X_test)

# Check for any remaining non-numeric values
X_train_cleaned.info(), X_test_cleaned.info()

```

```



# Function to clean the data by converting comma-separated numbers to floats and coercing errors to NaN
def clean_data(df):
    for col in df.columns:
        if df[col].dtype == 'object':
            df[col] = df[col].str.replace(',', '')
            df[col] = pd.to_numeric(df[col], errors='coerce')
    return df

# Clean the training and test datasets
X_train_cleaned = clean_data(X_train)
X_test_cleaned = clean_data(X_test)

# Check for any remaining non-numeric values
X_train_cleaned.info(), X_test_cleaned.info()

```

```


from sklearn.linear_model import LinearRegression

# Initialize the linear regression model
model = LinearRegression()

# Fit the model on the cleaned training data
model.fit(X_train_cleaned, y_train)

# Print the coefficients and intercept of the model
model.coef_, model.intercept_

```

```



# Drop rows with missing values in the training and test datasets
X_train_cleaned = X_train_cleaned.dropna()
y_train = y_train[X_train_cleaned.index]

X_test_cleaned = X_test_cleaned.dropna()
y_test = y_test[X_test_cleaned.index]

# Check the shape of the cleaned datasets
X_train_cleaned.shape, y_train.shape, X_test_cleaned.shape, y_test.shape
```

```



from sklearn.linear_model import LinearRegression

# Initialize the linear regression model
model = LinearRegression()

# Fit the model on the cleaned training data
model.fit(X_train_cleaned, y_train)

# Print the coefficients and intercept of the model
model.coef_, model.intercept_
```

```



# Make predictions on the test set
y_pred = model.predict(X_test_cleaned)

# Display the first few predictions
y_pred[:5]

```

```




from sklearn.metrics import mean_squared_error
import numpy as np

# Calculate the RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

# Print the RMSE
rmse

```
