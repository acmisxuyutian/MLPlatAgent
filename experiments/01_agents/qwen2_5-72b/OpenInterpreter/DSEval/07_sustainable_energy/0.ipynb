```

import pandas as pd

# Load the training and test datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\07_sustainable_energy\\train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\07_sustainable_energy\\test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of each dataset
print("Training Data:")
print(train_df.head())

print("\nTest Data:")
print(test_df.head())
```

```

import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

# Load the datasets
train_df = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\07_sustainable_energy\\train.csv')
test_df = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\07_sustainable_energy\\test.csv')

# Identify the target variable
target_column = 'Renewable energy share in the total final energy consumption (%)'

# Drop 'Entity' and 'Year' columns from the feature set
X_train = train_df.drop(columns=['Entity', 'Year', target_column])
y_train = train_df[target_column]

X_test = test_df.drop(columns=['Entity', 'Year', target_column])
y_test = test_df[target_column]

# Display the first few rows of the feature and target sets
print("X_train (first few rows):")
print(X_train.head())
print("\ny_train (first few rows):")
print(y_train.head())
print("\nX_test (first few rows):")
print(X_test.head())
print("\ny_test (first few rows):")
print(y_test.head())
```

```


# Check the columns in the training and test datasets
print("Training Data Columns:")
print(train_df.columns)

print("\nTest Data Columns:")
print(test_df.columns)

# Drop only the 'Entity' column from the feature set
X_train = train_df.drop(columns=['Entity', target_column])
y_train = train_df[target_column]

X_test = test_df.drop(columns=['Entity', target_column])
y_test = test_df[target_column]

# Display the first few rows of the feature and target sets
print("\nX_train (first few rows):")
print(X_train.head())
print("\ny_train (first few rows):")
print(y_train.head())
print("\nX_test (first few rows):")
print(X_test.head())
print("\ny_test (first few rows):")
print(y_test.head())

```

```


# Train the Linear Regression Model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Display the first few predictions
print("First few predictions:")
print(y_pred[:5])
```

```


# Preprocess the data by removing commas from numeric values
X_train = X_train.apply(lambda x: x.str.replace(',', '.').astype(float) if x.dtype == 'object' else x)
X_test = X_test.apply(lambda x: x.str.replace(',', '.').astype(float) if x.dtype == 'object' else x)

# Display the first few rows of the preprocessed feature sets
print("Preprocessed X_train (first few rows):")
print(X_train.head())
print("\nPreprocessed X_test (first few rows):")
print(X_test.head())
```

```



# Train the Linear Regression Model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Display the first few predictions
print("First few predictions:")
print(y_pred[:5])

```

```



from sklearn.impute import SimpleImputer

# Initialize the imputer to fill missing values with the mean of each column
imputer = SimpleImputer(strategy='mean')

# Fit the imputer on the training data and transform both training and test data
X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)

# Display the first few rows of the imputed feature sets
print("Imputed X_train (first few rows):")
print(X_train_imputed[:5])

print("\nImputed X_test (first few rows):")
print(X_test_imputed[:5])

```

```



# Train the Linear Regression Model on the imputed data
model = LinearRegression()
model.fit(X_train_imputed, y_train)

# Make predictions on the imputed test data
y_pred = model.predict(X_test_imputed)

# Display the first few predictions
print("First few predictions:")
print(y_pred[:5])

# Calculate the Root Mean Squared Error (RMSE)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print("\nRoot Mean Squared Error (RMSE):", rmse)

```

```




# Remove rows with NaN values in the target variable from both the training and test sets
train_df_clean = train_df.dropna(subset=[target_column])
test_df_clean = test_df.dropna(subset=[target_column])

# Reassign the cleaned feature and target sets
X_train_clean = train_df_clean.drop(columns=['Entity', target_column])
y_train_clean = train_df_clean[target_column]

X_test_clean = test_df_clean.drop(columns=['Entity', target_column])
y_test_clean = test_df_clean[target_column]

# Preprocess the data by removing commas from numeric values and converting to float
X_train_clean = X_train_clean.apply(lambda x: x.str.replace(',', '.').astype(float) if x.dtype == 'object' else x)
X_test_clean = X_test_clean.apply(lambda x: x.str.replace(',', '.').astype(float) if x.dtype == 'object' else x)

# Impute missing values in the feature sets
X_train_imputed = imputer.fit_transform(X_train_clean)
X_test_imputed = imputer.transform(X_test_clean)

# Display the first few rows of the cleaned and imputed feature sets
print("Cleaned and Imputed X_train (first few rows):")
print(X_train_imputed[:5])

print("\nCleaned and Imputed X_test (first few rows):")
print(X_test_imputed[:5])

# Display the first few rows of the cleaned target sets
print("\nCleaned y_train (first few rows):")
print(y_train_clean.head())

print("\nCleaned y_test (first few rows):")
print(y_test_clean.head())
```

```




# Train the Linear Regression Model on the cleaned and imputed data
model = LinearRegression()
model.fit(X_train_imputed, y_train_clean)

# Make predictions on the cleaned and imputed test data
y_pred = model.predict(X_test_imputed)

# Display the first few predictions
print("First few predictions:")
print(y_pred[:5])

# Calculate the Root Mean Squared Error (RMSE)
rmse = np.sqrt(mean_squared_error(y_test_clean, y_pred))
print("\nRoot Mean Squared Error (RMSE):", rmse)


```
