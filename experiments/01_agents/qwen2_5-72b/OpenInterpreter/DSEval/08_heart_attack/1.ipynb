```

import pandas as pd

# Load the data
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/08_heart_attack/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/08_heart_attack/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Convert the 'Diet' column
diet_mapping = {'Healthy': 1, 'Average': 2, 'Unhealthy': 3}
train_df['Diet'] = train_df['Diet'].map(diet_mapping)
test_df['Diet'] = test_df['Diet'].map(diet_mapping)

# Display the first few rows of the datasets to verify
train_df.head(), test_df.head()
```

```


# Drop the 'Patient ID' column
train_df = train_df.drop('Patient ID', axis=1)
test_df = test_df.drop('Patient ID', axis=1)

# Separate features and target
X_train = train_df.drop('Heart Attack Risk', axis=1)
y_train = train_df['Heart Attack Risk']

X_test = test_df.drop('Heart Attack Risk', axis=1)
y_test = test_df['Heart Attack Risk']

# Display the first few rows of the processed datasets to verify
X_train.head(), X_test.head(), y_train.head(), y_test.head()

```

```


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

# Initialize the RandomForest model
rf_model = RandomForestClassifier(random_state=42)

# Fit the model on the training data
rf_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_proba = rf_model.predict_proba(X_test)[:, 1]

# Evaluate the performance of the model
roc_auc = roc_auc_score(y_test, y_pred_proba)

roc_auc

```

```


# Encode categorical variables
X_train_encoded = pd.get_dummies(X_train, columns=['Sex', 'Country', 'Continent', 'Hemisphere'], drop_first=True)
X_test_encoded = pd.get_dummies(X_test, columns=['Sex', 'Country', 'Continent', 'Hemisphere'], drop_first=True)

# Display the first few rows of the encoded datasets to verify
X_train_encoded.head(), X_test_encoded.head()
```

```



# Initialize the RandomForest model
rf_model = RandomForestClassifier(random_state=42)

# Fit the model on the training data
rf_model.fit(X_train_encoded, y_train)

# Make predictions on the test set
y_pred_proba = rf_model.predict_proba(X_test_encoded)[:, 1]

# Evaluate the performance of the model
roc_auc = roc_auc_score(y_test, y_pred_proba)

roc_auc

```
