```

import pandas as pd

# Load the training and test datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/08_heart_attack/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/08_heart_attack/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the training and test datasets
train_df.head(), test_df.head()
```

```


# Define the mapping for the 'Diet' column
diet_mapping = {'Healthy': 1, 'Average': 2, 'Unhealthy': 3}

# Apply the mapping to the 'Diet' column in both datasets
train_df['Diet'] = train_df['Diet'].map(diet_mapping)
test_df['Diet'] = test_df['Diet'].map(diet_mapping)

# Display the first few rows of the training and test datasets to verify the conversion
train_df.head(), test_df.head()

```

```


# Display the 'Diet' column values from both datasets to verify the conversion
train_df['Diet'].head(), test_df['Diet'].head()
```

```



from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score

# Drop the 'Patient ID' column as it is not a predictive feature
train_df = train_df.drop(columns=['Patient ID'])
test_df = test_df.drop(columns=['Patient ID'])

# Separate the features and target variable
X_train = train_df.drop(columns=['Heart Attack Risk'])
y_train = train_df['Heart Attack Risk']
X_test = test_df.drop(columns=['Heart Attack Risk'])
y_test = test_df['Heart Attack Risk']

# Initialize and fit the RandomForest model
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_proba = rf_model.predict_proba(X_test)[:, 1]

# Display the predicted probabilities
y_pred_proba[:5]

```

```



# Convert categorical columns to numerical values using one-hot encoding
train_df_encoded = pd.get_dummies(train_df, drop_first=True)
test_df_encoded = pd.get_dummies(test_df, drop_first=True)

# Ensure both datasets have the same columns
missing_cols = set(train_df_encoded.columns) - set(test_df_encoded.columns)
for c in missing_cols:
    test_df_encoded[c] = 0

# Reorder the columns to match
test_df_encoded = test_df_encoded[train_df_encoded.columns]

# Separate the features and target variable
X_train = train_df_encoded.drop(columns=['Heart Attack Risk'])
y_train = train_df_encoded['Heart Attack Risk']
X_test = test_df_encoded.drop(columns=['Heart Attack Risk'])
y_test = test_df_encoded['Heart Attack Risk']

# Initialize and fit the RandomForest model
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_proba = rf_model.predict_proba(X_test)[:, 1]

# Display the predicted probabilities
y_pred_proba[:5]

```

```



# Calculate the ROC AUC score
roc_auc = roc_auc_score(y_test, y_pred_proba)

# Return the ROC AUC score as a tuple
(roc_auc,)
```
