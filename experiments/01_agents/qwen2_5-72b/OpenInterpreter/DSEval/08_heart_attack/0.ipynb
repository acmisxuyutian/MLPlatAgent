```

import pandas as pd

# Define the file paths
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\08_heart_attack\\train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\08_heart_attack\\test.csv'

# Load the datasets
train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Convert 'Diet' column values
diet_mapping = {'Healthy': 1, 'Average': 2, 'Unhealthy': 3}
train_df['Diet'] = train_df['Diet'].map(diet_mapping)
test_df['Diet'] = test_df['Diet'].map(diet_mapping)

# Print the first few rows of the datasets to verify the conversion
print("Train Data (first 5 rows):")
print(train_df.head())
print("\nTest Data (first 5 rows):")
print(test_df.head())
```

```

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split

# Split the datasets into features and target variable
X_train = train_df.drop(columns=['Patient ID', 'Heart Attack Risk'])
y_train = train_df['Heart Attack Risk']

X_test = test_df.drop(columns=['Patient ID', 'Heart Attack Risk'])
y_test = test_df['Heart Attack Risk']

# Initialize the RandomForest model
rf_model = RandomForestClassifier(random_state=42)

# Fit the model on the training data
rf_model.fit(X_train, y_train)

# Check the feature names and the model's feature importances
rf_model.feature_importances_
```

```


from sklearn.preprocessing import LabelEncoder

# Initialize LabelEncoder
label_encoder = LabelEncoder()

# Encode categorical columns
categorical_columns = ['Sex', 'Continent', 'Hemisphere']

# Fit and transform the training data
for col in categorical_columns:
    train_df[col] = label_encoder.fit_transform(train_df[col])
    test_df[col] = label_encoder.transform(test_df[col])

# Split the datasets into features and target variable again
X_train = train_df.drop(columns=['Patient ID', 'Heart Attack Risk'])
y_train = train_df['Heart Attack Risk']

X_test = test_df.drop(columns=['Patient ID', 'Heart Attack Risk'])
y_test = test_df['Heart Attack Risk']

# Initialize the RandomForest model
rf_model = RandomForestClassifier(random_state=42)

# Fit the model on the training data
rf_model.fit(X_train, y_train)

# Check the feature names and the model's feature importances
rf_model.feature_importances_

```

```



# Identify the unique values in the 'Continent' column for both training and test datasets
unique_train_continents = train_df['Continent'].unique()
unique_test_continents = test_df['Continent'].unique()

# Combine the unique values to ensure we have all possible values
all_unique_continents = set(unique_train_continents) | set(unique_test_continents)

# Create a mapping from the unique values to integers
continent_mapping = {value: index for index, value in enumerate(all_unique_continents)}

# Apply the mapping to the 'Continent' column in both the training and test datasets
train_df['Continent'] = train_df['Continent'].map(continent_mapping).fillna(-1).astype(int)
test_df['Continent'] = test_df['Continent'].map(continent_mapping).fillna(-1).astype(int)

# Split the datasets into features and target variable
X_train = train_df.drop(columns=['Patient ID', 'Heart Attack Risk'])
y_train = train_df['Heart Attack Risk']

X_test = test_df.drop(columns=['Patient ID', 'Heart Attack Risk'])
y_test = test_df['Heart Attack Risk']

# Initialize the RandomForest model
rf_model = RandomForestClassifier(random_state=42)

# Fit the model on the training data
rf_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_proba = rf_model.predict_proba(X_test)[:, 1]

# Evaluate the performance of the RandomForest model on the test set
roc_auc = roc_auc_score(y_test, y_pred_proba)

# Return the ROC AUC score
roc_auc

```

```



from sklearn.model_selection import GridSearchCV

# Define the parameter grid to search over
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['auto', 'sqrt']
}

# Initialize the RandomForest model
rf_model = RandomForestClassifier(random_state=42)

# Initialize GridSearchCV
grid_search = GridSearchCV(estimator=rf_model, param_grid=param_grid, cv=5, scoring='roc_auc', n_jobs=-1)

# Fit the GridSearchCV model on the training data
grid_search.fit(X_train, y_train)

# Get the best parameters and best score
best_params = grid_search.best_params_
best_score = grid_search.best_score_

# Use the best model to make predictions on the test set
best_model = grid_search.best_estimator_
y_pred_proba = best_model.predict_proba(X_test)[:, 1]

# Evaluate the performance of the best model on the test set
roc_auc = roc_auc_score(y_test, y_pred_proba)

# Return the best parameters, best score, and test set ROC AUC score
best_params, best_score, roc_auc


```
