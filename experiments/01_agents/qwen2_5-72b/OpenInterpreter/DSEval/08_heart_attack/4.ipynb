```

import pandas as pd

# Load the training and test data
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\08_heart_attack\\train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\08_heart_attack\\test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Convert the 'Diet' column
diet_mapping = {'Healthy': 1, 'Average': 2, 'Unhealthy': 3}
train_df['Diet'] = train_df['Diet'].map(diet_mapping)
test_df['Diet'] = test_df['Diet'].map(diet_mapping)

# Display the first few rows of the training and test data to verify the conversion
train_df.head(), test_df.head()
```

```


# Split the data into features (X) and target (y)
X_train = train_df.drop(columns=['Patient ID', 'Heart Attack Risk'])
y_train = train_df['Heart Attack Risk']

X_test = test_df.drop(columns=['Patient ID', 'Heart Attack Risk'])
y_test = test_df['Heart Attack Risk']

# Display the shapes of the training and test sets to verify the split
X_train.shape, y_train.shape, X_test.shape, y_test.shape

```

```


from sklearn.ensemble import RandomForestClassifier

# Initialize the RandomForest model
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)

# Train the model on the training set
rf_model.fit(X_train, y_train)

# Display the model to confirm it has been trained
rf_model


```

```



# Convert the 'Sex' column
sex_mapping = {'Male': 1, 'Female': 0}
X_train['Sex'] = X_train['Sex'].map(sex_mapping)
X_test['Sex'] = X_test['Sex'].map(sex_mapping)

# Check for other categorical columns and convert them if necessary
categorical_columns = X_train.select_dtypes(include=['object']).columns.tolist()

for col in categorical_columns:
    unique_values = X_train[col].unique()
    value_mapping = {value: i for i, value in enumerate(unique_values)}
    X_train[col] = X_train[col].map(value_mapping)
    X_test[col] = X_test[col].map(value_mapping)

# Display the first few rows of the training and test sets to verify the conversion
X_train.head(), X_test.head()

```

```



from sklearn.ensemble import RandomForestClassifier

# Initialize the RandomForest model
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)

# Train the model on the training set
rf_model.fit(X_train, y_train)

# Display the model to confirm it has been trained
rf_model


```

```




# Make predictions on the test set
y_pred = rf_model.predict_proba(X_test)[:, 1]

# Display the first few predicted probabilities
y_pred[:5]


```

```



from sklearn.metrics import roc_auc_score

# Calculate the ROC AUC score
roc_auc = roc_auc_score(y_test, y_pred)

# Return the ROC AUC score in a tuple
(roc_auc,)


```
