```

import pandas as pd

# Load the training and test datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/10_life_expectancy/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/10_life_expectancy/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the training dataset
train_df.head()
```

```

# Perform feature construction to build the 'GDP_bins' feature
def categorize_gdp(gdp):
    if gdp <= 1000:
        return 'Low GDP'
    elif 1000 < gdp <= 10000:
        return 'Medium GDP'
    else:
        return 'High GDP'

# Apply the function to create the 'GDP_bins' feature in the training and test datasets
train_df['GDP_bins'] = train_df['GDP'].apply(categorize_gdp)
test_df['GDP_bins'] = test_df['GDP'].apply(categorize_gdp)

# Display the first few rows of the training dataset with the new feature
train_df.head()
```

```


from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Select the features and target variable
features = train_df.columns.difference(['Life expectancy', 'Country', 'GDP_bins'])
X_train = train_df[features]
y_train = train_df['Life expectancy']

X_test = test_df[features]
y_test = test_df['Life expectancy']

# Initialize and train the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions on the test dataset
y_pred = model.predict(X_test)

# Evaluate the model's performance using the R-squared score
r2 = r2_score(y_test, y_pred)
r2

```

```



from sklearn.preprocessing import OneHotEncoder

# Identify categorical columns
categorical_columns = ['Status']

# Initialize the OneHotEncoder
encoder = OneHotEncoder(handle_unknown='ignore', sparse=False)

# Fit and transform the categorical columns in the training dataset
X_train_encoded = encoder.fit_transform(train_df[categorical_columns])

# Transform the categorical columns in the test dataset
X_test_encoded = encoder.transform(test_df[categorical_columns])

# Combine the encoded categorical features with the numerical features
X_train = pd.concat([train_df[features.difference(categorical_columns)], pd.DataFrame(X_train_encoded)], axis=1)
X_test = pd.concat([test_df[features.difference(categorical_columns)], pd.DataFrame(X_test_encoded)], axis=1)

# Initialize and train the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions on the test dataset
y_pred = model.predict(X_test)

# Evaluate the model's performance using the R-squared score
r2 = r2_score(y_test, y_pred)
r2
```

```



from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# Identify categorical and numerical columns
categorical_columns = ['Status']
numerical_columns = features.difference(categorical_columns)

# Create a ColumnTransformer to handle both categorical and numerical columns
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_columns),
        ('num', 'passthrough', numerical_columns)
    ]
)

# Apply the preprocessing to the training and test datasets
X_train_preprocessed = preprocessor.fit_transform(train_df)
X_test_preprocessed = preprocessor.transform(test_df)

# Initialize and train the linear regression model
model = LinearRegression()
model.fit(X_train_preprocessed, y_train)

# Make predictions on the test dataset
y_pred = model.predict(X_test_preprocessed)

# Evaluate the model's performance using the R-squared score
r2 = r2_score(y_test, y_pred)
r2

```
