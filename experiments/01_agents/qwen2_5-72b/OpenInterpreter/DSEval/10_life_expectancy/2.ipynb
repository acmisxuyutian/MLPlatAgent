```


import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import r2_score

# Load the training and test datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/10_life_expectancy/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/10_life_expectancy/test.csv'

# Read the CSV files
train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Define the GDP bins
bins = [0, 1000, 10000, float('inf')]
labels = ['Low GDP', 'Medium GDP', 'High GDP']

# Create the GDP_bins feature
train_df['GDP_bins'] = pd.cut(train_df['GDP'], bins=bins, labels=labels, right=False)
test_df['GDP_bins'] = pd.cut(test_df['GDP'], bins=bins, labels=labels, right=False)

# Define the features and target
features = ['GDP_bins', 'Income composition of resources', 'Schooling']
target = 'Life expectancy '

# One-hot encode the GDP_bins feature
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(), ['GDP_bins'])
    ],
    remainder='passthrough'
)

# Create the pipeline
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', LinearRegression())
])

# Fit the model on the training data
pipeline.fit(train_df[features], train_df[target])

# Make predictions on the test data
test_predictions = pipeline.predict(test_df[features])

# Evaluate the model using R-squared score
r2 = r2_score(test_df[target], test_predictions)

r2
```

```



import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import r2_score

# Load the training and test datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/10_life_expectancy/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/10_life_expectancy/test.csv'

# Read the CSV files
train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Correct the column name in the training dataset
train_df.rename(columns={'Life expectancy ': 'Life expectancy'}, inplace=True)

# Define the GDP bins
bins = [0, 1000, 10000, float('inf')]
labels = ['Low GDP', 'Medium GDP', 'High GDP']

# Create the GDP_bins feature
train_df['GDP_bins'] = pd.cut(train_df['GDP'], bins=bins, labels=labels, right=False)
test_df['GDP_bins'] = pd.cut(test_df['GDP'], bins=bins, labels=labels, right=False)

# Define the features and target
features = ['GDP_bins', 'Income composition of resources', 'Schooling']
target = 'Life expectancy'

# One-hot encode the GDP_bins feature
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(), ['GDP_bins'])
    ],
    remainder='passthrough'
)

# Create the pipeline
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', LinearRegression())
])

# Fit the model on the training data
pipeline.fit(train_df[features], train_df[target])

# Make predictions on the test data
test_predictions = pipeline.predict(test_df[features])

# Evaluate the model using R-squared score
r2 = r2_score(test_df[target], test_predictions)

r2


```
