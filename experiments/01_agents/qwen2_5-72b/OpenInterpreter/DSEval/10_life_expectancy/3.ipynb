```


import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Define the file paths
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark\DSEval\datasets\10_life_expectancy\train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark\DSEval\datasets\10_life_expectancy\test.csv'

# Load the datasets
train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Define the GDP bins
bins = [0, 1000, 10000, float('inf')]
labels = ['Low GDP', 'Medium GDP', 'High GDP']

# Create the 'GDP_bins' feature
train_df['GDP_bins'] = pd.cut(train_df['GDP'], bins=bins, labels=labels, right=False)
test_df['GDP_bins'] = pd.cut(test_df['GDP'], bins=bins, labels=labels, right=False)

# Convert 'GDP_bins' to dummy variables
train_df = pd.get_dummies(train_df, columns=['GDP_bins'])
test_df = pd.get_dummies(test_df, columns=['GDP_bins'])

# Select features and target variable
features = ['GDP', 'Schooling', 'GDP_bins_Low GDP', 'GDP_bins_Medium GDP', 'GDP_bins_High GDP']
target = 'Life expectancy '

# Prepare the training and test datasets
X_train = train_df[features]
y_train = train_df[target]
X_test = test_df[features]
y_test = test_df[target]

# Initialize and fit the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model using R-squared score
r2 = r2_score(y_test, y_pred)

r2

```
