```

import pandas as pd

# Load the training and test datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/10_life_expectancy/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/10_life_expectancy/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the training and test datasets
print("Training Data:")
print(train_df.head())
print("\nTest Data:")
print(test_df.head())
```

```


# Define the GDP bins
def categorize_gdp(gdp):
    if gdp <= 1000:
        return "Low GDP"
    elif gdp <= 10000:
        return "Medium GDP"
    else:
        return "High GDP"

# Apply the function to create the new feature "GDP_bins" in the training and test datasets
train_df['GDP_bins'] = train_df['GDP'].apply(categorize_gdp)
test_df['GDP_bins'] = test_df['GDP'].apply(categorize_gdp)

# Display the first few rows of the updated training and test datasets
print("Updated Training Data:")
print(train_df.head())
print("\nUpdated Test Data:")
print(test_df.head())

```

```


from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.preprocessing import OneHotEncoder
import numpy as np

# Select the numerical features and the target variable
numerical_features = [
    'Year', 'Adult Mortality', 'infant deaths', 'Alcohol', 'percentage expenditure', 'Hepatitis B',
    'Measles ', ' BMI ', 'under-five deaths ', 'Polio', 'Total expenditure', 'Diphtheria ', ' HIV/AIDS',
    'GDP', 'Population', ' thinness  1-19 years', ' thinness 5-9 years', 'Income composition of resources', 'Schooling'
]

# One-hot encode the 'GDP_bins' feature
encoder = OneHotEncoder(sparse=False)
gdp_bins_train = encoder.fit_transform(train_df[['GDP_bins']])
gdp_bins_test = encoder.transform(test_df[['GDP_bins']])

# Combine the numerical features and the one-hot encoded GDP_bins
X_train = np.concatenate((train_df[numerical_features], gdp_bins_train), axis=1)
X_test = np.concatenate((test_df[numerical_features], gdp_bins_test), axis=1)

# Target variable
y_train = train_df['Life expectancy ']

# Train the Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Display the first few predictions
print("Predictions on the test set:")
print(y_pred[:5])

```
