```

import pandas as pd

# Load the train and test datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/06_employee/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/06_employee/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the train and test datasets
train_df.head(), test_df.head()
```

```


# Perform feature construction to build the "YearsInCompany" feature
train_df['YearsInCompany'] = 2023 - train_df['JoiningYear']
test_df['YearsInCompany'] = 2023 - test_df['JoiningYear']

# Display the first few rows of the train and test datasets with the new feature
train_df.head(), test_df.head()

```

```


# Encode the categorical columns using one-hot encoding
categorical_columns = ['Education']

# Perform one-hot encoding
train_df_encoded = pd.get_dummies(train_df, columns=categorical_columns)
test_df_encoded = pd.get_dummies(test_df, columns=categorical_columns)

# Display the first few rows of the train and test datasets with the one-hot encoded features
train_df_encoded.head(), test_df_encoded.head()
```

```



# Separate features and target variable
X_train = train_df_encoded.drop(columns=['LeaveOrNot'])
y_train = train_df_encoded['LeaveOrNot']

X_test = test_df_encoded.drop(columns=['LeaveOrNot'])
y_test = test_df_encoded['LeaveOrNot']

# Display the shapes of the feature and target datasets
X_train.shape, y_train.shape, X_test.shape, y_test.shape

```

```


from sklearn.ensemble import RandomForestClassifier

# Initialize the random forest classifier
rf_classifier = RandomForestClassifier(random_state=42)

# Train the random forest classifier
rf_classifier.fit(X_train, y_train)

# Make predictions on the test set
y_pred = rf_classifier.predict(X_test)

# Display the first few predictions
y_pred[:5]

```
