````

import pandas as pd

# Load the training and test data
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\06_employee\\train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\06_employee\\test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the training and test datasets
train_df.head(), test_df.head()
````

````

# Feature Construction: Create "YearsInCompany" feature
train_df['YearsInCompany'] = 2023 - train_df['JoiningYear']
test_df['YearsInCompany'] = 2023 - test_df['JoiningYear']

# Display the updated first few rows of the training and test datasets
train_df.head(), test_df.head()
````

````

# One-hot encode categorical columns
categorical_cols = ['Education']

train_df_encoded = pd.get_dummies(train_df, columns=categorical_cols, drop_first=True)
test_df_encoded = pd.get_dummies(test_df, columns=categorical_cols, drop_first=True)

# Display the updated first few rows of the training and test datasets
train_df_encoded.head(), test_df_encoded.head()
````

````


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, confusion_matrix
from sklearn.preprocessing import StandardScaler

# Separate features and target variable
X_train = train_df_encoded.drop(columns=['LeaveOrNot'])
y_train = train_df_encoded['LeaveOrNot']

X_test = test_df_encoded.drop(columns=['LeaveOrNot'])
y_test = test_df_encoded['LeaveOrNot']

# Standardize the features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train a Random Forest Classifier
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train_scaled, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test_scaled)
y_pred_proba = model.predict_proba(X_test_scaled)[:, 1]

# Compute AUC-ROC
auc_roc = roc_auc_score(y_test, y_pred_proba)

# Compute confusion matrix
conf_matrix = confusion_matrix(y_test, y_pred)

auc_roc, conf_matrix

````
