```

import pandas as pd

# Load the training and test datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\06_employee\\train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\06_employee\\test.csv'

# Read the datasets
train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the training and test datasets
train_df.head(), test_df.head()
```

```

# Calculate 'YearsInCompany' for both training and test datasets
train_df['YearsInCompany'] = 2023 - train_df['JoiningYear']
test_df['YearsInCompany'] = 2023 - test_df['JoiningYear']

# Display the first few rows to verify the new feature
train_df.head(), test_df.head()
```

```


# Identify categorical columns
categorical_columns = ['Education']

# Apply one-hot encoding to the categorical columns
train_df = pd.get_dummies(train_df, columns=categorical_columns, drop_first=True)
test_df = pd.get_dummies(test_df, columns=categorical_columns, drop_first=True)

# Display the first few rows to verify the one-hot encoding
train_df.head(), test_df.head()

```

```


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, confusion_matrix

# Define the features and target variable
X_train = train_df.drop(columns=['LeaveOrNot', 'JoiningYear', 'City'])
y_train = train_df['LeaveOrNot']

X_test = test_df.drop(columns=['LeaveOrNot', 'JoiningYear', 'City'])
y_test = test_df['LeaveOrNot']

# Initialize and train the random forest classifier
rf_clf = RandomForestClassifier(n_estimators=100, random_state=42)
rf_clf.fit(X_train, y_train)

# Make predictions on the test set
y_pred = rf_clf.predict(X_test)
y_pred_proba = rf_clf.predict_proba(X_test)[:, 1]

# Display the first few predictions
y_pred[:5], y_pred_proba[:5]

```

```



# Identify all categorical columns
categorical_columns = train_df.select_dtypes(include=['object']).columns

# Apply one-hot encoding to all categorical columns
train_df = pd.get_dummies(train_df, columns=categorical_columns, drop_first=True)
test_df = pd.get_dummies(test_df, columns=categorical_columns, drop_first=True)

# Ensure that both training and test datasets have the same columns
missing_cols = set(train_df.columns) - set(test_df.columns)
for c in missing_cols:
    test_df[c] = 0

# Ensure that the order of columns is the same
test_df = test_df[train_df.columns]

# Define the features and target variable
X_train = train_df.drop(columns=['LeaveOrNot'])
y_train = train_df['LeaveOrNot']

X_test = test_df.drop(columns=['LeaveOrNot'])
y_test = test_df['LeaveOrNot']

# Initialize and train the random forest classifier
rf_clf = RandomForestClassifier(n_estimators=100, random_state=42)
rf_clf.fit(X_train, y_train)

# Make predictions on the test set
y_pred = rf_clf.predict(X_test)
y_pred_proba = rf_clf.predict_proba(X_test)[:, 1]

# Display the first few predictions
y_pred[:5], y_pred_proba[:5]


```

```


from sklearn.metrics import roc_auc_score

# Compute the area under the ROC curve
auc_roc = roc_auc_score(y_test, y_pred_proba)

auc_roc


```
