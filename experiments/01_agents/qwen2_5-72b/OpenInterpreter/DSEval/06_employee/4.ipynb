```

import pandas as pd

# Load the training and test datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\06_employee\\train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\06_employee\\test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and test datasets
train_df.head(), test_df.head()
```

```


# Add a new feature 'YearsInCompany' to both the training and test datasets
train_df['YearsInCompany'] = 2023 - train_df['JoiningYear']
test_df['YearsInCompany'] = 2023 - test_df['JoiningYear']

# Display the first few rows of the datasets with the new feature
train_df.head(), test_df.head()

```

```


# Perform one-hot encoding on the categorical columns
categorical_columns = ['Education', 'City', 'Gender', 'EverBenched']

train_df_encoded = pd.get_dummies(train_df, columns=categorical_columns)
test_df_encoded = pd.get_dummies(test_df, columns=categorical_columns)

# Display the first few rows of the encoded datasets
train_df_encoded.head(), test_df_encoded.head()

```

```



from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score, confusion_matrix

# Split the training data into features (X) and target variable (y)
X_train = train_df_encoded.drop(columns=['LeaveOrNot', 'JoiningYear'])
y_train = train_df_encoded['LeaveOrNot']

# Split the test data into features (X) and target variable (y)
X_test = test_df_encoded.drop(columns=['LeaveOrNot', 'JoiningYear'])
y_test = test_df_encoded['LeaveOrNot']

# Initialize and train the Random Forest Classifier
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
rf_classifier.fit(X_train, y_train)

# Make predictions on the test set
y_pred = rf_classifier.predict(X_test)
y_pred_proba = rf_classifier.predict_proba(X_test)[:, 1]

# Display the first few predictions
y_pred[:5], y_pred_proba[:5]


```

```




# Compute the area under the ROC curve
roc_auc = roc_auc_score(y_test, y_pred_proba)
roc_auc
```

```




# Compute the confusion matrix
conf_matrix = confusion_matrix(y_test, y_pred)

# Display the confusion matrix
conf_matrix

```
