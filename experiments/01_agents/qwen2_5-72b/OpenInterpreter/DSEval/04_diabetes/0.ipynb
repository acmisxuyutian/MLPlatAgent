```

import pandas as pd

# Load the training and test datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\04_diabetes\\train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\04_diabetes\\test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Drop duplicate rows
train_df.drop_duplicates(inplace=True)
test_df.drop_duplicates(inplace=True)

# Display the first few rows of the cleaned datasets
train_df.head(), test_df.head()
```

```


# Perform one-hot encoding on the 'gender' and 'smoking_history' columns
train_df_encoded = pd.get_dummies(train_df, columns=['gender', 'smoking_history'])
test_df_encoded = pd.get_dummies(test_df, columns=['gender', 'smoking_history'])

# Display the first few rows of the encoded datasets
train_df_encoded.head(), test_df_encoded.head()

```

```


from sklearn.ensemble import RandomForestClassifier

# Separate features and target variable
X_train = train_df_encoded.drop(columns=['diabetes'])
y_train = train_df_encoded['diabetes']

# Initialize and fit a random forest model
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Get feature importances
feature_importances = rf_model.feature_importances_

# Create a DataFrame to display feature importances
importance_df = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': feature_importances
}).sort_values(by='Importance', ascending=False)

# Display the feature importances
importance_df

```

```



# Initialize and fit the random forest classifier
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
rf_classifier.fit(X_train, y_train)


```

```



# Separate features and target variable for the test set
X_test = test_df_encoded.drop(columns=['diabetes'])
y_test = test_df_encoded['diabetes']

# Make predictions on the test set
y_pred = rf_classifier.predict(X_test)
y_pred_proba = rf_classifier.predict_proba(X_test)[:, 1]  # Probability of the positive class

# Display the first few predictions
y_pred[:5], y_pred_proba[:5]


```

```



from sklearn.metrics import roc_auc_score

# Compute the AUC-ROC
auc_roc = roc_auc_score(y_test, y_pred_proba)

# Display the AUC-ROC
auc_roc


```

```



from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Compute the confusion matrix
cm = confusion_matrix(y_test, y_pred)

# Display the confusion matrix
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['No Diabetes', 'Diabetes'])
disp.plot(cmap=plt.cm.Blues)
plt.show()


```
