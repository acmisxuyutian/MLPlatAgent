```

import pandas as pd

# Load the training and test datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/04_diabetes/train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/04_diabetes/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Drop duplicate rows
train_df = train_df.drop_duplicates()
test_df = test_df.drop_duplicates()

# Display the first few rows of the training and test datasets
train_df.head(), test_df.head()
```

```


# Perform one-hot encoding on 'gender' and 'smoking_history' columns
train_df = pd.get_dummies(train_df, columns=['gender', 'smoking_history'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['gender', 'smoking_history'], drop_first=True)

# Display the first few rows of the training and test datasets after one-hot encoding
train_df.head(), test_df.head()

```

```


from sklearn.ensemble import RandomForestClassifier

# Separate features and target variable
X_train = train_df.drop(columns=['diabetes'])
y_train = train_df['diabetes']

# Initialize and fit a random forest classifier
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)

# Get feature importances
feature_importances = rf.feature_importances_

# Create a DataFrame to display feature importances
importances_df = pd.DataFrame({'Feature': X_train.columns, 'Importance': feature_importances})
importances_df = importances_df.sort_values(by='Importance', ascending=False)

importances_df
```

```



# Initialize and fit the random forest classifier on the training data
rf_final = RandomForestClassifier(n_estimators=100, random_state=42)
rf_final.fit(X_train, y_train)

# Display the fitted model
rf_final
```

```



# Separate features and target variable in the test set
X_test = test_df.drop(columns=['diabetes'])
y_test = test_df['diabetes']

# Make predictions on the test set
y_pred = rf_final.predict(X_test)

# Display the first few predictions
y_pred[:5]

```

```



from sklearn.metrics import roc_auc_score

# Compute the predicted probabilities for the positive class
y_pred_proba = rf_final.predict_proba(X_test)[:, 1]

# Compute the AUC-ROC score
auc_roc = roc_auc_score(y_test, y_pred_proba)

auc_roc

```

```




from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Compute the confusion matrix
cm = confusion_matrix(y_test, y_pred)

# Create a ConfusionMatrixDisplay object
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Non-Diabetic', 'Diabetic'])

# Plot the confusion matrix
plt.figure(figsize=(8, 6))
disp.plot(cmap='Blues')
plt.title('Confusion Matrix')
plt.show()

# Display the confusion matrix as a DataFrame
pd.DataFrame(cm, index=['Non-Diabetic', 'Diabetic'], columns=['Predicted Non-Diabetic', 'Predicted Diabetic'])

```
