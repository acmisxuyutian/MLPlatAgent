```

import pandas as pd

# Load the training and test datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\01_bank_customer_churn\\train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\01_bank_customer_churn\\test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

# Drop uninteresting columns
columns_to_drop = ['CustomerId', 'Complain', 'Surname']
train_data = train_data.drop(columns=columns_to_drop)
test_data = test_data.drop(columns=columns_to_drop)

# Display the first few rows of the training and test data
train_data.head(), test_data.head()
```

```


# Perform one-hot encoding for categorical variables
train_data = pd.get_dummies(train_data, columns=['Geography', 'Card Type'], drop_first=True)
test_data = pd.get_dummies(test_data, columns=['Geography', 'Card Type'], drop_first=True)

# Display the first few rows of the training and test data after encoding
train_data.head(), test_data.head()

```

```


import xgboost as xgb
from sklearn.metrics import roc_auc_score

# Split the training data into features and target variable
X_train = train_data.drop(columns=['Exited', 'RowNumber'])
y_train = train_data['Exited']

# Split the test data into features (test set does not have the target variable)
X_test = test_data.drop(columns=['RowNumber'])

# Initialize and train the XGBoost classifier
xgb_model = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss')
xgb_model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_proba = xgb_model.predict_proba(X_test)[:, 1]

# Since the test set does not have the target variable, we can't compute AUC-ROC directly.
# However, we can still make predictions and save them for further analysis if needed.
y_pred_proba
```
