```

import pandas as pd
import xgboost as xgb
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import roc_auc_score

# Load the training and test datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\01_bank_customer_churn\\train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\01_bank_customer_churn\\test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and test datasets
print("Training Data:")
print(train_df.head())
print("\nTest Data:")
print(test_df.head())
```

```


# Drop the unnecessary columns from the training and test datasets
columns_to_drop = ['CustomerId', 'Complain', 'Surname']

train_df.drop(columns=columns_to_drop, inplace=True)
test_df.drop(columns=columns_to_drop, inplace=True)

# Display the first few rows of the training and test datasets after dropping columns
print("Training Data after Dropping Columns:")
print(train_df.head())
print("\nTest Data after Dropping Columns:")
print(test_df.head())

```

```



# Perform one-hot encoding for the 'Geography' and 'Card Type' columns
train_df = pd.get_dummies(train_df, columns=['Geography', 'Card Type'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['Geography', 'Card Type'], drop_first=True)

# Display the first few rows of the training and test datasets after one-hot encoding
print("Training Data after One-Hot Encoding:")
print(train_df.head())
print("\nTest Data after One-Hot Encoding:")
print(test_df.head())

```

```



# Separate features and target variable for the training dataset
X_train = train_df.drop(columns=['RowNumber', 'Exited'])
y_train = train_df['Exited']

# Drop the 'RowNumber' column from the test dataset
X_test = test_df.drop(columns=['RowNumber'])

# Display the shapes of the prepared datasets
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)


```

```




# Align columns in X_test to match X_train
# Get the set of column names for X_train and X_test
train_columns = set(X_train.columns)
test_columns = set(X_test.columns)

# Find columns in X_train that are not in X_test
missing_columns = train_columns - test_columns

# Add missing columns to X_test with default value 0
for col in missing_columns:
    X_test[col] = 0

# Ensure the order of columns in X_test matches X_train
X_test = X_test[X_train.columns]

# Display the shapes of the prepared datasets after alignment
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)


```

```




# Train the XGBoost model
model = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss')
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_proba = model.predict_proba(X_test)[:, 1]

# Display the first few predictions
print("Predicted Probabilities for the Test Set:")
print(y_pred_proba[:5])


```

```




# Perform one-hot encoding for the 'Gender' column
train_df = pd.get_dummies(train_df, columns=['Gender'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['Gender'], drop_first=True)

# Separate features and target variable for the training dataset
X_train = train_df.drop(columns=['RowNumber', 'Exited'])
y_train = train_df['Exited']

# Drop the 'RowNumber' column from the test dataset
X_test = test_df.drop(columns=['RowNumber'])

# Align columns in X_test to match X_train
train_columns = set(X_train.columns)
test_columns = set(X_test.columns)

missing_columns = train_columns - test_columns
for col in missing_columns:
    X_test[col] = 0

X_test = X_test[X_train.columns]

# Display the shapes of the prepared datasets after alignment
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)

# Train the XGBoost model
model = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss')
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_proba = model.predict_proba(X_test)[:, 1]

# Display the first few predictions
print("Predicted Probabilities for the Test Set:")
print(y_pred_proba[:5])

```

```





# Assuming the test set has the 'Exited' column for true labels
y_test = test_df['Exited']

# Compute the AUC-ROC score
auc_roc = roc_auc_score(y_test, y_pred_proba)

# Display the AUC-ROC score
print("AUC-ROC Score:", auc_roc)


```
