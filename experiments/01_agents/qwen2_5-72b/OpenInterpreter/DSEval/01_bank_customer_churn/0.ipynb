```

import pandas as pd

# Load the training and test datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/01_bank_customer_churn/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/01_bank_customer_churn/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the training and test datasets
train_df.head(), test_df.head()
```

```


# Drop unnecessary columns
columns_to_drop = ['CustomerId', 'Complain', 'Surname']
train_df = train_df.drop(columns=columns_to_drop)
test_df = test_df.drop(columns=columns_to_drop)

# One-hot encode the categorical columns
train_df = pd.get_dummies(train_df, columns=['Geography', 'Card Type'])
test_df = pd.get_dummies(test_df, columns=['Geography', 'Card Type'])

# Display the first few rows of the preprocessed datasets
train_df.head(), test_df.head()

```

```


from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score

# Split the training data into features and target variable
X_train = train_df.drop(columns=['Exited'])
y_train = train_df['Exited']

# Split the test data into features
X_test = test_df.drop(columns=['Exited'])
y_test = test_df['Exited']

# Initialize and train the XGBoost classifier
model = XGBClassifier(use_label_encoder=False, eval_metric='logloss')
model.fit(X_train, y_train)

# Make predictions on the test data
y_pred_proba = model.predict_proba(X_test)[:, 1]

# Compute the AUC-ROC score
auc_roc = roc_auc_score(y_test, y_pred_proba)

auc_roc
```
