```

import pandas as pd

# Load the training and test datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\01_bank_customer_churn\\train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\01_bank_customer_churn\\test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Drop the columns that will not be interesting to models
columns_to_drop = ['CustomerId', 'Complain', 'Surname']
train_df = train_df.drop(columns=columns_to_drop)
test_df = test_df.drop(columns=columns_to_drop)

# Display the first few rows of the training and test data
train_df.head(), test_df.head()
```

```

# One-hot encode the 'Geography' and 'Card Type' columns
train_df = pd.get_dummies(train_df, columns=['Geography', 'Card Type'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['Geography', 'Card Type'], drop_first=True)

# Display the first few rows of the training and test data after encoding
train_df.head(), test_df.head()
```

```


# Separate features and target variable from the training data
X_train = train_df.drop(columns=['RowNumber', 'Exited'])
y_train = train_df['Exited']

# Separate features from the test data
X_test = test_df.drop(columns=['RowNumber'])

# Display the shapes of the training and test data
X_train.shape, y_train.shape, X_test.shape

```

```


# Ensure consistent features in training and test sets
X_test = X_test[X_train.columns]

# Display the shapes of the training and test data again
X_train.shape, X_test.shape

```

```


import xgboost as xgb
from sklearn.metrics import roc_auc_score

# Initialize the XGBoost classifier
xgb_clf = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss')

# Train the model on the training data
xgb_clf.fit(X_train, y_train)

# Make predictions on the test set
y_pred = xgb_clf.predict_proba(X_test)[:, 1]  # Get the probability of the positive class

# Display the first few predictions
y_pred[:5]

```
