```

import pandas as pd

# Load the training and test datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\03_coffee_quality\\train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\03_coffee_quality\\test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training dataset to understand its structure
train_df.head()
```

```


# Drop duplicated rows from the training and test datasets
train_df = train_df.drop_duplicates()
test_df = test_df.drop_duplicates()

# Drop specified columns from the datasets
columns_to_drop = [
    "ID", "ICO Number", "Owner", "Region", "Certification Contact", "Certification Address", 
    "Farm Name", "Lot Number", "Mill", "ICO Number", "Producer", "Company", "Expiration", 
    "Harvest Year", "Unnamed: 0", "Number of Bags", "Bag Weight", "In-Country Partner", 
    "Country of Origin", "Processing Method", "Grading Date", "Variety", "Status", "Defects", 
    "Uniformity", "Clean Cup", "Sweetness"
]

train_df = train_df.drop(columns=columns_to_drop, errors='ignore')
test_df = test_df.drop(columns=columns_to_drop, errors='ignore')

# Perform one-hot encoding for the "Color" column
train_df = pd.get_dummies(train_df, columns=['Color'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['Color'], drop_first=True)

# Display the first few rows of the training dataset after preprocessing
train_df.head()

```

```


from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

# Split the training data into features (X) and target (y)
X_train = train_df.drop(columns=['Total Cup Points'])
y_train = train_df['Total Cup Points']

# Split the test data into features (X_test) and target (y_test)
X_test = test_df.drop(columns=['Total Cup Points'], errors='ignore')
y_test = test_df['Total Cup Points']

# Initialize and train the RandomForestRegressor
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the mean squared error (MSE) on the test set
mse = mean_squared_error(y_test, y_pred)

mse
```
