```

import pandas as pd

# Load the training and test datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\03_coffee_quality\\train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\03_coffee_quality\\test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Drop duplicated rows
train_df = train_df.drop_duplicates()
test_df = test_df.drop_duplicates()

# Display the first few rows of the cleaned datasets
train_df.head(), test_df.head()
```

```


# List of columns to drop
columns_to_drop = [
    "ID", "ICO Number", "Owner", "Region", "Certification Contact", "Certification Address", "Farm Name",
    "Lot Number", "Mill", "ICO Number", "Producer", "Company", "Expiration", "Harvest Year", "Unnamed: 0",
    "Number of Bags", "Bag Weight", "In-Country Partner", "Country of Origin", "Processing Method", "Grading Date",
    "Variety", "Status", "Defects", "Uniformity", "Clean Cup", "Sweetness", "Certification Body"
]

# Drop the specified columns
train_df = train_df.drop(columns=columns_to_drop, errors='ignore')
test_df = test_df.drop(columns=columns_to_drop, errors='ignore')

# Display the first few rows of the cleaned datasets
train_df.head(), test_df.head()

```

```



# One-hot encode the 'Color' column
train_df = pd.get_dummies(train_df, columns=['Color'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['Color'], drop_first=True)

# Display the first few rows of the cleaned datasets
train_df.head(), test_df.head()

```

```



from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error

# Define the target variable
target_column = 'Total Cup Points'

# Split the datasets into features (X) and target (y)
X_train = train_df.drop(columns=[target_column])
y_train = train_df[target_column]

X_test = test_df.drop(columns=[target_column])
y_test = test_df[target_column]

# Initialize and train the RandomForestRegressor
model = RandomForestRegressor(random_state=42)
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Display the first few predictions
y_pred[:5]


```

```




# Calculate the mean squared error (MSE)
mse = mean_squared_error(y_test, y_pred)

# Display the mean squared error
mse


```
