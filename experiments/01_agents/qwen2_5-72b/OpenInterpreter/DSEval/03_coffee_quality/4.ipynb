```

import pandas as pd

# Load the training and test datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\03_coffee_quality\\train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\03_coffee_quality\\test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the datasets to understand the structure
print("Training Data:")
print(train_df.head())
print("\nTest Data:")
print(test_df.head())
```

```


# Step 1: Drop duplicated rows
train_df = train_df.drop_duplicates()
test_df = test_df.drop_duplicates()

# Step 2: Drop unnecessary columns
columns_to_drop = [
    "ID", "ICO Number", "Owner", "Region", "Certification Contact", "Certification Address", 
    "Farm Name", "Lot Number", "Mill", "Producer", "Company", "Expiration", "Harvest Year", 
    "Unnamed: 0", "Number of Bags", "Bag Weight", "In-Country Partner", "Country of Origin", 
    "Processing Method", "Grading Date", "Variety", "Status", "Defects", "Uniformity", 
    "Clean Cup", "Sweetness", "Certification Body"
]

train_df = train_df.drop(columns=columns_to_drop)
test_df = test_df.drop(columns=columns_to_drop)

# Display the first few rows of the cleaned datasets to verify
print("Cleaned Training Data:")
print(train_df.head())
print("\nCleaned Test Data:")
print(test_df.head())

```

```


# Check the columns present in the datasets
print("Columns in Training Data:")
print(train_df.columns.tolist())
print("\nColumns in Test Data:")
print(test_df.columns.tolist())

```

```



# Update the list of columns to drop by removing 'Unnamed: 0'
columns_to_drop = [
    "ID", "ICO Number", "Owner", "Region", "Certification Contact", "Certification Address", 
    "Farm Name", "Lot Number", "Mill", "Producer", "Company", "Expiration", "Harvest Year", 
    "Number of Bags", "Bag Weight", "In-Country Partner", "Country of Origin", 
    "Processing Method", "Grading Date", "Variety", "Status", "Defects", "Uniformity", 
    "Clean Cup", "Sweetness", "Certification Body"
]

# Drop duplicated rows
train_df = train_df.drop_duplicates()
test_df = test_df.drop_duplicates()

# Drop unnecessary columns
train_df = train_df.drop(columns=columns_to_drop)
test_df = test_df.drop(columns=columns_to_drop)

# Display the first few rows of the cleaned datasets to verify
print("Cleaned Training Data:")
print(train_df.head())
print("\nCleaned Test Data:")
print(test_df.head())


```

```


import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# One-hot encode the 'Color' column
encoder = OneHotEncoder(sparse=False, drop='first')
train_color_encoded = encoder.fit_transform(train_df[['Color']])
test_color_encoded = encoder.transform(test_df[['Color']])

# Create DataFrame for the encoded columns
train_color_df = pd.DataFrame(train_color_encoded, columns=encoder.get_feature_names_out(['Color']))
test_color_df = pd.DataFrame(test_color_encoded, columns=encoder.get_feature_names_out(['Color']))

# Concatenate the encoded columns back to the original DataFrames
train_df = pd.concat([train_df, train_color_df], axis=1)
test_df = pd.concat([test_df, test_color_df], axis=1)

# Drop the original 'Color' column
train_df = train_df.drop(columns=['Color'])
test_df = test_df.drop(columns=['Color'])

# Display the first few rows of the datasets to verify
print("Training Data after One-Hot Encoding:")
print(train_df.head())
print("\nTest Data after One-Hot Encoding:")
print(test_df.head())

```

```



from sklearn.preprocessing import OneHotEncoder

# One-hot encode the 'Color' column
encoder = OneHotEncoder(sparse_output=False, drop='first')
train_color_encoded = encoder.fit_transform(train_df[['Color']])
test_color_encoded = encoder.transform(test_df[['Color']])

# Create DataFrame for the encoded columns
train_color_df = pd.DataFrame(train_color_encoded, columns=encoder.get_feature_names_out(['Color']))
test_color_df = pd.DataFrame(test_color_encoded, columns=encoder.get_feature_names_out(['Color']))

# Concatenate the encoded columns back to the original DataFrames
train_df = pd.concat([train_df, train_color_df], axis=1)
test_df = pd.concat([test_df, test_color_df], axis=1)

# Drop the original 'Color' column
train_df = train_df.drop(columns=['Color'])
test_df = test_df.drop(columns=['Color'])

# Display the first few rows of the datasets to verify
print("Training Data after One-Hot Encoding:")
print(train_df.head())
print("\nTest Data after One-Hot Encoding:")
print(test_df.head())


```

```



from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error

# Separate features and target variable for the training data
X_train = train_df.drop(columns=['Total Cup Points'])
y_train = train_df['Total Cup Points']

# Separate features and target variable for the test data
X_test = test_df.drop(columns=['Total Cup Points'])
y_test = test_df['Total Cup Points']

# Train the Random Forest Regressor
rf_regressor = RandomForestRegressor(n_estimators=100, random_state=42)
rf_regressor.fit(X_train, y_train)

# Make predictions on the test data
y_pred = rf_regressor.predict(X_test)

# Evaluate the model
mse = mean_squared_error(y_test, y_pred)
print(f"Mean Squared Error: {mse}")


```
