```

# Correct the training data path
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\03_coffee_quality\\train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\03_coffee_quality\\test.csv'

# Load the datasets
train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows and the shape of the datasets
print("Training Dataset:")
print(train_df.head())
print(train_df.shape)

print("\nTesting Dataset:")
print(test_df.head())
print(test_df.shape)
```

```


# Drop duplicated rows from the training dataset
train_df.drop_duplicates(inplace=True)

# Drop duplicated rows from the test dataset
test_df.drop_duplicates(inplace=True)

# Display the updated shape of the datasets
print("Training Dataset Shape after dropping duplicates:", train_df.shape)
print("Testing Dataset Shape after dropping duplicates:", test_df.shape)

```

```


# List of columns to drop
columns_to_drop = [
    "ID", "ICO Number", "Owner", "Region", "Certification Contact", "Certification Address", "Farm Name", 
    "Lot Number", "Mill", "ICO Number", "Producer", "Company", "Expiration", "Harvest Year", "Unnamed: 0", 
    "Number of Bags", "Bag Weight", "In-Country Partner", "Country of Origin", "Processing Method", "Grading Date", 
    "Variety", "Status", "Defects", "Uniformity", "Clean Cup", "Sweetness", "Certification Body"
]

# Drop the specified columns from the training dataset
train_df.drop(columns=columns_to_drop, inplace=True, errors='ignore')

# Drop the specified columns from the test dataset
test_df.drop(columns=columns_to_drop, inplace=True, errors='ignore')

# Display the updated shape of the datasets
print("Training Dataset Shape after dropping columns:", train_df.shape)
print("Testing Dataset Shape after dropping columns:", test_df.shape)

# Display the first few rows of the updated datasets
print("Training Dataset Head:")
print(train_df.head())

print("Testing Dataset Head:")
print(test_df.head())

```

```


# One-hot encode the 'Color' column
train_df = pd.get_dummies(train_df, columns=['Color'])
test_df = pd.get_dummies(test_df, columns=['Color'])

# Display the updated shape of the datasets
print("Training Dataset Shape after one-hot encoding:", train_df.shape)
print("Testing Dataset Shape after one-hot encoding:", test_df.shape)

# Display the first few rows of the updated datasets
print("Training Dataset Head:")
print(train_df.head())

print("Testing Dataset Head:")
print(test_df.head())

```

```



# Separate features and target variable for the training dataset
X_train = train_df.drop(columns=['Total Cup Points'])
y_train = train_df['Total Cup Points']

# Separate features and target variable for the test dataset
X_test = test_df.drop(columns=['Total Cup Points'])
y_test = test_df['Total Cup Points']

# Display the shapes of the feature and target datasets
print("Training Feature Shape:", X_train.shape)
print("Training Target Shape:", y_train.shape)
print("Testing Feature Shape:", X_test.shape)
print("Testing Target Shape:", y_test.shape)
```

```



from sklearn.ensemble import RandomForestRegressor

# Initialize the RandomForestRegressor
rf_regressor = RandomForestRegressor(n_estimators=100, random_state=42)

# Train the model on the training data
rf_regressor.fit(X_train, y_train)

# Display the trained model
rf_regressor


```

```



# Make predictions on the test set
y_pred = rf_regressor.predict(X_test)

# Display the first few predictions
y_pred[:5]


```

```



from sklearn.metrics import mean_squared_error

# Calculate the mean squared error
mse = mean_squared_error(y_test, y_pred)

# Display the mean squared error
mse


```
