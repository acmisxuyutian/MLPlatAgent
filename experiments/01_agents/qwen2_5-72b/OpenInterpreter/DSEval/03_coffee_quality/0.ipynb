```

import pandas as pd

# Load the training and test datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\03_coffee_quality\\train.csv'
test_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\03_coffee_quality\\test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the training and test datasets
print("Training Data:")
print(train_df.head())
print("\nTest Data:")
print(test_df.head())
```

```

# Drop duplicated rows from the training and test datasets
train_df.drop_duplicates(inplace=True)
test_df.drop_duplicates(inplace=True)

# Check the shape of the datasets after dropping duplicates
print("Training Data Shape after dropping duplicates:", train_df.shape)
print("Test Data Shape after dropping duplicates:", test_df.shape)
```

```

# List of columns to drop
columns_to_drop = [
    "ID", "ICO Number", "Owner", "Region", "Certification Contact", "Certification Address", "Farm Name",
    "Lot Number", "Mill", "ICO Number", "Producer", "Company", "Expiration", "Harvest Year", "Unnamed: 0",
    "Number of Bags", "Bag Weight", "In-Country Partner", "Country of Origin", "Processing Method", "Grading Date",
    "Variety", "Status", "Defects", "Uniformity", "Clean Cup", "Sweetness", "Certification Body"
]

# Drop the specified columns from the training and test datasets
train_df.drop(columns=columns_to_drop, inplace=True)
test_df.drop(columns=columns_to_drop, inplace=True)

# Check the shape of the datasets after dropping columns
print("Training Data Shape after dropping columns:", train_df.shape)
print("Test Data Shape after dropping columns:", test_df.shape)
```

```


# Check the column names in the training and test datasets
print("Training Data Columns:")
print(train_df.columns.tolist())
print("\nTest Data Columns:")
print(test_df.columns.tolist())

# Filter the columns to drop to only include those that are present in the datasets
columns_to_drop_train = [col for col in columns_to_drop if col in train_df.columns]
columns_to_drop_test = [col for col in columns_to_drop if col in test_df.columns]

# Drop the specified columns from the training and test datasets
train_df.drop(columns=columns_to_drop_train, inplace=True)
test_df.drop(columns=columns_to_drop_test, inplace=True)

# Check the shape of the datasets after dropping columns
print("\nTraining Data Shape after dropping columns:", train_df.shape)
print("Test Data Shape after dropping columns:", test_df.shape)

```

```



# One-hot encode the 'Color' column
train_df = pd.get_dummies(train_df, columns=['Color'])
test_df = pd.get_dummies(test_df, columns=['Color'])

# Check the shape of the datasets after one-hot encoding
print("Training Data Shape after one-hot encoding:", train_df.shape)
print("Test Data Shape after one-hot encoding:", test_df.shape)

# Check the columns to ensure one-hot encoding was successful
print("\nTraining Data Columns after one-hot encoding:")
print(train_df.columns.tolist())
print("\nTest Data Columns after one-hot encoding:")
print(test_df.columns.tolist())

```

```


# Separate features and target variable
X_train = train_df.drop(columns=['Total Cup Points'])
y_train = train_df['Total Cup Points']

X_test = test_df.drop(columns=['Total Cup Points'])
y_test = test_df['Total Cup Points']

# Check the shapes of the feature and target datasets
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)

```

```



from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error

# Train a RandomForestRegressor
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model using mean squared error
mse = mean_squared_error(y_test, y_pred)

# Print the mean squared error
print("Mean Squared Error:", mse)


```
