```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

# Load the train and test datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark/DSEval/datasets/02_cardiovascular_diseases/test.csv'

# Read the datasets
train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the train and test datasets
print("Train Data:")
train_df.head()
print("\nTest Data:")
test_df.head()
```

```

# Drop rows with missing values and duplicates from the train and test datasets
train_df = train_df.dropna().drop_duplicates()
test_df = test_df.dropna().drop_duplicates()

# Display the shape of the datasets after dropping missing values and duplicates
print("Train Data Shape after dropping missing values and duplicates:", train_df.shape)
print("Test Data Shape after dropping missing values and duplicates:", test_df.shape)
```

```


# Define the BMI categories
bmi_categories = [
    (train_df['BMI'] < 18.5, 'Underweight'),
    ((train_df['BMI'] >= 18.5) & (train_df['BMI'] < 24.9), 'Normal weight'),
    ((train_df['BMI'] >= 24.9) & (train_df['BMI'] < 30), 'Overweight'),
    (train_df['BMI'] >= 30, 'Obesity')
]

# Create the BMI_Category column for the train dataset
train_df['BMI_Category'] = pd.Categorical(pd.DataFrame(
    [cat for condition, cat in bmi_categories if condition.any()],
    index=[cat for _, cat in bmi_categories]
).mode(axis=1)[0], categories=['Underweight', 'Normal weight', 'Overweight', 'Obesity'], ordered=True)

# Define the BMI categories for the test dataset
bmi_categories_test = [
    (test_df['BMI'] < 18.5, 'Underweight'),
    ((test_df['BMI'] >= 18.5) & (test_df['BMI'] < 24.9), 'Normal weight'),
    ((test_df['BMI'] >= 24.9) & (test_df['BMI'] < 30), 'Overweight'),
    (test_df['BMI'] >= 30, 'Obesity')
]

# Create the BMI_Category column for the test dataset
test_df['BMI_Category'] = pd.Categorical(pd.DataFrame(
    [cat for condition, cat in bmi_categories_test if condition.any()],
    index=[cat for _, cat in bmi_categories_test]
).mode(axis=1)[0], categories=['Underweight', 'Normal weight', 'Overweight', 'Obesity'], ordered=True)

# Display the first few rows of the train and test datasets with the new BMI_Category column
print("Train Data with BMI_Category:")
train_df.head()
print("\nTest Data with BMI_Category:")
test_df.head()

```

```


# Define the BMI categories and their corresponding labels
bins = [0, 18.5, 24.9, 29.9, float('inf')]
labels = ['Underweight', 'Normal weight', 'Overweight', 'Obesity']

# Create the BMI_Category column for the train dataset
train_df['BMI_Category'] = pd.cut(train_df['BMI'], bins=bins, labels=labels, include_lowest=True, ordered=True)

# Create the BMI_Category column for the test dataset
test_df['BMI_Category'] = pd.cut(test_df['BMI'], bins=bins, labels=labels, include_lowest=True, ordered=True)

# Display the first few rows of the train and test datasets with the new BMI_Category column
print("Train Data with BMI_Category:")
train_df.head()
print("\nTest Data with BMI_Category:")
test_df.head()

```
