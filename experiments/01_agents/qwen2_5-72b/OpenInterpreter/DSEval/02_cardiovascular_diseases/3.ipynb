```

import pandas as pd

# Load the training and test datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\02_cardiovascular_diseases\\train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\02_cardiovascular_diseases\\test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the training and test datasets
print("Training Data:")
print(train_df.head())
print("\nTest Data:")
print(test_df.head())
```

```


# Drop rows with missing values
train_df = train_df.dropna()
test_df = test_df.dropna()

# Drop duplicate rows
train_df = train_df.drop_duplicates()
test_df = test_df.drop_duplicates()

# Display the first few rows of the cleaned datasets
print("Cleaned Training Data:")
print(train_df.head())
print("\nCleaned Test Data:")
print(test_df.head())

# Check the number of rows before and after cleaning
print("\nNumber of rows in training data before cleaning:", len(train_df) + train_df.isna().any(axis=1).sum())
print("Number of rows in training data after cleaning:", len(train_df))
print("\nNumber of rows in test data before cleaning:", len(test_df) + test_df.isna().any(axis=1).sum())
print("Number of rows in test data after cleaning:", len(test_df))
```

```



# Function to calculate BMI and categorize it
def bmi_category(row):
    bmi = row['Weight_kg'] / (row['Height_cm'] / 100) ** 2
    if bmi < 18.5:
        return 'Underweight'
    elif 18.5 <= bmi < 24.9:
        return 'Normal weight'
    elif 24.9 <= bmi < 29.9:
        return 'Overweight'
    else:
        return 'Obesity'

# Apply the function to create the BMI_Category column
train_df['BMI_Category'] = train_df.apply(bmi_category, axis=1)
test_df['BMI_Category'] = test_df.apply(bmi_category, axis=1)

# Convert BMI_Category to category dtype
train_df['BMI_Category'] = train_df['BMI_Category'].astype('category')
test_df['BMI_Category'] = test_df['BMI_Category'].astype('category')

# Display the first few rows of the datasets with the new BMI_Category column
print("Training Data with BMI_Category:")
print(train_df.head())
print("\nTest Data with BMI_Category:")
print(test_df.head())

```

```



# Function to calculate BMI and categorize it
def bmi_category(row):
    height_m = row['Height_(cm)'] / 100
    bmi = row['Weight_(kg)'] / (height_m ** 2)
    if bmi < 18.5:
        return 'Underweight'
    elif 18.5 <= bmi < 24.9:
        return 'Normal weight'
    elif 24.9 <= bmi < 29.9:
        return 'Overweight'
    else:
        return 'Obesity'

# Apply the function to create the BMI_Category column
train_df['BMI_Category'] = train_df.apply(bmi_category, axis=1)
test_df['BMI_Category'] = test_df.apply(bmi_category, axis=1)

# Convert BMI_Category to category dtype
train_df['BMI_Category'] = train_df['BMI_Category'].astype('category')
test_df['BMI_Category'] = test_df['BMI_Category'].astype('category')

# Display the first few rows of the datasets with the new BMI_Category column
print("Training Data with BMI_Category:")
print(train_df.head())
print("\nTest Data with BMI_Category:")
print(test_df.head())

```

```



import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score

# Encode categorical variables using one-hot encoding
categorical_columns = ['General_Health', 'Checkup', 'Exercise', 'Heart_Disease', 'Skin_Cancer', 'Other_Cancer', 'Depression', 'Diabetes', 'Arthritis', 'Sex', 'Age_Category', 'Smoking_History', 'BMI_Category']

# One-hot encode the categorical columns
encoder = OneHotEncoder(drop='first', sparse_output=False)
train_encoded = pd.DataFrame(encoder.fit_transform(train_df[categorical_columns]))
test_encoded = pd.DataFrame(encoder.transform(test_df[categorical_columns]))

# Align the encoded data with the original data
train_encoded.columns = encoder.get_feature_names_out(categorical_columns)
test_encoded.columns = encoder.get_feature_names_out(categorical_columns)

# Drop the original categorical columns from the datasets
train_df = train_df.drop(columns=categorical_columns)
test_df = test_df.drop(columns=categorical_columns)

# Concatenate the encoded data with the remaining numerical data
train_df = pd.concat([train_df, train_encoded], axis=1)
test_df = pd.concat([test_df, test_encoded], axis=1)

# Split the data into features (X) and target (y)
X_train = train_df.drop(columns=['Heart_Disease'])
y_train = train_df['Heart_Disease']
X_test = test_df.drop(columns=['Heart_Disease'])
y_test = test_df['Heart_Disease']

# Train a logistic regression model
model = LogisticRegression()
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred_proba = model.predict_proba(X_test)[:, 1]

# Compute the area under the ROC curve
roc_auc = roc_auc_score(y_test, y_pred_proba)

# Print the ROC AUC score
print(f"Area Under ROC Curve: {roc_auc:.4f}")

```

```



from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

# Remove rows with missing values in the target variable (y)
train_df = train_df.dropna(subset=['Heart_Disease_Yes'])
test_df = test_df.dropna(subset=['Heart_Disease_Yes'])

# Split the data into features (X) and target (y) using the correct column names
X_train = train_df.drop(columns=['Heart_Disease_Yes'])
y_train = train_df['Heart_Disease_Yes']
X_test = test_df.drop(columns=['Heart_Disease_Yes'])
y_test = test_df['Heart_Disease_Yes']

# Impute missing values with the median
imputer = SimpleImputer(strategy='median')
X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)

# Scale the data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_imputed)
X_test_scaled = scaler.transform(X_test_imputed)

# Train a logistic regression model with increased max_iter
model = LogisticRegression(max_iter=1000)
model.fit(X_train_scaled, y_train)

# Make predictions on the test set
y_pred_proba = model.predict_proba(X_test_scaled)[:, 1]

# Compute the area under the ROC curve
roc_auc = roc_auc_score(y_test, y_pred_proba)

# Print the ROC AUC score
print(f"Area Under ROC Curve: {roc_auc:.4f}")


```
