```

import pandas as pd

# Load the training and test data
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark\DSEval\datasets\09_housing_price\train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark\DSEval\datasets\09_housing_price\test.csv'

train_data = pd.read_csv(train_data_path)
test_data = pd.read_csv(test_data_path)

# Display the first few rows of the training and test datasets
train_data.head(), test_data.head()
```

```


from sklearn.linear_model import LinearRegression

# Assuming the target column is 'price' and the rest are features
X_train = train_data.drop(columns='price')
y_train = train_data['price']

# Initialize and fit the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Display the coefficients and intercept of the model
model.coef_, model.intercept_

```

```


from sklearn.preprocessing import OneHotEncoder

# Identify categorical columns
categorical_columns = train_data.select_dtypes(include=['object']).columns

# Initialize the OneHotEncoder
encoder = OneHotEncoder(sparse=False)

# Fit and transform the training data
X_train_encoded = pd.DataFrame(encoder.fit_transform(train_data[categorical_columns]), columns=encoder.get_feature_names_out(categorical_columns))

# Concatenate the encoded categorical features with the numerical features
X_train_encoded = pd.concat([train_data.drop(columns=categorical_columns), X_train_encoded], axis=1)

# Display the first few rows of the encoded training data
X_train_encoded.head()

```

```


from sklearn.preprocessing import OneHotEncoder

# Identify categorical columns
categorical_columns = train_data.select_dtypes(include=['object']).columns

# Initialize the OneHotEncoder with sparse_output=False
encoder = OneHotEncoder(sparse_output=False)

# Fit and transform the training data
X_train_encoded = pd.DataFrame(encoder.fit_transform(train_data[categorical_columns]), columns=encoder.get_feature_names_out(categorical_columns))

# Concatenate the encoded categorical features with the numerical features
X_train_encoded = pd.concat([train_data.drop(columns=categorical_columns), X_train_encoded], axis=1)

# Display the first few rows of the encoded training data
X_train_encoded.head()
```

```



from sklearn.linear_model import LinearRegression

# Separate the target variable from the features
X_train = X_train_encoded.drop(columns='price')
y_train = X_train_encoded['price']

# Initialize and fit the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Display the coefficients and intercept of the model
model.coef_, model.intercept_


```

```



# Identify categorical columns in the test data
categorical_columns_test = test_data.select_dtypes(include=['object']).columns

# Transform the test data using the same encoder
X_test_encoded = pd.DataFrame(encoder.transform(test_data[categorical_columns_test]), columns=encoder.get_feature_names_out(categorical_columns_test))

# Concatenate the encoded categorical features with the numerical features
X_test_encoded = pd.concat([test_data.drop(columns=categorical_columns_test), X_test_encoded], axis=1)

# Make predictions on the test set
predictions = model.predict(X_test_encoded)

# Display the first few predictions
predictions[:5]

```

```



# Drop the 'price' column from the test data
X_test_encoded = X_test_encoded.drop(columns=['price'], errors='ignore')

# Make predictions on the test set
predictions = model.predict(X_test_encoded)

# Display the first few predictions
predictions[:5]


```

```



from sklearn.metrics import mean_squared_error
import numpy as np

# Assuming the test data has a 'price' column for the true prices
y_test = test_data['price']

# Calculate the RMSE
rmse = np.sqrt(mean_squared_error(y_test, predictions))

# Display the RMSE
rmse


```
