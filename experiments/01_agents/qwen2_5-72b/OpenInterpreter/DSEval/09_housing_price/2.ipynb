```

import pandas as pd

# Load the training and test datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\09_housing_price\\train.csv'
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\09_housing_price\\test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of each dataset
train_df.head(), test_df.head()
```

```

# Import necessary libraries
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

# Load the training and test datasets
train_df = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\09_housing_price\\train.csv')
test_df = pd.read_csv('D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\09_housing_price\\test.csv')

# Display the first few rows of each dataset to verify loading
print("Training Data:")
print(train_df.head())

print("\nTest Data:")
print(test_df.head())

# Preprocess the data
# Encode categorical variables
label_encoder = LabelEncoder()
train_df['furnishingstatus'] = label_encoder.fit_transform(train_df['furnishingstatus'])
test_df['furnishingstatus'] = label_encoder.transform(test_df['furnishingstatus'])

# Separate features (X) and target (y)
X_train = train_df.drop(columns=['price'])
y_train = train_df['price']

X_test = test_df.drop(columns=['price'])
y_test = test_df['price']

# Display the preprocessed data
print("\nPreprocessed Training Data Features (X_train):")
print(X_train.head())

print("\nPreprocessed Training Data Target (y_train):")
print(y_train.head())

print("\nPreprocessed Test Data Features (X_test):")
print(X_test.head())

print("\nPreprocessed Test Data Target (y_test):")
print(y_test.head())
```

```


# Import the necessary libraries
from sklearn.linear_model import LinearRegression

# Initialize the linear regression model
model = LinearRegression()

# Fit the model on the training data
model.fit(X_train, y_train)

# Make predictions on the test data
y_pred = model.predict(X_test)

# Display the first few predictions
y_pred[:5]

```

```


# Import necessary libraries
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

# Encode the 'prefarea' column
label_encoder = LabelEncoder()
train_df['prefarea'] = label_encoder.fit_transform(train_df['prefarea'])
test_df['prefarea'] = label_encoder.transform(test_df['prefarea'])

# Re-separate features (X) and target (y)
X_train = train_df.drop(columns=['price'])
y_train = train_df['price']

X_test = test_df.drop(columns=['price'])
y_test = test_df['price']

# Re-fit the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions on the test data
y_pred = model.predict(X_test)

# Evaluate the model using RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
rmse
```

```



# Check the data types of the columns in the training and test datasets
print("Training Data Data Types:")
print(train_df.dtypes)

print("\nTest Data Data Types:")
print(test_df.dtypes)

# Identify and encode any remaining categorical variables
# We will encode all columns that are of type 'object'
categorical_columns = train_df.select_dtypes(include=['object']).columns

label_encoder = LabelEncoder()
for col in categorical_columns:
    train_df[col] = label_encoder.fit_transform(train_df[col])
    test_df[col] = label_encoder.transform(test_df[col])

# Re-separate features (X) and target (y)
X_train = train_df.drop(columns=['price'])
y_train = train_df['price']

X_test = test_df.drop(columns=['price'])
y_test = test_df['price']

# Re-fit the linear regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions on the test data
y_pred = model.predict(X_test)

# Evaluate the model using RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
rmse

```
