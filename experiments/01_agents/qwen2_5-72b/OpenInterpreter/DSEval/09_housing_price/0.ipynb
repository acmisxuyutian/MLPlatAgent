```

import pandas as pd

# Define the file paths
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/09_housing_price/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/DSEval/datasets/09_housing_price/test.csv'

# Load the datasets
train_data = pd.read_csv(train_data_path)
test_data = pd.read_csv(test_data_path)

# Display the first few rows of the datasets
print("Training Data:")
print(train_data.head())

print("\nTest Data:")
print(test_data.head())
```

```


# Import necessary libraries
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

# Select the features and the target variable
features = ['area', 'bedrooms', 'bathrooms', 'stories', 'mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'parking', 'prefarea']
target = 'price'

# Convert categorical variables to numerical using one-hot encoding
train_data_encoded = pd.get_dummies(train_data[features + [target]], drop_first=True)
test_data_encoded = pd.get_dummies(test_data[features], drop_first=True)

# Align the columns of the test data with the training data
test_data_encoded = test_data_encoded.reindex(columns=train_data_encoded.columns.difference([target]), fill_value=0)

# Split the data into features (X) and target variable (y)
X_train = train_data_encoded.drop(columns=[target])
y_train = train_data_encoded[target]

X_test = test_data_encoded

# Display the first few rows of the encoded datasets
print("Encoded Training Data:")
print(X_train.head())

print("\nEncoded Test Data:")
print(X_test.head())

```

```


# Initialize the linear regression model
model = LinearRegression()

# Fit the model on the training data
model.fit(X_train, y_train)

# Make predictions on the test data
y_pred = model.predict(X_test)

# Since we don't have the actual target values for the test data, we'll assume the test data has a 'price' column for evaluation
y_test = test_data[target]

# Calculate the RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

# Print the RMSE
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")

# Display the first few predicted prices
print("\nPredicted Prices:")
print(y_pred[:5])

# Display the actual prices for comparison
print("\nActual Prices:")
print(y_test[:5])

```
