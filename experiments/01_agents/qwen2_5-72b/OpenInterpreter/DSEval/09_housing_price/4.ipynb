```

import pandas as pd

# Load the training data
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\09_housing_price\\train.csv'
train_data = pd.read_csv(train_data_path)

# Load the test data
test_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\DSEval\\datasets\\09_housing_price\\test.csv'
test_data = pd.read_csv(test_data_path)

# Display the first few rows of the training and test data
print("Training Data:")
print(train_data.head())
print("\nTest Data:")
print(test_data.head())
```

```


import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Load the data again (for clarity)
train_data = pd.read_csv(train_data_path)
test_data = pd.read_csv(test_data_path)

# Identify categorical columns
categorical_columns = ['prefarea', 'furnishingstatus']

# Create a ColumnTransformer to apply one-hot encoding to categorical columns
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(), categorical_columns)
    ],
    remainder='passthrough'
)

# Preprocess the training data
X_train = train_data.drop(columns=['price'])
y_train = train_data['price']
X_train_processed = preprocessor.fit_transform(X_train)

# Preprocess the test data
X_test = test_data.drop(columns=['price'])
y_test = test_data['price']
X_test_processed = preprocessor.transform(X_test)

# Display the first few rows of the processed training data
print("Processed Training Data (First 5 rows):")
print(X_train_processed[:5])

# Display the first few rows of the processed test data
print("\nProcessed Test Data (First 5 rows):")
print(X_test_processed[:5])

```

```


# Initialize the Linear Regression model
model = LinearRegression()

# Train the model on the processed training data
model.fit(X_train_processed, y_train)

# Display the coefficients of the model
print("Model Coefficients:")
print(model.coef_)
print("Model Intercept:")
print(model.intercept_)

# Make predictions on the test set
y_pred = model.predict(X_test_processed)

# Display the first few predictions
print("\nPredicted Prices (First 5):")
print(y_pred[:5])

# Evaluate the model using RMSE
rmse = mean_squared_error(y_test, y_pred, squared=False)
print("\nRoot Mean Squared Error (RMSE):", rmse)

```

```




import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

# Load the data again (for clarity)
train_data = pd.read_csv(train_data_path)
test_data = pd.read_csv(test_data_path)

# Identify columns with string values
string_columns = train_data.select_dtypes(include=['object']).columns

# One-hot encode the categorical columns
train_data = pd.get_dummies(train_data, columns=string_columns)
test_data = pd.get_dummies(test_data, columns=string_columns)

# Ensure all columns are in the same order and have the same names
# This is important for the test data to match the training data
train_columns = train_data.columns
test_data = test_data.reindex(columns=train_columns, fill_value=0)

# Split the data into features (X) and target variable (y)
X_train = train_data.drop(columns=['price'])
y_train = train_data['price']
X_test = test_data.drop(columns=['price'])
y_test = test_data['price']

# Display the first few rows of the processed training data
print("Processed Training Data (First 5 rows):")
print(X_train.head())

# Display the first few rows of the processed test data
print("\nProcessed Test Data (First 5 rows):")
print(X_test.head())

# Initialize the Linear Regression model
model = LinearRegression()

# Train the model on the processed training data
model.fit(X_train, y_train)

# Display the coefficients of the model
print("\nModel Coefficients:")
print(model.coef_)
print("Model Intercept:")
print(model.intercept_)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Display the first few predictions
print("\nPredicted Prices (First 5):")
print(y_pred[:5])

# Calculate the Mean Squared Error (MSE)
mse = mean_squared_error(y_test, y_pred)

# Calculate the Root Mean Squared Error (RMSE)
rmse = np.sqrt(mse)

# Display the RMSE
print("\nRoot Mean Squared Error (RMSE):", rmse)


```
