```

import pandas as pd

# 加载训练集和测试集
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/obesity_CVD_risk/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/obesity_CVD_risk/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# 显示前几条数据
train_df.head()
```

```


# 查看数据的基本信息
train_df.info()

# 查看数据的描述统计
train_df.describe(include='all')
```

```


# 查看数据的基本信息
train_df.info()

# 查看数据的描述统计
train_df.describe(include='all')
```

```


# 特征编码
from sklearn.preprocessing import LabelEncoder

# 选择需要编码的特征
categorical_features = ['Gender', 'family_history_with_overweight', 'FAVC', 'CAEC', 'SMOKE', 'SCC', 'CALC', 'MTRANS', 'NObeyesdad']

# 初始化LabelEncoder
le = LabelEncoder()

# 对训练集和测试集的分类特征进行编码
for feature in categorical_features:
    train_df[feature] = le.fit_transform(train_df[feature])
    test_df[feature] = le.fit_transform(test_df[feature])

# 查看编码后的前几行
train_df.head()
```

```


# 特征构建：计算 BMI
train_df['BMI'] = train_df['Weight'] / (train_df['Height'] ** 2)
test_df['BMI'] = test_df['Weight'] / (test_df['Height'] ** 2)

# 查看包含 BMI 的前几行
train_df.head()

```

```



import matplotlib.pyplot as plt
import seaborn as sns

# 计算特征与目标变量的相关性
correlation_matrix = train_df.corr()
correlation_with_target = correlation_matrix['NObeyesdad'].sort_values(ascending=False)

# 绘制相关性热力图
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()

# 显示特征与目标变量的相关性
correlation_with_target
```
