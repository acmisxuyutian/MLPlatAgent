```

import pandas as pd

# 读取训练集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/obesity_CVD_risk/train.csv'
train_df = pd.read_csv(train_path)

# 读取测试集
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/obesity_CVD_risk/test.csv'
test_df = pd.read_csv(test_path)

# 查看数据前几行
train_df.head()
```

```


# 检查数据的基本统计信息
train_df.describe(include='all')

# 检查缺失值
train_df.isnull().sum()
```

```


# 检查数据的基本统计信息
train_df.describe(include='all')

# 检查分类特征的类别分布
categorical_columns = ['Gender', 'family_history_with_overweight', 'FAVC', 'CAEC', 'SMOKE', 'SCC', 'CALC', 'MTRANS', 'NObeyesdad']
for col in categorical_columns:
    print(f'{col} distribution:')
    print(train_df[col].value_counts(normalize=True))
    print()
```

```



# 创建BMI特征
train_df['BMI'] = train_df['Weight'] / (train_df['Height'] ** 2)

# 查看新的特征
train_df[['Weight', 'Height', 'BMI']].head()
```

```



import matplotlib.pyplot as plt
import seaborn as sns

# 计算相关性矩阵
correlation_matrix = train_df.corr()

# 绘制相关性矩阵热图
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()

# 显示与目标变量NObeyesdad相关的特征
target_corr = correlation_matrix['NObeyesdad'].sort_values(ascending=False)
target_corr

```

```



# 对分类特征进行独热编码
categorical_columns = ['Gender', 'family_history_with_overweight', 'FAVC', 'CAEC', 'SMOKE', 'SCC', 'CALC', 'MTRANS']
train_df_encoded = pd.get_dummies(train_df, columns=categorical_columns, drop_first=True)

# 计算相关性矩阵
correlation_matrix = train_df_encoded.corr()

# 绘制相关性矩阵热图
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()

# 显示与目标变量NObeyesdad相关的特征
target_corr = correlation_matrix['NObeyesdad_Obesity_Type_III'].sort_values(ascending=False)
target_corr
```

```



from sklearn.preprocessing import LabelEncoder

# 对目标变量进行编码
label_encoder = LabelEncoder()
train_df_encoded['NObeyesdad_encoded'] = label_encoder.fit_transform(train_df_encoded['NObeyesdad'])

# 计算相关性矩阵
correlation_matrix = train_df_encoded.corr()

# 绘制相关性矩阵热图
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()

# 显示与目标变量NObeyesdad相关的特征
target_corr = correlation_matrix['NObeyesdad_encoded'].sort_values(ascending=False)
target_corr

```

```




# 确保只使用数值型列来计算相关性矩阵
numeric_columns = train_df_encoded.select_dtypes(include=['number']).columns
correlation_matrix = train_df_encoded[numeric_columns].corr()

# 绘制相关性矩阵热图
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()

# 显示与目标变量NObeyesdad相关的特征
target_corr = correlation_matrix['NObeyesdad_encoded'].sort_values(ascending=False)
target_corr
```
