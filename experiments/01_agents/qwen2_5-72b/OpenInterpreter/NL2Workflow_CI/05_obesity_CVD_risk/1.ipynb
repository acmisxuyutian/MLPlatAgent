```

import pandas as pd

# 载入训练集和测试集
train_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/obesity_CVD_risk/train.csv'
test_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/obesity_CVD_risk/test.csv'

train_df = pd.read_csv(train_file_path)
test_df = pd.read_csv(test_file_path)

# 查看数据前5行
train_df.head(5)
```

```

# 数据探索
# 查看数据的基本统计信息
train_df.describe(include='all')

# 检查缺失值
train_df.isnull().sum()
```

```


# 查看数值型特征的基本统计信息
train_df.describe()

# 查看分类特征的基本统计信息
train_df.describe(include=['object', 'category'])

```

```


# 构建新的特征：BMI (Body Mass Index)
train_df['BMI'] = train_df['Weight'] / (train_df['Height'] ** 2)
test_df['BMI'] = test_df['Weight'] / (test_df['Height'] ** 2)

# 查看新的数据集
train_df.head(5)

```

```


# 导入必要的库
from sklearn.preprocessing import LabelEncoder

# 编码分类变量
label_encoders = {}

# 获取所有分类变量的列名
categorical_columns = train_df.select_dtypes(include=['object', 'category']).columns

# 对每个分类变量进行编码
for col in categorical_columns:
    le = LabelEncoder()
    train_df[col] = le.fit_transform(train_df[col])
    test_df[col] = le.transform(test_df[col])
    label_encoders[col] = le

# 查看编码后的数据
train_df.head(5)

```

```



# 计算特征与目标变量的相关性
correlation_matrix = train_df.corr()
correlation_with_target = correlation_matrix['NObeyesdad'].sort_values(ascending=False)

# 显示相关性
correlation_with_target
```

```



# 选择相关性较高的特征
selected_features = ['BMI', 'Weight', 'family_history_with_overweight', 'CAEC', 'Age']

# 分离特征和目标变量
X_train = train_df[selected_features]
y_train = train_df['NObeyesdad']

X_test = test_df[selected_features]
y_test = test_df['NObeyesdad']

# 检查特征和目标变量的形状
X_train.shape, y_train.shape, X_test.shape, y_test.shape

```

```


# 导入必要的库
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# 选择模型：随机森林分类器
model = RandomForestClassifier(n_estimators=100, random_state=42)

# 训练模型
model.fit(X_train, y_train)

# 在测试集上进行预测
y_pred = model.predict(X_test)

# 计算模型的准确性
accuracy = accuracy_score(y_test, y_pred)
print(f'模型准确率: {accuracy:.4f}')

# 打印分类报告
classification_rep = classification_report(y_test, y_pred)
print('分类报告:\n', classification_rep)


```
