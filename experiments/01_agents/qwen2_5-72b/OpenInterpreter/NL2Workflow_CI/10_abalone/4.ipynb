```



import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_log_error
import numpy as np

# 读取训练集和测试集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 独热编码
train_df = pd.get_dummies(train_df, columns=['Sex'])
test_df = pd.get_dummies(test_df, columns=['Sex'])

# 数据划分
X_train = train_df.drop(columns=['id', 'Rings'])
y_train = train_df['Rings']
X_test = test_df.drop(columns=['id', 'Rings'])
y_test = test_df['Rings']

# 划分训练集和验证集
X_train_split, X_val, y_train_split, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# 模型训练
model = GradientBoostingRegressor(random_state=42)
model.fit(X_train_split, y_train_split)

# 模型评估
y_pred = model.predict(X_val)
val_rmlse = np.sqrt(mean_squared_log_error(y_val, y_pred))
val_rmlse
```

```



import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_log_error
import numpy as np

# 读取训练集和测试集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 独热编码
train_df = pd.get_dummies(train_df, columns=['Sex'])
test_df = pd.get_dummies(test_df, columns=['Sex'])

# 数据划分
X_train = train_df.drop(columns=['id', 'Rings'])
y_train = train_df['Rings']
X_test = test_df.drop(columns=['id', 'Rings'])
y_test = test_df['Rings']

# 划分训练集和验证集
X_train_split, X_val, y_train_split, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# 模型训练
model = GradientBoostingRegressor(random_state=42)
model.fit(X_train_split, y_train_split)

# 在测试集上进行预测
y_pred_test = model.predict(X_test)
test_rmlse = np.sqrt(mean_squared_log_error(y_test, y_pred_test))
test_rmlse

```
