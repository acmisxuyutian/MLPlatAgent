```

import pandas as pd

# 加载训练集和测试集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 查看数据集的前几行
train_df.head()
```

```


# 对分类特征进行独热编码
train_df = pd.get_dummies(train_df, columns=['Sex'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['Sex'], drop_first=True)

# 查看独热编码后的数据
train_df.head()

```

```


# 分割特征和目标变量
X_train = train_df.drop(columns=['id', 'Rings'])
y_train = train_df['Rings']

X_test = test_df.drop(columns=['id', 'Rings'])
y_test = test_df['Rings']

# 将训练集进一步划分为训练子集和验证子集
from sklearn.model_selection import train_test_split

X_train_sub, X_val, y_train_sub, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# 查看划分后的数据形状
print(f"X_train_sub shape: {X_train_sub.shape}")
print(f"y_train_sub shape: {y_train_sub.shape}")
print(f"X_val shape: {X_val.shape}")
print(f"y_val shape: {y_val.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_test shape: {y_test.shape}")

```

```


from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_log_error
import numpy as np

# 选择模型
model = RandomForestRegressor(n_estimators=100, random_state=42)

# 训练模型
model.fit(X_train_sub, y_train_sub)

# 在验证集上进行预测
y_val_pred = model.predict(X_val)

# 计算验证集上的RMLSE
rmlse_val = np.sqrt(mean_squared_log_error(y_val, y_val_pred))
print(f"Validation RMLSE: {rmlse_val}")

# 在测试集上进行预测
y_test_pred = model.predict(X_test)

# 计算测试集上的RMLSE
rmlse_test = np.sqrt(mean_squared_log_error(y_test, y_test_pred))
print(f"Test RMLSE: {rmlse_test}")

```
