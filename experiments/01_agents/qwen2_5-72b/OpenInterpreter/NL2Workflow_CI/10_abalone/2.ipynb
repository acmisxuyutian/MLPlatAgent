```

import pandas as pd

# 加载训练集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/train.csv'
train_data = pd.read_csv(train_path)

# 加载测试集
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/test.csv'
test_data = pd.read_csv(test_path)

# 查看数据
train_data.head()
```

```


# 删除重复列
train_data = train_data.drop(columns=['Whole weight.1', 'Whole weight.2'])
test_data = test_data.drop(columns=['Whole weight.1', 'Whole weight.2'])

# 对类别特征进行独热编码
train_data = pd.get_dummies(train_data, columns=['Sex'])
test_data = pd.get_dummies(test_data, columns=['Sex'])

# 查看处理后的数据
train_data.head()
```

```


from sklearn.model_selection import train_test_split

# 将特征和目标变量分开
X_train = train_data.drop(columns=['id', 'Rings'])
y_train = train_data['Rings']

X_test = test_data.drop(columns=['id', 'Rings'])
y_test = test_data['Rings']

# 将训练集进一步划分为训练集和验证集
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# 查看数据集的形状
print(f'X_train shape: {X_train.shape}')
print(f'y_train shape: {y_train.shape}')
print(f'X_val shape: {X_val.shape}')
print(f'y_val shape: {y_val.shape}')
print(f'X_test shape: {X_test.shape}')
print(f'y_test shape: {y_test.shape}')

```

```


from sklearn.ensemble import RandomForestRegressor

# 初始化模型
model = RandomForestRegressor(n_estimators=100, random_state=42)

# 训练模型
model.fit(X_train, y_train)

# 在验证集上评估模型
y_val_pred = model.predict(X_val)

# 计算验证集上的RMLSE
import numpy as np

def calculate_rmlse(y_true, y_pred):
    return np.sqrt(np.mean((np.log1p(y_pred) - np.log1p(y_true)) ** 2))

rmlse_val = calculate_rmlse(y_val, y_val_pred)
print(f'Validation RMLSE: {rmlse_val:.4f}')

```

```



# 在测试集上评估模型
y_test_pred = model.predict(X_test)

# 计算测试集上的RMLSE
rmlse_test = calculate_rmlse(y_test, y_test_pred)
print(f'Test RMLSE: {rmlse_test:.4f}')


```
