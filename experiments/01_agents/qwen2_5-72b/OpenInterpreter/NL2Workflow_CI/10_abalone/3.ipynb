```


import pandas as pd
from sklearn.model_selection import train_test_split

# 加载训练集和测试集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/abalone/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 对类别特征进行独热编码
train_df = pd.get_dummies(train_df, columns=['Sex'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['Sex'], drop_first=True)

# 删除 id 列
train_df = train_df.drop(columns=['id'])
test_df = test_df.drop(columns=['id'])

# 分割特征和标签
X_train = train_df.drop(columns=['Rings'])
y_train = train_df['Rings']

X_test = test_df.drop(columns=['Rings'])
y_test = test_df['Rings']

# 查看预处理后的数据
print("预处理后的训练集前5行：")
print(X_train.head())
print("\n预处理后的测试集前5行：")
print(X_test.head())

```

```

from sklearn.model_selection import train_test_split

# 将训练集划分为训练集和验证集
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# 查看划分后的数据集大小
print(f"训练集大小: {X_train.shape[0]}")
print(f"验证集大小: {X_val.shape[0]}")
print(f"测试集大小: {X_test.shape[0]}")
```

```

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_log_error
import numpy as np

# 初始化线性回归模型
model = LinearRegression()

# 训练模型
model.fit(X_train, y_train)

# 在验证集上进行预测
y_val_pred = model.predict(X_val)

# 计算验证集上的RMLSE
rmlse_val = np.sqrt(mean_squared_log_error(y_val, y_val_pred))
print(f"验证集上的RMLSE: {rmlse_val}")

# 在测试集上进行预测
y_test_pred = model.predict(X_test)

# 计算测试集上的RMLSE
rmlse_test = np.sqrt(mean_squared_log_error(y_test, y_test_pred))
print(f"测试集上的RMLSE: {rmlse_test}")
```

```


from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_log_error
import numpy as np

# 初始化线性回归模型
model = LinearRegression()

# 训练模型
model.fit(X_train, y_train)

# 在验证集上进行预测
y_val_pred = model.predict(X_val)

# 确保预测值和真实值都是非负的
y_val_pred = np.maximum(y_val_pred, 0)
y_val = np.maximum(y_val, 0)

# 计算验证集上的RMLSE
rmlse_val = np.sqrt(mean_squared_log_error(y_val + 1, y_val_pred + 1))
print(f"验证集上的RMLSE: {rmlse_val}")

# 在测试集上进行预测
y_test_pred = model.predict(X_test)

# 确保预测值和真实值都是非负的
y_test_pred = np.maximum(y_test_pred, 0)
y_test = np.maximum(y_test, 0)

# 计算测试集上的RMLSE
rmlse_test = np.sqrt(mean_squared_log_error(y_test + 1, y_test_pred + 1))
print(f"测试集上的RMLSE: {rmlse_test}")

```
