```

import pandas as pd

# 定义训练集和测试集的文件路径
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/mohs_hardness/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/mohs_hardness/test.csv'

# 加载数据
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 查看数据集的基本信息
print("训练集基本信息：")
print(train_df.info())
print("\n测试集基本信息：")
print(test_df.info())

# 显示前几行数据，了解数据
print("\n训练集头几行：")
print(train_df.head())
print("\n测试集头几行：")
print(test_df.head())
```

```


from sklearn.preprocessing import StandardScaler

# 移除id列
train_df = train_df.drop(columns=['id'])
test_df = test_df.drop(columns=['id'])

# 检查缺失值
print("训练集缺失值：")
print(train_df.isnull().sum())
print("\n测试集缺失值：")
print(test_df.isnull().sum())

# 分离特征和目标变量
X_train = train_df.drop(columns=['Hardness'])
y_train = train_df['Hardness']
X_test = test_df.drop(columns=['Hardness'])
y_test = test_df['Hardness']

# 标准化特征
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 查看处理后的训练集和测试集的前几行
print("\n处理后的训练集前几行：")
print(X_train_scaled[:5])
print("\n处理后的测试集前几行：")
print(X_test_scaled[:5])

```

```


import torch
import torch.nn as nn
import torch.optim as optim

# 将数据转换为PyTorch张量
X_train_tensor = torch.tensor(X_train_scaled, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train.values, dtype=torch.float32).unsqueeze(1)
X_test_tensor = torch.tensor(X_test_scaled, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test.values, dtype=torch.float32).unsqueeze(1)

# 定义神经网络模型
class MineralHardnessPredictor(nn.Module):
    def __init__(self):
        super(MineralHardnessPredictor, self).__init__()
        self.fc1 = nn.Linear(12, 64)
        self.fc2 = nn.Linear(64, 32)
        self.fc3 = nn.Linear(32, 1)
        self.tanh = nn.Tanh()
    
    def forward(self, x):
        x = self.tanh(self.fc1(x))
        x = self.tanh(self.fc2(x))
        x = self.fc3(x)
        return x

# 初始化模型、损失函数和优化器
model = MineralHardnessPredictor()
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

# 检查模型结构
print(model)


```

```



# 定义训练参数
num_epochs = 100
batch_size = 64

# 定义训练循环
for epoch in range(num_epochs):
    # 打乱数据
    permutation = torch.randperm(X_train_tensor.size()[0])
    for i in range(0, X_train_tensor.size()[0], batch_size):
        indices = permutation[i:i+batch_size]
        batch_X, batch_y = X_train_tensor[indices], y_train_tensor[indices]
        
        # 前向传播
        outputs = model(batch_X)
        loss = criterion(outputs, batch_y)
        
        # 反向传播和优化
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
    
    # 打印每个epoch的损失
    if (epoch+1) % 10 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')


```
