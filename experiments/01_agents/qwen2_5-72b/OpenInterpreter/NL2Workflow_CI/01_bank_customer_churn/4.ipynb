```

import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, roc_auc_score
from sklearn.preprocessing import LabelEncoder
import numpy as np

# 加载训练数据
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/train.csv'
train_data = pd.read_csv(train_path)

# 加载测试数据
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/test.csv'
test_data = pd.read_csv(test_path)

# 查看训练数据和测试数据
train_data.head(), test_data.head()
```

```


# 检查缺失值
print("训练数据缺失值：\n", train_data.isnull().sum())
print("测试数据缺失值：\n", test_data.isnull().sum())

# 处理缺失值（如果有）
# 假设没有缺失值，如果有缺失值，可以使用fillna()方法进行填充

# 编码分类变量
label_encoders = {}

for column in train_data.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    le.fit(train_data[column].fillna('NaN'))
    train_data[column] = le.transform(train_data[column].fillna('NaN'))
    test_data[column] = le.transform(test_data[column].fillna('NaN'))
    label_encoders[column] = le

# 查看编码后的数据
train_data.head(), test_data.head()

```

```


# 合并训练集和测试集进行标签编码
combined_data = pd.concat([train_data, test_data], axis=0)

# 编码分类变量
label_encoders = {}

for column in combined_data.select_dypes(include=['object']).columns:
    le = LabelEncoder()
    le.fit(combined_data[column].fillna('NaN'))
    combined_data[column] = le.transform(combined_data[column].fillna('NaN'))
    label_encoders[column] = le

# 将编码后的数据重新分割为训练集和测试集
train_data = combined_data.iloc[:len(train_data)]
test_data = combined_data.iloc[len(train_data):]

# 查看编码后的数据
train_data.head(), test_data.head()
```

```



# 合并训练集和测试集进行标签编码
combined_data = pd.concat([train_data, test_data], axis=0)

# 将所有分类变量转换为字符串类型
for column in combined_data.select_dtypes(include=['object']).columns:
    combined_data[column] = combined_data[column].astype(str)

# 编码分类变量
label_encoders = {}

for column in combined_data.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    le.fit(combined_data[column].fillna('NaN'))
    combined_data[column] = le.transform(combined_data[column].fillna('NaN'))
    label_encoders[column] = le

# 将编码后的数据重新分割为训练集和测试集
train_data = combined_data.iloc[:len(train_data)]
test_data = combined_data.iloc[len(train_data):]

# 查看编码后的数据
train_data.head(), test_data.head()

```

```



# 划分特征和标签
X_train = train_data.drop(columns=['Exited', 'id'])
y_train = train_data['Exited']

X_test = test_data.drop(columns=['Exited', 'id'])
y_test = test_data['Exited']

# 构建随机森林模型
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# 预测测试集
y_pred = rf_model.predict(X_test)
y_pred_proba = rf_model.predict_proba(X_test)[:, 1]

# 保存预测结果
prediction_df = pd.DataFrame({'id': test_data['id'], 'Exited': y_pred})
prediction_df.to_csv('predictions.csv', index=False)

# 计算混淆矩阵
conf_matrix = confusion_matrix(y_test, y_pred)

# 计算AUC-ROC分数
auc_roc = roc_auc_score(y_test, y_pred_proba)

conf_matrix, auc_roc

```
