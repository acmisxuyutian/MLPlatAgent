```

# 导入必要的库
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, roc_auc_score
from sklearn.preprocessing import LabelEncoder
import joblib
```

```


# 读取训练集和测试集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 预览数据集
train_df.head(), test_df.head()
```

```


# 对训练集进行预处理
# 假设我们的目标变量是 'Exited'，标识客户是否会流失
# 首先，我们移除 'id', 'CustomerId', 和 'Surname' 因为这些特征通常是不相关的
train_df_processed = train_df.drop(['id', 'CustomerId', 'Surname'], axis=1)
test_df_processed = test_df.drop(['id', 'CustomerId', 'Surname'], axis=1)

# 对类别特征进行编码
label_encoders = {}

# 查找所有非数值类型的列
for col in train_df_processed.select_dtypes(exclude=['number']).columns:
    label_encoders[col] = LabelEncoder()
    train_df_processed[col] = label_encoders[col].fit_transform(train_df_processed[col])
    test_df_processed[col] = label_encoders[col].transform(test_df_processed[col])

# 分离特征和目标变量
X_train = train_df_processed.drop('Exited', axis=1)
y_train = train_df_processed['Exited']

X_test = test_df_processed.drop('Exited', axis=1)
y_test = test_df_processed['Exited']

# 查看处理后的数据
X_train.head()
```

```



# 构建随机森林模型
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# 对测试集进行预测
y_pred = rf_model.predict(X_test)
y_pred_proba = rf_model.predict_proba(X_test)[:, 1]

# 保存预测结果
joblib.dump(y_pred, 'predictions.pkl')

# 计算混淆矩阵
conf_matrix = confusion_matrix(y_test, y_pred)
conf_matrix
```
