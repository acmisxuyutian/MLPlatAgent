```

import pandas as pd

# Load the train and test datasets
train_file = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/train.csv'
test_file = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/test.csv'

train_data = pd.read_csv(train_file)
test_data = pd.read_csv(test_file)

# Display the first few rows of the train and test datasets
print("Training Data:")
print(train_data.head())
print("\nTesting Data:")
print(test_data.head())
```

```


# Check for missing values in the train and test datasets
print("Missing values in training data:")
print(train_data.isnull().sum())
print("\nMissing values in test data:")
print(test_data.isnull().sum())

# If there are missing values, we can handle them here
# For now, let's assume there are no missing values or they are handled appropriately

# Separate features and target variable
X_train = train_data.drop(columns=['id', 'CustomerId', 'Surname', 'Exited'])
y_train = train_data['Exited']

X_test = test_data.drop(columns=['id', 'CustomerId', 'Surname', 'Exited'])
y_test = test_data['Exited']

# Display the first few rows of the feature and target variables
print("\nFeatures in training data:")
print(X_train.head())
print("\nTarget variable in training data:")
print(y_train.head())
print("\nFeatures in test data:")
print(X_test.head())
print("\nTarget variable in test data:")
print(y_test.head())

```

```


import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, roc_auc_score, accuracy_score

# One-hot encode categorical variables
categorical_features = ['Geography', 'Gender']

# Fit and transform the training data
encoder = OneHotEncoder()
encoded_train = encoder.fit_transform(X_train[categorical_features])

# Convert the encoded data to a DataFrame
encoded_train_df = pd.DataFrame(encoded_train.toarray(), columns=encoder.get_feature_names_out(categorical_features))

# Drop the original categorical features from the training data
X_train = X_train.drop(columns=categorical_features)

# Concatenate the encoded features with the numerical features
X_train = pd.concat([X_train, encoded_train_df], axis=1)

# Transform the test data using the same encoder
encoded_test = encoder.transform(X_test[categorical_features])
encoded_test_df = pd.DataFrame(encoded_test.toarray(), columns=encoder.get_feature_names_out(categorical_features))

# Drop the original categorical features from the test data
X_test = X_test.drop(columns=categorical_features)

# Concatenate the encoded features with the numerical features
X_test = pd.concat([X_test, encoded_test_df], axis=1)

# Display the first few rows of the transformed feature data
print("Transformed Features in training data:")
print(X_train.head())
print("\nTransformed Features in test data:")
print(X_test.head())
```

```



# Train a Random Forest model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Make predictions on the test data
y_pred = model.predict(X_test)
y_pred_proba = model.predict_proba(X_test)[:, 1]

# Display the first few predictions
print("Predicted labels for the test data:")
print(y_pred[:5])
print("\nPredicted probabilities for the positive class (Exited=1):")
print(y_pred_proba[:5])

```

```



from sklearn.metrics import confusion_matrix, roc_auc_score, accuracy_score

# Compute the confusion matrix
conf_matrix = confusion_matrix(y_test, y_pred)

# Compute the AUC-ROC score
auc_roc = roc_auc_score(y_test, y_pred_proba)

# Compute the accuracy
accuracy = accuracy_score(y_test, y_pred)

# Display the results
print("Confusion Matrix:")
print(conf_matrix)
print("\nAUC-ROC Score:")
print(auc_roc)
print("\nAccuracy:")
print(accuracy)
```

```




# Save the predictions to a CSV file
predictions = pd.DataFrame({'id': test_data['id'], 'Exited': y_pred, 'Exited_Probability': y_pred_proba})
predictions.to_csv('predictions.csv', index=False)

# Display the first few rows of the predictions DataFrame
print("First few rows of the predictions DataFrame:")
print(predictions.head())
```
