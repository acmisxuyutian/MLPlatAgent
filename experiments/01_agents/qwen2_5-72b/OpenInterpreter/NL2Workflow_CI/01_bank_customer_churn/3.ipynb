```

import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Load the train and test datasets
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/bank_customer_churn/test.csv'

# Load datasets
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# Display the first few rows of the datasets to inspect the data
train_df.head(), test_df.head()
```

```


# Check for missing values in the train and test datasets
train_missing = train_df.isnull().sum()
test_missing = test_df.isnull().sum()

# Display missing values
train_missing, test_missing

# Drop unnecessary columns
train_df = train_df.drop(columns=['id', 'CustomerId', 'Surname'])
test_df = test_df.drop(columns=['id', 'CustomerId', 'Surname'])

# Encode categorical variables
train_df = pd.get_dummies(train_df, columns=['Geography', 'Gender'], drop_first=True)
test_df = pd.get_dummies(test_df, columns=['Geography', 'Gender'], drop_first=True)

# Display the first few rows of the preprocessed datasets to inspect the data
train_df.head(), test_df.head()

```

```



# Split the data into features (X) and target (y)
X_train = train_df.drop(columns=['Exited'])
y_train = train_df['Exited']

X_test = test_df.drop(columns=['Exited'])
y_test = test_df['Exited']

# Initialize and train the Random Forest model
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# Predict on the test set
y_pred = model.predict(X_test)

# Calculate AUC-ROC score
auc_roc = roc_auc_score(y_test, y_pred)

# Display AUC-ROC score
auc_roc


```

```


# Save the predictions
predictions_df = pd.DataFrame({
    'id': test_df.index,
    'Exited': y_pred
})
predictions_df.to_csv('predictions.csv', index=False)

# Display the confusion matrix
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model.classes_)
disp.plot(cmap='Blues')
plt.show()
```
