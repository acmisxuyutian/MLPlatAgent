```

import pandas as pd

# 定义训练集和测试集的路径
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/test.csv'

# 加载训练集
train_df = pd.read_csv(train_path)
# 加载测试集
test_df = pd.read_csv(test_path)

# 查看前几行数据，确认加载成功
train_df.head()
test_df.head()
```

```


# 对训练数据进行独热编码
train_df_encoded = pd.get_dummies(train_df, columns=['Sex'])

# 对测试数据进行独热编码
test_df_encoded = pd.get_dummies(test_df, columns=['Sex'])

# 检查编码后的前几行数据
train_df_encoded.head()
test_df_encoded.head()
```

```


# 准备特征与标签
X_train = train_df_encoded.drop(columns=['id', 'Age'])
y_train = train_df_encoded['Age']

# 测试集中可能不存在某些类别，因此需要创建与训练集相同的特征
X_test = test_df_encoded.reindex(columns=X_train.columns, fill_value=0)

# 显示处理后的部分数据以便确认
X_train.head(), X_test.head()

```

```



from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

# 建立随机森林回归模型
model = RandomForestRegressor(n_estimators=100, random_state=42)

# 训练模型
model.fit(X_train, y_train)

# 预测测试集
y_pred = model.predict(X_test)

# 评估模型的MAE
mae = mean_absolute_error(y_true=test_df['Age'], y_pred=y_pred)

mae, y_pred

```

```



import numpy as np

# 创建包含id和预测结果的DataFrame
predictions_df = pd.DataFrame({
    'id': test_df['id'],
    'predicted_age': y_pred
})

# 保存预测结果到CSV文件
predictions_df.to_csv('crab_age_predictions.csv', index=False)

# 确认文件已成功创建
predictions_df.head()


```
