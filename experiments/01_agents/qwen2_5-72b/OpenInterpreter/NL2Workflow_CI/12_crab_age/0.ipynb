```

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

# 加载训练数据和测试数据
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 查看数据结构
train_df.head()
```

```


# 对类别特征进行独热编码
categorical_features = ['Sex']
one_hot_encoder = OneHotEncoder(sparse=False, drop='first')

# 对训练数据进行独热编码
train_encoded = one_hot_encoder.fit_transform(train_df[categorical_features])
train_encoded_df = pd.DataFrame(train_encoded, columns=one_hot_encoder.get_feature_names_out(categorical_features))

# 对测试数据进行独热编码
test_encoded = one_hot_encoder.transform(test_df[categorical_features])
test_encoded_df = pd.DataFrame(test_encoded, columns=one_hot_encoder.get_feature_names_out(categorical_features))

# 将编码后的特征与原始数据合并
train_df = pd.concat([train_df.drop(columns=categorical_features), train_encoded_df], axis=1)
test_df = pd.concat([test_df.drop(columns=categorical_features), test_encoded_df], axis=1)

# 查看编码后的训练数据结构
train_df.head()
```

```



# 对类别特征进行独热编码
categorical_features = ['Sex']
one_hot_encoder = OneHotEncoder(drop='first')

# 对训练数据进行独热编码
train_encoded = one_hot_encoder.fit_transform(train_df[categorical_features])

# 获取编码后的特征名称
train_encoded_feature_names = one_hot_encoder.get_feature_names_out(categorical_features)

# 将编码后的特征转换为密集矩阵，然后再转换为DataFrame
train_encoded_df = pd.DataFrame(train_encoded.toarray(), columns=train_encoded_feature_names)

# 对测试数据进行独热编码
test_encoded = one_hot_encoder.transform(test_df[categorical_features])
test_encoded_df = pd.DataFrame(test_encoded.toarray(), columns=train_encoded_feature_names)

# 将编码后的特征与原始数据合并
train_df = pd.concat([train_df.drop(columns=categorical_features), train_encoded_df], axis=1)
test_df = pd.concat([test_df.drop(columns=categorical_features), test_encoded_df], axis=1)

# 查看编码后的训练数据结构
train_df.head()

```

```



# 分离特征和目标变量
X_train = train_df.drop(columns=['id', 'Age'])
y_train = train_df['Age']

X_test = test_df.drop(columns=['id', 'Age'])
y_test = test_df['Age']

# 查看特征和目标变量的形状
X_train.shape, y_train.shape, X_test.shape, y_test.shape

```

```



# 建立随机森林模型
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)

# 训练模型
rf_model.fit(X_train, y_train)

# 在测试集上进行预测
y_pred = rf_model.predict(X_test)

# 计算MAE
mae = mean_absolute_error(y_test, y_pred)
mae


```

```




# 创建一个包含 id 和预测年龄的 DataFrame
predictions_df = pd.DataFrame({'id': test_df['id'], 'Age': y_pred})

# 保存预测结果到 CSV 文件
predictions_df.to_csv('crab_age_predictions.csv', index=False)

# 查看预测结果的前几行
predictions_df.head()


```
