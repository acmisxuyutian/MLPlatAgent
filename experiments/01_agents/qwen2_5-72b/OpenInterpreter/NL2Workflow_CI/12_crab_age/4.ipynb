```

import pandas as pd

# 训练数据集路径
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/train.csv'
# 测试数据集路径
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/test.csv'

# 读取数据
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 显示训练集和测试集的前几行
train_df.head(), test_df.head()
```

```


# 显示训练集和测试集的基本信息
train_df.info()
test_df.info()

# 查看训练集的统计信息
train_df.describe()

# 查看我们需要进行独热编码的列
train_df['Sex'].value_counts()

```

```


from sklearn.preprocessing import OneHotEncoder

# 初始化 OneHotEncoder
encoder = OneHotEncoder()

# 为训练集生成独热编码
train_sex_encoded = encoder.fit_transform(train_df[['Sex']]).toarray()
train_sex_encoded = pd.DataFrame(train_sex_encoded, columns=encoder.get_feature_names(['Sex']))

# 保持特征集的完整性
train_df = train_df.drop('Sex', axis=1).join(train_sex_encoded)

# 为测试集生成独热编码
test_sex_encoded = encoder.transform(test_df[['Sex']]).toarray()
test_sex_encoded = pd.DataFrame(test_sex_encoded, columns=encoder.get_feature_names(['Sex']))

# 保持特征集的完整性
test_df = test_df.drop('Sex', axis=1).join(test_sex_encoded)

train_df.head(), test_df.head()
```

```


# 为了兼容较新版本的scikit-learn
train_sex_encoded = encoder.fit_transform(train_df[['Sex']]).toarray()
train_sex_encoded = pd.DataFrame(train_sex_encoded, columns=encoder.get_feature_names_out(['Sex']))

# 保持训练集特征集的完整性
train_df = train_df.drop('Sex', axis=1).join(train_sex_encoded)

# 为测试集生成独热编码
test_sex_encoded = encoder.transform(test_df[['Sex']]).toarray()
test_sex_encoded = pd.DataFrame(test_sex_encoded, columns=encoder.get_feature_names_out(['Sex']))

# 保持测试集特征集的完整性
test_df = test_df.drop('Sex', axis=1).join(test_sex_encoded)

# 检查新的数据集
train_df.head(), test_df.head()

```

```


from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

# 定义特征（X）和目标变量（y）
X_train = train_df.drop(['id', 'Age'], axis=1)
y_train = train_df['Age']

X_test = test_df.drop(['id', 'Age'], axis=1)
y_test = test_df['Age']

# 定义并训练随机森林模型
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)

rf_model.fit(X_train, y_train)

# 在测试集上评估模型
y_pred = rf_model.predict(X_test)

# 计算测试集上的MAE
mae = mean_absolute_error(y_test, y_pred)
mae

```

```


# 为预测创建一个 data frame
predictions_df = test_df[['id']].copy()
predictions_df['Age'] = y_pred

# 保存预测结果到CSV文件
predictions_df.to_csv('crab_age_predictions.csv', index=False)
print('预测结果已保存至文件 "crab_age_predictions.csv" 。')
```
