```

import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

# 加载处理后的数据
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/train.csv'
train_data = pd.read_csv(train_path)

test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/test.csv'
test_data = pd.read_csv(test_path)

# 独热编码
encoder = OneHotEncoder(sparse_output=False, drop='first')

# 对训练数据进行编码
sex_encoded_train = encoder.fit_transform(train_data[['Sex']])
sex_df_train = pd.DataFrame(sex_encoded_train, columns=encoder.get_feature_names_out(['Sex']))

# 对测试数据进行编码
sex_encoded_test = encoder.transform(test_data[['Sex']])
sex_df_test = pd.DataFrame(sex_encoded_test, columns=encoder.get_feature_names_out(['Sex']))

# 合并编码后的数据和原始数据
train_data = pd.concat([train_data.drop(columns=['Sex', 'id']), sex_df_train], axis=1)
test_data = pd.concat([test_data.drop(columns=['Sex', 'id']), sex_df_test], axis=1)

# 分离特征和目标变量
X_train = train_data.drop(columns=['Age'])
y_train = train_data['Age']
X_test = test_data.drop(columns=['Age'])
y_test = test_data['Age']

# 训练随机森林模型
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 预测
y_pred = model.predict(X_test)

# 计算MAE
mae = mean_absolute_error(y_test, y_pred)
print(f"模型在测试集上的平均绝对误差（MAE）: {mae:.2f}")
```
