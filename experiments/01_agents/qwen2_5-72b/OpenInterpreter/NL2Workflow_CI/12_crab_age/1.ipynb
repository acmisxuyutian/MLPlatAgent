```

import pandas as pd

# 加载训练数据
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/train.csv'
train_df = pd.read_csv(train_data_path)

# 查看训练数据的前几行
train_df.head()
```

```


# 对性别列进行独热编码
train_df_encoded = pd.get_dummies(train_df, columns=['Sex'])

# 查看独热编码后的训练数据的前几行
train_df_encoded.head()

```

```



# 分离特征和标签
X_train = train_df_encoded.drop(columns=['id', 'Age'])
y_train = train_df_encoded['Age']

# 查看特征和标签的前几行
X_train.head(), y_train.head()


```

```



from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

# 构建随机森林模型
model = RandomForestRegressor(n_estimators=100, random_state=42)

# 训练模型
model.fit(X_train, y_train)

# 模型训练完成后，我们加载测试数据
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crab_age/test.csv'
test_df = pd.read_csv(test_data_path)

# 查看测试数据的前几行
test_df.head()


```

```




# 对测试数据进行独热编码
test_df_encoded = pd.get_dummies(test_df, columns=['Sex'])

# 查看独热编码后的测试数据的前几行
test_df_encoded.head()


```

```




# 分离特征和标签
X_test = test_df_encoded.drop(columns=['id', 'Age'])
y_test = test_df_encoded['Age']

# 使用训练好的模型进行预测
y_pred = model.predict(X_test)

# 计算平均绝对误差 (MAE)
mae = mean_absolute_error(y_test, y_pred)
print(f'Mean Absolute Error (MAE): {mae:.2f}')

# 保存预测结果
predictions_df = pd.DataFrame({'id': test_df_encoded['id'], 'Predicted_Age': y_pred})
predictions_df.to_csv('crab_age_predictions.csv', index=False)

# 查看预测结果的前几行
predictions_df.head()



```
