```

import pandas as pd

# 文件路径
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/monster/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/monster/test.csv'

# 加载数据
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 查看数据基本信息
print("训练集数据预览:")
print(train_df.head())
print("\n测试集数据预览:")
print(test_df.head())
print("\n训练集的基本信息:")
train_df.info()
print("\n测试集的基本信息:")
test_df.info()
```

```

from sklearn.preprocessing import OrdinalEncoder

# 预处理: 对颜色特征进行编码
ordinal_encoder = OrdinalEncoder()

# 拟合与转换训练集的颜色列
train_df['color'] = ordinal_encoder.fit_transform(train_df[['color']])

# 只转换测试集的颜色列
test_df['color'] = ordinal_encoder.transform(test_df[['color']])

# 查看前几行以确认编码是否正确
print('训练集颜色列编码后预览:')
print(train_df.head())

print('\n测试集颜色列编码后预览:')
print(test_df.head())
```

```


# 移除 id 列
train_features = train_df.drop(columns=['id', 'type'])
train_labels = train_df['type']
test_features = test_df.drop(columns=['id', 'type'])
test_labels = test_df['type']

# 查看划分后的数据
print('训练特征预览:')
print(train_features.head())
print('\n训练标签预览:')
print(train_labels.head())
print('\n测试特征预览:')
print(test_features.head())
print('\n测试标签预览:')
print(test_labels.head())

# 导入必要的库
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# 选择并训练模型: 随机森林分类器
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(train_features, train_labels)

# 进行预测
predictions = model.predict(test_features)

# 评估模型
accuracy = accuracy_score(test_labels, predictions)
print(f'\n模型在测试集上的准确率: {accuracy:.2f}')

# 保存预测结果
predictions_df = pd.DataFrame({'id': test_df['id'], 'predicted_type': predictions})
predictions_df.to_csv('monster_predictions.csv', index=False)
print('\n预测结果已保存至 monster_predictions.csv')
```
