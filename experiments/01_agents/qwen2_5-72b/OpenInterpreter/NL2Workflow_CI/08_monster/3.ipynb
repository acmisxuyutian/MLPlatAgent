```

import pandas as pd
from sklearn.model_selection import train_test_split

# Load the datasets
train_file = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/monster/train.csv'
test_file = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/monster/test.csv'

# Read the CSV files
train_data = pd.read_csv(train_file)
test_data = pd.read_csv(test_file)

# Display the first few rows of the datasets
train_data.head(), test_data.head()
```

```


# Check for missing values
train_data.info(), test_data.info()

```

```


# Explore the distribution of the target variable in the training set
train_data['type'].value_counts(), test_data['type'].value_counts()

```

```


from sklearn.preprocessing import LabelEncoder

# Encode the 'color' column
label_encoder = LabelEncoder()
train_data['color'] = label_encoder.fit_transform(train_data['color'])
test_data['color'] = label_encoder.transform(test_data['color'])

# Display the first few rows of the datasets after encoding
train_data.head(), test_data.head()

```

```



# Separate features and target variable
X_train = train_data[['bone_length', 'rotting_flesh', 'hair_length', 'color']]
y_train = train_data['type']

X_test = test_data[['bone_length', 'rotting_flesh', 'hair_length', 'color']]
y_test = test_data['type']

# Display the shape of the feature and target variables
X_train.shape, y_train.shape, X_test.shape, y_test.shape


```

```



from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Initialize the Random Forest Classifier
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)

# Train the model
rf_classifier.fit(X_train, y_train)

# Make predictions on the test set
y_pred = rf_classifier.predict(X_test)

# Calculate the accuracy
accuracy = accuracy_score(y_test, y_pred)
accuracy



```

```




# Save the predictions to a CSV file
predictions = pd.DataFrame({'id': test_data['id'], 'predicted_type': y_pred})
predictions['predicted_type'] = label_encoder.inverse_transform(predictions['predicted_type'])
predictions.to_csv('monster_predictions.csv', index=False)

# Report the accuracy
accuracy




```
