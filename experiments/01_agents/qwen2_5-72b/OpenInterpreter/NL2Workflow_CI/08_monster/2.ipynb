```

import pandas as pd

# Load the training and testing datasets
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/monster/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/monster/test.csv'

train_data = pd.read_csv(train_data_path)
test_data = pd.read_csv(test_data_path)

# Display the first few rows of the datasets
train_data.head(), test_data.head()
```

```


# Import necessary libraries for preprocessing
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

# Encode categorical variables
label_encoder = LabelEncoder()
train_data['color'] = label_encoder.fit_transform(train_data['color'])
test_data['color'] = label_encoder.transform(test_data['color'])

# Separate features and target for the training set
X_train = train_data.drop(columns=['id', 'type'])
y_train = train_data['type']

# Separate features for the test set
X_test = test_data.drop(columns=['id', 'type'])

# Display the preprocessed data
X_train.head(), X_test.head(), y_train.head()
```

```


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Initialize the Random Forest classifier
rf_clf = RandomForestClassifier(n_estimators=100, random_state=42)

# Train the model
rf_clf.fit(X_train, y_train)

# Predict on the test set
y_pred = rf_clf.predict(X_test)

# Since we don't have the actual labels for the test set, we can't compute the accuracy directly.
# However, we can save the predictions and report the accuracy on a validation set if available.

# Save predictions to a CSV file
predictions = test_data[['id']].copy()
predictions['predicted_type'] = y_pred

# Save the predictions
predictions.to_csv('monster_predictions.csv', index=False)

# Display the first few rows of the predictions
predictions.head()
```

```

# Load the true labels for the test set (if available)
test_labels_path = r'path_to_test_labels.csv'
test_labels = pd.read_csv(test_labels_path)

# Compute accuracy
accuracy = accuracy_score(test_labels['type'], predictions['predicted_type'])
print(f'Test set accuracy: {accuracy:.4f}')
```
