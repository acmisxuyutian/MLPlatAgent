```

import pandas as pd
import numpy as np

# 加载训练集和测试集
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/kidney_stone/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/kidney_stone/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# 检查数据集基本信息
print("训练集基本信息:")
print(train_df.info())
print("\n训练集前几行:")
print(train_df.head())

print("\n测试集基本信息:")
print(test_df.info())
print("\n测试集前几行:")
print(test_df.head())

# 检查缺失值
print("\n训练集缺失值情况:")
print(train_df.isnull().sum())
print("\n测试集缺失值情况:")
print(test_df.isnull().sum())
```

```


from sklearn.model_selection import train_test_split

# 移除id列
train_df = train_df.drop(columns=['id'])
test_df = test_df.drop(columns=['id'])

# 分离特征和标签
X_train = train_df.drop(columns=['target'])
y_train = train_df['target']

X_test = test_df.drop(columns=['target'])
y_test = test_df['target']

# 划分训练集和验证集
X_train_split, X_val, y_train_split, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# 打印数据集的形状
print(f"训练集特征形状: {X_train_split.shape}")
print(f"训练集标签形状: {y_train_split.shape}")
print(f"验证集特征形状: {X_val.shape}")
print(f"验证集标签形状: {y_val.shape}")
print(f"测试集特征形状: {X_test.shape}")
print(f"测试集标签形状: {y_test.shape}")

```

```


from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

# 初始化逻辑回归模型
model = LogisticRegression(max_iter=1000, random_state=42)

# 训练模型
model.fit(X_train_split, y_train_split)

# 预测验证集概率
y_val_probs = model.predict_proba(X_val)[:, 1]

# 计算验证集的AUC-ROC
val_auc_roc = roc_auc_score(y_val, y_val_probs)
print(f"验证集的AUC-ROC: {val_auc_roc:.4f}")

# 预测测试集概率
y_test_probs = model.predict_proba(X_test)[:, 1]

# 计算测试集的AUC-ROC
test_auc_roc = roc_auc_score(y_test, y_test_probs)
print(f"测试集的AUC-ROC: {test_auc_roc:.4f}")

```
