```

import pandas as pd

# Load the datasets
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/kidney_stone/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/kidney_stone/test.csv'

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display dataframes' first few rows
train_df.head(), test_df.head()
```

```


# Checking for missing values
train_df.isnull().sum(), test_df.isnull().sum()

# Exploring basic statistics
train_df.describe(), test_df.describe()

```

```



# Drop the id column as it is not a predictive feature
train_df = train_df.drop('id', axis=1)
test_df = test_df.drop('id', axis=1)

# Display the first few rows of the data without the id column
train_df.head(), test_df.head()


```

```


from sklearn.preprocessing import StandardScaler

# Initialize the StandardScaler
scaler = StandardScaler()

# fit the scaler on the training data (excluding the target column)
scaler.fit(train_df.drop('target', axis=1))

# transform the training and test data
X_train_scaled = scaler.transform(train_df.drop('target', axis=1))
X_test_scaled = scaler.transform(test_df.drop('target', axis=1))

# Prepare the target variables
y_train = train_df['target']
y_test = test_df['target']

# Convert the numpy arrays back to DataFrame to keep the column names (for readability)
X_train_scaled = pd.DataFrame(X_train_scaled, columns=train_df.columns[:-1])
X_test_scaled = pd.DataFrame(X_test_scaled, columns=train_df.columns[:-1])

# Display the first few rows of the scaled training data
X_train_scaled.head(), X_test_scaled.head()

```

```


from sklearn.model_selection import train_test_split

# Split the training data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_train_scaled, y_train, test_size=0.2, random_state=42)

# Display the shapes of the training, validation, and test sets
X_train.shape, y_train.shape, X_val.shape, y_val.shape, X_test_scaled.shape, y_test.shape

```

```



from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

# Initialize the Logistic Regression model
model = LogisticRegression()

# Train the model on the training data
model.fit(X_train, y_train)

# Predict probabilities on the validation set
y_val_pred_proba = model.predict_proba(X_val)[:, 1]

# Calculate the AUC-ROC score on the validation set
val_auc_roc = roc_auc_score(y_val, y_val_pred_proba)

val_auc_roc


```

```


# Predict probabilities on the test set
y_test_pred_proba = model.predict_proba(X_test_scaled)[:, 1]

# Calculate the AUC-ROC score on the test set
test_auc_roc = roc_auc_score(y_test, y_test_pred_proba)

test_auc_roc

```
