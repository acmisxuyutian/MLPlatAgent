```

import pandas as pd

# 定义文件路径
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/kidney_stone/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/kidney_stone/test.csv'

# 加载数据集
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 查看数据集的前几行
train_df.head(), test_df.head()
```

```


# Check for missing values in the training and testing datasets
train_df.isnull().sum(), test_df.isnull().sum()

```

```


from sklearn.model_selection import train_test_split

# Drop the 'id' column as it is not needed for modeling
train_df = train_df.drop(columns=['id'])
test_df = test_df.drop(columns=['id'])

# Separate features and target variable
X_train = train_df.drop(columns=['target'])
y_train = train_df['target']
X_test = test_df.drop(columns=['target'])
y_test = test_df['target']

# Split the training data into training and validation sets
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# Check the shapes of the resulting datasets
X_train.shape, X_val.shape, y_train.shape, y_val.shape, X_test.shape, y_test.shape

```

```



from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

# Initialize the logistic regression model
model = LogisticRegression(random_state=42)

# Train the model
model.fit(X_train, y_train)

# Predict probabilities on the validation set
y_val_pred_probs = model.predict_proba(X_val)[:, 1]

# Calculate the AUC-ROC score on the validation set
val_auc_roc = roc_auc_score(y_val, y_val_pred_probs)
val_auc_roc


```

```


from sklearn.preprocessing import StandardScaler

# Scale the data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

# Initialize the logistic regression model with increased max_iter
model = LogisticRegression(max_iter=1000, random_state=42)

# Train the model
model.fit(X_train_scaled, y_train)

# Predict probabilities on the validation set
y_val_pred_probs = model.predict_proba(X_val_scaled)[:, 1]

# Calculate the AUC-ROC score on the validation set
val_auc_roc = roc_auc_score(y_val, y_val_pred_probs)
val_auc_roc


```

```



# Predict probabilities on the test set
y_test_pred_probs = model.predict_proba(X_test_scaled)[:, 1]

# Calculate the AUC-ROC score on the test set
test_auc_roc = roc_auc_score(y_test, y_test_pred_probs)
test_auc_roc


```
