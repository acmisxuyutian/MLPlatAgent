```

import pandas as pd

# 定义数据文件路径
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/kidney_stone/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/kidney_stone/test.csv'

# 加载数据
train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# 检查数据的前几行
train_df.head(), test_df.head()
```

```


# 检查数据的缺失值情况
train_df.info(), test_df.info()

```

```


# 删除 'id' 列
train_df = train_df.drop(columns=['id'])
test_df = test_df.drop(columns=['id'])

# 查看数据的基本统计信息
train_df.describe(), test_df.describe()

```

```


from sklearn.model_selection import train_test_split

# 选择特征和目标变量
X_train = train_df.drop(columns=['target'])
y_train = train_df['target']
X_test = test_df.drop(columns=['target'])
y_test = test_df['target']

# 将训练集进一步划分为训练集和验证集
X_train_split, X_val, y_train_split, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# 检查分割后的数据集大小
X_train_split.shape, X_val.shape, X_test.shape
```

```



from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

# 初始化随机森林分类器
rf_classifier = RandomForestClassifier(random_state=42)

# 训练模型
rf_classifier.fit(X_train_split, y_train_split)

# 在验证集上进行预测
y_val_pred = rf_classifier.predict_proba(X_val)[:, 1]

# 计算验证集上的AUC-ROC分数
val_auc_roc = roc_auc_score(y_val, y_val_pred)

# 输出验证集上的AUC-ROC分数
val_auc_roc
```

```



# 在测试集上进行预测
y_test_pred = rf_classifier.predict_proba(X_test)[:, 1]

# 计算测试集上的AUC-ROC分数
test_auc_roc = roc_auc_score(y_test, y_test_pred)

# 输出测试集上的AUC-ROC分数
test_auc_roc

```
