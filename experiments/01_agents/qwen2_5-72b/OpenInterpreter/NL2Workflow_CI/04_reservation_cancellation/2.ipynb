```

import pandas as pd
import lightgbm as lgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score
import os
```

```


# Load the training and testing datasets
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/reservation_cancellation/train.csv'
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/reservation_cancellation/test.csv'

# Check the file paths to ensure files exist
print(f"Training data file exists: {os.path.exists(train_data_path)}")
print(f"Testing data file exists: {os.path.exists(test_data_path)}")

train_df = pd.read_csv(train_data_path)
test_df = pd.read_csv(test_data_path)

# Display the first few rows of the datasets
print("Sample from the training set:")
print(train_df.head())
print("\nSample from the testing set:")
print(test_df.head())

```

```


# Check for missing values in the training and testing datasets
print("Missing values in the training set:")
print(train_df.isnull().sum())
print("\nMissing values in the testing set:")
print(test_df.isnull().sum())

# Display the data types of the columns
print("\nData types of the columns in the training set:")
print(train_df.dtypes)
print("\nData types of the columns in the testing set:")
print(test_df.dypes)

```

```


# Check for missing values in the training and testing datasets
print("Missing values in the training set:")
print(train_df.isnull().sum())
print("\nMissing values in the testing set:")
print(test_df.isnull().sum())

# Display the data types of the columns
print("\nData types of the columns in the training set:")
print(train_df.dtypes)
print("\nData types of the columns in the testing set:")
print(test_df.dtypes)

```

```



# Split the datasets into features (X) and target (y)
X_train = train_df.drop(columns=['id', 'booking_status'])
y_train = train_df['booking_status']

X_test = test_df.drop(columns=['id', 'booking_status'])
y_test = test_df['booking_status']

# Display the shapes of the datasets
print(f"X_train shape: {X_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_test shape: {y_test.shape}")


```

```



# Train the LightGBM model
lgb_train = lgb.Dataset(X_train, y_train)

# Define the parameters for the LightGBM model
params = {
    'objective': 'binary',
    'metric': 'auc',
    'boosting_type': 'gbdt',
    'num_leaves': 31,
    'learning_rate': 0.05,
    'feature_fraction': 0.9,
    'bagging_fraction': 0.8,
    'bagging_freq': 5,
    'verbose': 0
}

# Train the model
model = lgb.train(params, lgb_train, num_boost_round=100)

# Save the model
model.save_model('lightgbm_model.txt')

# Make predictions on the test set
y_pred = model.predict(X_test)

# Calculate the AUC-ROC score
auc_roc = roc_auc_score(y_test, y_pred)
print(f"AUC-ROC on the test set: {auc_roc:.4f}")

# Save the predictions to a CSV file
predictions_df = pd.DataFrame({'id': test_df['id'], 'predicted_booking_status': y_pred})
predictions_df.to_csv('predictions.csv', index=False)

print("Predictions saved to predictions.csv")
```
