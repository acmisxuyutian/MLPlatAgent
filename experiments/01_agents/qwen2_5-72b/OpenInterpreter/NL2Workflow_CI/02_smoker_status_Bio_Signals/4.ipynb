```

import pandas as pd

# 设置数据集路径
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/smoker_status_Bio_Signals/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/smoker_status_Bio_Signals/test.csv'

# 加载数据
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 检视数据前几行
train_df.head(), test_df.head()
```

```


# 检查缺失值
train_df.isnull().sum(), test_df.isnull().sum()

```

```


# 删除不需要的列（如id）
train_df = train_df.drop(columns=['id'])
test_df = test_df.drop(columns=['id'])

# 分离特征和目标变量
X_train = train_df.drop(columns=['smoking'])
y_train = train_df['smoking']

X_test = test_df.drop(columns=['smoking'])
y_test = test_df['smoking']

# 检查特征和目标变量的形状
X_train.shape, y_train.shape, X_test.shape, y_test.shape

```

```


from sklearn.preprocessing import StandardScaler
import xgboost as xgb
from sklearn.metrics import roc_auc_score

# 特征缩放
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 创建 DART 模型
xgb_classifier = xgb.XGBClassifier(booster='dart', learning_rate=0.1, n_estimators=100, random_state=42)

# 训练模型
xgb_classifier.fit(X_train_scaled, y_train)

# 在测试集上进行预测
y_pred_proba = xgb_classifier.predict_proba(X_test_scaled)[:, 1]

# 计算 AUC-ROC
auc_roc = roc_auc_score(y_test, y_pred_proba)
auc_roc


```
