```

import pandas as pd

# 加载训练集和测试集
train_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/smoker_status_Bio_Signals/train.csv'
test_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/smoker_status_Bio_Signals/test.csv'

train_df = pd.read_csv(train_file_path)
test_df = pd.read_csv(test_file_path)

# 查看训练集的前几行
train_df.head()
```

```


# 检查训练集和测试集中是否存在缺失值
print("训练集缺失值情况：")
print(train_df.isnull().sum())

print("\n测试集缺失值情况：")
print(test_df.isnull().sum())

# 如果存在缺失值，可以进行填充或删除
# 例如，使用中位数填充缺失值
train_df.fillna(train_df.median(), inplace=True)
test_df.fillna(test_df.median(), inplace=True)

# 重新检查缺失值
print("\n处理后的训练集缺失值情况：")
print(train_df.isnull().sum())

print("\n处理后的测试集缺失值情况：")
print(test_df.isnull().sum())

```

```


# 选择特征列
features = [col for col in train_df.columns if col not in ['id', 'smoking']]
target = 'smoking'

# 分割训练集和测试集的数据
X_train = train_df[features]
y_train = train_df[target]
X_test = test_df[features]
y_test = test_df[target]

# 检查数据形状
print("训练集特征形状：", X_train.shape)
print("训练集目标形状：", y_train.shape)
print("测试集特征形状：", X_test.shape)
print("测试集目标形状：", y_test.shape)

```

```


import xgboost as xgb
from sklearn.metrics import roc_auc_score

# 构建XGBoost DART模型
params = {
    'booster': 'dart',
    'objective': 'binary:logistic',
    'learning_rate': 0.1,
    'n_estimators': 100,  # 模型的树的数量
    'subsample': 0.8,
    'colsample_bytree': 0.8,
    'reg_alpha': 0.1,
    'reg_lambda': 0.1,
    'gamma': 0.1,
    'min_child_weight': 1,
    'max_depth': 6,
    'sample_type': 'uniform',
    'normalize_type': 'tree',
    'rate_drop': 0.1,
    'skip_drop': 0.5,
    'seed': 42
}

# 训练模型
model = xgb.XGBClassifier(**params)
model.fit(X_train, y_train)

# 预测测试集
y_pred_proba = model.predict_proba(X_test)[:, 1]

# 计算AUC-ROC
auc_roc = roc_auc_score(y_test, y_pred_proba)
print(f"AUC-ROC: {auc_roc:.4f}")


```
