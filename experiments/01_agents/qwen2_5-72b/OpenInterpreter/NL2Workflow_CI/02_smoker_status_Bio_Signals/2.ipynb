```

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score
from xgboost import XGBClassifier

# Load the datasets
train_df = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/smoker_status_Bio_Signals/train.csv')
test_df = pd.read_csv(r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/smoker_status_Bio_Signals/test.csv')

# Display the first few rows of both datasets to understand the structure
train_df.head(), test_df.head()
```

```


# Check for missing values in the training and test datasets
train_missing = train_df.isnull().sum()
test_missing = test_df.isnull().sum()

train_missing, test_missing

```

```


# Separate the features and the target variable
X_train = train_df.drop(columns=['id', 'smoking'])
y_train = train_df['smoking']
X_test = test_df.drop(columns=['id', 'smoking'])
y_test = test_df['smoking']

# Standardize the features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Convert the scaled data back to DataFrames to keep the column names
X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_scaled_df = pd.DataFrame(X_test_scaled, columns=X_test.columns)

# Display the first few rows of the scaled training data
X_train_scaled_df.head()
```

```



# Define the XGBoost model with DART booster and learning rate 0.1
xgb_model = XGBClassifier(booster='dart', learning_rate=0.1, use_label_encoder=False, eval_metric='auc')

# Train the model
xgb_model.fit(X_train_scaled_df, y_train)

# Predict probabilities on the test set
y_pred_proba = xgb_model.predict_proba(X_test_scaled_df)[:, 1]

# Calculate the AUC-ROC score
auc_roc = roc_auc_score(y_test, y_pred_proba)
auc_roc

```
