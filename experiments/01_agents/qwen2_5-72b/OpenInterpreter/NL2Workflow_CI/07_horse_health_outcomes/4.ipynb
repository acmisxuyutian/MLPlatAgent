```

import pandas as pd

# 定义训练数据集和测试数据集的路径
train_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/train.csv'
test_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/test.csv'

# 加载数据
train_df = pd.read_csv(train_file_path)
test_df = pd.read_csv(test_file_path)

# 查看训练数据的前5行
train_df.head()
```

```

# 查看测试数据的前5行
test_df.head()
```

```


# 检查训练数据集的缺失值
train_df.isnull().sum()
```

```


# 处理训练数据集中的缺失值
train_df['peripheral_pulse'].fillna(train_df['peripheral_pulse'].mode()[0], inplace=True)
train_df['mucous_membrane'].fillna(train_df['mucous_membrane'].mode()[0], inplace=True)
train_df['capillary_refill_time'].fillna(train_df['capillary_refill_time'].mode()[0], inplace=True)

# 检查训练数据集是否还有缺失值
train_df.isnull().sum()

```

```



# 检查测试数据集的缺失值
test_df.isnull().sum()

```

```



# 处理测试数据集中的缺失值
test_df['peripheral_pulse'].fillna(test_df['peripheral_pulse'].mode()[0], inplace=True)
test_df['mucous_membrane'].fillna(test_df['mucous_membrane'].mode()[0], inplace=True)
test_df['capillary_refill_time'].fillna(test_df['capillary_refill_time'].mode()[0], inplace=True)

# 检查测试数据集是否还有缺失值
test_df.isnull().sum()


```

```



import pandas as pd

# 删除无关列
train_df.drop(columns=['hospital_number'], inplace=True)
test_df.drop(columns=['hospital_number'], inplace=True)

# 对分类变量进行 one-hot 编码
train_df = pd.get_dummies(train_df, columns=['surgery', 'peripheral_pulse', 'mucous_membrane', 'capillary_refill_time'])
test_df = pd.get_dummies(test_df, columns=['surgery', 'peripheral_pulse', 'mucous_membrane', 'capillary_refill_time'])

# 检查编码后的训练数据集
train_df.head()
```

```




# 检查编码后的测试数据集
test_df.head()

```

```


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score
from sklearn.model_selection import train_test_split

# 将数据集拆分为特征和目标变量
X_train = train_df.drop(columns=['outcome'])
y_train = train_df['outcome']
X_test = test_df.drop(columns=['outcome'])
y_test = test_df['outcome']

# 训练随机森林分类器
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)

# 预测测试集
y_pred = rf.predict(X_test)

# 计算F1分数
f1 = f1_score(y_test, y_pred, average='weighted')

# 输出F1分数
print(f'F1 Score: {f1:.4f}')
```

```



# 从原始测试数据集中获取hospital_number列
hospital_number = pd.read_csv(test_file_path)['hospital_number']

# 创建包含预测结果的DataFrame
predictions_df = pd.DataFrame({
    'hospital_number': hospital_number,
    'predicted_outcome': y_pred
})

# 保存预测结果到CSV文件
predictions_df.to_csv('predictions.csv', index=False)

# 确认文件已保存
print("预测结果已保存到 'predictions.csv'")
```
