```

import pandas as pd

# 加载训练数据
train_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/train.csv'
train_df = pd.read_csv(train_data_path)

# 加载测试数据
test_data_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/test.csv'
test_df = pd.read_csv(test_data_path)

# 查看训练数据的前几行
train_df.head()
```

```


# 查看训练数据的基本统计信息
train_df.info()
test_df.info()

# 查看训练数据的描述性统计信息
train_df.describe(include='all')
```

```


# 查看训练数据的基本统计信息
train_df.info()
test_df.info()

# 查看训练数据的描述性统计信息
train_df.describe(include='all')
```

```


# 处理训练数据中的缺失值
train_df['peripheral_pulse'].fillna('missing', inplace=True)
train_df['mucous_membrane'].fillna('missing', inplace=True)
train_df['capillary_refill_time'].fillna('missing', inplace=True)

# 处理测试数据中的缺失值
test_df['peripheral_pulse'].fillna('missing', inplace=True)
test_df['mucous_membrane'].fillna('missing', inplace=True)
test_df['capillary_refill_time'].fillna('missing', inplace=True)

# 检查处理后的缺失值情况
train_df.isnull().sum(), test_df.isnull().sum()
```

```


# 对训练数据进行独热编码
train_df_encoded = pd.get_dummies(train_df, drop_first=True)

# 对测试数据进行独热编码
test_df_encoded = pd.get_dummies(test_df, drop_first=True)

# 确保训练集和测试集的特征列一致
missing_cols = set(train_df_encoded.columns) - set(test_df_encoded.columns)
for c in missing_cols:
    test_df_encoded[c] = 0

# 确保列顺序一致
test_df_encoded = test_df_encoded[train_df_encoded.columns]

# 查看编码后的数据
train_df_encoded.head(), test_df_encoded.head()

```

```



# 选择特征和目标变量
X_train = train_df_encoded.drop(columns=['outcome_died', 'outcome_euthanized', 'outcome_lived'])
y_train = train_df_encoded[['outcome_died', 'outcome_euthanized', 'outcome_lived']]

X_test = test_df_encoded.drop(columns=['outcome_died', 'outcome_euthanized', 'outcome_lived'])
y_test = test_df_encoded[['outcome_died', 'outcome_euthanized', 'outcome_lived']]

# 查看特征和目标变量
X_train.head(), y_train.head(), X_test.head(), y_test.head()
```

```



# 重新检查独热编码后的列
print(train_df_encoded.columns)
print(test_df_encoded.columns)
```

```



# 选择特征
X_train = train_df_encoded.drop(columns=['outcome_euthanized', 'outcome_lived'])
X_test = test_df_encoded.drop(columns=['outcome_euthanized', 'outcome_lived'])

# 选择目标变量，并将 outcome_euthanized 和 outcome_lived 合并为一个二分类目标变量
y_train = (train_df_encoded['outcome_euthanized'] | train_df_encoded['outcome_lived']).astype(int)
y_test = (test_df_encoded['outcome_euthanized'] | test_df_encoded['outcome_lived']).astype(int)

# 查看特征和目标变量
X_train.head(), y_train.head(), X_test.head(), y_test.head()

```

```



from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, classification_report

# 初始化模型
model = RandomForestClassifier(n_estimators=100, random_state=42)

# 训练模型
model.fit(X_train, y_train)

# 进行预测
y_pred = model.predict(X_test)

# 计算 F1 分数
f1 = f1_score(y_test, y_pred)

# 输出分类报告
print(classification_report(y_test, y_pred))

# 输出 F1 分数
f1

```

```



# 保存预测结果到文件
predictions = pd.DataFrame({'id': test_df['hospital_number'], 'prediction': y_pred})
predictions.to_csv('predictions.csv', index=False)

# 查看保存的文件
predictions.head()
```
