```

import pandas as pd

# 训练集路径
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/train.csv'
# 测试集路径
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/test.csv'

# 加载训练集
train_df = pd.read_csv(train_path)
# 加载测试集
test_df = pd.read_csv(test_path)

# 显示训练集和测试集的前几行
print("Training Data:")
print(train_df.head())
print("\nTesting Data:")
print(test_df.head())
```

```


# 检查训练集和测试集中的缺失值
print("Training Data Missing Values:")
print(train_df.isnull().sum())
print("\nTesting Data Missing Values:")
print(test_df.isnull().sum())

```

```


# 处理训练集中的缺失值
train_df['peripheral_pulse'].fillna(train_df['peripheral_pulse'].median(), inplace=True)
train_df['mucous_membrane'].fillna(train_df['mucous_membrane'].mode()[0], inplace=True)
train_df['capillary_refill_time'].fillna(train_df['capillary_refill_time'].mode()[0], inplace=True)

# 处理测试集中的缺失值
test_df['peripheral_pulse'].fillna(test_df['peripheral_pulse'].median(), inplace=True)
test_df['mucous_membrane'].fillna(test_df['mucous_membrane'].mode()[0], inplace=True)
test_df['capillary_refill_time'].fillna(test_df['capillary_refill_time'].mode()[0], inplace=True)

# 再次检查缺失值
print("Training Data Missing Values After Handling:")
print(train_df.isnull().sum())
print("\nTesting Data Missing Values After Handling:")
print(test_df.isnull().sum())

```

```


# 进行独热编码
train_df_encoded = pd.get_dummies(train_df, drop_first=True)
test_df_encoded = pd.get_dummies(test_df, drop_first=True)

# 显示编码后的训练集和测试集的前几行
print("Encoded Training Data:")
print(train_df_encoded.head())
print("\nEncoded Testing Data:")
print(test_df_encoded.head())

# 分离特征和标签
X_train = train_df_encoded.drop(columns=['outcome_died', 'outcome_euthanized'])
y_train = train_df_encoded[['outcome_died', 'outcome_euthanized']]

X_test = test_df_encoded.drop(columns=['outcome_died', 'outcome_euthanized'])
y_test = test_df_encoded[['outcome_died', 'outcome_euthanized']]

# 显示特征和标签的形状
print("\nFeature and Label Shapes:")
print(f"X_train shape: {X_train.shape}, y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}, y_test shape: {y_test.shape}")
```

```



# 检查编码后的训练集和测试集的列名
print("Encoded Training Data Columns:")
print(train_df_encoded.columns)
print("\nEncoded Testing Data Columns:")
print(test_df_encoded.columns)
```

```



# 创建新的标签列
train_df_encoded['outcome'] = train_df['outcome'].map({'died': 0, 'lived': 1, 'euthanized': 2})
test_df_encoded['outcome'] = test_df['outcome'].map({'died': 0, 'lived': 1, 'euthanized': 2})

# 分离特征和标签
X_train = train_df_encoded.drop(columns=['outcome', 'outcome_euthanized', 'outcome_lived'])
y_train = train_df_encoded['outcome']

X_test = test_df_encoded.drop(columns=['outcome', 'outcome_euthanized', 'outcome_lived'])
y_test = test_df_encoded['outcome']

# 显示特征和标签的形状
print("Feature and Label Shapes:")
print(f"X_train shape: {X_train.shape}, y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}, y_test shape: {y_test.shape}")

```

```



from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, classification_report

# 训练随机森林分类器
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 在测试集上进行预测
y_pred = model.predict(X_test)

# 计算F1分数
f1 = f1_score(y_test, y_pred, average='weighted')
print(f"F1 Score: {f1}")

# 生成分类报告
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```



# 创建包含预测结果的数据框
results_df = pd.DataFrame({
    'hospital_number': test_df['hospital_number'],
    'predicted_outcome': y_pred
})

# 将预测结果保存为CSV文件
results_df.to_csv('horse_health_predictions.csv', index=False)

# 显示保存的文件路径
print("Predictions saved to: horse_health_predictions.csv")

```
