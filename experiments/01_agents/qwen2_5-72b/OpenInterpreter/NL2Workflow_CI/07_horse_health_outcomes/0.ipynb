```

import pandas as pd

# 加载训练集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/train.csv'
train_df = pd.read_csv(train_path)

# 加载测试集
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/test.csv'
test_df = pd.read_csv(test_path)

# 查看数据的基本信息
print("训练集基本信息:")
print(train_df.info())
print("\n训练集前几行数据:")
print(train_df.head())

print("\n测试集基本信息:")
print(test_df.info())
print("\n测试集前几行数据:")
print(test_df.head())
```

```


# 处理训练集中的缺失值
train_df['peripheral_pulse'] = train_df['peripheral_pulse'].fillna('unknown')
train_df['mucous_membrane'] = train_df['mucous_membrane'].fillna('unknown')
train_df['capillary_refill_time'] = train_df['capillary_refill_time'].fillna('unknown')

# 处理测试集中的缺失值
test_df['peripheral_pulse'] = test_df['peripheral_pulse'].fillna('unknown')
test_df['mucous_membrane'] = test_df['mucous_membrane'].fillna('unknown')
test_df['capillary_refill_time'] = test_df['capillary_refill_time'].fillna('unknown')

# 查看处理后的数据基本信息
print("处理后训练集基本信息:")
print(train_df.info())

print("\n处理后测试集基本信息:")
print(test_df.info())

```

```

from sklearn.preprocessing import LabelEncoder

# 创建LabelEncoder实例
label_encoder = LabelEncoder()

# 编码分类变量
categorical_columns = ['surgery', 'peripheral_pulse', 'mucous_membrane', 'capillary_refill_time']

# 对训练集和测试集中的分类变量进行编码
for col in categorical_columns:
    train_df[col] = label_encoder.fit_transform(train_df[col])
    test_df[col] = label_encoder.transform(test_df[col])

# 转换标签
train_df['outcome'] = train_df['outcome'].map({'died': 0, 'lived': 1, 'euthanized': 2})
test_df['outcome'] = test_df['outcome'].map({'died': 0, 'lived': 1, 'euthanized': 2})

# 查看处理后的数据
print("处理后训练集基本信息:")
print(train_df.info())
print("\n处理后训练集前几行数据:")
print(train_df.head())

print("\n处理后测试集基本信息:")
print(test_df.info())
print("\n处理后测试集前几行数据:")
print(test_df.head())
```

```


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score
from sklearn.model_selection import train_test_split

# 定义特征和标签
X_train = train_df.drop(columns=['outcome', 'hospital_number'])
y_train = train_df['outcome']

X_test = test_df.drop(columns=['outcome', 'hospital_number'])
y_test = test_df['outcome']

# 初始化随机森林分类器
clf = RandomForestClassifier(n_estimators=100, random_state=42)

# 训练模型
clf.fit(X_train, y_train)

# 在测试集上进行预测
y_pred = clf.predict(X_test)

# 计算F1分数
f1 = f1_score(y_test, y_pred, average='weighted')

# 输出F1分数
print(f"测试集上的F1分数: {f1}")

# 保存预测结果
test_df['predicted_outcome'] = y_pred
test_df[['hospital_number', 'outcome', 'predicted_outcome']].to_csv('predictions.csv', index=False)

print("预测结果已保存到 'predictions.csv' 文件中。")

```
