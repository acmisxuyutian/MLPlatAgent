```

import pandas as pd

# 加载训练集和测试集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/horse_health_outcomes/test.csv'

train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

# 查看数据的前几行
print("训练集前几行:")
print(train_data.head())
print("\n测试集前几行:")
print(test_data.head())
```

```


# 数据预处理

# 检查缺失值
print("训练集缺失值情况:")
print(train_data.isnull().sum())
print("\n测试集缺失值情况:")
print(test_data.isnull().sum())

# 处理缺失值
# 对于数值型特征，可以使用中位数或均值填充
# 对于分类特征，可以使用众数填充

train_data.fillna(train_data.median(), inplace=True)
test_data.fillna(test_data.median(), inplace=True)

# 对分类变量进行编码
from sklearn.preprocessing import LabelEncoder

# 获取所有分类变量的列名
categorical_vars = ['surgery', 'temperature_of_extremities', 'peripheral_pulse', 'mucous_membranes', 'capillary_refill_time', 'outcome']

# 初始化LabelEncoder
le = LabelEncoder()

# 对训练集和测试集的分类变量进行编码
for var in categorical_vars:
    train_data[var] = le.fit_transform(train_data[var])
    test_data[var] = le.transform(test_data[var])

# 查看处理后的数据
print("\n处理后的训练集前几行:")
print(train_data.head())
print("\n处理后的测试集前几行:")
print(test_data.head())

```

```



# 分离特征和目标变量
X_train = train_data.drop(columns=['outcome', 'hospital_number'])
y_train = train_data['outcome']
X_test = test_data.drop(columns=['outcome', 'hospital_number'])
y_test = test_data['outcome']

# 特征选择
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import SelectFromModel

# 训练一个随机森林模型以选择特征
selector = RandomForestClassifier(n_estimators=100, random_state=42)
selector.fit(X_train, y_train)

# 选择特征
sfm = SelectFromModel(selector, prefit=True)
X_train_selected = sfm.transform(X_train)
X_test_selected = sfm.transform(X_test)

# 查看选择的特征
selected_features = X_train.columns[sfm.get_support()]
print("选择的特征:", selected_features)

# 训练最终模型
final_model = RandomForestClassifier(n_estimators=100, random_state=42)
final_model.fit(X_train_selected, y_train)

# 预测测试集
y_pred = final_model.predict(X_test_selected)

# 评估模型性能
from sklearn.metrics import f1_score

f1 = f1_score(y_test, y_pred, average='weighted')
print("测试集上的F1分数:", f1)

# 保存预测结果
predictions = pd.DataFrame({'hospital_number': test_data['hospital_number'], 'outcome': y_pred})
predictions.to_csv('horse_health_predictions.csv', index=False)
print("预测结果已保存到 horse_health_predictions.csv")

```
