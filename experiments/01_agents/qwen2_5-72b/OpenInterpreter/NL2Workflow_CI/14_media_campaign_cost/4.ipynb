```

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
import warnings
warnings.filterwarnings('ignore')
```

```


# 定义数据集的路径
train_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/train.csv'
test_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/test.csv'

# 加载数据
train_data = pd.read_csv(train_file_path)
test_data = pd.read_csv(test_file_path)

# 查看训练集前5行
train_data.head()

# 查看训练集的基本信息
train_data.info()

# 查看测试集前5行
test_data.head()

# 查看测试集的基本信息
test_data.info()
```

```


# 查看训练数据的基本统计信息
train_data.describe()

# 查看测试数据的基本统计信息
test_data.describe()

```

```


# 删除无用的列
train_data = train_data.drop(columns=['id'])
test_data = test_data.drop(columns=['id'])

# 检查异常值
train_data.describe().T
```

```


from sklearn.preprocessing import StandardScaler

# 分离特征和目标变量
X_train = train_data.drop(columns=['cost'])
y_train = train_data['cost']

X_test = test_data.drop(columns=['cost'])
y_test = test_data['cost']

# 标准化数值特征
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 将标准化后的数据转换回DataFrame
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_scaled = pd.DataFrame(X_test_scaled, columns=X_test.columns)

X_train_scaled.head()
```

```



from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_score
from sklearn.metrics import mean_squared_log_error
import numpy as np

# 初始化随机森林回归器
model = RandomForestRegressor(n_estimators=100, random_state=42)

# 交叉验证
scores = cross_val_score(model, X_train_scaled, y_train, cv=5, scoring='neg_mean_squared_log_error')
rmse_scores = np.sqrt(-scores)

# 打印交叉验证的RMLSE
print(f"Cross-validated RMLSE: {rmse_scores.mean():.4f} (std: {rmse_scores.std():.4f})")

# 训练模型
model.fit(X_train_scaled, y_train)

# 预测测试集
y_pred = model.predict(X_test_scaled)

# 计算测试集上的RMLSE
rmlse = np.sqrt(mean_squared_log_error(y_test, y_pred))
print(f"Test RMLSE: {rmlse:.4f}")

# 将预测结果保存到文件
submission = pd.DataFrame({
    'id': test_data.index,
    'predicted_cost': y_pred
})
submission.to_csv('submission.csv', index=False)

submission.head()
```
