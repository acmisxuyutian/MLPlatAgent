```

import pandas as pd

# 读取训练数据
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/train.csv'
train_data = pd.read_csv(train_path)

# 读取测试数据
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/test.csv'
test_data = pd.read_csv(test_path)

# 查看数据的前几行
print("训练数据的前几行：")
print(train_data.head())

print("\n测试数据的前几行：")
print(test_data.head())
```

```

# 检查训练数据和测试数据中的缺失值
print("训练数据中的缺失值情况：")
print(train_data.isnull().sum())

print("\n测试数据中的缺失值情况：")
print(test_data.isnull().sum())
```

```


import matplotlib.pyplot as plt
import seaborn as sns

# 检查每个特征的分布
plt.figure(figsize=(20, 15))
for i, column in enumerate(train_data.columns):
    plt.subplot(5, 4, i + 1)
    sns.boxplot(x=train_data[column])
    plt.title(column)

plt.tight_layout()
plt.show()

# 检查目标变量的分布
sns.histplot(train_data['cost'], kde=True)
plt.title('Distribution of Cost')
plt.show()

```
