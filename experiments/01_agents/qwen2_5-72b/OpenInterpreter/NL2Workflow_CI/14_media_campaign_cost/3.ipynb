```

import pandas as pd

# 定义文件路径
train_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/train.csv'
test_file_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/test.csv'

# 从文件路径加载数据
train_df = pd.read_csv(train_file_path)
test_df = pd.read_csv(test_file_path)

# 查看数据集的前几行
train_df.head(), test_df.head()
```

```



# 从测试数据中删除cost列
X_test = test_df.drop(columns=['id', 'cost'], errors='ignore')

# 选择特征列和标签列
X_train = train_df.drop(columns=['id', 'cost'])
y_train = train_df['cost']

# 特征缩放
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 检查特征缩放后的数据
X_train_scaled[:5], X_test_scaled[:5]


```

```


from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

# 选择模型
model = LinearRegression()

# 训练模型
model.fit(X_train_scaled, y_train)

# 预测测试集
y_test_pred = model.predict(X_test_scaled)

# 计算 RMLSE
rmlse = np.sqrt(mean_squared_error(y_train, model.predict(X_train_scaled), squared=False))
print(f"RMLSE on the training set: {rmlse}")

# 保存预测结果
test_df['predicted_cost'] = y_test_pred
test_df.to_csv('predicted_cost.csv', index=False)

# 查看预测结果的前几行
test_df[['id', 'predicted_cost']].head()
```

```



# 计算 RMLSE
def rmlse(y_true, y_pred):
    log_true = np.log1p(y_true)
    log_pred = np.log1p(y_pred)
    return np.sqrt(np.mean((log_true - log_pred) ** 2))

# 计算 RMLSE
rmlse_train = rmlse(y_train, model.predict(X_train_scaled))
print(f"RMLSE on the training set: {rmlse_train}")

# 保存预测结果
test_df['predicted_cost'] = y_test_pred
test_df.to_csv('predicted_cost.csv', index=False)

# 查看预测结果的前几行
test_df[['id', 'predicted_cost']].head()

```
