```

import pandas as pd

# 指定数据文件路径
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/media_campaign_cost/test.csv'

# 读取数据
train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)

# 显示前几行数据，检查是否正确读取
train_data.head(), test_data.head()
```

```


# 检查数据集中是否有空值
train_data.isnull().sum(), test_data.isnull().sum()

```

```


# 查看数据的描述统计
train_data.describe(include='all')

```

```



from sklearn.preprocessing import StandardScaler

# 选择数值特征
numeric_features = ['store_sales', 'unit_sales', 'total_children', 'num_children_at_home', 'avg_cars_at_home', 'gross_weight', 'store_sqft']

# 初始化标准化器
scaler = StandardScaler()

# 对训练数据进行标准化
train_data[numeric_features] = scaler.fit_transform(train_data[numeric_features])

# 对测试数据进行标准化
test_data[numeric_features] = scaler.transform(test_data[numeric_features])

# 查看标准化后的数据
train_data.head(), test_data.head()


```

```



from sklearn.model_selection import train_test_split

# 拆分训练数据为训练集和验证集
X_train = train_data.drop(columns=['id', 'cost'])
y_train = train_data['cost']
X_test = test_data.drop(columns=['id', 'cost'])

# 拆分训练集和验证集
X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# 查看拆分后的数据
X_train_split.shape, X_val_split.shape, y_train_split.shape, y_val_split.shape


```

```


from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_log_error

# 初始化随机森林回归模型
model = RandomForestRegressor(n_estimators=100, random_state=42)

# 训练模型
model.fit(X_train_split, y_train_split)

# 在验证集上进行预测
y_val_pred = model.predict(X_val_split)

# 计算验证集上的RMLSE
rmlse = mean_squared_log_error(y_val_split, y_val_pred, squared=False)
rmlse
```

```



import numpy as np

# 计算RMLSE
def calculate_rmlse(y_true, y_pred):
    log_true = np.log1p(y_true)
    log_pred = np.log1p(y_pred)
    mse = np.mean((log_true - log_pred) ** 2)
    rmlse = np.sqrt(mse)
    return rmlse

# 计算验证集上的RMLSE
rmlse = calculate_rmlse(y_val_split, y_val_pred)
rmlse
```

```



# 在测试集上进行预测
y_test_pred = model.predict(X_test)

# 将预测结果保存为CSV文件
test_data['predicted_cost'] = y_test_pred
test_data[['id', 'predicted_cost']].to_csv('predicted_media_campaign_cost.csv', index=False)

# 计算测试集上的RMLSE（假设测试集中有真实成本）
if 'cost' in test_data.columns:
    test_rmlse = calculate_rmlse(test_data['cost'], y_test_pred)
    test_rmlse
else:
    print("测试集中没有真实成本，无法计算RMLSE。")
```
