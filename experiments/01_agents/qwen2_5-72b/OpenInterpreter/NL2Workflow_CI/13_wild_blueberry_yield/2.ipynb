```

import pandas as pd

# 加载训练集和测试集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/test.csv'

# 读取数据
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 显示训练集和测试集的前几行
print('训练集样例:')
print(train_df.head())
print('\n测试集样例:')
print(test_df.head())

# 显示训练集和测试集的基本信息
print('\n训练集基本信息:')
train_df.info()
print('\n测试集基本信息:')
test_df.info()
```

```


# 计算基本统计量
train_stats = train_df.describe()
test_stats = test_df.describe()

# 显示统计量
print('训练集基本统计量：')
print(train_stats)
print('\n测试集基本统计量：')
print(test_stats)

```

```


# 计算特征与目标变量 'yield' 的相关性
correlations = train_df.corr()['yield'].sort_values(ascending=False)

# 显示相关性
print('与目标变量 "yield" 的相关性：')
print(correlations)

# 绘制相关性热图
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
sns.heatmap(train_df.corr(), annot=True, fmt=".2f")
plt.show()
```
