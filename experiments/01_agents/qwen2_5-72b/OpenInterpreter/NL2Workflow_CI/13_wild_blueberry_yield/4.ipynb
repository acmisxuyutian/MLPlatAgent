```

import pandas as pd

# 加载训练集和测试集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 查看数据集的前几行
train_df.head(), test_df.head()
```

```

# 查看训练集的统计信息
train_df.describe()
```

```


import numpy as np
from sklearn.preprocessing import PolynomialFeatures

# 特征构建
train_df['honeybee_bumbles'] = train_df['honeybee'] * train_df['bumbles']
train_df['clonesize_honeybee'] = train_df['clonesize'] * train_df['honeybee']
train_df['AverageOfUpperTRange_AverageOfLowerTRange'] = train_df['AverageOfUpperTRange'] * train_df['AverageOfLowerTRange']

test_df['honeybee_bumbles'] = test_df['honeybee'] * test_df['bumbles']
test_df['clonesize_honeybee'] = test_df['clonesize'] * test_df['honeybee']
test_df['AverageOfUpperTRange_AverageOfLowerTRange'] = test_df['AverageOfUpperTRange'] * test_df['AverageOfLowerTRange']

# 查看新的特征
train_df.head()
```

```


import seaborn as sns
import matplotlib.pyplot as plt

# 计算相关系数矩阵
corr_matrix = train_df.corr()

# 选择与目标变量 'yield' 最相关的特征
corr_with_yield = corr_matrix['yield'].abs().sort_values(ascending=False)

# 打印与 'yield' 最相关的特征
corr_with_yield

# 绘制相关性热力图
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.show()

```
