```

import pandas as pd

# 定义数据集路径
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/wild_blueberry_yield/test.csv'

# 加载数据集
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 查看前几行数据
train_df.head(), test_df.head()
```

```


# 查看数据概要
train_df.describe(include='all'), test_df.describe(include='all')

# 检查有无缺失值
train_df.isnull().sum(), test_df.isnull().sum()

```

```


# 数据概要统计
train_summary = train_df.describe()
test_summary = test_df.describe()

train_summary, test_summary

```

```


# 特征构建
train_df['temp_range_diff'] = train_df['MaxOfUpperTRange'] - train_df['MinOfUpperTRange']
train_df['rain_days_diff'] = train_df['MaxOfLowerTRange'] - train_df['MinOfLowerTRange']

test_df['temp_range_diff'] = test_df['MaxOfUpperTRange'] - test_df['MinOfUpperTRange']
test_df['rain_days_diff'] = test_df['MaxOfLowerTRange'] - test_df['MinOfLowerTRange']

# 查看构建后的特征
train_df[['temp_range_diff', 'rain_days_diff']].head(), test_df[['temp_range_diff', 'rain_days_diff']].head()

```

```



# 计算相关性矩阵
correlation_matrix = train_df.corr()

# 查看目标变量 yield 与各个特征的相关性
correlation_with_yield = correlation_matrix['yield'].abs().sort_values(ascending=False)

correlation_with_yield


```

```



# 导入必要的库
from sklearn.model_selection import train_test_split

# 选择相关性较高的特征
selected_features = ['fruitset', 'seeds', 'fruitmass', 'AverageRainingDays', 'RainingDays', 'clonesize']

# 分离特征和目标变量
X_train = train_df[selected_features]
y_train = train_df['yield']

X_test = test_df[selected_features]
y_test = test_df['yield']

# 数据抽样（80% 训练，20% 验证）
X_train_sample, X_val, y_train_sample, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

# 查看抽样后的数据集大小
X_train_sample.shape, X_val.shape, X_test.shape


```

```



# 导入必要的库
import xgboost as xgb
from sklearn.metrics import mean_absolute_error

# 初始化XGBoost回归模型
model = xgb.XGBRegressor(objective='reg:squarederror', random_state=42)

# 训练模型
model.fit(X_train_sample, y_train_sample)

# 在验证集上进行预测
y_val_pred = model.predict(X_val)

# 计算验证集上的MAE
val_mae = mean_absolute_error(y_val, y_val_pred)

val_mae


```

```




# 在测试集上进行预测
y_test_pred = model.predict(X_test)

# 计算测试集上的MAE
test_mae = mean_absolute_error(y_test, y_test_pred)

test_mae



```
