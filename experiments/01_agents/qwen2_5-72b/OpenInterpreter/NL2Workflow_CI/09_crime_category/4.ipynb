```

import pandas as pd

# 加载训练集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/train.csv'
train_data = pd.read_csv(train_path)

# 加载测试集
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/test.csv'
test_data = pd.read_csv(test_path)

# 显示前几行数据以检查数据
train_data.head(), test_data.head()
```

```


from sklearn.preprocessing import LabelEncoder

# 初始化编码器
label_encoder = LabelEncoder()

# 对训练集和测试集中的类别变量进行编码
categorical_columns = ['DayOfWeek', 'PdDistrict']

for column in categorical_columns:
    train_data[column] = label_encoder.fit_transform(train_data[column])
    test_data[column] = label_encoder.transform(test_data[column])

# 显示预处理后的训练集和测试集的前几行
train_data.head(), test_data.head()
```

```


# 拆分训练数据
X_train = train_data.drop('Category', axis=1)
y_train = train_data['Category']

# 检查拆分后的训练数据
X_train.head(), y_train.head()
```

```


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import log_loss

# 选择模型
model = RandomForestClassifier(n_estimators=100, random_state=42)

# 训练模型
model.fit(X_train, y_train)

# 对测试数据进行预测
X_test = test_data.drop('Category', axis=1)
y_test_pred = model.predict_proba(X_test)

# 检查预测结果
y_test_pred[:5]

```

```



# 假设测试集中的类别标签已经编码为数字
y_test = label_encoder.transform(test_data['Category'])

# 计算logloss
logloss = log_loss(y_test, y_test_pred)
logloss

```

```



from sklearn.preprocessing import LabelBinarizer

# 初始化编码器
label_binarizer = LabelBinarizer()

# 将训练集中的类别标签编码为二进制形式
y_train_encoded = label_binarizer.fit_transform(y_train)

# 将测试集中的类别标签编码为二进制形式，处理未见过的标签
y_test_encoded = label_binarizer.transform(test_data['Category'])

# 检查编码后的标签
y_train_encoded[:5], y_test_encoded[:5]

```

```



# 重新训练模型
model.fit(X_train, y_train_encoded)

# 对测试数据进行预测
y_test_pred = model.predict_proba(X_test)

# 检查预测结果
y_test_pred[:5]


```
