```

import pandas as pd

# 训练集路径
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/train.csv'
# 测试集路径
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/test.csv'

# 加载数据
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 查看数据前几行
print("训练集前几行：")
print(train_df.head())

print("测试集前几行：")
print(test_df.head())
```

```


# 检查缺失值
print("训练集缺失值情况：")
print(train_df.isnull().sum())

print("测试集缺失值情况：")
print(test_df.isnull().sum())

```

```

import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 加载数据
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 对 DayOfWeek 和 PdDistrict 进行独热编码
train_df = pd.get_dummies(train_df, columns=['DayOfWeek', 'PdDistrict'])
test_df = pd.get_dummies(test_df, columns=['DayOfWeek', 'PdDistrict'])

# 对 Category 进行标签编码
label_encoder = LabelEncoder()
train_df['Category'] = label_encoder.fit_transform(train_df['Category'])

# 查看编码后的数据
print("编码后的训练集前几行：")
print(train_df.head())

print("编码后的测试集前几行：")
print(test_df.head())
```

```


# 特征和目标变量
X_train = train_df.drop(columns=['Category'])
y_train = train_df['Category']

X_test = test_df.drop(columns=['Category'])
y_test = label_encoder.transform(test_df['Category'])

# 查看特征和目标变量的形状
print("训练集特征形状：", X_train.shape)
print("训练集目标形状：", y_train.shape)
print("测试集特征形状：", X_test.shape)
print("测试集目标形状：", y_test.shape)

```

```


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import log_loss

# 初始化模型
model = RandomForestClassifier(n_estimators=100, random_state=42)

# 训练模型
model.fit(X_train, y_train)

# 在测试集上进行预测
y_pred_proba = model.predict_proba(X_test)

# 计算 logloss
logloss = log_loss(y_test, y_pred_proba)
print("测试集上的 logloss: ", logloss)

```

```


import pandas as pd

# 创建一个 DataFrame 来保存预测结果
result_df = pd.DataFrame(y_pred_proba, columns=label_encoder.classes_)

# 保存预测结果到文件
result_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/predictions.csv'
result_df.to_csv(result_path, index=False)

print(f"预测结果已保存到 {result_path}")
```
