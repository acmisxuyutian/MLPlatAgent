```

import pandas as pd

# 加载训练数据
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/train.csv'
train_data = pd.read_csv(train_path)

# 加载测试数据
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/test.csv'
test_data = pd.read_csv(test_path)

# 显示训练数据的前几行
print("训练数据前几行:")
print(train_data.head())

# 显示测试数据的前几行
print("测试数据前几行:")
print(test_data.head())
```

```


# 处理类别变量
from sklearn.preprocessing import LabelEncoder

# 初始化LabelEncoder
label_encoder = LabelEncoder()

# 对训练数据进行编码
train_data['DayOfWeek'] = label_encoder.fit_transform(train_data['DayOfWeek'])
train_data['PdDistrict'] = label_encoder.fit_transform(train_data['PdDistrict'])

# 对测试数据进行编码
test_data['DayOfWeek'] = label_encoder.transform(test_data['DayOfWeek'])
test_data['PdDistrict'] = label_encoder.transform(test_data['PdDistrict'])

# 检查是否有缺失值
print("训练数据缺失值情况:")
print(train_data.isnull().sum())

print("测试数据缺失值情况:")
print(test_data.isnull().sum())

# 分离特征和目标变量
X_train = train_data.drop(columns=['Category'])
y_train = train_data['Category']

X_test = test_data.drop(columns=['Category'])
y_test = test_data['Category']

# 显示预处理后的数据
print("预处理后的训练数据前几行:")
print(X_train.head())
print("预处理后的测试数据前几行:")
print(X_test.head())

```

```


from sklearn.preprocessing import OneHotEncoder

# 初始化OneHotEncoder
encoder = OneHotEncoder(sparse=False, handle_unknown='ignore')

# 将类别变量转换为适合编码的格式
categorical_features = ['DayOfWeek', 'PdDistrict']
train_categorical = train_data[categorical_features]
test_categorical = test_data[categorical_features]

# 对训练数据进行编码
train_encoded = encoder.fit_transform(train_categorical)
train_encoded_df = pd.DataFrame(train_encoded, columns=encoder.get_feature_names_out(categorical_features))

# 对测试数据进行编码
test_encoded = encoder.transform(test_categorical)
test_encoded_df = pd.DataFrame(test_encoded, columns=encoder.get_feature_names_out(categorical_features))

# 将编码后的特征与原始数据合并
X_train = pd.concat([train_data.drop(columns=categorical_features), train_encoded_df], axis=1)
X_test = pd.concat([test_data.drop(columns=categorical_features), test_encoded_df], axis=1)

# 显示预处理后的数据
print("预处理后的训练数据前几行:")
print(X_train.head())
print("预处理后的测试数据前几行:")
print(X_test.head())

# 分离特征和目标变量
y_train = train_data['Category']
y_test = test_data['Category']

# 检查是否有缺失值
print("训练数据缺失值情况:")
print(X_train.isnull().sum())

print("测试数据缺失值情况:")
print(X_test.isnull().sum())

```

```



from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score

# 初始化随机森林分类器
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)

# 训练模型
rf_classifier.fit(X_train, y_train)

# 预测测试数据
y_pred = rf_classifier.predict(X_test)

# 评估模型
print("模型评估结果:")
print(classification_report(y_test, y_pred))
print("准确率:", accuracy_score(y_test, y_pred))
```

```




from sklearn.preprocessing import LabelEncoder

# 初始化LabelEncoder
label_encoder = LabelEncoder()

# 对训练数据的Category列进行编码
y_train_encoded = label_encoder.fit_transform(y_train)

# 对测试数据的Category列进行编码
y_test_encoded = label_encoder.transform(y_test)

# 初始化随机森林分类器
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)

# 训练模型
rf_classifier.fit(X_train, y_train_encoded)

# 预测测试数据
y_pred_encoded = rf_classifier.predict(X_test)

# 将预测结果反编码为原始类别
y_pred = label_encoder.inverse_transform(y_pred_encoded)

# 评估模型
print("模型评估结果:")
print(classification_report(y_test, y_pred))
print("准确率:", accuracy_score(y_test, y_pred))


```
