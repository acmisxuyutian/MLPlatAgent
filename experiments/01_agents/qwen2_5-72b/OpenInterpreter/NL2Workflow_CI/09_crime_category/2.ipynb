```


import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import log_loss

# 加载训练集和测试集
train_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/train.csv'
test_path = r'D:\LLM-Driven_AI-Studio\MLAgent\data\benchmark/NL2Workflow/datasets/tabular/crime_category/test.csv'

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

# 对 'DayOfWeek' 和 'PdDistrict' 进行 one-hot 编码
train_df = pd.get_dummies(train_df, columns=['DayOfWeek', 'PdDistrict'])
test_df = pd.get_dummies(test_df, columns=['DayOfWeek', 'PdDistrict'])

# 对 'Category' 进行标签编码
label_encoder = LabelEncoder()
train_df['Category'] = label_encoder.fit_transform(train_df['Category'])

# 选择特征
features = ['X', 'Y'] + [col for col in train_df.columns if 'DayOfWeek_' in col or 'PdDistrict_' in col]
target = 'Category'

# 分割特征和目标变量
X_train = train_df[features]
y_train = train_df[target]
X_test = test_df[features]

# 训练模型
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 预测
y_pred_proba = model.predict_proba(X_test)

# 将预测结果保存到文件
test_df['Predicted_Category'] = label_encoder.inverse_transform(y_pred_proba.argmax(axis=1))
test_df[['Predicted_Category']].to_csv('predicted_categories.csv', index=False)

# 计算logloss
logloss = log_loss(test_df['Category'], y_pred_proba)
print(f'Log Loss on the test set: {logloss:.4f}')

```
