```

import pandas as pd

# Load the train and eval datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\08_santander_value\\split_train.csv'
eval_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\08_santander_value\\split_eval.csv'

train_df = pd.read_csv(train_path)
eval_df = pd.read_csv(eval_path)

# Display the first few rows of the train and eval datasets
print("Train Data:")
print(train_df.head())
print("\nEval Data:")
print(eval_df.head())
```

```

# Get a summary of the train and eval datasets
print("Train Data Summary:")
print(train_df.describe())

print("\nEval Data Summary:")
print(eval_df.describe())

# Check for missing values
print("\nMissing Values in Train Data:")
print(train_df.isnull().sum())

print("\nMissing Values in Eval Data:")
print(eval_df.isnull().sum())
```

```


import matplotlib.pyplot as plt
import seaborn as sns

# Set the style for the plots
sns.set(style="whitegrid")

# Plot the distribution of the target
plt.figure(figsize=(12, 6))
sns.histplot(train_df['target'], kde=True, bins=50)
plt.title('Distribution of Target')
plt.xlabel('Target')
plt.ylabel('Frequency')
plt.show()

# Plot the distribution of a few key features
key_features = ['48df886f9', '0deb4b6a8', '96f83a237', 'b8a716ebf']

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
for i, feature in enumerate(key_features):
    ax = axes[i // 2, i % 2]
    sns.histplot(train_df[feature], kde=True, bins=50, ax=ax)
    ax.set_title(f'Distribution of {feature}')
    ax.set_xlabel(feature)
    ax.set_ylabel('Frequency')
plt.tight_layout()
plt.show()

```
