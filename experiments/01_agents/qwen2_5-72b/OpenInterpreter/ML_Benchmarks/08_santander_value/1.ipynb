```

import pandas as pd

# Load the training and evaluation datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\08_santander_value\\split_train.csv'
eval_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\08_santander_value\\split_eval.csv'

train_df = pd.read_csv(train_path)
eval_df = pd.read_csv(eval_path)

# Display the first few rows of the training dataset
train_df.head()
```

```


# Check for missing values in the training dataset
print("Missing values in the training dataset:")
print(train_df.isnull().sum())

# Check for missing values in the evaluation dataset
print("\nMissing values in the evaluation dataset:")
print(eval_df.isnull().sum())

# Examine data types
print("\nData types in the training dataset:")
print(train_df.dtypes)

# Summary statistics
print("\nSummary statistics for the training dataset:")
print(train_df.describe())

# Distribution of the target variable
print("\nDistribution of the target variable in the training dataset:")
print(train_df['target'].describe())

```

```


import matplotlib.pyplot as plt
import numpy as np

# Visualize the distribution of the target variable
plt.figure(figsize=(10, 6))
plt.hist(train_df['target'], bins=50, color='blue', alpha=0.7)
plt.title('Distribution of Target Variable')
plt.xlabel('Target Value')
plt.ylabel('Frequency')
plt.yscale('log')  # Use log scale for y-axis to better visualize the distribution
plt.show()

# Apply log transformation to the target variable
train_df['log_target'] = np.log1p(train_df['target'])

# Visualize the distribution of the log-transformed target variable
plt.figure(figsize=(10, 6))
plt.hist(train_df['log_target'], bins=50, color='blue', alpha=0.7)
plt.title('Distribution of Log-Transformed Target Variable')
plt.xlabel('Log-Transformed Target Value')
plt.ylabel('Frequency')
plt.show()

# Summary statistics for the log-transformed target variable
print(train_df['log_target'].describe())
```
