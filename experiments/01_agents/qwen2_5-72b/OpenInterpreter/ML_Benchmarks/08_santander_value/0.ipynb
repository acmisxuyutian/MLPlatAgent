```

import pandas as pd

# Load the datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\08_santander_value\\split_train.csv'
eval_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\08_santander_value\\split_eval.csv'

train_df = pd.read_csv(train_path)
eval_df = pd.read_csv(eval_path)

# Display the first few rows of the training data
print("Training Data:")
print(train_df.head())

# Display the first few rows of the evaluation data
print("\nEvaluation Data:")
print(eval_df.head())

# Summary statistics for the training data
print("\nSummary Statistics for Training Data:")
print(train_df.describe())

# Check for missing values in the training data
print("\nMissing Values in Training Data:")
print(train_df.isnull().sum())

# Summary statistics for the evaluation data
print("\nSummary Statistics for Evaluation Data:")
print(eval_df.describe())

# Check for missing values in the evaluation data
print("\nMissing Values in Evaluation Data:")
print(eval_df.isnull().sum())
```

```


from sklearn.preprocessing import StandardScaler

# Separate features and target
X_train = train_df.drop(columns=['target'])
y_train = train_df['target']
X_eval = eval_df.drop(columns=['target'])
y_eval = eval_df['target']

# Standardize the features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_eval_scaled = scaler.transform(X_eval)

# Convert back to DataFrame for easier handling
X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_eval_scaled_df = pd.DataFrame(X_eval_scaled, columns=X_eval.columns)

# Display the first few rows of the scaled training data
print("Scaled Training Data:")
print(X_train_scaled_df.head())

# Display the first few rows of the scaled evaluation data
print("\nScaled Evaluation Data:")
print(X_eval_scaled_df.head())

```

```


from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_log_error
import numpy as np

# Initialize the Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X_train_scaled_df, y_train)

# Predict on the evaluation set
y_pred = model.predict(X_eval_scaled_df)

# Calculate RMSLE
rmsle = np.sqrt(mean_squared_log_error(y_eval, y_pred))

# Print the RMSLE
print(f"RMSLE: {rmsle}")
```

```


# Check for non-positive values in the target columns
print("Non-positive values in y_train:", (y_train <= 0).sum())
print("Non-positive values in y_eval:", (y_eval <= 0).sum())

# Add a small constant to the target values to ensure they are all positive
epsilon = 1e-6
y_train_adjusted = y_train + epsilon
y_eval_adjusted = y_eval + epsilon

# Recalculate the predictions and RMSLE
y_pred_adjusted = model.predict(X_eval_scaled_df)
rmsle_adjusted = np.sqrt(mean_squared_log_error(y_eval_adjusted, y_pred_adjusted))

# Print the adjusted RMSLE
print(f"Adjusted RMSLE: {rmsle_adjusted}")
```

```



# Clip the predicted values to ensure they are all positive
y_pred_clipped = np.clip(y_pred_adjusted, a_min=epsilon, a_max=None)

# Recalculate the RMSLE
rmsle_clipped = np.sqrt(mean_squared_log_error(y_eval_adjusted, y_pred_clipped))

# Print the adjusted RMSLE
print(f"Clipped RMSLE: {rmsle_clipped}")

```

```



from sklearn.ensemble import RandomForestRegressor

# Initialize the Random Forest Regressor
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)

# Train the model
rf_model.fit(X_train_scaled_df, y_train_adjusted)

# Predict on the evaluation set
y_pred_rf = rf_model.predict(X_eval_scaled_df)

# Clip the predicted values to ensure they are all positive
y_pred_rf_clipped = np.clip(y_pred_rf, a_min=epsilon, a_max=None)

# Recalculate the RMSLE
rmsle_rf = np.sqrt(mean_squared_log_error(y_eval_adjusted, y_pred_rf_clipped))

# Print the RMSLE
print(f"Random Forest RMSLE: {rmsle_rf}")

```

```




from sklearn.model_selection import GridSearchCV

# Define the parameter grid
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# Initialize the Random Forest Regressor
rf_model = RandomForestRegressor(random_state=42)

# Initialize Grid Search
grid_search = GridSearchCV(estimator=rf_model, param_grid=param_grid, cv=3, scoring='neg_mean_squared_log_error', n_jobs=-1)

# Fit the Grid Search
grid_search.fit(X_train_scaled_df, y_train_adjusted)

# Get the best parameters and best score
best_params = grid_search.best_params_
best_score = np.sqrt(-grid_search.best_score_)

# Print the best parameters and the best score
print(f"Best Parameters: {best_params}")
print(f"Best RMSLE: {best_score}")

# Predict on the evaluation set using the best model
y_pred_best = grid_search.best_estimator_.predict(X_eval_scaled_df)

# Clip the predicted values to ensure they are all positive
y_pred_best_clipped = np.clip(y_pred_best, a_min=epsilon, a_max=None)

# Recalculate the RMSLE
rmsle_best = np.sqrt(mean_squared_log_error(y_eval_adjusted, y_pred_best_clipped))

# Print the RMSLE
print(f"Best Model RMSLE: {rmsle_best}")

```

```



# Get the feature importances
feature_importances = grid_search.best_estimator_.feature_importances_

# Create a DataFrame to display feature importances
feature_importance_df = pd.DataFrame({
    'Feature': X_train_scaled_df.columns,
    'Importance': feature_importances
})

# Sort the features by importance
feature_importance_df = feature_importance_df.sort_values(by='Importance', ascending=False)

# Display the feature importances
print("Feature Importances:")
print(feature_importance_df)

# Visualize the feature importances
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))
plt.barh(feature_importance_df['Feature'], feature_importance_df['Importance'], color='skyblue')
plt.xlabel('Importance')
plt.title('Feature Importances')
plt.gca().invert_yaxis()  # Invert the y-axis to show the most important features at the top
plt.show()
```
