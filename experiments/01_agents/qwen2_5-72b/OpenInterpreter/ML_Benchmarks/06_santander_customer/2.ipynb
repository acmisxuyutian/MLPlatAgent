```

import pandas as pd

# Load the training and evaluation datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\06_santander_customer\\split_train.csv'
eval_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\06_santander_customer\\split_eval.csv'

train_df = pd.read_csv(train_data_path)
eval_df = pd.read_csv(eval_data_path)

# Display the first few rows of the training dataset
print("Training Data:")
print(train_df.head())

# Display the first few rows of the evaluation dataset
print("\nEvaluation Data:")
print(eval_df.head())

# Display the shape of the training and evaluation datasets
print("\nTraining Data Shape:", train_df.shape)
print("Evaluation Data Shape:", eval_df.shape)

# Display summary statistics of the training dataset
print("\nTraining Data Summary Statistics:")
print(train_df.describe())

# Display information about the training dataset
print("\nTraining Data Information:")
print(train_df.info())
```

```


from sklearn.preprocessing import StandardScaler

# Extract features and target from the training and evaluation datasets
X_train = train_df.drop(columns=['target'])
y_train = train_df['target']
X_eval = eval_df.drop(columns=['target'])
y_eval = eval_df['target']

# Standardize the features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_eval_scaled = scaler.transform(X_eval)

# Convert back to DataFrame to maintain column names
X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_eval_scaled_df = pd.DataFrame(X_eval_scaled, columns=X_eval.columns)

# Display the first few rows of the scaled training features
print("Scaled Training Features:")
print(X_train_scaled_df.head())

# Display the first few rows of the scaled evaluation features
print("\nScaled Evaluation Features:")
print(X_eval_scaled_df.head())

```

```


from sklearn.model_selection import train_test_split

# Split the training data into training and validation sets
X_train_split, X_val_split, y_train_split, y_val_split = train_test_split(
    X_train_scaled_df, y_train, test_size=0.3, random_state=42, stratify=y_train
)

# Display the shapes of the resulting datasets
print("Training Data (X_train_split) Shape:", X_train_split.shape)
print("Training Data (y_train_split) Shape:", y_train_split.shape)
print("Validation Data (X_val_split) Shape:", X_val_split.shape)
print("Validation Data (y_val_split) Shape:", y_val_split.shape)
```

```


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

# Initialize the Random Forest classifier
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)

# Train the model on the training set
rf_classifier.fit(X_train_split, y_train_split)

# Make predictions on the validation set
y_val_pred_proba = rf_classifier.predict_proba(X_val_split)[:, 1]

# Calculate the AUC-ROC score
val_auc = roc_auc_score(y_val_split, y_val_pred_proba)

# Display the AUC-ROC score on the validation set
print("AUC-ROC on Validation Set:", val_auc)
```

```



from sklearn.model_selection import GridSearchCV

# Define the parameter grid for grid search
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# Initialize the Random Forest classifier
rf_classifier = RandomForestClassifier(random_state=42)

# Initialize GridSearchCV
grid_search = GridSearchCV(estimator=rf_classifier, param_grid=param_grid, cv=3, scoring='roc_auc', n_jobs=-1)

# Perform grid search on the training set
grid_search.fit(X_train_split, y_train_split)

# Get the best parameters and best AUC-ROC score
best_params = grid_search.best_params_
best_auc = grid_search.best_score_

# Display the best parameters and best AUC-ROC score
print("Best Parameters:", best_params)
print("Best AUC-ROC Score on Validation Set:", best_auc)

# Evaluate the best model on the validation set
best_model = grid_search.best_estimator_
y_val_pred_proba = best_model.predict_proba(X_val_split)[:, 1]
val_auc_tuned = roc_auc_score(y_val_split, y_val_pred_proba)

# Display the AUC-ROC score on the validation set after tuning
print("AUC-ROC on Validation Set after Tuning:", val_auc_tuned)

```

```



# Define a simplified parameter grid for grid search
param_grid_simplified = {
    'n_estimators': [50, 100],
    'max_depth': [None, 10],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2]
}

# Initialize the Random Forest classifier
rf_classifier = RandomForestClassifier(random_state=42)

# Initialize GridSearchCV with a simplified parameter grid
grid_search_simplified = GridSearchCV(estimator=rf_classifier, param_grid=param_grid_simplified, cv=3, scoring='roc_auc', n_jobs=-1)

# Perform grid search on the training set
grid_search_simplified.fit(X_train_split, y_train_split)

# Get the best parameters and best AUC-ROC score
best_params_simplified = grid_search_simplified.best_params_
best_auc_simplified = grid_search_simplified.best_score_

# Display the best parameters and best AUC-ROC score
print("Best Parameters (Simplified Grid):", best_params_simplified)
print("Best AUC-ROC Score on Validation Set (Simplified Grid):", best_auc_simplified)

# Evaluate the best model on the validation set
best_model_simplified = grid_search_simplified.best_estimator_
y_val_pred_proba_simplified = best_model_simplified.predict_proba(X_val_split)[:, 1]
val_auc_tuned_simplified = roc_auc_score(y_val_split, y_val_pred_proba_simplified)

# Display the AUC-ROC score on the validation set after tuning
print("AUC-ROC on Validation Set after Tuning (Simplified Grid):", val_auc_tuned_simplified)


```

```



# Make predictions on the evaluation set
y_eval_pred_proba = best_model_simplified.predict_proba(X_eval_scaled_df)[:, 1]

# Calculate the AUC-ROC score on the evaluation set
eval_auc = roc_auc_score(y_eval, y_eval_pred_proba)

# Display the AUC-ROC score on the evaluation set
print("AUC-ROC on Evaluation Set:", eval_auc)

```
