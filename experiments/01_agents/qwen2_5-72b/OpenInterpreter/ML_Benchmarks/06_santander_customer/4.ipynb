```

import pandas as pd

# Load the training and evaluation datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\06_santander_customer\\split_train.csv'
eval_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\06_santander_customer\\split_eval.csv'

train_df = pd.read_csv(train_path)
eval_df = pd.read_csv(eval_path)

# Display the first few rows of the training dataset
train_df.head()
```

```


# Display the first few rows of the evaluation dataset
eval_df.head()

# Check the data types of the columns in the training dataset
train_df.dtypes

# Check for missing values in the training dataset
train_df.isnull().sum()

# Check the distribution of the target variable
train_df['target'].value_counts(normalize=True)

# Summary statistics of the features
train_df.describe()
```

```


from sklearn.preprocessing import StandardScaler

# Separate features and target in the training and evaluation datasets
X_train = train_df.drop(columns=['target'])
y_train = train_df['target']

X_eval = eval_df.drop(columns=['target'])
y_eval = eval_df['target']

# Standardize the features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_eval_scaled = scaler.transform(X_eval)

# Convert the scaled features back to DataFrames for easier handling
X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_eval_scaled_df = pd.DataFrame(X_eval_scaled, columns=X_eval.columns)

# Display the first few rows of the scaled training features
X_train_scaled_df.head()

```

```


from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

# Initialize the Random Forest classifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)

# Train the model on the scaled training data
rf_model.fit(X_train_scaled_df, y_train)

# Predict probabilities on the evaluation data
y_eval_pred_proba = rf_model.predict_proba(X_eval_scaled_df)[:, 1]

# Calculate the AUC score
auc_score = roc_auc_score(y_eval, y_eval_pred_proba)

auc_score
```

```



from sklearn.model_selection import GridSearchCV

# Define the parameter grid for hyperparameter tuning
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'class_weight': [None, 'balanced']
}

# Initialize the GridSearchCV object
grid_search = GridSearchCV(estimator=RandomForestClassifier(random_state=42), param_grid=param_grid, cv=3, n_jobs=-1, scoring='roc_auc')

# Fit the GridSearchCV object to the training data
grid_search.fit(X_train_scaled_df, y_train)

# Get the best parameters and the best model
best_params = grid_search.best_params_
best_model = grid_search.best_estimator_

# Predict probabilities on the evaluation data using the best model
y_eval_pred_proba_best = best_model.predict_proba(X_eval_scaled_df)[:, 1]

# Calculate the AUC score for the best model
auc_score_best = roc_auc_score(y_eval, y_eval_pred_proba_best)

auc_score_best, best_params

```

```


# Check if the grid search has completed
if grid_search.best_score_ is not None:
    # Get the best parameters and the best model
    best_params = grid_search.best_params_
    best_model = grid_search.best_estimator_

    # Predict probabilities on the evaluation data using the best model
    y_eval_pred_proba_best = best_model.predict_proba(X_eval_scaled_df)[:, 1]

    # Calculate the AUC score for the best model
    auc_score_best = roc_auc_score(y_eval, y_eval_pred_proba_best)

    auc_score_best, best_params
else:
    "Grid search has not completed yet."
```
