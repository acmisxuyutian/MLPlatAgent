```

import pandas as pd

# Load the train and evaluation datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\06_santander_customer\\split_train.csv'
eval_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\06_santander_customer\\split_eval.csv'

train_df = pd.read_csv(train_path)
eval_df = pd.read_csv(eval_path)

# Display basic information about the datasets
print("Train dataset info:")
print(train_df.info())
print("\nEvaluation dataset info:")
print(eval_df.info())

# Display the first few rows of the datasets
print("\nFirst few rows of the train dataset:")
print(train_df.head())
print("\nFirst few rows of the evaluation dataset:")
print(eval_df.head())
```

```

import matplotlib.pyplot as plt
import seaborn as sns

# Check the distribution of the target variable
plt.figure(figsize=(10, 5))
sns.countplot(data=train_df, x='target')
plt.title('Distribution of Target Variable')
plt.show()

# Check the distribution of feature variables
plt.figure(figsize=(20, 15))
for i, var in enumerate(train_df.columns[1:]):
    plt.subplot(5, 4, i+1)
    sns.histplot(train_df[var], kde=True)
    plt.title(f'Distribution of {var}')
plt.tight_layout()
plt.show()

# Check the correlation matrix
plt.figure(figsize=(10, 8))
sns.heatmap(train_df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```
