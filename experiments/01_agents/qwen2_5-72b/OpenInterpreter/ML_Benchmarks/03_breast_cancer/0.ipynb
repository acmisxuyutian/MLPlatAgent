```

from sklearn.datasets import load_breast_cancer

# Load the dataset
data = load_breast_cancer()

# Display the keys of the dataset
data.keys()
```

```


# Print the shape of the data
print("Data shape:", data.data.shape)

# Print the first few rows of the data
print("Data (first 5 rows):")
print(data.data[:5])

# Print the target names
print("Target names:", data.target_names)

# Print the first few target values
print("Target (first 5 values):")
print(data.target[:5])
```

```


import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Convert the dataset to a DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target

# Pair plot for the first few features
sns.pairplot(df.iloc[:, :5 + 1], hue='target')
plt.show()

# Histogram of the target classes
sns.countplot(x='target', data=df)
plt.xticks([0, 1], data.target_names)
plt.show()

```
