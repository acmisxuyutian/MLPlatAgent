```

from sklearn.datasets import load_breast_cancer

# Load the dataset
data = load_breast_cancer()

# Display the keys of the dataset
data.keys()
```

```


# Display the shape of the data
data_shape = data['data'].shape

# Display the feature names
feature_names = data['feature_names']

# Display the target names
target_names = data['target_names']

data_shape, feature_names, target_names
```

```


import matplotlib.pyplot as plt

# Extract the 'mean radius' feature
mean_radius = data['data'][:, data['feature_names'].tolist().index('mean radius')]

# Create a histogram of 'mean radius'
plt.hist(mean_radius, bins=20, alpha=0.7, color='blue')
plt.xlabel('Mean Radius')
plt.ylabel('Frequency')
plt.title('Histogram of Mean Radius')
plt.show()

```
