```

import pandas as pd

# Load the training and evaluation data
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\07_icr_identify\\split_train.csv'
eval_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\07_icr_identify\\split_eval.csv'

train_df = pd.read_csv(train_data_path)
eval_df = pd.read_csv(eval_data_path)

# Display the first few rows and summary statistics
print("Training Data:")
print(train_df.head())
print(train_df.info())

print("\nEvaluation Data:")
print(eval_df.head())
print(eval_df.info())
```

```


# Handle missing values in the training and evaluation datasets
train_df.fillna(train_df.mean(), inplace=True)
eval_df.fillna(eval_df.mean(), inplace=True)

# Convert the Class column to a categorical type
train_df['Class'] = train_df['Class'].astype('category')
eval_df['Class'] = eval_df['Class'].astype('category')

# Check if any columns are still missing values
print("Training Data Missing Values:")
print(train_df.isnull().sum().sum())

print("\nEvaluation Data Missing Values:")
print(eval_df.isnull().sum().sum())

# Display the first few rows of the cleaned data
print("\nCleaned Training Data:")
print(train_df.head())

print("\nCleaned Evaluation Data:")
print(eval_df.head())

```

```


# Drop the 'Id' column as it is not a feature
train_df.drop(columns=['Id'], inplace=True)
eval_df.drop(columns=['Id'], inplace=True)

# Convert the 'Class' column to a binary target (0 or 1)
train_df['Class'] = train_df['Class'].cat.codes
eval_df['Class'] = eval_df['Class'].cat.codes

# Standardize the numerical features
from sklearn.preprocessing import StandardScaler

# Initialize the StandardScaler
scaler = StandardScaler()

# Fit the scaler on the training data and transform both training and evaluation data
train_df_scaled = pd.DataFrame(scaler.fit_transform(train_df.drop(columns=['Class'])), columns=train_df.drop(columns=['Class']).columns)
eval_df_scaled = pd.DataFrame(scaler.transform(eval_df.drop(columns=['Class'])), columns=eval_df.drop(columns=['Class']).columns)

# Add the 'Class' column back to the scaled data
train_df_scaled['Class'] = train_df['Class']
eval_df_scaled['Class'] = eval_df['Class']

# Display the first few rows of the cleaned and scaled data
print("Cleaned and Scaled Training Data:")
print(train_df_scaled.head())

print("\nCleaned and Scaled Evaluation Data:")
print(eval_df_scaled.head())
```

```



# Drop the 'Id' column as it is not a feature
train_df.drop(columns=['Id'], inplace=True)
eval_df.drop(columns=['Id'], inplace=True)

# Standardize the numerical features
from sklearn.preprocessing import StandardScaler

# Initialize the StandardScaler
scaler = StandardScaler()

# Fit the scaler on the training data and transform both training and evaluation data
train_df_scaled = pd.DataFrame(scaler.fit_transform(train_df.drop(columns=['Class'])), columns=train_df.drop(columns=['Class']).columns)
eval_df_scaled = pd.DataFrame(scaler.transform(eval_df.drop(columns=['Class'])), columns=eval_df.drop(columns=['Class']).columns)

# Add the 'Class' column back to the scaled data
train_df_scaled['Class'] = train_df['Class']
eval_df_scaled['Class'] = eval_df['Class']

# Display the first few rows of the cleaned and scaled data
print("Cleaned and Scaled Training Data:")
print(train_df_scaled.head())

print("\nCleaned and Scaled Evaluation Data:")
print(eval_df_scaled.head())

```

```



# Verify the columns in the DataFrame
print("Training Data Columns:")
print(train_df.columns)

print("\nEvaluation Data Columns:")
print(eval_df.columns)

# Standardize the numerical features
from sklearn.preprocessing import StandardScaler

# Initialize the StandardScaler
scaler = StandardScaler()

# Fit the scaler on the training data and transform both training and evaluation data
train_features = train_df.drop(columns=['Class'])
eval_features = eval_df.drop(columns=['Class'])

train_df_scaled = pd.DataFrame(scaler.fit_transform(train_features), columns=train_features.columns)
eval_df_scaled = pd.DataFrame(scaler.transform(eval_features), columns=eval_features.columns)

# Add the 'Class' column back to the scaled data
train_df_scaled['Class'] = train_df['Class']
eval_df_scaled['Class'] = eval_df['Class']

# Display the first few rows of the cleaned and scaled data
print("\nCleaned and Scaled Training Data:")
print(train_df_scaled.head())

print("\nCleaned and Scaled Evaluation Data:")
print(eval_df_scaled.head())


```

```



# Identify non-numeric columns
non_numeric_columns = train_df.select_dtypes(include=['object']).columns
print("Non-numeric columns in the training data:", non_numeric_columns)

# Encode non-numeric columns using one-hot encoding
train_df_encoded = pd.get_dummies(train_df, columns=non_numeric_columns, drop_first=True)
eval_df_encoded = pd.get_dummies(eval_df, columns=non_numeric_columns, drop_first=True)

# Standardize the numerical features
from sklearn.preprocessing import StandardScaler

# Initialize the StandardScaler
scaler = StandardScaler()

# Fit the scaler on the training data and transform both training and evaluation data
train_features = train_df_encoded.drop(columns=['Class'])
eval_features = eval_df_encoded.drop(columns=['Class'])

train_df_scaled = pd.DataFrame(scaler.fit_transform(train_features), columns=train_features.columns)
eval_df_scaled = pd.DataFrame(scaler.transform(eval_features), columns=eval_features.columns)

# Add the 'Class' column back to the scaled data
train_df_scaled['Class'] = train_df_encoded['Class']
eval_df_scaled['Class'] = eval_df_encoded['Class']

# Display the first few rows of the cleaned and scaled data
print("\nCleaned and Scaled Training Data:")
print(train_df_scaled.head())

print("\nCleaned and Scaled Evaluation Data:")
print(eval_df_scaled.head())


```

```



# Split the training data into training and validation sets
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(
    train_df_scaled.drop(columns=['Class']), 
    train_df_scaled['Class'], 
    test_size=0.2, 
    random_state=42
)

# Train a logistic regression model
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(random_state=42)
model.fit(X_train, y_train)

# Evaluate the model on the validation set
from sklearn.metrics import f1_score

y_val_pred = model.predict(X_val)
val_f1_score = f1_score(y_val, y_val_pred)
print(f"Validation F1 Score: {val_f1_score}")

# Evaluate the model on the evaluation set
y_eval_pred = model.predict(eval_df_scaled.drop(columns=['Class']))
eval_f1_score = f1_score(eval_df_scaled['Class'], y_eval_pred)
print(f"Evaluation F1 Score: {eval_f1_score}")


```

```




# Handle missing values using mean imputation
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='mean')

# Fit the imputer on the training features and transform both training and validation sets
X_train_imputed = imputer.fit_transform(X_train)
X_val_imputed = imputer.transform(X_val)

# Transform the evaluation set
X_eval_imputed = imputer.transform(eval_df_scaled.drop(columns=['Class']))

# Train a logistic regression model
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(random_state=42)
model.fit(X_train_imputed, y_train)

# Evaluate the model on the validation set
from sklearn.metrics import f1_score

y_val_pred = model.predict(X_val_imputed)
val_f1_score = f1_score(y_val, y_val_pred)
print(f"Validation F1 Score: {val_f1_score}")

# Evaluate the model on the evaluation set
y_eval_pred = model.predict(X_eval_imputed)
eval_f1_score = f1_score(eval_df_scaled['Class'], y_eval_pred)
print(f"Evaluation F1 Score: {eval_f1_score}")

```
