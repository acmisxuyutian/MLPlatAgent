```

import pandas as pd

# Load the training and evaluation datasets
train_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\07_icr_identify\\split_train.csv'
eval_data_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\07_icr_identify\\split_eval.csv'

train_df = pd.read_csv(train_data_path)
eval_df = pd.read_csv(eval_data_path)

# Display the first few rows of the training and evaluation datasets
train_df.head(), eval_df.head()
```

```

# Perform initial data analysis
print("Training Data Information:")
print(train_df.info())
print("\nEvaluation Data Information:")
print(eval_df.info())

# Check for missing values
print("\nMissing values in training data:")
print(train_df.isnull().sum())
print("\nMissing values in evaluation data:")
print(eval_df.isnull().sum())

# Summary statistics
print("\nSummary statistics for training data:")
print(train_df.describe(include='all'))
print("\nSummary statistics for evaluation data:")
print(eval_df.describe(include='all'))

# Distribution of the target variable
print("\nDistribution of the target variable in training data:")
print(train_df['Class'].value_counts())
print("\nDistribution of the target variable in evaluation data:")
print(eval_df['Class'].value_counts())
```

```


# Handle missing values in the training and evaluation datasets
from sklearn.impute import SimpleImputer

# Impute missing values with the mean of each column
imputer = SimpleImputer(strategy='mean')

# Fit the imputer on the training data and transform both datasets
train_df_imputed = pd.DataFrame(imputer.fit_transform(train_df.drop(columns=['Id', 'Class'])), columns=train_df.columns.drop(['Id', 'Class']))
eval_df_imputed = pd.DataFrame(imputer.transform(eval_df.drop(columns=['Id', 'Class'])), columns=eval_df.columns.drop(['Id', 'Class']))

# Add the target variable back to the datasets
train_df_imputed['Class'] = train_df['Class']
eval_df_imputed['Class'] = eval_df['Class']

# Check for missing values again
print("Missing values in training data after imputation:")
print(train_df_imputed.isnull().sum())
print("\nMissing values in evaluation data after imputation:")
print(eval_df_imputed.isnull().sum())

```

```


# Identify non-numeric columns
non_numeric_columns = train_df.select_dtypes(exclude=['number']).columns
non_numeric_columns
```

```


# Drop the 'Id' column
train_df = train_df.drop(columns=['Id'])
eval_df = eval_df.drop(columns=['Id'])

# Check the unique values in the 'EJ' column to decide how to handle it
print("Unique values in 'EJ' column in training data:")
print(train_df['EJ'].unique())

print("\nUnique values in 'EJ' column in evaluation data:")
print(eval_df['EJ'].unique())

# Drop the 'EJ' column if it's not relevant or encode it if it is
# For now, let's drop it
train_df = train_df.drop(columns=['EJ'])
eval_df = eval_df.drop(columns=['EJ'])

# Now, handle missing values in the numeric columns
imputer = SimpleImputer(strategy='mean')
train_df_imputed = pd.DataFrame(imputer.fit_transform(train_df.drop(columns=['Class'])), columns=train_df.columns.drop(['Class']))
eval_df_imputed = pd.DataFrame(imputer.transform(eval_df.drop(columns=['Class'])), columns=eval_df.columns.drop(['Class']))

# Add the target variable back to the datasets
train_df_imputed['Class'] = train_df['Class']
eval_df_imputed['Class'] = eval_df['Class']

# Check for missing values again
print("\nMissing values in training data after imputation:")
print(train_df_imputed.isnull().sum())
print("\nMissing values in evaluation data after imputation:")
print(eval_df_imputed.isnull().sum())

```

```



# Scale the features
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Fit the scaler on the training data and transform both datasets
train_df_scaled = pd.DataFrame(scaler.fit_transform(train_df_imputed.drop(columns=['Class'])), columns=train_df_imputed.columns.drop(['Class']))
eval_df_scaled = pd.DataFrame(scaler.transform(eval_df_imputed.drop(columns=['Class'])), columns=eval_df_imputed.columns.drop(['Class']))

# Add the target variable back to the datasets
train_df_scaled['Class'] = train_df_imputed['Class']
eval_df_scaled['Class'] = eval_df_imputed['Class']

# Display the first few rows of the scaled datasets
train_df_scaled.head(), eval_df_scaled.head()

```

```


from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score

# Split the data into features and target
X_train = train_df_scaled.drop(columns=['Class'])
y_train = train_df_scaled['Class']

X_eval = eval_df_scaled.drop(columns=['Class'])
y_eval = eval_df_scaled['Class']

# Train a Logistic Regression model
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

# Predict on the evaluation data
y_pred = model.predict(X_eval)

# Calculate the F1 score
f1 = f1_score(y_eval, y_pred)
f1

```

```



from sklearn.ensemble import RandomForestClassifier

# Train a Random Forest model
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Predict on the evaluation data
y_pred_rf = rf_model.predict(X_eval)

# Calculate the F1 score
f1_rf = f1_score(y_eval, y_pred_rf)
f1_rf


```

```




from sklearn.ensemble import GradientBoostingClassifier

# Train a Gradient Boosting model
gb_model = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, random_state=42)
gb_model.fit(X_train, y_train)

# Predict on the evaluation data
y_pred_gb = gb_model.predict(X_eval)

# Calculate the F1 score
f1_gb = f1_score(y_eval, y_pred_gb)
f1_gb



```

```



from sklearn.model_selection import GridSearchCV

# Define the parameter grid for GridSearchCV
param_grid = {
    'n_estimators': [50, 100, 200],
    'learning_rate': [0.01, 0.1, 0.2],
    'max_depth': [3, 4, 5]
}

# Initialize GridSearchCV with the Gradient Boosting model
grid_search = GridSearchCV(GradientBoostingClassifier(random_state=42), param_grid, scoring='f1', cv=5, n_jobs=-1)

# Fit the GridSearchCV to the training data
grid_search.fit(X_train, y_train)

# Get the best parameters and the best model
best_params = grid_search.best_params_
best_model = grid_search.best_estimator_

# Predict on the evaluation data using the best model
y_pred_best = best_model.predict(X_eval)

# Calculate the F1 score
f1_best = f1_score(y_eval, y_pred_best)
f1_best, best_params


```

```




# Get the feature importances from the best model
feature_importances = best_model.feature_importances_

# Create a DataFrame to display the feature importances
feature_importance_df = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': feature_importances
})

# Sort the DataFrame by importance in descending order
feature_importance_df = feature_importance_df.sort_values(by='Importance', ascending=False)

# Display the top 10 most important features
feature_importance_df.head(10)


```

```




from sklearn.metrics import precision_score, recall_score, confusion_matrix, classification_report

# Calculate precision, recall, and the confusion matrix
precision = precision_score(y_eval, y_pred_best)
recall = recall_score(y_eval, y_pred_best)
conf_matrix = confusion_matrix(y_eval, y_pred_best)
class_report = classification_report(y_eval, y_pred_best)

precision, recall, conf_matrix, class_report



```
