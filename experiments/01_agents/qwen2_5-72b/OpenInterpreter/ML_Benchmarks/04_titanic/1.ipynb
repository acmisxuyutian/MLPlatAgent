```

# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets
train_df = pd.read_csv(r'D:\PythonCodes\benchmark/ml_benchmark/04_titanic/split_train.csv')
eval_df = pd.read_csv(r'D:\PythonCodes\benchmark/ml_benchmark/04_titanic/split_eval.csv')

# Display basic information about the datasets
print("Training Data Information:")
train_df.info()

print("\nEvaluation Data Information:")
eval_df.info()

# Display summary statistics
print("\nTraining Data Summary Statistics:")
train_df.describe()

print("\nEvaluation Data Summary Statistics:")
eval_df.describe()

# Check for missing values
print("\nMissing Values in Training Data:")
print(train_df.isnull().sum())

print("\nMissing Values in Evaluation Data:")
print(eval_df.isnull().sum())
```

```


# Handle missing values in the training data
# Impute Age with the median
train_df['Age'].fillna(train_df['Age'].median(), inplace=True)
eval_df['Age'].fillna(eval_df['Age'].median(), inplace=True)

# Impute Cabin with 'Unknown'
train_df['Cabin'].fillna('Unknown', inplace=True)
eval_df['Cabin'].fillna('Unknown', inplace=True)

# Impute Embarked with the most common value
most_common_embarked = train_df['Embarked'].mode()[0]
train_df['Embarked'].fillna(most_common_embarked, inplace=True)

# Check for remaining missing values
print("Remaining missing values in Training Data:")
print(train_df.isnull().sum())

print("\nRemaining missing values in Evaluation Data:")
print(eval_df.isnull().sum())

```

```


# Verify handling of missing values
print("Remaining missing values in Training Data:")
print(train_df.isnull().sum())

print("\nRemaining missing values in Evaluation Data:")
print(eval_df.isnull().sum())

```

```


# Feature Engineering
# Create a new feature 'FamilySize' from 'SibSp' and 'Parch'
train_df['FamilySize'] = train_df['SibSp'] + train_df['Parch'] + 1
eval_df['FamilySize'] = eval_df['SibSp'] + eval_df['Parch'] + 1

# Extract the first letter of 'Cabin' to create a simpler categorical feature
train_df['Cabin'] = train_df['Cabin'].apply(lambda x: x[0] if x != 'Unknown' else 'U')
eval_df['Cabin'] = eval_df['Cabin'].apply(lambda x: x[0] if x != 'Unknown' else 'U')

# Convert categorical variables to numerical using one-hot encoding
train_df = pd.get_dummies(train_df, columns=['Sex', 'Embarked', 'Cabin'])
eval_df = pd.get_dummies(eval_df, columns=['Sex', 'Embarked', 'Cabin'])

# Drop unnecessary columns
train_df.drop(['Name', 'Ticket', 'Cabin_U'], axis=1, inplace=True)
eval_df.drop(['Name', 'Ticket', 'Cabin_U'], axis=1, inplace=True)

# Display the first few rows of the modified datasets
train_df.head(), eval_df.head()

```

```



# Import necessary libraries
from sklearn.preprocessing import StandardScaler

# Define the features and target
X_train = train_df.drop(['Survived', 'PassengerId'], axis=1)
y_train = train_df['Survived']

X_eval = eval_df.drop(['Survived', 'PassengerId'], axis=1)
y_eval = eval_df['Survived']

# Scale numerical features
scaler = StandardScaler()
X_train[['Age', 'Fare', 'FamilySize']] = scaler.fit_transform(X_train[['Age', 'Fare', 'FamilySize']])
X_eval[['Age', 'Fare', 'FamilySize']] = scaler.transform(X_eval[['Age', 'Fare', 'FamilySize']])

# Display the first few rows of the scaled datasets
X_train.head(), X_eval.head()


```

```



# Import necessary libraries
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Train a Logistic Regression model
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

# Make predictions on the evaluation data
y_pred = model.predict(X_eval)

# Calculate accuracy
accuracy = accuracy_score(y_eval, y_pred)
accuracy


```

```



# Align feature sets
# Get the set of features in the training data
train_features = set(X_train.columns)

# Get the set of features in the evaluation data
eval_features = set(X_eval.columns)

# Find the missing features in the evaluation data
missing_features = train_features - eval_features

# Add the missing features to the evaluation data and set their values to 0
for feature in missing_features:
    X_eval[feature] = 0

# Reorder the columns of the evaluation data to match the training data
X_eval = X_eval[X_train.columns]

# Verify the alignment
X_train.head(), X_eval.head()


```
