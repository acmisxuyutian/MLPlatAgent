```

import pandas as pd

# Load the training and evaluation datasets
train_path = 'D:\\PythonCodes\\benchmark\\ml_benchmark\\04_titanic\\split_train.csv'
eval_path = 'D:\\PythonCodes\\benchmark\\ml_benchmark\\04_titanic\\split_eval.csv'

train_df = pd.read_csv(train_path)
eval_df = pd.read_csv(eval_path)

# Display the first few rows of the training dataset
print("Training dataset head:")
print(train_df.head())

# Display the first few rows of the evaluation dataset
print("\nEvaluation dataset head:")
print(eval_df.head())

# Display basic information about the training dataset
print("\nTraining dataset info:")
train_df.info()

# Display basic information about the evaluation dataset
print("\nEvaluation dataset info:")
eval_df.info()

# Display summary statistics of the training dataset
print("\nTraining dataset description:")
print(train_df.describe())
```

```

# Handling missing values in the training dataset
train_df['Age'].fillna(train_df['Age'].median(), inplace=True)
train_df['Cabin'].fillna('Unknown', inplace=True)
train_df['Embarked'].fillna(train_df['Embarked'].mode()[0], inplace=True)

# Handling missing values in the evaluation dataset
eval_df['Age'].fillna(eval_df['Age'].median(), inplace=True)
eval_df['Cabin'].fillna('Unknown', inplace=True)

# Display the first few rows of the training dataset after handling missing values
print("Training dataset head after handling missing values:")
print(train_df.head())

# Display the first few rows of the evaluation dataset after handling missing values
print("\nEvaluation dataset head after handling missing values:")
print(eval_df.head())
```

```


# Verify the first few rows of the training dataset
print("Training dataset head after handling missing values:")
print(train_df.head())

# Verify the first few rows of the evaluation dataset
print("\nEvaluation dataset head after handling missing values:")
print(eval_df.head())

# Check for missing values in the training dataset
print("\nMissing values in the training dataset:")
print(train_df.isnull().sum())

# Check for missing values in the evaluation dataset
print("\nMissing values in the evaluation dataset:")
print(eval_df.isnull().sum())

```

```


import re

# Extract titles from the Name column
def extract_title(name):
    title_search = re.search(' ([A-Za-z]+)\.', name)
    if title_search:
        return title_search.group(1)
    return ""

# Apply the function to the training and evaluation datasets
train_df['Title'] = train_df['Name'].apply(extract_title)
eval_df['Title'] = eval_df['Name'].apply(extract_title)

# Create a FamilySize feature
train_df['FamilySize'] = train_df['SibSp'] + train_df['Parch'] + 1
eval_df['FamilySize'] = eval_df['SibSp'] + eval_df['Parch'] + 1

# Display the first few rows of the training dataset after feature engineering
print("Training dataset head after feature engineering:")
print(train_df.head())

# Display the first few rows of the evaluation dataset after feature engineering
print("\nEvaluation dataset head after feature engineering:")
print(eval_df.head())

```

```



# Group rare titles into a single category 'Rare'
def group_titles(title):
    if title in ['Mr', 'Miss', 'Mrs', 'Master']:
        return title
    else:
        return 'Rare'

# Apply the function to the training and evaluation datasets
train_df['Title'] = train_df['Title'].apply(group_titles)
eval_df['Title'] = eval_df['Title'].apply(group_titles)

# Display the value counts of the Title column to verify the grouping
print("Title value counts in the training dataset:")
print(train_df['Title'].value_counts())

print("\nTitle value counts in the evaluation dataset:")
print(eval_df['Title'].value_counts())

# Display the first few rows of the training dataset after grouping titles
print("\nTraining dataset head after grouping titles:")
print(train_df.head())

# Display the first few rows of the evaluation dataset after grouping titles
print("\nEvaluation dataset head after grouping titles:")
print(eval_df.head())
```

```



# One-hot encode categorical variables
train_df = pd.get_dummies(train_df, columns=['Sex', 'Embarked', 'Title', 'Cabin'], drop_first=True)
eval_df = pd.get_dummies(eval_df, columns=['Sex', 'Embarked', 'Title', 'Cabin'], drop_first=True)

# Display the first few rows of the training dataset after one-hot encoding
print("Training dataset head after one-hot encoding:")
print(train_df.head())

# Display the first few rows of the evaluation dataset after one-hot encoding
print("\nEvaluation dataset head after one-hot encoding:")
print(eval_df.head())
```

```



from sklearn.preprocessing import StandardScaler

# List of numerical features to normalize
numerical_features = ['Age', 'Fare']

# Initialize the StandardScaler
scaler = StandardScaler()

# Fit and transform the training dataset
train_df[numerical_features] = scaler.fit_transform(train_df[numerical_features])

# Transform the evaluation dataset
eval_df[numerical_features] = scaler.transform(eval_df[numerical_features])

# Display the first few rows of the training dataset after normalization
print("Training dataset head after normalization:")
print(train_df.head())

# Display the first few rows of the evaluation dataset after normalization
print("\nEvaluation dataset head after normalization:")
print(eval_df.head())

# Ensure consistent columns in both datasets
train_columns = set(train_df.columns)
eval_columns = set(eval_df.columns)

# Find columns in the training dataset that are not in the evaluation dataset
missing_cols = train_columns - eval_columns

# Add missing columns to the evaluation dataset with 0 values
for col in missing_cols:
    eval_df[col] = 0

# Reorder the columns in the evaluation dataset to match the training dataset
eval_df = eval_df[train_df.columns]

# Display the first few rows of the evaluation dataset after ensuring consistent columns
print("\nEvaluation dataset head after ensuring consistent columns:")
print(eval_df.head())

```

```



from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Separate features and target variable
X_train = train_df.drop(columns=['Survived', 'Name', 'Ticket'])
y_train = train_df['Survived']

X_eval = eval_df.drop(columns=['Survived', 'Name', 'Ticket'])
y_eval = eval_df['Survived']

# Initialize the Logistic Regression model
model = LogisticRegression(max_iter=1000)

# Train the model
model.fit(X_train, y_train)

# Predict on the evaluation dataset
y_pred = model.predict(X_eval)

# Calculate the accuracy
accuracy = accuracy_score(y_eval, y_pred)

# Print the accuracy
print(f"Accuracy of the model on the evaluation dataset: {accuracy:.2f}")

```

```



# Initialize the Logistic Regression model with increased max_iter and a different solver
model = LogisticRegression(max_iter=2000, solver='saga')

# Train the model
model.fit(X_train, y_train)

# Predict on the evaluation dataset
y_pred = model.predict(X_eval)

# Calculate the accuracy
accuracy = accuracy_score(y_eval, y_pred)

# Print the accuracy
print(f"Accuracy of the model on the evaluation dataset: {accuracy:.2f}")
```

```




from sklearn.ensemble import RandomForestClassifier

# Initialize the Random Forest Classifier
model = RandomForestClassifier(n_estimators=100, random_state=42)

# Train the model
model.fit(X_train, y_train)

# Predict on the evaluation dataset
y_pred = model.predict(X_eval)

# Calculate the accuracy
accuracy = accuracy_score(y_eval, y_pred)

# Print the accuracy
print(f"Accuracy of the model on the evaluation dataset: {accuracy:.2f}")

# Calculate additional metrics
from sklearn.metrics import precision_score, recall_score, f1_score

precision = precision_score(y_eval, y_pred)
recall = recall_score(y_eval, y_pred)
f1 = f1_score(y_eval, y_pred)

print(f"Precision: {precision:.2f}")
print(f"Recall: {recall:.2f}")
print(f"F1 Score: {f1:.2f}")

```

```




from sklearn.model_selection import GridSearchCV

# Define the parameter grid
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# Initialize the Random Forest Classifier
rf = RandomForestClassifier(random_state=42)

# Initialize GridSearchCV
grid_search = GridSearchCV(estimator=rf, param_grid=param_grid, cv=5, n_jobs=-1, scoring='accuracy')

# Train the model
grid_search.fit(X_train, y_train)

# Get the best parameters
best_params = grid_search.best_params_

# Evaluate the best model on the evaluation dataset
best_model = grid_search.best_estimator_
y_pred = best_model.predict(X_eval)

# Calculate the accuracy
accuracy = accuracy_score(y_eval, y_pred)

# Print the best parameters and the accuracy
print(f"Best Parameters: {best_params}")
print(f"Accuracy of the best model on the evaluation dataset: {accuracy:.2f}")

# Calculate additional metrics
precision = precision_score(y_eval, y_pred)
recall = recall_score(y_eval, y_pred)
f1 = f1_score(y_eval, y_pred)

print(f"Precision: {precision:.2f}")
print(f"Recall: {recall:.2f}")
print(f"F1 Score: {f1:.2f}")

# Feature importance
feature_importances = best_model.feature_importances_
feature_importance_df = pd.DataFrame({'Feature': X_train.columns, 'Importance': feature_importances})
feature_importance_df = feature_importance_df.sort_values(by='Importance', ascending=False)

print("\nFeature Importance:")
print(feature_importance_df)
```

```




import joblib

# Save the best model to a file
model_filename = 'titanic_survival_model.pkl'
joblib.dump(best_model, model_filename)

print(f"Model saved to {model_filename}")

# Load the model from the file
loaded_model = joblib.load(model_filename)

# Verify the loaded model
y_pred_loaded = loaded_model.predict(X_eval)
accuracy_loaded = accuracy_score(y_eval, y_pred_loaded)

print(f"Accuracy of the loaded model on the evaluation dataset: {accuracy_loaded:.2f}")
```
