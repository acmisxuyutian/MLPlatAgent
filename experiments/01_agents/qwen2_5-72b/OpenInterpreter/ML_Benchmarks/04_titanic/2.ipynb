```


import pandas as pd
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split

# Load the training and evaluation datasets
train_path = 'D:\\PythonCodes\\benchmark\\ml_benchmark\\04_titanic\\split_train.csv'
eval_path = 'D:\\PythonCodes\\benchmark\\ml_benchmark\\04_titanic\\split_eval.csv'

train_df = pd.read_csv(train_path)
eval_df = pd.read_csv(eval_path)

# Extract deck from Cabin
train_df['Cabin'] = train_df['Cabin'].apply(lambda x: x[0] if pd.notnull(x) else 'Unknown')
eval_df['Cabin'] = eval_df['Cabin'].apply(lambda x: x[0] if pd.notnull(x) else 'Unknown')

# Separate features and target
X_train = train_df.drop(columns=['PassengerId', 'Survived', 'Name', 'Ticket'])
y_train = train_df['Survived']
X_eval = eval_df.drop(columns=['PassengerId', 'Survived', 'Name', 'Ticket'])
y_eval = eval_df['Survived']

# Define preprocessing for numerical and categorical data
numeric_features = ['Age', 'SibSp', 'Parch', 'Fare']
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())])

categorical_features = ['Sex', 'Embarked', 'Cabin']
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='Unknown')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)])

# Preprocess the training and evaluation data
X_train_preprocessed = preprocessor.fit_transform(X_train)
X_eval_preprocessed = preprocessor.transform(X_eval)

# Convert preprocessed data to DataFrame for easier inspection
X_train_preprocessed_df = pd.DataFrame(X_train_preprocessed, columns=preprocessor.get_feature_names_out())
X_eval_preprocessed_df = pd.DataFrame(X_eval_preprocessed, columns=preprocessor.get_feature_names_out())

# Display the first few rows of the preprocessed training data
print("Preprocessed Training Data:")
print(X_train_preprocessed_df.head())

# Display the first few rows of the preprocessed evaluation data
print("\nPreprocessed Evaluation Data:")
print(X_eval_preprocessed_df.head())

```

```


from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Train a logistic regression model
model = LogisticRegression(max_iter=1000)
model.fit(X_train_preprocessed, y_train)

# Make predictions on the evaluation data
y_pred = model.predict(X_eval_preprocessed)

# Calculate and report the accuracy
accuracy = accuracy_score(y_eval, y_pred)
print(f"Accuracy on the evaluation data: {accuracy:.4f}")
```

```



from sklearn.ensemble import RandomForestClassifier

# Train a Random Forest model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train_preprocessed, y_train)

# Make predictions on the evaluation data
y_pred = model.predict(X_eval_preprocessed)

# Calculate and report the accuracy
accuracy = accuracy_score(y_eval, y_pred)
print(f"Accuracy on the evaluation data using Random Forest: {accuracy:.4f}")

```

```


from sklearn.model_selection import GridSearchCV

# Define the parameter grid for the Random Forest model
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# Create a Random Forest model
rf_model = RandomForestClassifier(random_state=42)

# Set up the grid search
grid_search = GridSearchCV(estimator=rf_model, param_grid=param_grid, cv=5, n_jobs=-1, scoring='accuracy')

# Fit the grid search to the data
grid_search.fit(X_train_preprocessed, y_train)

# Get the best parameters and the best model
best_params = grid_search.best_params_
best_model = grid_search.best_estimator_

# Make predictions on the evaluation data using the best model
y_pred = best_model.predict(X_eval_preprocessed)

# Calculate and report the accuracy
accuracy = accuracy_score(y_eval, y_pred)
print(f"Best Hyperparameters: {best_params}")
print(f"Accuracy on the evaluation data using the best model: {accuracy:.4f}")

```
