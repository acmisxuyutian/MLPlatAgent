```

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load the train and eval datasets
train_path = r'D:\PythonCodes\benchmark/ml_benchmark/04_titanic/split_train.csv'
eval_path = r'D:\PythonCodes\benchmark/ml_benchmark/04_titanic/split_eval.csv'

train_df = pd.read_csv(train_path)
eval_df = pd.read_csv(eval_path)

# Display basic statistics
print("Train Data Summary:")
print(train_df.describe())

print("\nEval Data Summary:")
print(eval_df.describe())

# Check for missing values
print("\nMissing Values in Train Data:")
print(train_df.isnull().sum())

print("\nMissing Values in Eval Data:")
print(eval_df.isnull().sum())

# Visualize the distribution of the target variable
plt.figure(figsize=(8, 6))
sns.countplot(x='Survived', data=train_df)
plt.title('Survival Count in Train Data')
plt.show()
```
