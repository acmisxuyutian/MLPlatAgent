```

import pandas as pd
import numpy as np

# Load the datasets
train_path = 'D:\\PythonCodes\\benchmark\\ml_benchmark\\04_titanic\\split_train.csv'
eval_path = 'D:\\PythonCodes\\benchmark\\ml_benchmark\\04_titanic\\split_eval.csv'

train_data = pd.read_csv(train_path)
eval_data = pd.read_csv(eval_path)

# Display the first few rows of the training data
print(train_data.head())

# Check for missing values
print(train_data.isnull().sum())

# Handle missing values
# For Age, we can fill with the median
train_data['Age'].fillna(train_data['Age'].median(), inplace=True)
eval_data['Age'].fillna(eval_data['Age'].median(), inplace=True)

# For Cabin, we can fill with 'Unknown'
train_data['Cabin'].fillna('Unknown', inplace=True)
eval_data['Cabin'].fillna('Unknown', inplace=True)

# For Embarked, we can fill with the mode
train_data['Embarked'].fillna(train_data['Embarked'].mode()[0], inplace=True)
eval_data['Embarked'].fillna(eval_data['Embarked'].mode()[0], inplace=True)

# Convert categorical variables to numerical values
train_data['Sex'] = train_data['Sex'].map({'male': 0, 'female': 1})
eval_data['Sex'] = eval_data['Sex'].map({'male': 0, 'female': 1})

train_data['Embarked'] = train_data['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})
eval_data['Embarked'] = eval_data['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})

# Drop non-numeric columns that won't be used for modeling
train_data = train_data.drop(['Name', 'Ticket', 'Cabin'], axis=1)
eval_data = eval_data.drop(['Name', 'Ticket', 'Cabin'], axis=1)

# Display the first few rows of the preprocessed training data
print(train_data.head())
```

```


# Display the first few rows of the preprocessed training data
print(train_data.head())

# Check for missing values in the preprocessed training data
print(train_data.isnull().sum())

# Display the first few rows of the preprocessed evaluation data
print(eval_data.head())

# Check for missing values in the preprocessed evaluation data
print(eval_data.isnull().sum())

```

```


import matplotlib.pyplot as plt
import seaborn as sns

# Set the style for seaborn plots
sns.set(style="whitegrid")

# Plot distributions of numerical features
plt.figure(figsize=(12, 6))

# Age distribution
plt.subplot(2, 2, 1)
sns.histplot(train_data['Age'], kde=True)
plt.title('Age Distribution')

# Fare distribution
plt.subplot(2, 2, 2)
sns.histplot(train_data['Fare'], kde=True)
plt.title('Fare Distribution')

# SibSp distribution
plt.subplot(2, 2, 3)
sns.countplot(x='SibSp', data=train_data)
plt.title('SibSp Distribution')

# Parch distribution
plt.subplot(2, 2, 4)
sns.countplot(x='Parch', data=train_data)
plt.title('Parch Distribution')

plt.tight_layout()
plt.show()

# Correlation matrix
correlation_matrix = train_data.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()

```
