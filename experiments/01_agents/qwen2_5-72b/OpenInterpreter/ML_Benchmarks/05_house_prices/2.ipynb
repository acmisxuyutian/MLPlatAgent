```

import pandas as pd
import numpy as np

# Load the train and eval datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\05_house_prices\\split_train.csv'
eval_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\05_house_prices\\split_eval.csv'

train_df = pd.read_csv(train_path)
eval_df = pd.read_csv(eval_path)

# Display the first few rows of the train dataset
train_df.head()
```

```


# Summary statistics for the train dataset
train_summary = train_df.describe(include='all')

# Check for missing values in the train dataset
missing_values = train_df.isnull().sum().sort_values(ascending=False)

# Display the summary statistics and missing values
train_summary, missing_values[missing_values > 0]
```

```


import matplotlib.pyplot as plt
import seaborn as sns

# Handle missing values
# For simplicity, let's fill numerical missing values with the median and categorical missing values with 'None'
train_df.fillna(train_df.median(), inplace=True)
train_df.fillna('None', inplace=True)

# Check the distribution of the target variable 'SalePrice'
plt.figure(figsize=(10, 6))
sns.histplot(train_df['SalePrice'], kde=True)
plt.title('Distribution of SalePrice')
plt.xlabel('SalePrice')
plt.ylabel('Frequency')
plt.show()

# Check the distribution of the target variable's log
plt.figure(figsize=(10, 6))
sns.histplot(np.log(train_df['SalePrice']), kde=True)
plt.title('Distribution of Log(SalePrice)')
plt.xlabel('Log(SalePrice)')
plt.ylabel('Frequency')
plt.show()

```

```



# Separate numerical and categorical columns
numerical_cols = train_df.select_dtypes(include=['number']).columns
categorical_cols = train_df.select_dtypes(include=['object']).columns

# Fill missing values in numerical columns with the median
train_df[numerical_cols] = train_df[numerical_cols].fillna(train_df[numerical_cols].median())

# Fill missing values in categorical columns with 'None'
train_df[categorical_cols] = train_df[categorical_cols].fillna('None')

# Check the distribution of the target variable 'SalePrice'
plt.figure(figsize=(10, 6))
sns.histplot(train_df['SalePrice'], kde=True)
plt.title('Distribution of SalePrice')
plt.xlabel('SalePrice')
plt.ylabel('Frequency')
plt.show()

# Check the distribution of the target variable's log
plt.figure(figsize=(10, 6))
sns.histplot(np.log(train_df['SalePrice']), kde=True)
plt.title('Distribution of Log(SalePrice)')
plt.xlabel('Log(SalePrice)')
plt.ylabel('Frequency')
plt.show()

```
