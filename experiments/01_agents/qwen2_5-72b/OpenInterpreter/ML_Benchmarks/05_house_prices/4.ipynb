```


import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the train and evaluation datasets
train_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\05_house_prices\\split_train.csv'
eval_path = 'D:\\LLM-Driven_AI-Studio\\MLAgent\\data\\benchmark\\ml_benchmark\\05_house_prices\\split_eval.csv'

train_df = pd.read_csv(train_path)
eval_df = pd.read_csv(eval_path)

# Display basic statistics
print("Train Data Statistics:")
print(train_df.describe())

print("\nEvaluation Data Statistics:")
print(eval_df.describe())

# Check for missing values
print("\nMissing Values in Train Data:")
print(train_df.isnull().sum())

print("\nMissing Values in Evaluation Data:")
print(eval_df.isnull().sum())

# Visualize the distribution of SalePrice
plt.figure(figsize=(10, 6))
sns.histplot(train_df['SalePrice'], kde=True)
plt.title('Distribution of SalePrice')
plt.xlabel('SalePrice')
plt.ylabel('Frequency')
plt.show()

```
