```

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine

# Load the Wine dataset
wine = load_wine()

# Convert the dataset to a DataFrame for easier manipulation
wine_df = pd.DataFrame(data=np.c_[wine['data'], wine['target']], 
                       columns=wine['feature_names'] + ['target'])

# Display the first few rows of the dataset
wine_df.head()
```

```


# Summarize the data
summary_stats = wine_df.describe()
summary_stats

# Plot the data to visualize relationships
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
wine_df.hist(column='alcohol', bins=20, edgecolor='black')
plt.title('Alcohol Content Distribution')

plt.subplot(1, 2, 2)
wine_df.boxplot(column='alcohol', by='target', grid=False)
plt.title('Alcohol Content by Wine Class')
plt.suptitle('')

plt.tight_layout()
plt.show()
```
